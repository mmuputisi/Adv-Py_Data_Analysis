```python
# standalone_optimal_decay_model.py
"""
Optimal Decay Rate Model Analysis
Finds optimal exponential decay rates for ESG effects over time
"""

# ============================================================================
# SETUP AND IMPORTS
# ============================================================================
from google.colab import drive
drive.mount('/content/drive')

!pip install -q linearmodels pandas numpy matplotlib seaborn scipy statsmodels

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from linearmodels.panel import PanelOLS
from scipy import stats
from scipy.optimize import minimize_scalar
import warnings
warnings.filterwarnings('ignore')

# ============================================================================
# CONFIGURATION
# ============================================================================
class Config:
    """Configuration for optimal decay rate analysis"""
    PROJECT_DIR = '/content/drive/MyDrive/Colab Notebooks/FTSE Data'
    DATA_PATH = os.path.join(PROJECT_DIR, 'FTSE data.xlsx')
    OUTPUT_DIR = os.path.join(PROJECT_DIR, 'optimal_decay_outputs')

    # Consumer Staples firms
    CONSUMER_STAPLES_FIRMS = [
        'Associated British Foods', 'British American Tobacco', 'Coca Cola HBC AG',
        'Diageo', 'J. Sainsbury', 'Reckitt Benckiser', 'Tesco', 'Unilever', 'Haleon Plc'
    ]

    # Decay analysis settings
    MAX_LAGS = 5  # Maximum lags to consider
    DECAY_GRID = np.linspace(0.1, 0.9, 9)  # Grid for decay rate search
    OPTIMIZATION_METHOD = 'bounded'  # 'grid' or 'bounded' optimization

    # Model settings
    CONTROL_VARS = ['ROA', 'Size', 'Leverage']
    USE_LAGGED_CONTROLS = False  # Use current controls as in original

    # Output settings
    CREATE_PLOTS = True
    SAVE_RESULTS = True
    VERBOSE = True

# ============================================================================
# DATA LOADER
# ============================================================================
class DecayDataLoader:
    """Load data for decay rate analysis"""

    def __init__(self, config):
        self.config = config
        self.df = None
        self.df_cs = None
        self.esg_means = {}

    def load_and_prepare_data(self):
        """Load data and prepare for decay analysis"""
        print("Loading data for decay rate analysis...")

        # Load data
        self.df = pd.read_excel(self.config.DATA_PATH)
        self.df = self.df.set_index(['Firm', 'Year']).sort_index()
        self.df = self.df.rename(columns={"Tobin's Q": "Tobin_Q"})

        # Basic transformations
        self.df['Size'] = np.log(self.df['Total Assets'])
        self.df['Leverage'] = self.df['Total Liabilities'] / self.df['Total Assets']
        self.df['Tobin_Q_log'] = np.log(self.df['Tobin_Q'] + 0.001)

        # Create lagged ESG variables
        print("Creating lagged ESG variables...")
        for lag in range(1, self.config.MAX_LAGS + 1):
            self.df[f'E_lag{lag}'] = self.df.groupby(level=0)['E'].shift(lag)
            self.df[f'S_lag{lag}'] = self.df.groupby(level=0)['S'].shift(lag)
            self.df[f'G_lag{lag}'] = self.df.groupby(level=0)['G'].shift(lag)

        # Also create simple lagged variables
        self.df['E_lag'] = self.df.groupby(level=0)['E'].shift(1)
        self.df['S_lag'] = self.df.groupby(level=0)['S'].shift(1)
        self.df['G_lag'] = self.df.groupby(level=0)['G'].shift(1)

        # Create control variables
        if self.config.USE_LAGGED_CONTROLS:
            self.df['ROA_lag'] = self.df.groupby(level=0)['ROA'].shift(1)
            self.df['Size_lag'] = self.df.groupby(level=0)['Size'].shift(1)
            self.df['Leverage_lag'] = self.df.groupby(level=0)['Leverage'].shift(1)
        else:
            # Use current controls as in original
            pass

        # Create sector variable
        sectors = []
        for firm in self.df.index.get_level_values(0):
            if firm in self.config.CONSUMER_STAPLES_FIRMS:
                sectors.append('Consumer Staples')
            else:
                sectors.append('Other')

        self.df['Sector'] = sectors
        self.df['ConsumerStaples'] = (self.df['Sector'] == 'Consumer Staples').astype(int)

        # Create Consumer Staples dataset
        self.df_cs = self.df[self.df['ConsumerStaples'] == 1].copy()

        # Store means for interpretation
        for component in ['E', 'S', 'G']:
            self.esg_means[component] = self.df_cs[f'{component}_lag'].mean()

        print(f"✓ Loaded {len(self.df)} total observations")
        print(f"✓ Consumer Staples: {len(self.df_cs)} observations")
        print(f"✓ Created ESG lags 1-{self.config.MAX_LAGS}")

        return self.df, self.df_cs

    def create_decaying_average_vectorized(self, df, component, decay_rate, max_lag=3, start_from_lag1=True):
        """
        Create exponentially decaying average for an ESG component

        Parameters:
        -----------
        df : DataFrame
            Input data
        component : str
            'E', 'S', or 'G'
        decay_rate : float
            Decay rate (0 < δ ≤ 1)
        max_lag : int
            Maximum lag to consider
        start_from_lag1 : bool
            If True, start from t-1 (correct for causality)
            If False, include t-0 (original, problematic)
        """
        df_copy = df.copy()
        decay_col = f'{component}_decay{decay_rate:.2f}'

        # Get all lagged variables
        lag_data = {}
        for lag in range(1, max_lag + 1):
            lag_var = f'{component}_lag{lag}'
            if lag_var in df_copy.columns:
                lag_data[lag] = df_copy[lag_var].values

        # Calculate weights - adjust based on start_from_lag1
        if start_from_lag1:
            # CORRECT: Start from t-1 (lag 1)
            weights = np.array([decay_rate ** (lag-1) for lag in range(1, max_lag + 1)])
            lags_used = list(range(1, max_lag + 1))
            weight_labels = [f't-{lag}' for lag in range(1, max_lag + 1)]
        else:
            # ORIGINAL (problematic): Include t-0
            weights = np.array([decay_rate ** lag for lag in range(max_lag + 1)])
            lags_used = list(range(max_lag + 1))
            weight_labels = [f't-{lag}' for lag in range(max_lag + 1)]

        weights = weights / weights.sum()  # Normalize

        # Initialize result array
        n_obs = len(df_copy)
        result = np.full(n_obs, np.nan)

        # For each observation, calculate weighted average
        for i in range(n_obs):
            valid_weights = []
            values = []

            for idx, lag in enumerate(lags_used):
                if lag == 0:
                    # This case only happens when start_from_lag1=False
                    if f'{component}' in df_copy.columns:
                        val = df_copy.iloc[i][component]
                        if not pd.isna(val):
                            values.append(val)
                            valid_weights.append(weights[idx])
                else:
                    # Lagged values (t-1, t-2, etc.)
                    if lag in lag_data and not pd.isna(lag_data[lag][i]):
                        values.append(lag_data[lag][i])
                        valid_weights.append(weights[idx])

            if values and valid_weights:
                # Normalize weights for available data
                valid_weights = np.array(valid_weights)
                valid_weights = valid_weights / valid_weights.sum()
                result[i] = np.dot(values, valid_weights)

        df_copy[decay_col] = result

        # Print weight structure for verification
        if self.config.VERBOSE and decay_rate in [0.1, 0.5, 0.9]:
            print(f"\n  Weight structure for {component} (δ={decay_rate:.3f}, start_from_lag1={start_from_lag1}):")
            for label, weight in zip(weight_labels, weights):
                print(f"    {label}: {weight:.4f}")

        return df_copy, decay_col

# ============================================================================
# DECAY RATE OPTIMIZER
# ============================================================================
class DecayRateOptimizer:
    """Optimize decay rates for ESG components"""

    def __init__(self, config, data_loader):
        self.config = config
        self.loader = data_loader
        self.optimal_decays = {}
        self.optimization_results = {}
        self.model_results = {}

    def find_optimal_decay_rates(self, df, sector='cs', y_var='Tobin_Q',
                              method=None, max_lag=3):
      """
      Find optimal decay rates for each ESG component

      Parameters:
      -----------
      df : DataFrame
          Input data
      sector : str
          'cs' for Consumer Staples, 'all' for All Sectors
      y_var : str
          Dependent variable
      method : str or None
          'grid' for grid search, 'bounded' for bounded optimization
      max_lag : int
          Maximum lag to consider in decay calculation
      """
      if method is None:
          method = self.config.OPTIMIZATION_METHOD

      print(f"\n{'='*80}")
      print(f"OPTIMAL DECAY RATE SEARCH")
      print(f"Sector: {'Consumer Staples' if sector == 'cs' else 'All Sectors'}")
      print(f"Method: {method.upper()}")
      print(f"Max lag: {max_lag}")
      print(f"{'='*80}")

      # Prepare data based on sector
      if sector == 'cs':
          analysis_df = df[df['ConsumerStaples'] == 1].copy()
          control_vars = self.config.CONTROL_VARS
          if self.config.USE_LAGGED_CONTROLS:
              control_vars = [f'{var}_lag' for var in control_vars]
          print(f"✓ Using Consumer Staples subset: {len(analysis_df)} observations")
      else:  # 'all' sector
          analysis_df = df.copy()  # Use ALL data
          control_vars = self.config.CONTROL_VARS + ['ConsumerStaples']  # Add sector control
          if self.config.USE_LAGGED_CONTROLS:
              control_vars = [f'{var}_lag' for var in self.config.CONTROL_VARS] + ['ConsumerStaples']
          print(f"✓ Using ALL sectors data: {len(analysis_df)} observations")
          print(f"  • Consumer Staples: {analysis_df['ConsumerStaples'].sum()} observations")
          print(f"  • Other sectors: {len(analysis_df) - analysis_df['ConsumerStaples'].sum()} observations")

      # Find optimal decay for each component
      components = ['E', 'S', 'G']

      for component in components:
          print(f"\nOptimizing decay rate for {component} component:")
          print("-" * 50)

          if method == 'grid':
              optimal_decay = self._grid_search_optimization(
                  analysis_df, component, y_var, control_vars, max_lag
              )
          elif method == 'bounded':
              optimal_decay = self._bounded_optimization(
                  analysis_df, component, y_var, control_vars, max_lag
              )
          else:
              raise ValueError(f"Unknown method: {method}")

          if optimal_decay is not None:
              self.optimal_decays[component] = {
                  'optimal_decay': optimal_decay,
                  'sector': sector,
                  'y_var': y_var,
                  'max_lag': max_lag
              }

              print(f"✓ Optimal decay rate for {component}: δ = {optimal_decay:.4f}")

              # Run final model with optimal decay
              final_results = self._run_decay_model(
                  analysis_df, component, optimal_decay, y_var, control_vars, max_lag
              )

              if final_results:
                  key = f"{component}_{sector}_{y_var}"
                  self.model_results[key] = final_results
          else:
              print(f"✗ Could not find optimal decay for {component}")

      return self.optimal_decays

    def _grid_search_optimization(self, df, component, y_var, control_vars, max_lag):
        """Find optimal decay rate using grid search"""
        decay_grid = self.config.DECAY_GRID
        r2_values = []
        decay_results = []

        print(f"  Testing {len(decay_grid)} decay rates: {[f'{d:.1f}' for d in decay_grid]}")

        for decay_rate in decay_grid:
            # Create decaying average - CORRECTED: start_from_lag1=True
            df_with_decay, decay_col = self.loader.create_decaying_average_vectorized(
                df.copy(), component, decay_rate, max_lag, start_from_lag1=True
            )

            # Prepare data for regression
            all_vars = [decay_col] + control_vars + [y_var]
            df_clean = df_with_decay.dropna(subset=all_vars)

            if len(df_clean) < 10:
                r2_values.append(-np.inf)
                decay_results.append({'decay': decay_rate, 'r2': -np.inf, 'nobs': 0})
                continue

            # Run regression
            y = df_clean[y_var]
            X = df_clean[[decay_col] + control_vars]
            X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

            try:
                model = PanelOLS(y, X, entity_effects=False, time_effects=False)
                results = model.fit(cov_type='robust')

                r2 = results.rsquared
                r2_values.append(r2)
                decay_results.append({
                    'decay': decay_rate,
                    'r2': r2,
                    'nobs': results.nobs,
                    'coeff': results.params.get(decay_col, np.nan),
                    'pval': results.pvalues.get(decay_col, 1.0)
                })

                if self.config.VERBOSE:
                    sig = "***" if decay_results[-1]['pval'] < 0.01 else "**" if decay_results[-1]['pval'] < 0.05 else "*" if decay_results[-1]['pval'] < 0.10 else ""
                    print(f"    δ={decay_rate:.2f}: R²={r2:.6f}, β={decay_results[-1]['coeff']:.6f}{sig}, N={results.nobs}")

            except Exception as e:
                r2_values.append(-np.inf)
                decay_results.append({'decay': decay_rate, 'r2': -np.inf, 'error': str(e)})

        # Find best decay rate
        if r2_values and max(r2_values) > -np.inf:
            best_idx = np.argmax(r2_values)
            best_decay = decay_grid[best_idx]
            best_r2 = r2_values[best_idx]

            print(f"\n  Best: δ={best_decay:.4f} (R²={best_r2:.6f})")

            # Store optimization results
            self.optimization_results[f"{component}_grid"] = {
                'method': 'grid',
                'decay_grid': decay_grid,
                'r2_values': r2_values,
                'decay_results': decay_results,
                'best_decay': best_decay,
                'best_r2': best_r2,
                'best_idx': best_idx
            }

            # Create optimization plot
            if self.config.CREATE_PLOTS:
                self._plot_decay_optimization(decay_grid, r2_values, component, 'grid')

            return best_decay
        else:
            print(f"  No valid models found")
            return None

    def _bounded_optimization(self, df, component, y_var, control_vars, max_lag):
        """Find optimal decay rate using bounded optimization"""

        def objective_function(decay_rate):
            """Objective: maximize R-squared"""
            # Create decaying average - CORRECTED: start_from_lag1=True
            df_with_decay, decay_col = self.loader.create_decaying_average_vectorized(
                df.copy(), component, decay_rate, max_lag, start_from_lag1=True
            )

            # Prepare data
            all_vars = [decay_col] + control_vars + [y_var]
            df_clean = df_with_decay.dropna(subset=all_vars)

            if len(df_clean) < 10:
                return 1e10  # Return large value for minimization

            # Run regression
            y = df_clean[y_var]
            X = df_clean[[decay_col] + control_vars]
            X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

            try:
                model = PanelOLS(y, X, entity_effects=False, time_effects=False)
                results = model.fit(cov_type='robust')

                # Return negative R-squared (we maximize R² by minimizing -R²)
                return -results.rsquared

            except:
                return 1e10

        print(f"  Running bounded optimization (0.01 < δ < 0.99)...")

        try:
            # Use bounded optimization
            result = minimize_scalar(
                objective_function,
                bounds=(0.01, 0.99),
                method='bounded',
                options={'xatol': 1e-4, 'maxiter': 50}
            )

            if result.success:
                optimal_decay = result.x
                optimal_r2 = -result.fun  # Convert back to R²

                print(f"  Optimal: δ={optimal_decay:.6f} (R²={optimal_r2:.6f})")
                print(f"  Function evaluations: {result.nfev}")

                # Also evaluate on grid for comparison
                decay_grid = np.linspace(0.1, 0.9, 17)
                r2_values = []
                for d in decay_grid:
                    r2_values.append(-objective_function(d))

                # Store results
                self.optimization_results[f"{component}_bounded"] = {
                    'method': 'bounded',
                    'optimal_decay': optimal_decay,
                    'optimal_r2': optimal_r2,
                    'success': result.success,
                    'message': result.message,
                    'nfev': result.nfev,
                    'decay_grid': decay_grid,
                    'r2_values': r2_values
                }

                # Create plot
                if self.config.CREATE_PLOTS:
                    self._plot_decay_optimization(decay_grid, r2_values, component, 'bounded', optimal_decay)

                return optimal_decay
            else:
                print(f"  Optimization failed: {result.message}")
                return None

        except Exception as e:
            print(f"  Optimization error: {e}")
            return None

    def _run_decay_model(self, df, component, decay_rate, y_var, control_vars, max_lag):
        """Run final model with optimal decay rate"""
        print(f"\n  Running final model with δ={decay_rate:.4f}:")

        # Create decaying average - CORRECTED: start_from_lag1=True
        df_with_decay, decay_col = self.loader.create_decaying_average_vectorized(
            df.copy(), component, decay_rate, max_lag, start_from_lag1=True
        )

        # Prepare data
        all_vars = [decay_col] + control_vars + [y_var]
        df_clean = df_with_decay.dropna(subset=all_vars)

        if len(df_clean) < 10:
            print(f"    ⚠ Insufficient data: {len(df_clean)} observations")
            return None

        # Run regression
        y = df_clean[y_var]
        X = df_clean[[decay_col] + control_vars]
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

        try:
            model = PanelOLS(y, X, entity_effects=False, time_effects=False)
            results = model.fit(cov_type='robust')

            # Calculate half-life
            if decay_rate > 0:
                half_life = np.log(0.5) / np.log(decay_rate)
            else:
                half_life = np.inf

            # Calculate effective weight of each lag - CORRECTED: starting from t-1
            weights = np.array([decay_rate ** lag for lag in range(max_lag)])
            weights = weights / weights.sum()

            result_info = {
                'component': component,
                'decay_rate': decay_rate,
                'decay_col': decay_col,
                'results': results,
                'nobs': results.nobs,
                'r2': results.rsquared,
                'coeff': results.params.get(decay_col, np.nan),
                'coeff_se': results.std_errors.get(decay_col, np.nan),
                'coeff_pval': results.pvalues.get(decay_col, 1.0),
                'half_life': half_life,
                'weights': weights,
                'max_lag': max_lag,
                'data': df_clean
            }

            # Print results
            coeff = result_info['coeff']
            pval = result_info['coeff_pval']
            sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""

            print(f"    ✓ N={results.nobs}, R²={results.rsquared:.6f}")
            print(f"    ✓ Coefficient: {coeff:.6f}{sig} (p={pval:.4f})")
            print(f"    ✓ Half-life: {half_life:.2f} years")
            print(f"    ✓ Lag weights (t-1 to t-{max_lag}): {', '.join([f't-{l+1}:{w:.3f}' for l, w in enumerate(weights)])}")

            return result_info

        except Exception as e:
            print(f"    ✗ Error: {e}")
            return None

    def _plot_decay_optimization(self, decay_grid, r2_values, component, method, optimal_decay=None):
        """Create plot of decay rate optimization"""
        fig, axes = plt.subplots(1, 2, figsize=(14, 5))
        fig.suptitle(f'Decay Rate Optimization: {component} Component ({method.upper()})',
                    fontsize=14, fontweight='bold')

        # Plot 1: R² vs Decay Rate
        axes[0].plot(decay_grid, r2_values, 'b-o', linewidth=2, markersize=6)

        if optimal_decay is not None:
            # Find corresponding R²
            idx = np.abs(decay_grid - optimal_decay).argmin()
            optimal_r2 = r2_values[idx]
            axes[0].plot(optimal_decay, optimal_r2, 'ro', markersize=10,
                        label=f'Optimal: δ={optimal_decay:.3f}')
            axes[0].axvline(x=optimal_decay, color='r', linestyle='--', alpha=0.5)

        axes[0].set_xlabel('Decay Rate (δ)', fontsize=12)
        axes[0].set_ylabel('R-squared', fontsize=12)
        axes[0].set_title('R² as Function of Decay Rate', fontsize=13)
        axes[0].grid(True, alpha=0.3)
        axes[0].legend()

        # Plot 2: Weight Structure - CORRECTED: starting from t-1
        if optimal_decay is not None:
            max_lag = 5
            lags = np.arange(1, max_lag + 1)
            weights = np.array([optimal_decay ** (lag-1) for lag in lags])
            weights = weights / weights.sum()  # Normalize

            axes[1].bar(lags, weights, alpha=0.7, color='steelblue')
            axes[1].set_xlabel('Years Lag', fontsize=12)
            axes[1].set_ylabel('Weight', fontsize=12)
            axes[1].set_title(f'Weight Structure (δ={optimal_decay:.3f})', fontsize=13)
            axes[1].set_xticks(lags)
            axes[1].set_xticklabels([f't-{lag}' for lag in lags])
            axes[1].grid(True, alpha=0.3, axis='y')

            # Add weight values on bars
            for lag, weight in zip(lags, weights):
                axes[1].text(lag, weight + 0.01, f'{weight:.3f}',
                           ha='center', va='bottom', fontsize=9)

            # Calculate and display half-life
            half_life = np.log(0.5) / np.log(optimal_decay)
            axes[1].text(0.95, 0.95, f'Half-life: {half_life:.2f} years',
                        transform=axes[1].transAxes, ha='right', va='top',
                        bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

        plt.tight_layout()

        # Save plot
        os.makedirs(self.config.OUTPUT_DIR, exist_ok=True)
        plt.savefig(os.path.join(self.config.OUTPUT_DIR,
                                f'decay_optimization_{component}_{method}.png'),
                   dpi=300, bbox_inches='tight')
        plt.show()

        return fig

# ============================================================================
# COMPREHENSIVE DECAY MODEL ANALYZER
# ============================================================================
class DecayModelAnalyzer:
    """Comprehensive analyzer for decay rate models"""

    def __init__(self, config):
        self.config = config
        self.loader = DecayDataLoader(config)
        self.optimizer = None
        self.comparison_results = {}

    def run_comprehensive_decay_analysis(self, sector='cs', y_var='Tobin_Q'):
      """
      Run comprehensive decay rate analysis

      Includes:
      1. Optimal decay rate search for each ESG component
      2. Comparison with simple lagged models
      3. Comparison with moving averages
      4. Model diagnostics
      """
      print("\n" + "="*80)
      print("COMPREHENSIVE DECAY RATE ANALYSIS")
      print(f"Sector: {'Consumer Staples' if sector == 'cs' else 'All Sectors'}")
      print(f"Dependent: {y_var}")
      print("="*80)

      # Load data
      df, df_cs = self.loader.load_and_prepare_data()

      # Select dataset based on sector parameter
      if sector == 'cs':
          analysis_df = df_cs.copy()  # Consumer Staples only
          sector_name = 'Consumer Staples'
          print(f"✓ Using Consumer Staples data: {len(analysis_df)} observations")
      else:
          analysis_df = df.copy()  # ALL sectors (full dataset)
          sector_name = 'All Sectors'
          print(f"✓ Using ALL sectors data: {len(analysis_df)} observations")

      # Verify we're using the right dataset
      print(f"  • Firms in dataset: {analysis_df.index.get_level_values(0).nunique()}")
      print(f"  • Consumer Staples firms: {analysis_df['ConsumerStaples'].sum() if 'ConsumerStaples' in analysis_df.columns else 'N/A'}")

      # Initialize optimizer
      self.optimizer = DecayRateOptimizer(self.config, self.loader)

      # 1. Find optimal decay rates
      print(f"\n1. FINDING OPTIMAL DECAY RATES")
      print("-" * 60)

      optimal_decays = self.optimizer.find_optimal_decay_rates(
          analysis_df, sector, y_var,
          method=self.config.OPTIMIZATION_METHOD,
          max_lag=3
      )

      # 2. Run combined model with optimal decays
      print(f"\n2. COMBINED MODEL WITH OPTIMAL DECAYS")
      print("-" * 60)

      if optimal_decays:
          combined_results = self._run_combined_decay_model(
              analysis_df, optimal_decays, y_var, sector
          )

          # Store
          key = f"combined_optimal_{sector}_{y_var}"
          self.comparison_results[key] = combined_results
      else:
          print("No optimal decays found")
          combined_results = None

      # 3. Compare with alternative specifications
      print(f"\n3. MODEL COMPARISON")
      print("-" * 60)

      comparison_df = self._compare_decay_models(
          analysis_df, optimal_decays, y_var, sector
      )

      # 4. Economic interpretation
      print(f"\n4. ECONOMIC INTERPRETATION")
      print("-" * 60)

      self._economic_interpretation(optimal_decays, combined_results)

      # 5. Save results
      if self.config.SAVE_RESULTS:
          self._save_all_results(optimal_decays, comparison_df, sector_name, y_var)

      print("\n" + "="*80)
      print("ANALYSIS COMPLETE")
      print("="*80)

      return {
          'optimal_decays': optimal_decays,
          'combined_results': combined_results,
          'comparison': comparison_df,
          'sector': sector_name,
          'y_var': y_var
      }
    def _run_combined_decay_model(self, df, optimal_decays, y_var, sector):
        """Run combined model with optimal decays for all components"""
        print("Running combined model with optimal decay rates...")

        # Prepare control variables
        if sector == 'cs':
            control_vars = self.config.CONTROL_VARS
            if self.config.USE_LAGGED_CONTROLS:
                control_vars = [f'{var}_lag' for var in control_vars]
        else:
            control_vars = self.config.CONTROL_VARS + ['ConsumerStaples']
            if self.config.USE_LAGGED_CONTROLS:
                control_vars = [f'{var}_lag' for var in self.config.CONTROL_VARS] + ['ConsumerStaples']

        # Create decaying averages with optimal rates - CORRECTED: start_from_lag1=True
        df_with_decays = df.copy()
        decay_cols = []

        for component, info in optimal_decays.items():
            decay_rate = info['optimal_decay']
            max_lag = info.get('max_lag', 3)

            df_with_decays, decay_col = self.loader.create_decaying_average_vectorized(
                df_with_decays, component, decay_rate, max_lag, start_from_lag1=True
            )
            decay_cols.append(decay_col)

        # Prepare data
        all_vars = decay_cols + control_vars + [y_var]
        df_clean = df_with_decays.dropna(subset=all_vars)

        if len(df_clean) < 10:
            print(f"  ⚠ Insufficient data: {len(df_clean)} observations")
            return None

        # Run regression
        y = df_clean[y_var]
        X = df_clean[decay_cols + control_vars]
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

        try:
            model = PanelOLS(y, X, entity_effects=False, time_effects=False)
            results = model.fit(cov_type='robust')

            # Calculate model statistics
            result_info = {
                'model_type': 'combined_optimal_decay',
                'results': results,
                'nobs': results.nobs,
                'r2': results.rsquared,
                'decay_cols': decay_cols,
                'optimal_decays': optimal_decays,
                'coefficients': {},
                'half_lives': {}
            }

            # Extract coefficients and calculate half-lives
            for component, info in optimal_decays.items():
                decay_rate = info['optimal_decay']
                decay_col = f'{component}_decay{decay_rate:.2f}'

                if decay_col in results.params:
                    result_info['coefficients'][component] = {
                        'coeff': results.params[decay_col],
                        'se': results.std_errors.get(decay_col, np.nan),
                        'pval': results.pvalues.get(decay_col, 1.0)
                    }

                    # Calculate half-life
                    if decay_rate > 0:
                        half_life = np.log(0.5) / np.log(decay_rate)
                        result_info['half_lives'][component] = half_life

            # Print results
            print(f"  Combined Model Results:")
            print(f"    N = {results.nobs}, R² = {results.rsquared:.6f}")

            print(f"\n    ESG Coefficients with Optimal Decay:")
            for component, coeff_info in result_info['coefficients'].items():
                coeff = coeff_info['coeff']
                pval = coeff_info['pval']
                sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                decay_rate = optimal_decays[component]['optimal_decay']
                half_life = result_info['half_lives'].get(component, np.inf)

                print(f"    {component}: δ={decay_rate:.4f}, β={coeff:.6f}{sig}, "
                      f"half-life={half_life:.2f} years")

            # Create equation
            equation = self._create_decay_equation(results, decay_cols, control_vars, y_var)
            print(f"\n    Regression Equation:")
            print(f"    {equation}")

            return result_info

        except Exception as e:
            print(f"  ✗ Error: {e}")
            return None

    def _compare_decay_models(self, df, optimal_decays, y_var, sector):
        """Compare decay models with alternative specifications"""
        print("Comparing different model specifications...")

        # Define control variables
        if sector == 'cs':
            control_vars = self.config.CONTROL_VARS
            if self.config.USE_LAGGED_CONTROLS:
                control_vars = [f'{var}_lag' for var in control_vars]
        else:
            control_vars = self.config.CONTROL_VARS + ['ConsumerStaples']
            if self.config.USE_LAGGED_CONTROLS:
                control_vars = [f'{var}_lag' for var in self.config.CONTROL_VARS] + ['ConsumerStaples']

        # Define model specifications to compare
        specifications = []

        # 1. Simple lagged model (baseline)
        specifications.append({
            'name': 'Simple Lagged (t-1 only)',
            'esg_vars': ['E_lag', 'S_lag', 'G_lag'],
            'description': 'Baseline: only 1-year lag'
        })

        # 2. Distributed lag model (1-3 years)
        specifications.append({
            'name': 'Distributed Lag (1-3 years)',
            'esg_vars': ['E_lag1', 'E_lag2', 'E_lag3',
                        'S_lag1', 'S_lag2', 'S_lag3',
                        'G_lag1', 'G_lag2', 'G_lag3'],
            'description': 'Individual lags 1-3 years'
        })

        # 3. Moving average (2-year)
        # Create if not exists
        for component in ['E', 'S', 'G']:
            ma_col = f'{component}_lag_ma2'
            if ma_col not in df.columns:
                df[ma_col] = df.groupby(level=0)[f'{component}_lag'].rolling(
                    window=2, min_periods=2).mean().reset_index(level=0, drop=True)

        specifications.append({
            'name': '2-Year Moving Average',
            'esg_vars': ['E_lag_ma2', 'S_lag_ma2', 'G_lag_ma2'],
            'description': 'Equal weights for t-1 and t-2'
        })

        # 4. Optimal decay model (if available)
        if optimal_decays:
            decay_cols = []
            for component, info in optimal_decays.items():
                decay_rate = info['optimal_decay']
                decay_cols.append(f'{component}_decay{decay_rate:.2f}')

            specifications.append({
                'name': 'Optimal Decay',
                'esg_vars': decay_cols,
                'description': f'Exponential decay (δ_E={optimal_decays.get("E",{}).get("optimal_decay","N/A"):.3f}, '
                             f'δ_S={optimal_decays.get("S",{}).get("optimal_decay","N/A"):.3f}, '
                             f'δ_G={optimal_decays.get("G",{}).get("optimal_decay","N/A"):.3f})'
            })

        # 5. Fixed decay (δ=0.5 for all) - CORRECTED: start_from_lag1=True
        df_fixed = df.copy()
        fixed_decay_cols = []
        for component in ['E', 'S', 'G']:
            df_fixed, decay_col = self.loader.create_decaying_average_vectorized(
                df_fixed, component, 0.5, 3, start_from_lag1=True
            )
            fixed_decay_cols.append(decay_col)

        specifications.append({
            'name': 'Fixed Decay (δ=0.5)',
            'esg_vars': fixed_decay_cols,
            'description': 'Exponential decay with δ=0.5 for all components'
        })

        # Run all specifications
        comparison_data = []

        for spec in specifications:
            print(f"\n  Testing: {spec['name']}")
            print(f"    {spec['description']}")

            # Prepare data
            all_vars = spec['esg_vars'] + control_vars + [y_var]

            # For optimal decay, we need to ensure decay variables exist
            if spec['name'] == 'Optimal Decay':
                df_temp = df.copy()
                for component, info in optimal_decays.items():
                    decay_rate = info['optimal_decay']
                    df_temp, _ = self.loader.create_decaying_average_vectorized(
                        df_temp, component, decay_rate, 3, start_from_lag1=True
                    )
                df_clean = df_temp.dropna(subset=all_vars)
            elif spec['name'] == 'Fixed Decay (δ=0.5)':
                df_clean = df_fixed.dropna(subset=all_vars)
            else:
                df_clean = df.dropna(subset=all_vars)

            if len(df_clean) < 10:
                print(f"    ⚠ Insufficient data")
                continue

            # Run regression
            y = df_clean[y_var]
            X = df_clean[spec['esg_vars'] + control_vars]
            X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

            try:
                model = PanelOLS(y, X, entity_effects=False, time_effects=False)
                results = model.fit(cov_type='robust')

                # Calculate AIC/BIC
                n = results.nobs
                k = len(results.params)

                # Get SSR
                y_pred = results.predict().fitted_values
                y_actual = results.model.dependent.dataframe.iloc[:, 0]
                ssr = np.sum((y_actual - y_pred) ** 2)

                aic = n * np.log(ssr/n) + 2 * k
                bic = n * np.log(ssr/n) + k * np.log(n)

                # Calculate average ESG coefficient
                esg_coeffs = []
                for var in spec['esg_vars']:
                    if var in results.params:
                        esg_coeffs.append(results.params[var])

                avg_coeff = np.mean(esg_coeffs) if esg_coeffs else np.nan

                # Store results
                row = {
                    'Model': spec['name'],
                    'Description': spec['description'],
                    'N': results.nobs,
                    'R²': results.rsquared,
                    'AIC': aic,
                    'BIC': bic,
                    'K': k,
                    'Avg_ESG_Coeff': avg_coeff,
                    'results': results
                }

                comparison_data.append(row)

                print(f"    ✓ N={results.nobs}, R²={results.rsquared:.6f}, "
                      f"AIC={aic:.1f}, BIC={bic:.1f}")

                # Store in comparison results
                key = f"{spec['name'].replace(' ', '_').lower()}_{sector}_{y_var}"
                self.comparison_results[key] = {
                    'spec': spec,
                    'results': results,
                    'nobs': results.nobs,
                    'r2': results.rsquared
                }

            except Exception as e:
                print(f"    ✗ Error: {str(e)[:50]}")

        # Create comparison DataFrame
        if comparison_data:
            comparison_df = pd.DataFrame(comparison_data)

            # Sort by BIC (most conservative)
            comparison_df = comparison_df.sort_values('BIC')

            print(f"\n  Model Comparison (sorted by BIC):")
            print(f"  {'-'*100}")
            print(f"  {'Model':<30} {'Description':<40} {'N':>6} {'R²':>10} {'AIC':>10} {'BIC':>10}")
            print(f"  {'-'*100}")

            for _, row in comparison_df.iterrows():
                print(f"  {row['Model']:<30.30} {row['Description']:<40.40} "
                      f"{row['N']:>6} {row['R²']:>10.6f} "
                      f"{row['AIC']:>10.1f} {row['BIC']:>10.1f}")

            # Best model
            best_by_bic = comparison_df.iloc[0]
            best_by_r2 = comparison_df.loc[comparison_df['R²'].idxmax()]

            print(f"\n  Best by BIC: {best_by_bic['Model']} (BIC={best_by_bic['BIC']:.1f})")
            print(f"  Best by R²: {best_by_r2['Model']} (R²={best_by_r2['R²']:.6f})")

            return comparison_df
        else:
            print(f"  No valid models for comparison")
            return None

    def _economic_interpretation(self, optimal_decays, combined_results):
        """Provide economic interpretation of decay rates"""
        print("Economic Interpretation of Decay Rates:")
        print("-" * 60)

        if not optimal_decays:
            print("No optimal decays to interpret")
            return

        print("\n1. DECAY RATE INTERPRETATION:")
        print("   δ = 1.0: Permanent effects (no decay)")
        print("   δ = 0.9: Slow decay (long-lasting effects)")
        print("   δ = 0.5: Moderate decay")
        print("   δ = 0.1: Rapid decay (short-lived effects)")

        print("\n2. HALF-LIFE INTERPRETATION:")
        print("   Half-life = years for effect to reduce by 50%")

        for component, info in optimal_decays.items():
            decay_rate = info['optimal_decay']

            if decay_rate > 0:
                half_life = np.log(0.5) / np.log(decay_rate)

                print(f"\n   {component} component:")
                print(f"   • Optimal decay rate: δ = {decay_rate:.4f}")
                print(f"   • Half-life: {half_life:.2f} years")

                # Weight of each year - CORRECTED: starting from t-1
                print(f"   • Weight distribution (starting from t-1):")
                total_weight = sum(decay_rate ** lag for lag in range(4))
                cumulative = 0
                for lag in range(4):
                    weight = (decay_rate ** lag) / total_weight
                    cumulative += weight
                    print(f"     Year t-{lag+1}: {weight:.3f} ({cumulative:.1%} cumulative)")

        if combined_results and 'coefficients' in combined_results:
            print("\n3. ECONOMIC MAGNITUDE:")
            print("   Effect of 1-point increase in ESG (with optimal decay):")

            for component, coeff_info in combined_results['coefficients'].items():
                coeff = coeff_info['coeff']
                pval = coeff_info['pval']

                if pval < 0.10:
                    print(f"   • {component}: {coeff:.6f} units increase in Tobin's Q")

        print("\n4. POLICY IMPLICATIONS:")
        print("   • Slow decay (high δ): Long-term ESG investments valuable")
        print("   • Fast decay (low δ): Need continuous ESG investment")
        print("   • Different δ by component: Tailored investment strategies")

    def _create_decay_equation(self, results, decay_cols, control_vars, y_var):
        """Create regression equation for decay model"""
        equation_parts = [f"{y_var}(t) = "]

        # Constant
        if 'const' in results.params:
            const = results.params['const']
            if const >= 0:
                equation_parts.append(f"{const:.4f}")
            else:
                equation_parts.append(f"- {abs(const):.4f}")

        # Decay terms
        for decay_col in decay_cols:
            if decay_col in results.params:
                coeff = results.params[decay_col]
                pval = results.pvalues.get(decay_col, 1.0)
                sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""

                # Parse component from decay column name
                component = decay_col.split('_')[0]

                if coeff >= 0:
                    equation_parts.append(f"+ {coeff:.6f}{sig}·{component}_decay")
                else:
                    equation_parts.append(f"- {abs(coeff):.6f}{sig}·{component}_decay")

        # Control variables
        for control in control_vars:
            if control in results.params:
                coeff = results.params[control]
                pval = results.pvalues.get(control, 1.0)
                sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""

                if coeff >= 0:
                    equation_parts.append(f"+ {coeff:.4f}{sig}·{control}")
                else:
                    equation_parts.append(f"- {abs(coeff):.4f}{sig}·{control}")

        equation = " ".join(equation_parts) + " + ε"
        return equation

    def _save_all_results(self, optimal_decays, comparison_df, sector_name, y_var):
        """Save all results to Excel"""
        os.makedirs(self.config.OUTPUT_DIR, exist_ok=True)

        filename = f"decay_analysis_{sector_name.replace(' ', '_')}_{y_var}.xlsx"
        filepath = os.path.join(self.config.OUTPUT_DIR, filename)

        with pd.ExcelWriter(filepath, engine='openpyxl') as writer:
            # 1. Optimal decay rates
            if optimal_decays:
                decay_data = []
                for component, info in optimal_decays.items():
                    decay_data.append({
                        'Component': component,
                        'Optimal_Decay_Rate': info['optimal_decay'],
                        'Sector': info['sector'],
                        'Y_Variable': info['y_var'],
                        'Max_Lag': info.get('max_lag', 3)
                    })

                decay_df = pd.DataFrame(decay_data)
                decay_df.to_excel(writer, sheet_name='Optimal_Decays', index=False)

            # 2. Model comparison
            if comparison_df is not None:
                # Create clean comparison table (without results objects)
                clean_comparison = comparison_df.copy()
                if 'results' in clean_comparison.columns:
                    clean_comparison = clean_comparison.drop(columns=['results'])
                clean_comparison.to_excel(writer, sheet_name='Model_Comparison', index=False)

            # 3. Individual model results
            for key, result_info in self.comparison_results.items():
                if 'results' in result_info:
                    results = result_info['results']

                    # Create coefficient table
                    coeff_df = pd.DataFrame({
                        'Variable': results.params.index,
                        'Coefficient': results.params.values,
                        'Std_Error': [results.std_errors.get(v, np.nan) for v in results.params.index],
                        'P_Value': [results.pvalues.get(v, np.nan) for v in results.params.index]
                    })

                    # Shorten sheet name
                    sheet_name = key[:31]
                    coeff_df.to_excel(writer, sheet_name=sheet_name, index=False)

            # 4. Optimization results
            if hasattr(self.optimizer, 'optimization_results'):
                for opt_key, opt_info in self.optimizer.optimization_results.items():
                    if 'decay_grid' in opt_info:
                        opt_df = pd.DataFrame({
                            'Decay_Rate': opt_info['decay_grid'],
                            'R_Squared': opt_info['r2_values']
                        })

                        sheet_name = f"Opt_{opt_key}"[:31]
                        opt_df.to_excel(writer, sheet_name=sheet_name, index=False)

        print(f"\n✓ All results saved to: {filepath}")
        return filepath

# ============================================================================
# DISPLAY COMPREHENSIVE RESULTS
# ============================================================================
def display_comprehensive_results(analyzer, results, sector_name):
    """Display comprehensive results including regression tables and equations"""

    print("\n" + "="*80)
    print(f"COMPREHENSIVE RESULTS: {sector_name}")
    print("="*80)

    if not results or 'optimal_decays' not in results:
        print("No results to display")
        return

    # 1. OPTIMAL DECAY RATES TABLE
    print("\n1. OPTIMAL DECAY RATES")
    print("-" * 60)

    decay_table = []
    for component, info in results['optimal_decays'].items():
        decay_rate = info['optimal_decay']
        half_life = np.log(0.5) / np.log(decay_rate) if decay_rate > 0 else np.inf

        decay_table.append({
            'Component': component,
            'Optimal δ': f"{decay_rate:.4f}",
            'Half-Life (years)': f"{half_life:.2f}",
            'Interpretation': 'Slow decay' if decay_rate > 0.7 else
                             'Moderate decay' if decay_rate > 0.4 else 'Fast decay'
        })

    decay_df = pd.DataFrame(decay_table)
    print(decay_df.to_string(index=False))

    # 2. WEIGHT STRUCTURE TABLE - CORRECTED: starting from t-1
    print("\n2. WEIGHT STRUCTURE (by lag, starting from t-1)")
    print("-" * 60)

    for component, info in results['optimal_decays'].items():
        decay_rate = info['optimal_decay']
        print(f"\n{component} Component (δ={decay_rate:.4f}):")

        weights = []
        for lag in range(1, 5):  # Show lags 1-4
            weight = decay_rate ** (lag-1)
            weights.append(weight)

        # Normalize
        weights = np.array(weights)
        weights_norm = weights / weights.sum()

        weight_df = pd.DataFrame({
            'Lag': [f't-{lag}' for lag in range(1, 5)],
            'Raw Weight': [f"{w:.4f}" for w in weights],
            'Normalized': [f"{w:.4f}" for w in weights_norm],
            'Cumulative %': [f"{weights_norm[:lag].sum():.1%}" for lag in range(1, 5)]
        })
        print(weight_df.to_string(index=False))

    # 3. COMBINED MODEL REGRESSION TABLE
    if results.get('combined_results'):
        print("\n3. COMBINED MODEL REGRESSION RESULTS")
        print("-" * 60)

        combined = results['combined_results']
        if combined and 'results' in combined:
            res = combined['results']

            # Create coefficient table
            coeff_data = []
            for var in res.params.index:
                coeff = res.params[var]
                se = res.std_errors.get(var, np.nan)
                tstat = coeff / se if se not in [np.nan, 0] and not np.isnan(se) else np.nan
                pval = res.pvalues.get(var, 1.0)

                # Significance stars
                stars = ""
                if pval < 0.01:
                    stars = "***"
                elif pval < 0.05:
                    stars = "**"
                elif pval < 0.10:
                    stars = "*"

                coeff_data.append({
                    'Variable': var,
                    'Coefficient': f"{coeff:.6f}{stars}",
                    'Std.Error': f"{se:.6f}" if not np.isnan(se) else "N/A",
                    't-stat': f"{tstat:.3f}" if not np.isnan(tstat) else "N/A",
                    'p-value': f"{pval:.4f}"
                })

            coeff_df = pd.DataFrame(coeff_data)
            print(coeff_df.to_string(index=False))

            # Model statistics
            print(f"\nModel Statistics:")
            print(f"  • Observations: {res.nobs}")
            print(f"  • R-squared: {res.rsquared:.6f}")

            # Handle different R-squared metrics that PanelOLS provides
            if hasattr(res, 'rsquared_within'):
                print(f"  • R-squared (within): {res.rsquared_within:.6f}")
            if hasattr(res, 'rsquared_between'):
                print(f"  • R-squared (between): {res.rsquared_between:.6f}")
            if hasattr(res, 'rsquared_overall'):
                print(f"  • R-squared (overall): {res.rsquared_overall:.6f}")

            # Handle F-statistic
            if hasattr(res, 'f_statistic'):
                f_stat = res.f_statistic
                if hasattr(f_stat, 'stat') and hasattr(f_stat, 'pval'):
                    print(f"  • F-statistic: {f_stat.stat:.4f} (p={f_stat.pval:.4f})")

    # 4. REGRESSION EQUATIONS
    print("\n4. REGRESSION EQUATIONS")
    print("-" * 60)

    # Individual component equations
    if hasattr(analyzer.optimizer, 'model_results'):
        print("\nIndividual Component Models:")
        for key, model_info in analyzer.optimizer.model_results.items():
            if 'results' in model_info:
                res = model_info['results']
                component = model_info['component']
                decay_rate = model_info['decay_rate']

                # Build equation
                y_var = results['y_var']
                eq_parts = [f"{y_var}(t) = "]

                if 'const' in res.params:
                    const = res.params['const']
                    eq_parts.append(f"{const:.4f}")

                # Add decay term
                coeff = model_info['coeff']
                pval = model_info['coeff_pval']
                sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""

                if coeff >= 0:
                    eq_parts.append(f"+ {coeff:.6f}{sig}·{component}_decay(δ={decay_rate:.4f})")
                else:
                    eq_parts.append(f"- {abs(coeff):.6f}{sig}·{component}_decay(δ={decay_rate:.4f})")

                # Add controls (simplified)
                for control in analyzer.config.CONTROL_VARS:
                    if control in res.params:
                        c_coeff = res.params[control]
                        if c_coeff >= 0:
                            eq_parts.append(f"+ {c_coeff:.4f}·{control}")
                        else:
                            eq_parts.append(f"- {abs(c_coeff):.4f}·{control}")

                eq_parts.append("+ ε")
                equation = " ".join(eq_parts)

                print(f"\n  {component}:")
                print(f"  {equation}")
                print(f"  R² = {model_info['r2']:.6f}, N = {model_info['nobs']}")

    # Combined model equation
    if results.get('combined_results') and 'results' in results['combined_results']:
        print("\nCombined Model (All Components):")
        combined = results['combined_results']
        res = combined['results']

        # Get decay cols and control vars
        if results['sector'] == 'Consumer Staples':
            control_vars = analyzer.config.CONTROL_VARS
        else:
            control_vars = analyzer.config.CONTROL_VARS + ['ConsumerStaples']

        # Build equation
        y_var = results['y_var']
        eq_parts = [f"{y_var}(t) = "]

        if 'const' in res.params:
            const = res.params['const']
            eq_parts.append(f"{const:.4f}")

        # Add decay terms
        for decay_col in combined.get('decay_cols', []):
            if decay_col in res.params:
                coeff = res.params[decay_col]
                pval = res.pvalues.get(decay_col, 1.0)
                sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""

                component = decay_col.split('_')[0]
                decay_rate = results['optimal_decays'][component]['optimal_decay']

                if coeff >= 0:
                    eq_parts.append(f"+ {coeff:.6f}{sig}·{component}_decay(δ={decay_rate:.4f})")
                else:
                    eq_parts.append(f"- {abs(coeff):.6f}{sig}·{component}_decay(δ={decay_rate:.4f})")

        # Add controls
        for control in control_vars:
            if control in res.params:
                coeff = res.params[control]
                if coeff >= 0:
                    eq_parts.append(f"+ {coeff:.4f}·{control}")
                else:
                    eq_parts.append(f"- {abs(coeff):.4f}·{control}")

        eq_parts.append("+ ε")
        equation = " ".join(eq_parts)

        print(f"\n{equation}")

        # Display appropriate R-squared metrics
        print(f"R² = {res.rsquared:.6f}", end="")
        if hasattr(res, 'rsquared_within'):
            print(f", Within R² = {res.rsquared_within:.6f}", end="")
        if hasattr(res, 'rsquared_between'):
            print(f", Between R² = {res.rsquared_between:.6f}", end="")
        if hasattr(res, 'rsquared_overall'):
            print(f", Overall R² = {res.rsquared_overall:.6f}", end="")
        print(f", N = {res.nobs}")

    # 5. MODEL COMPARISON SUMMARY
    if results.get('comparison') is not None and results['comparison'] is not None:
        print("\n5. MODEL COMPARISON SUMMARY")
        print("-" * 60)

        comparison = results['comparison']

        # Create a display version without the 'results' column
        display_cols = ['Model', 'N', 'R²', 'AIC', 'BIC']
        if 'Avg_ESG_Coeff' in comparison.columns:
            display_cols.append('Avg_ESG_Coeff')

        # Filter to only columns that exist
        available_cols = [col for col in display_cols if col in comparison.columns]
        print(comparison[available_cols].to_string(index=False))

        # Best model by different criteria
        if 'BIC' in comparison.columns:
            best_bic = comparison.loc[comparison['BIC'].idxmin()]
            print(f"\nBest model by BIC: {best_bic['Model']} (BIC={best_bic['BIC']:.1f})")

        if 'R²' in comparison.columns:
            best_r2 = comparison.loc[comparison['R²'].idxmax()]
            print(f"Best model by R²: {best_r2['Model']} (R²={best_r2['R²']:.6f})")

# ============================================================================
# MAIN EXECUTION
# ============================================================================
def main():
    """Main execution function"""
    print("\n" + "="*80)
    print("OPTIMAL DECAY RATE ANALYSIS")
    print("Finding optimal exponential decay rates for ESG effects")
    print("="*80)

    # Initialize configuration
    config = Config()

    # Create output directory
    os.makedirs(config.OUTPUT_DIR, exist_ok=True)

    # Initialize analyzer
    analyzer = DecayModelAnalyzer(config)

    # Run analysis for Consumer Staples
    print("\n" + "="*80)
    print("ANALYSIS 1: CONSUMER STAPLES")
    print("="*80)

    results_cs = analyzer.run_comprehensive_decay_analysis(
        sector='cs',
        y_var='Tobin_Q'
    )

    # Display comprehensive results for Consumer Staples
    display_comprehensive_results(analyzer, results_cs, "Consumer Staples")

    # Run analysis for All Sectors
    print("\n" + "="*80)
    print("ANALYSIS 2: ALL SECTORS")
    print("="*80)

    results_all = analyzer.run_comprehensive_decay_analysis(
        sector='all',
        y_var='Tobin_Q'
    )

    # Display comprehensive results for All Sectors
    display_comprehensive_results(analyzer, results_all, "All Sectors")

    # Run robustness check with log-transformed Tobin's Q
    print("\n" + "="*80)
    print("ANALYSIS 3: LOG-TRANSFORMED TOBIN'S Q (Robustness)")
    print("="*80)

    results_log = analyzer.run_comprehensive_decay_analysis(
        sector='cs',
        y_var='Tobin_Q_log'
    )

    # Display comprehensive results for Log-transformed
    display_comprehensive_results(analyzer, results_log, "Consumer Staples (Log Tobin's Q)")

    print("\n" + "="*80)
    print("ANALYSIS COMPLETE")
    print("="*80)

    print(f"\nSummary:")
    print(f"  • Output directory: {config.OUTPUT_DIR}")
    print(f"  • Method used: {config.OPTIMIZATION_METHOD.upper()}")
    print(f"  • Maximum lags considered: {config.MAX_LAGS}")
    print(f"  • Models run: {len(analyzer.comparison_results)}")

    if results_cs and 'optimal_decays' in results_cs:
        print(f"\nOptimal Decay Rates (Consumer Staples):")
        for component, info in results_cs['optimal_decays'].items():
            print(f"  • {component}: δ = {info['optimal_decay']:.4f}")

    return analyzer, results_cs, results_all, results_log

# ============================================================================
# EXECUTION
# ============================================================================
if __name__ == "__main__":
    analyzer, results_cs, results_all, results_log = main()
```
