```python
# Colab setup, Drive paths, imports
from google.colab import drive
drive.mount('/content/drive')

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from linearmodels.panel import PanelOLS, compare
from statsmodels.stats.outliers_influence import variance_inflation_factor
import re
import warnings
warnings.filterwarnings('ignore')
import statsmodels.api as sm
from scipy import stats
from scipy.stats import shapiro, skew, kurtosis
from statsmodels.stats.diagnostic import het_breuschpagan
from statsmodels.stats.stattools import durbin_watson
from sklearn.preprocessing import PowerTransformer
from statsmodels.formula.api import ols

# ============================================================================
# DATA LOADING AND PREPARATION (MUST COME FIRST)
# ============================================================================
print("="*70)
print("ULTIMATE ESG IMPACT ANALYSIS WITH LAGGED SPECIFICATION")
print("="*70)

PROJECT_DIR = '/content/drive/MyDrive/Colab Notebooks/FTSE Data'
DATA_PATH = os.path.join(PROJECT_DIR, 'FTSE data.xlsx')
OUT_DIR = os.path.join(PROJECT_DIR, 'outputs_esg_tobinq')
os.makedirs(OUT_DIR, exist_ok=True)

print(f"Data path: {DATA_PATH}")
print(f"Output directory: {OUT_DIR}")

# Load the Excel file from Drive
df = pd.read_excel(DATA_PATH)
df = df.set_index(['Firm', 'Year']).sort_index()
df = df.rename(columns={"Tobin's Q": "Tobin_Q"})
df['Size'] = np.log(df['Total Assets'])
df['Leverage'] = df['Total Liabilities'] / df['Total Assets']
df['Tobin_Q_log'] = np.log(df['Tobin_Q'] + 0.001)

print(f"\nLoaded {len(df)} observations, {df.index.get_level_values(0).nunique()} firms")

# ============================================================================
# CRITICAL: CREATE LAGGED VARIABLES FOR CORRECT TIMING
# ============================================================================
print("\n" + "="*70)
print("CREATING LAGGED VARIABLES FOR CORRECT TIMING")
print("="*70)

# Create lagged ESG variables (t-1) - MUST BE DONE BEFORE ANY ANALYSIS
df['E_lag'] = df.groupby(level=0)['E'].shift(1)
df['S_lag'] = df.groupby(level=0)['S'].shift(1)
df['G_lag'] = df.groupby(level=0)['G'].shift(1)

# Create lagged control variables for consistency
df['ROA_lag'] = df.groupby(level=0)['ROA'].shift(1)
df['Size_lag'] = df.groupby(level=0)['Size'].shift(1)
df['Leverage_lag'] = df.groupby(level=0)['Leverage'].shift(1)

print(f"Created lagged variables for ESG and controls")
print(f"Missing observations due to lagging: {df[['E_lag', 'S_lag', 'G_lag']].isnull().sum().sum()}")

# ============================================================================
# CREATE TRANSFORMED TOBIN'S Q VARIABLES
# ============================================================================
print("\n" + "-"*70)
print("CREATING TRANSFORMED TOBIN'S Q VARIABLES")
print("-"*70)

# Create log transformation (Box-Cox λ ≈ 0)
df['Tobin_Q_log'] = np.log(df['Tobin_Q'] + 0.001)

# Create inverse transformation (for robustness check)
df['Tobin_Q_inv'] = 1 / (df['Tobin_Q'] + 0.001)

print("✓ Created log and inverse transformations of Tobin's Q")
print(f"  • Tobin_Q_log: log(Tobin's Q + 0.001)")
print(f"  • Tobin_Q_inv: 1/(Tobin's Q + 0.001)")

# ============================================================================
# SECTOR SETUP
# ============================================================================
# Define Consumer Staples firms
consumer_staples_firms = [
    'Associated British Foods', 'British American Tobacco', 'Coca Cola HBC AG',
    'Diageo', 'J. Sainsbury', 'Reckitt Benckiser', 'Tesco', 'Unilever'
]

# Create sector variable
firm_names = df.index.get_level_values(0)
sectors = []
for firm in firm_names:
    if firm in consumer_staples_firms:
        sectors.append('Consumer Staples')
    else:
        sectors.append('Other')

df['Sector'] = sectors
df['ConsumerStaples'] = (df['Sector'] == 'Consumer Staples').astype(int)

print("\n" + "="*70)
print("DATA SUMMARY WITH LAGGED VARIABLES")
print("="*70)
print(f"Total observations: {len(df)}")
print(f"Number of firms: {df.index.get_level_values(0).nunique()}")
print(f"Years: {sorted(df.index.get_level_values(1).unique())}")
print(f"\nSector Distribution:")
print(f"Consumer Staples firms: {df[df['ConsumerStaples']==1].index.get_level_values(0).nunique()}")
print(f"Other sectors firms: {df[df['ConsumerStaples']==0].index.get_level_values(0).nunique()}")
print(f"\nTiming Specification: ESG(t-1) → Tobin's Q(t)")

# ============================================================================
# CREATE ADDITIONAL TRANSFORMATIONS FOR ENHANCED ANALYSIS
# ============================================================================
print("\n" + "="*70)
print("CREATING ADDITIONAL TRANSFORMATIONS FOR ENHANCED ANALYSIS")
print("="*70)

# Create moving averages for enhanced analysis
for component in ['E_lag', 'S_lag', 'G_lag']:
    # 2-year moving averages
    df[f'{component}_ma2'] = df.groupby(level=0)[component].rolling(window=2, min_periods=2).mean().reset_index(level=0, drop=True)
    # 3-year moving averages
    df[f'{component}_ma3'] = df.groupby(level=0)[component].rolling(window=3, min_periods=3).mean().reset_index(level=0, drop=True)

print("✓ Created 2-year and 3-year moving averages of lagged ESG")

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================
def create_decaying_avg(series, decay_rate, max_lag=3):
    """Create exponentially decaying weighted average"""
    n = len(series)
    decaying_avg = np.zeros(n)

    for i in range(n):
        total_weight = 0
        weighted_sum = 0

        for lag in range(max_lag + 1):
            if i - lag >= 0:
                weight = decay_rate ** lag
                weighted_sum += weight * series.iloc[i - lag]
                total_weight += weight

        if total_weight > 0:
            decaying_avg[i] = weighted_sum / total_weight

    return decaying_avg

def create_cumulative(series, depreciation=0):
    """Create cumulative sum with optional depreciation"""
    cumulative = []
    current = 0

    for value in series:
        if pd.isna(value):
            cumulative.append(np.nan)
        else:
            current = (1 - depreciation) * current + value
            cumulative.append(current)

    return cumulative

def safe_assign(df, column_name, values):
    """Safely assign values to a dataframe column without warnings"""
    df = df.copy()
    df.loc[:, column_name] = values
    return df

def plot_model_diagnostics(model_name, y_actual, y_predicted, residuals, model_results=None):
    """Create comprehensive diagnostic plots for a model"""
    fig = plt.figure(figsize=(15, 10))
    fig.suptitle(f'{model_name} - Diagnostic Plots', fontsize=16, fontweight='bold')

    # Plot 1: Expected vs Observed
    ax1 = plt.subplot(2, 3, 1)
    ax1.scatter(y_actual, y_predicted, alpha=0.7, s=80, edgecolor='black', linewidth=0.5)

    # Add perfect fit line (y=x)
    min_val = min(y_actual.min(), y_predicted.min())
    max_val = max(y_actual.max(), y_predicted.max())
    ax1.plot([min_val, max_val], [min_val, max_val], 'r--', alpha=0.5, label='Perfect fit')

    # Calculate and plot regression line
    if len(y_actual) > 1:
        m, b = np.polyfit(y_actual, y_predicted, 1)
        ax1.plot(y_actual, m*y_actual + b, 'g-', alpha=0.7, label=f'Fit: y={m:.3f}x+{b:.3f}')

    ax1.set_xlabel('Observed Tobin\'s Q')
    ax1.set_ylabel('Expected Tobin\'s Q')
    ax1.set_title('Expected vs Observed Values')
    ax1.legend()
    ax1.grid(True, alpha=0.3)

    # Calculate R²
    ss_res = np.sum((y_actual - y_predicted) ** 2)
    ss_tot = np.sum((y_actual - np.mean(y_actual)) ** 2)
    r2 = 1 - (ss_res / ss_tot) if ss_tot != 0 else 0
    ax1.text(0.05, 0.95, f'R² = {r2:.4f}', transform=ax1.transAxes,
             fontsize=12, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

    # Plot 2: Residuals vs Fitted
    ax2 = plt.subplot(2, 3, 2)
    ax2.scatter(y_predicted, residuals, alpha=0.7, s=80, edgecolor='black', linewidth=0.5)
    ax2.axhline(y=0, color='r', linestyle='--', alpha=0.5)
    ax2.set_xlabel('Fitted Values')
    ax2.set_ylabel('Residuals')
    ax2.set_title('Residuals vs Fitted Values')
    ax2.grid(True, alpha=0.3)

    # Plot 3: Q-Q Plot of residuals
    ax3 = plt.subplot(2, 3, 3)
    if len(residuals) > 2:
        stats.probplot(residuals, dist="norm", plot=ax3)
    ax3.set_title('Q-Q Plot of Residuals')
    ax3.grid(True, alpha=0.3)

    # Plot 4: Histogram of residuals
    ax4 = plt.subplot(2, 3, 4)
    if len(residuals) > 1:
        ax4.hist(residuals, bins=min(10, len(residuals)//2), edgecolor='black', alpha=0.7, density=True)
        ax4.set_xlabel('Residuals')
        ax4.set_ylabel('Density')
        ax4.set_title('Distribution of Residuals')

        # Add normal curve
        xmin, xmax = ax4.get_xlim()
        x = np.linspace(xmin, xmax, 100)
        p = stats.norm.pdf(x, np.mean(residuals), np.std(residuals))
        ax4.plot(x, p, 'k', linewidth=2)

    # Plot 5: Time series of observed vs expected
    ax5 = plt.subplot(2, 3, 5)
    if hasattr(y_actual, 'index'):
        # Extract years if available
        if hasattr(y_actual.index, 'get_level_values'):
            years = y_actual.index.get_level_values(1)
            firms = y_actual.index.get_level_values(0)

            # Plot first 3 firms for clarity
            for firm in firms.unique()[:3]:
                mask = firms == firm
                if sum(mask) > 0:
                    ax5.plot(years[mask], y_actual[mask], 'o-', label=f'{firm} Observed', alpha=0.7)
                    ax5.plot(years[mask], y_predicted[mask], 's--', label=f'{firm} Expected', alpha=0.7)

        ax5.set_xlabel('Year')
        ax5.set_ylabel('Tobin\'s Q')
        ax5.set_title('Time Series: Observed vs Expected (Sample Firms)')
        ax5.legend(fontsize=9)
        ax5.grid(True, alpha=0.3)

    # Plot 6: Coefficient plot for ESG variables
    ax6 = plt.subplot(2, 3, 6)
    if model_results is not None and hasattr(model_results, 'params'):
        # Create coefficient plot for ESG variables
        esg_vars = [v for v in model_results.params.index if any(c in str(v) for c in ['E', 'S', 'G'])]

        if esg_vars:
            coefs = []
            errors = []
            labels = []

            for var in esg_vars[:6]:  # Limit to 6 for readability
                if var in model_results.params:
                    coefs.append(model_results.params[var])
                    if hasattr(model_results, 'std_errors') and var in model_results.std_errors:
                        errors.append(model_results.std_errors[var])
                    else:
                        errors.append(0)
                    labels.append(var)

            if coefs:
                x_pos = range(len(coefs))
                ax6.barh(x_pos, coefs, xerr=errors, capsize=5, alpha=0.7)
                ax6.axvline(x=0, color='black', linestyle='-', alpha=0.3)
                ax6.set_yticks(x_pos)
                ax6.set_yticklabels(labels, fontsize=9)
                ax6.set_xlabel('Coefficient Value')
                ax6.set_title('ESG Coefficients')
                ax6.invert_yaxis()
                ax6.grid(True, alpha=0.3, axis='x')
        else:
            ax6.text(0.5, 0.5, 'No ESG coefficients found',
                     ha='center', va='center', transform=ax6.transAxes)
            ax6.axis('off')
    else:
        ax6.text(0.5, 0.5, f'Model: {model_name}\nN = {len(y_actual)}',
                 ha='center', va='center', transform=ax6.transAxes, fontsize=12)
        ax6.axis('off')

    plt.tight_layout()
    return fig

def generate_regression_equation(model_name, results, variables, special_notes=None):
    """Generate a dynamic regression equation from model results"""

    print("\n" + "-"*70)
    print(f"REGRESSION EQUATION - {model_name}:")
    print("-"*70)

    # Start with constant
    eq_parts = []
    if 'const' in results.params:
        const = results.params['const']
        # Format with proper sign
        if const >= 0:
            eq_parts.append(f"{const:.4f}")
        else:
            eq_parts.append(f"- {abs(const):.4f}")

    # Add variables
    for var in variables:
        if var in results.params and var != 'const':
            coef = results.params[var]
            pval = results.pvalues[var] if var in results.pvalues else 1.0

            # Determine significance stars
            stars = ""
            if pval < 0.01:
                stars = "***"
            elif pval < 0.05:
                stars = "**"
            elif pval < 0.10:
                stars = "*"

            # Format coefficient
            if abs(coef) < 0.0001:  # Skip near-zero coefficients
                continue

            coef_str = f"{abs(coef):.4f}{stars}"

            # Format variable name for display
            if var.startswith('E_'):
                display_var = var.replace('E_', 'E.')
            elif var.startswith('S_'):
                display_var = var.replace('S_', 'S.')
            elif var.startswith('G_'):
                display_var = var.replace('G_', 'G.')
            elif var == 'ROA':
                display_var = 'ROA'
            elif var == 'Size':
                display_var = 'Size'
            elif var == 'Leverage':
                display_var = 'Leverage'
            elif var == 'ConsumerStaples':
                display_var = 'CS_Dummy'
            else:
                display_var = var

            if coef >= 0:
                eq_parts.append(f"+ {coef_str}·{display_var}")
            else:
                eq_parts.append(f"- {coef_str}·{display_var}")

    # Build equation
    equation = "Tobin's Q(t) = " + " ".join(eq_parts) + " + ε"
    print(equation)

    # Add R-squared and sample info
    print(f"\nR² = {results.rsquared:.4f}, N = {results.nobs}")

    # Add significance legend
    print("Significance: *** p<0.01, ** p<0.05, * p<0.10")

    # Add special notes if provided
    if special_notes:
        print("\nNotes:")
        for note in special_notes:
            print(f"  • {note}")

    return equation

# ============================================================================
# ENHANCED ANALYSIS FUNCTIONS
# ============================================================================
def run_comprehensive_regression(df, y_var, x_vars, model_name,
                                 entity_effects=False, time_effects=False,
                                 cluster_by='firm', include_constant=True,
                                 transformation=None):
    """Run regression with comprehensive diagnostics and multiple SE choices"""

    print(f"\n{'='*80}")
    print(f"{model_name}")
    print(f"{'='*80}")

    # Prepare data
    df_clean = df.dropna(subset=[y_var] + x_vars)

    if len(df_clean) < 10:
        print(f"Insufficient data: {len(df_clean)} observations")
        return None

    y = df_clean[y_var]
    X = df_clean[x_vars]

    if include_constant:
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

    # Run different specifications with different SE
    results_dict = {}

    # Standard errors to try
    se_methods = ['robust', 'clustered', 'unadjusted']

    for se_method in se_methods:
        try:
            if se_method == 'clustered':
                # Firm-clustered standard errors
                model = PanelOLS(y, X, entity_effects=entity_effects,
                                time_effects=time_effects)
                results = model.fit(cov_type='clustered',
                                   cluster_entity=True)
            else:
                model = PanelOLS(y, X, entity_effects=entity_effects,
                                time_effects=time_effects)
                results = model.fit(cov_type=se_method)

            results_dict[se_method] = results

            print(f"\n{se_method.upper()} Standard Errors:")
            print(f"R-squared: {results.rsquared:.4f}")
            print(f"Observations: {results.nobs}")
            print(f"F-statistic: {results.f_statistic.stat:.4f} (p={results.f_statistic.pval:.4f})")

        except Exception as e:
            print(f"Failed to run with {se_method} SE: {str(e)}")

    return results_dict

def test_fe_structure(df, y_var, x_vars, entity_var='Firm', time_var='Year'):
    """Test FE structure using Hausman and F-tests"""
    print(f"\n{'='*80}")
    print("FIXED EFFECTS STRUCTURE TESTING")
    print(f"{'='*80}")

    # Reset index for statsmodels
    df_reset = df.reset_index()

    # Prepare data
    y = df_reset[y_var]
    X = df_reset[x_vars]
    X = sm.add_constant(X)

    # Pooled OLS (no FE)
    pooled = sm.OLS(y, X).fit(cov_type='HC1')

    # Entity FE using statsmodels
    entity_dummies = pd.get_dummies(df_reset[entity_var], prefix='firm', drop_first=True)
    X_fe = pd.concat([X, entity_dummies], axis=1)
    fe_entity = sm.OLS(y, X_fe).fit(cov_type='HC1')

    # Time FE
    time_dummies = pd.get_dummies(df_reset[time_var], prefix='year', drop_first=True)
    X_timefe = pd.concat([X, time_dummies], axis=1)
    fe_time = sm.OLS(y, X_timefe).fit(cov_type='HC1')

    # Two-way FE
    X_twoway = pd.concat([X, entity_dummies, time_dummies], axis=1)
    fe_twoway = sm.OLS(y, X_twoway).fit(cov_type='HC1')

    # F-tests for FE significance
    print("\nF-TESTS FOR FIXED EFFECTS:")
    print("-" * 50)

    # Test entity FE vs pooled
    ssr_pooled = pooled.ssr
    ssr_fe = fe_entity.ssr
    n_entities = df_reset[entity_var].nunique() - 1
    n_obs = len(y)
    k = len(x_vars) + 1  # +1 for constant

    f_stat_entity = ((ssr_pooled - ssr_fe) / n_entities) / (ssr_fe / (n_obs - k - n_entities))
    p_val_entity = 1 - stats.f.cdf(f_stat_entity, n_entities, n_obs - k - n_entities)

    print(f"Entity FE vs Pooled:")
    print(f"  F-statistic: {f_stat_entity:.4f}")
    print(f"  p-value: {p_val_entity:.4f}")
    print(f"  Conclusion: {'Entity FE preferred' if p_val_entity < 0.05 else 'Pooled OLS preferred'}")

    # Test time FE vs pooled
    ssr_time = fe_time.ssr
    n_time = df_reset[time_var].nunique() - 1
    f_stat_time = ((ssr_pooled - ssr_time) / n_time) / (ssr_time / (n_obs - k - n_time))
    p_val_time = 1 - stats.f.cdf(f_stat_time, n_time, n_obs - k - n_time)

    print(f"\nTime FE vs Pooled:")
    print(f"  F-statistic: {f_stat_time:.4f}")
    print(f"  p-value: {p_val_time:.4f}")
    print(f"  Conclusion: {'Time FE preferred' if p_val_time < 0.05 else 'Pooled OLS preferred'}")

    # Test two-way FE vs pooled
    n_total_dummies = n_entities + n_time
    ssr_twoway = fe_twoway.ssr
    f_stat_twoway = ((ssr_pooled - ssr_twoway) / n_total_dummies) / (ssr_twoway / (n_obs - k - n_total_dummies))
    p_val_twoway = 1 - stats.f.cdf(f_stat_twoway, n_total_dummies, n_obs - k - n_total_dummies)

    print(f"\nTwo-way FE vs Pooled:")
    print(f"  F-statistic: {f_stat_twoway:.4f}")
    print(f"  p-value: {p_val_twoway:.4f}")
    print(f"  Conclusion: {'Two-way FE preferred' if p_val_twoway < 0.05 else 'Pooled OLS preferred'}")

    # Simplified Hausman test (FE vs RE)
    print("\n" + "="*50)
    print("HAUSMAN-TYPE TEST (FE vs RE GUIDANCE):")
    print("="*50)
    print("Note: Formal Hausman test requires random effects model")
    print("\nDecision Rules:")
    print("1. If entity F-test significant (p < 0.05): Use entity FE")
    print("2. If time F-test significant (p < 0.05): Use time FE")
    print("3. If both significant: Use two-way FE")
    print("4. If neither significant: Pooled OLS may be sufficient")

    # Recommendation
    print("\n" + "="*50)
    print("RECOMMENDED SPECIFICATION:")
    print("="*50)

    if p_val_entity < 0.05 and p_val_time < 0.05:
        print("Two-way Fixed Effects (firm and year)")
        recommendation = 'two-way'
    elif p_val_entity < 0.05:
        print("Entity Fixed Effects")
        recommendation = 'entity'
    elif p_val_time < 0.05:
        print("Time Fixed Effects")
        recommendation = 'time'
    else:
        print("Pooled OLS (no fixed effects)")
        recommendation = 'pooled'

    return recommendation

def analyze_tobinq_distribution(df, tobinq_var='Tobin_Q'):
    """Analyze and transform Tobin's Q for potential right skewness"""
    print(f"\n{'='*80}")
    print("TOBIN'S Q DISTRIBUTION ANALYSIS")
    print(f"{'='*80}")

    tobinq_data = df[tobinq_var].dropna()

    # Original distribution
    print("\nORIGINAL DISTRIBUTION:")
    print(f"Mean: {tobinq_data.mean():.4f}")
    print(f"Median: {tobinq_data.median():.4f}")
    print(f"Std Dev: {tobinq_data.std():.4f}")
    print(f"Skewness: {stats.skew(tobinq_data):.4f}")
    print(f"Kurtosis: {stats.kurtosis(tobinq_data):.4f}")

    # Shapiro-Wilk test for normality
    shapiro_stat, shapiro_p = shapiro(tobinq_data)
    print(f"Shapiro-Wilk test p-value: {shapiro_p:.6f}")
    print(f"Normality: {'Rejected' if shapiro_p < 0.05 else 'Not rejected'}")

    # Plot original distribution
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))

    # Histogram
    axes[0].hist(tobinq_data, bins=30, edgecolor='black', alpha=0.7)
    axes[0].axvline(tobinq_data.mean(), color='red', linestyle='--', label=f'Mean: {tobinq_data.mean():.2f}')
    axes[0].axvline(tobinq_data.median(), color='green', linestyle='--', label=f'Median: {tobinq_data.median():.2f}')
    axes[0].set_xlabel('Tobin\'s Q')
    axes[0].set_ylabel('Frequency')
    axes[0].set_title('Original Tobin\'s Q Distribution')
    axes[0].legend()
    axes[0].grid(True, alpha=0.3)

    # Q-Q plot
    stats.probplot(tobinq_data, dist="norm", plot=axes[1])
    axes[1].set_title('Q-Q Plot (Original)')
    axes[1].grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

    # Test transformations
    transformations = {
        'log': np.log(tobinq_data + 1 - tobinq_data.min()),  # Log transformation
        'sqrt': np.sqrt(tobinq_data),  # Square root
        'inverse': 1 / (tobinq_data + 1),  # Inverse
        'boxcox': None  # Will calculate below
    }

    # Box-Cox transformation (requires positive values)
    if (tobinq_data > 0).all():
        try:
            transformed_data, lambda_val = stats.boxcox(tobinq_data)
            transformations['boxcox'] = transformed_data
            print(f"\nBox-Cox lambda: {lambda_val:.4f}")
        except:
            pass

    print(f"\n{'='*50}")
    print("TRANSFORMATION COMPARISON:")
    print(f"{'='*50}")
    print(f"{'Transformation':<15} {'Skewness':<15} {'K-S Normality p-value':<20}")
    print("-" * 50)

    best_transform = None
    best_skewness = float('inf')

    for name, data in transformations.items():
        if data is not None:
            # Calculate skewness
            current_skewness = abs(stats.skew(data))

            # Kolmogorov-Smirnov test for normality
            ks_stat, ks_p = stats.kstest(data, 'norm',
                                        args=(data.mean(), data.std()))

            print(f"{name:<15} {current_skewness:<15.4f} {ks_p:<20.6f}")

            if current_skewness < best_skewness:
                best_skewness = current_skewness
                best_transform = name

    print(f"\nRecommended transformation: {best_transform}")

    # Apply best transformation
    if best_transform == 'log':
        df[f'{tobinq_var}_transformed'] = np.log(df[tobinq_var] + 1 - df[tobinq_var].min())
    elif best_transform == 'sqrt':
        df[f'{tobinq_var}_transformed'] = np.sqrt(df[tobinq_var])
    elif best_transform == 'inverse':
        df[f'{tobinq_var}_transformed'] = 1 / (df[tobinq_var] + 1)
    elif best_transform == 'boxcox' and (df[tobinq_var] > 0).all():
        transformed, _ = stats.boxcox(df[tobinq_var])
        df[f'{tobinq_var}_transformed'] = transformed

    return df, best_transform

def binned_scatterplot_fe(df, x_var, y_var, control_vars=None,
                         entity_fe=True, time_fe=True, n_bins=20):
    """Create binned scatterplot with FE partialling out controls"""
    print(f"\n{'='*80}")
    print(f"BINNED SCATTERPLOT: {x_var} vs {y_var}")
    print(f"Controls partialled out, FE-adjusted")
    print(f"{'='*80}")

    # Prepare data
    df_temp = df.reset_index()

    # Create FE dummies if requested
    fe_formula = ""
    if entity_fe:
        df_temp['entity_dummy'] = df_temp['Firm']
        fe_formula += " + C(entity_dummy)"

    if time_fe:
        df_temp['time_dummy'] = df_temp['Year']
        fe_formula += " + C(time_dummy)"

    # Create formula
    if control_vars:
        controls_str = " + ".join(control_vars)
        formula = f"{y_var} ~ {x_var} + {controls_str}{fe_formula}"
    else:
        formula = f"{y_var} ~ {x_var}{fe_formula}"

    # Run regression
    model = ols(formula, data=df_temp.dropna(subset=[x_var, y_var] + (control_vars or []))).fit()

    # Get residuals (y after partialling out controls and FE)
    df_temp['y_resid'] = model.resid + model.params.get('Intercept', 0)

    # Also get x residuals if controls are specified
    if control_vars:
        x_formula = f"{x_var} ~ {controls_str}{fe_formula}"
        x_model = ols(x_formula, data=df_temp.dropna(subset=[x_var] + control_vars)).fit()
        df_temp['x_resid'] = x_model.resid + x_model.params.get('Intercept', 0)
    else:
        df_temp['x_resid'] = df_temp[x_var]

    # Bin the data
    df_temp['x_bin'] = pd.qcut(df_temp['x_resid'], n_bins, duplicates='drop')
    binned = df_temp.groupby('x_bin').agg({
        'x_resid': 'mean',
        'y_resid': ['mean', 'std', 'count']
    }).round(4)

    binned.columns = ['x_mean', 'y_mean', 'y_std', 'count']
    binned = binned.reset_index()

    # Create scatterplot
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    # Binned scatterplot
    axes[0].errorbar(binned['x_mean'], binned['y_mean'],
                    yerr=binned['y_std']/np.sqrt(binned['count']),
                    fmt='o', capsize=5, alpha=0.7, color='steelblue')

    # Add regression line
    z = np.polyfit(binned['x_mean'], binned['y_mean'], 1)
    p = np.poly1d(z)
    x_range = np.linspace(binned['x_mean'].min(), binned['x_mean'].max(), 100)
    axes[0].plot(x_range, p(x_range), "r--", alpha=0.8, linewidth=2)

    axes[0].set_xlabel(f'{x_var} (residualized)')
    axes[0].set_ylabel(f'{y_var} (residualized)')
    axes[0].set_title(f'Binned Scatterplot: {x_var} vs {y_var}\nFE and Controls Partialled Out')
    axes[0].grid(True, alpha=0.3)

    # Add statistics
    slope = z[0]
    intercept = z[1]
    axes[0].text(0.05, 0.95, f'Slope: {slope:.4f}\nIntercept: {intercept:.4f}',
                transform=axes[0].transAxes, fontsize=10,
                verticalalignment='top',
                bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

    # Residual plot
    axes[1].scatter(binned['x_mean'], binned['y_mean'] - p(binned['x_mean']),
                   alpha=0.7, color='steelblue')
    axes[1].axhline(y=0, color='r', linestyle='--', alpha=0.5)
    axes[1].set_xlabel(f'{x_var} (residualized)')
    axes[1].set_ylabel('Residuals')
    axes[1].set_title('Residuals from Binned Regression')
    axes[1].grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

    print("\nBinned Statistics:")
    print(binned)

    return binned

def coefficient_plot_models(model_results_dict, target_vars=None):
    """Create coefficient plot comparing across model specifications"""
    print(f"\n{'='*80}")
    print("COEFFICIENT COMPARISON ACROSS MODEL SPECIFICATIONS")
    print(f"{'='*80}")

    # Prepare data for plotting
    plot_data = []

    for model_name, model_data in model_results_dict.items():
        if isinstance(model_data, dict) and 'results' in model_data:
            results = model_data['results']
        elif hasattr(model_data, 'params'):
            results = model_data
        else:
            continue

        # Extract coefficients for target variables
        for var in results.params.index:
            if target_vars and var not in target_vars:
                continue
            if var == 'const':
                continue

            plot_data.append({
                'Model': model_name,
                'Variable': var,
                'Coefficient': results.params[var],
                'StdError': results.std_errors.get(var, np.nan) if hasattr(results, 'std_errors') else np.nan,
                'PValue': results.pvalues.get(var, np.nan)
            })

    if not plot_data:
        print("No model results found for plotting")
        return

    plot_df = pd.DataFrame(plot_data)

    # Filter to variables that appear in multiple models
    var_counts = plot_df['Variable'].value_counts()
    common_vars = var_counts[var_counts > 1].index.tolist()

    if not common_vars:
        print("No common variables across models")
        return

    plot_df = plot_df[plot_df['Variable'].isin(common_vars)]

    # Create coefficient plot
    fig, ax = plt.subplots(figsize=(12, 8))

    # Get unique models and variables
    models = plot_df['Model'].unique()
    variables = plot_df['Variable'].unique()

    # Set positions
    n_models = len(models)
    n_vars = len(variables)

    # Color map
    colors = plt.cm.Set3(np.linspace(0, 1, n_models))

    # Plot each model
    for i, model in enumerate(models):
        model_data = plot_df[plot_df['Model'] == model]

        # Match variable order
        coeffs = []
        errors = []
        for var in variables:
            var_data = model_data[model_data['Variable'] == var]
            if len(var_data) > 0:
                coeffs.append(var_data['Coefficient'].values[0])
                errors.append(var_data['StdError'].values[0])
            else:
                coeffs.append(np.nan)
                errors.append(np.nan)

        # Plot with error bars
        x_pos = np.arange(n_vars) + i * 0.2
        ax.errorbar(x_pos, coeffs, yerr=errors, fmt='o',
                   color=colors[i], label=model, capsize=5,
                   alpha=0.8, markersize=8)

    ax.axhline(y=0, color='black', linestyle='-', alpha=0.3, linewidth=0.5)
    ax.set_xlabel('Variables')
    ax.set_ylabel('Coefficient Estimate')
    ax.set_title('Coefficient Comparison Across Model Specifications')
    ax.set_xticks(np.arange(n_vars) + 0.2 * (n_models - 1) / 2)
    ax.set_xticklabels(variables, rotation=45, ha='right')
    ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    ax.grid(True, alpha=0.3, axis='y')

    plt.tight_layout()
    plt.show()

    # Print comparison table
    print("\nCoefficient Comparison Table:")
    print("-" * 80)

    # Create pivot table
    pivot_coeff = plot_df.pivot(index='Variable', columns='Model', values='Coefficient')
    pivot_pval = plot_df.pivot(index='Variable', columns='Model', values='PValue')

    print("\nCoefficients:")
    print(pivot_coeff.round(4))

    print("\nP-values:")
    print(pivot_pval.round(4))

def run_robustness_specifications(df, y_var, x_vars, model_name):
    """Run high-impact robustness checks"""
    print(f"\n{'='*80}")
    print(f"ROBUSTNESS CHECKS: {model_name}")
    print(f"{'='*80}")

    robustness_results = {}

    # Specification 1: Two-way FE with firm-clustered SE
    print("\n1. TWO-WAY FIXED EFFECTS WITH FIRM-CLUSTERED SE")
    print("-" * 50)

    try:
        df_clean = df.dropna(subset=[y_var] + x_vars)
        y = df_clean[y_var]
        X = df_clean[x_vars]
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

        model = PanelOLS(y, X, entity_effects=True, time_effects=True)
        results = model.fit(cov_type='clustered', cluster_entity=True)

        robustness_results['Two-way FE + Firm-clustered SE'] = results

        print(f"Observations: {results.nobs}")
        print(f"Firms: {df_clean.index.get_level_values(0).nunique()}")
        print(f"Years: {df_clean.index.get_level_values(1).nunique()}")
        print(f"R-squared: {results.rsquared:.4f}")
        print(f"R-squared (within): {results.rsquared_within:.4f}")

        # Print key coefficients
        print("\nKey coefficients:")
        for var in ['E_lag', 'S_lag', 'G_lag']:
            if var in results.params:
                coeff = results.params[var]
                pval = results.pvalues.get(var, 1.0)
                sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                print(f"  {var}: {coeff:.4f}{sig} (p={pval:.4f})")

    except Exception as e:
        print(f"Failed: {str(e)}")

    # Specification 2: Winsorized variables
    print("\n2. WINSORIZED VARIABLES (1% level)")
    print("-" * 50)

    try:
        df_winsor = df.copy()

        # Winsorize Tobin's Q
        q_low = df_winsor[y_var].quantile(0.01)
        q_high = df_winsor[y_var].quantile(0.99)
        df_winsor[y_var] = df_winsor[y_var].clip(lower=q_low, upper=q_high)

        # Winsorize control variables
        for var in ['ROA', 'Size', 'Leverage']:
            if var in df_winsor.columns:
                q_low_var = df_winsor[var].quantile(0.01)
                q_high_var = df_winsor[var].quantile(0.99)
                df_winsor[var] = df_winsor[var].clip(lower=q_low_var, upper=q_high_var)

        df_clean = df_winsor.dropna(subset=[y_var] + x_vars)
        y = df_clean[y_var]
        X = df_clean[x_vars]
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

        model = PanelOLS(y, X, entity_effects=True, time_effects=False)
        results = model.fit(cov_type='robust')

        robustness_results['Winsorized (1%)'] = results

        print(f"Observations: {results.nobs}")
        print(f"R-squared: {results.rsquared:.4f}")

        print("\nKey coefficients:")
        for var in ['E_lag', 'S_lag', 'G_lag']:
            if var in results.params:
                coeff = results.params[var]
                pval = results.pvalues.get(var, 1.0)
                sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                print(f"  {var}: {coeff:.4f}{sig} (p={pval:.4f})")

    except Exception as e:
        print(f"Failed: {str(e)}")

    # Specification 3: Lagged controls
    print("\n3. LAGGED CONTROLS (t-1)")
    print("-" * 50)

    try:
        # Create lagged controls if not already present
        control_vars = ['ROA', 'Size', 'Leverage']
        lagged_controls = []

        for var in control_vars:
            if var in df.columns:
                lag_var = f'{var}_lag'
                df[lag_var] = df.groupby(level=0)[var].shift(1)
                lagged_controls.append(lag_var)

        # Use lagged controls in regression
        all_vars = x_vars + lagged_controls
        df_clean = df.dropna(subset=[y_var] + all_vars)
        y = df_clean[y_var]
        X = df_clean[all_vars]
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

        model = PanelOLS(y, X, entity_effects=True, time_effects=False)
        results = model.fit(cov_type='robust')

        robustness_results['Lagged Controls'] = results

        print(f"Observations: {results.nobs}")
        print(f"R-squared: {results.rsquared:.4f}")

        print("\nKey coefficients:")
        for var in ['E_lag', 'S_lag', 'G_lag']:
            if var in results.params:
                coeff = results.params[var]
                pval = results.pvalues.get(var, 1.0)
                sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                print(f"  {var}: {coeff:.4f}{sig} (p={pval:.4f})")

    except Exception as e:
        print(f"Failed: {str(e)}")

    # Specification 4: Industry-year FE
    print("\n4. INDUSTRY-YEAR INTERACTED FIXED EFFECTS")
    print("-" * 50)

    try:
        # Create industry-year dummies
        df_temp = df.reset_index()

        if 'Sector' in df_temp.columns:
            df_temp['industry_year'] = df_temp['Sector'].astype(str) + '_' + df_temp['Year'].astype(str)

            # Create dummy variables (simplified approach)
            dummies = pd.get_dummies(df_temp['industry_year'], prefix='indyear', drop_first=True)
            df_dummies = pd.concat([df_temp.set_index(['Firm', 'Year']), dummies.set_index(df_temp.index)], axis=1)

            # Run regression with industry-year dummies
            all_vars = x_vars + dummies.columns.tolist()
            df_clean = df_dummies.dropna(subset=[y_var] + x_vars)
            y = df_clean[y_var]
            X = df_clean[all_vars]
            X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

            model = PanelOLS(y, X, entity_effects=False, time_effects=False)
            results = model.fit(cov_type='robust')

            robustness_results['Industry-Year FE'] = results

            print(f"Observations: {results.nobs}")
            print(f"R-squared: {results.rsquared:.4f}")
            print(f"Industry-year groups: {len(dummies.columns)}")

            print("\nKey coefficients:")
            for var in ['E_lag', 'S_lag', 'G_lag']:
                if var in results.params:
                    coeff = results.params[var]
                    pval = results.pvalues.get(var, 1.0)
                    sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                    print(f"  {var}: {coeff:.4f}{sig} (p={pval:.4f})")
        else:
            print("Sector variable not found")

    except Exception as e:
        print(f"Failed: {str(e)}")

    return robustness_results

def enhanced_esg_analysis(df, consumer_staples=True):
    """Main function to run enhanced ESG analysis"""

    # Select data
    if consumer_staples:
        analysis_df = df[df['ConsumerStaples'] == 1].copy()
        print("Analyzing Consumer Staples sector only")
    else:
        analysis_df = df.copy()
        print("Analyzing all sectors")

    # 1. Test Tobin's Q transformation
    print("\n" + "="*80)
    print("STEP 1: TOBIN'S Q DISTRIBUTION ANALYSIS")
    print("="*80)

    analysis_df, best_transform = analyze_tobinq_distribution(analysis_df, 'Tobin_Q')

    # Use transformed Tobin's Q if recommended
    if best_transform:
        y_var = 'Tobin_Q_transformed'
        print(f"\nUsing {best_transform} transformation for Tobin's Q")
    else:
        y_var = 'Tobin_Q'
        print("\nUsing original Tobin's Q")

    # 2. Test FE structure
    print("\n" + "="*80)
    print("STEP 2: FIXED EFFECTS STRUCTURE TESTING")
    print("="*80)

    # Define variables
    esg_vars = ['E_lag', 'S_lag', 'G_lag']
    control_vars = ['ROA', 'Size', 'Leverage']

    fe_recommendation = test_fe_structure(
        analysis_df,
        y_var,
        esg_vars + control_vars
    )

    # 3. Run main regression with multiple SE choices
    print("\n" + "="*80)
    print("STEP 3: MAIN REGRESSION WITH MULTIPLE STANDARD ERRORS")
    print("="*80)

    # Set FE based on recommendation
    entity_fe = fe_recommendation in ['entity', 'two-way']
    time_fe = fe_recommendation in ['time', 'two-way']

    main_results = run_comprehensive_regression(
        analysis_df,
        y_var=y_var,
        x_vars=esg_vars + control_vars,
        model_name=f"MAIN MODEL (FE: {fe_recommendation})",
        entity_effects=entity_fe,
        time_effects=time_fe,
        cluster_by='firm'
    )

    # 4. Create binned scatterplot
    print("\n" + "="*80)
    print("STEP 4: BINNED SCATTERPLOTS")
    print("="*80)

    # Create binned scatterplot for each ESG component
    for esg_var in esg_vars:
        binned_scatterplot_fe(
            analysis_df,
            x_var=esg_var,
            y_var=y_var,
            control_vars=control_vars,
            entity_fe=entity_fe,
            time_fe=time_fe,
            n_bins=15
        )

    # 5. Run baseline to robust comparison
    print("\n" + "="*80)
    print("STEP 5: MODEL VARIANT COMPARISON")
    print("="*80)

    # Define model variants
    model_variants = [
        {
            'name': 'Baseline (no controls)',
            'x_vars': esg_vars,
            'entity_fe': False,
            'time_fe': False,
            'se_method': 'robust'
        },
        {
            'name': '+ Controls',
            'x_vars': esg_vars + control_vars,
            'entity_fe': False,
            'time_fe': False,
            'se_method': 'robust'
        },
        {
            'name': '+ Firm FE',
            'x_vars': esg_vars + control_vars,
            'entity_fe': True,
            'time_fe': False,
            'se_method': 'robust'
        },
        {
            'name': '+ Two-way FE',
            'x_vars': esg_vars + control_vars,
            'entity_fe': True,
            'time_fe': True,
            'se_method': 'robust'
        },
        {
            'name': '+ Firm-clustered SE',
            'x_vars': esg_vars + control_vars,
            'entity_fe': True,
            'time_fe': True,
            'se_method': 'clustered'
        }
    ]

    variant_results = {}

    for variant in model_variants:
        print(f"\nRunning: {variant['name']}")
        print("-" * 40)

        try:
            df_clean = analysis_df.dropna(subset=[y_var] + variant['x_vars'])
            y = df_clean[y_var]
            X = df_clean[variant['x_vars']]
            X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

            model = PanelOLS(y, X,
                           entity_effects=variant['entity_fe'],
                           time_effects=variant['time_fe'])

            if variant['se_method'] == 'clustered':
                results = model.fit(cov_type='clustered', cluster_entity=True)
            else:
                results = model.fit(cov_type=variant['se_method'])

            variant_results[variant['name']] = {
                'results': results,
                'type': 'variant'
            }

            print(f"Observations: {results.nobs}")
            print(f"R-squared: {results.rsquared:.4f}")

            # Print ESG coefficients
            for var in esg_vars:
                if var in results.params:
                    coeff = results.params[var]
                    pval = results.pvalues.get(var, 1.0)
                    sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                    print(f"  {var}: {coeff:.4f}{sig}")

        except Exception as e:
            print(f"Failed: {str(e)}")

    # 6. Create coefficient plot
    print("\n" + "="*80)
    print("STEP 6: COEFFICIENT PLOT ACROSS VARIANTS")
    print("="*80)

    coefficient_plot_models(variant_results, target_vars=esg_vars)

    # 7. Run robustness checks
    print("\n" + "="*80)
    print("STEP 7: ROBUSTNESS CHECKS")
    print("="*80)

    robustness_results = run_robustness_specifications(
        analysis_df,
        y_var=y_var,
        x_vars=esg_vars + control_vars,
        model_name="Main Specification Robustness"
    )

    # Combine all results
    all_results = {
        'main': main_results,
        'variants': variant_results,
        'robustness': robustness_results,
        'fe_recommendation': fe_recommendation,
        'tobinq_transform': best_transform
    }

    return all_results

def create_summary_table(results_dict):
    """Create publication-quality summary table"""
    print(f"\n{'='*80}")
    print("PUBLICATION-QUALITY SUMMARY TABLE")
    print(f"{'='*80}")

    table_data = []

    # Extract from main results
    if 'main' in results_dict and results_dict['main']:
        for se_method, results in results_dict['main'].items():
            row = {
                'Specification': f'Main ({se_method} SE)',
                'N': results.nobs,
                'R2': results.rsquared,
                'E(t-1)': results.params.get('E_lag', np.nan),
                'S(t-1)': results.params.get('S_lag', np.nan),
                'G(t-1)': results.params.get('G_lag', np.nan),
                'ROA': results.params.get('ROA', np.nan),
                'Size': results.params.get('Size', np.nan),
                'Leverage': results.params.get('Leverage', np.nan)
            }
            table_data.append(row)

    # Extract from variants
    if 'variants' in results_dict:
        for name, data in results_dict['variants'].items():
            if 'results' in data:
                results = data['results']
                row = {
                    'Specification': name,
                    'N': results.nobs,
                    'R2': results.rsquared,
                    'E(t-1)': results.params.get('E_lag', np.nan),
                    'S(t-1)': results.params.get('S_lag', np.nan),
                    'G(t-1)': results.params.get('G_lag', np.nan),
                    'ROA': results.params.get('ROA', np.nan),
                    'Size': results.params.get('Size', np.nan),
                    'Leverage': results.params.get('Leverage', np.nan)
                }
                table_data.append(row)

    # Create DataFrame
    summary_df = pd.DataFrame(table_data)

    # Format for display
    display_df = summary_df.copy()

    # Add significance stars
    for col in ['E(t-1)', 'S(t-1)', 'G(t-1)', 'ROA', 'Size', 'Leverage']:
        # This is simplified - in practice you'd need p-values
        display_df[col] = display_df[col].apply(lambda x: f"{x:.4f}" if not pd.isna(x) else "")

    print("\nSummary of Regression Results:")
    print("-" * 100)
    print(display_df.to_string(index=False))

    return summary_df

def create_diagnostic_summary(results_dict):
    """Create summary of diagnostic tests"""
    print(f"\n{'='*80}")
    print("MODEL DIAGNOSTICS SUMMARY")
    print(f"{'='*80}")

    diagnostics = []

    if results_dict and 'main' in results_dict and results_dict['main']:
        for se_method, results in results_dict['main'].items():
            diag = {
                'Model': f'Main ({se_method})',
                'FE Structure': results_dict.get('fe_recommendation', 'N/A'),
                'TobinQ Transform': results_dict.get('tobinq_transform', 'None'),
                'N': results.nobs,
                'R2': results.rsquared,
                'F-stat': results.f_statistic.stat if hasattr(results, 'f_statistic') else np.nan,
                'F-pval': results.f_statistic.pval if hasattr(results, 'f_statistic') else np.nan
            }
            diagnostics.append(diag)

    diag_df = pd.DataFrame(diagnostics)
    print("\nDiagnostics Table:")
    print("-" * 80)
    print(diag_df.to_string(index=False))

    return diag_df

# ============================================================================
# RUN ENHANCED ANALYSIS
# ============================================================================
print("\n" + "="*80)
print("ENHANCED ESG-FINANCIAL PERFORMANCE ANALYSIS")
print("="*80)

# Run enhanced analysis for Consumer Staples
print("\nANALYSIS 1: CONSUMER STAPLES SECTOR")
try:
    enhanced_results_cs = enhanced_esg_analysis(df, consumer_staples=True)
    print("✓ Consumer Staples analysis completed")
except Exception as e:
    print(f"✗ Error in Consumer Staples analysis: {e}")
    enhanced_results_cs = None

# Run enhanced analysis for All Sectors
print("\n" + "="*80)
print("ANALYSIS 2: ALL SECTORS")
try:
    enhanced_results_all = enhanced_esg_analysis(df, consumer_staples=False)
    print("✓ All Sectors analysis completed")
except Exception as e:
    print(f"✗ Error in All Sectors analysis: {e}")
    enhanced_results_all = None

# Create diagnostic summaries if analyses completed
if enhanced_results_cs:
    print("\nCONSUMER STAPLES DIAGNOSTICS:")
    cs_diagnostics = create_diagnostic_summary(enhanced_results_cs)

if enhanced_results_all:
    print("\nALL SECTORS DIAGNOSTICS:")
    all_diagnostics = create_diagnostic_summary(enhanced_results_all)

# ============================================================================
# RUN YOUR ORIGINAL MODELS AFTER ENHANCED ANALYSIS
# ============================================================================
print("\n" + "="*70)
print("PHASE 2: ORIGINAL CONSUMER STAPLES MODELS WITH LAGGED ESG")
print("="*70)

# Use only Consumer Staples data for original models
df_cs = df[df['ConsumerStaples'] == 1].copy()

# Initialize results dictionary for original models
original_model_results = {}

# ============================================================================
# MODEL 1A: 2-YEAR MOVING AVERAGES - CONSUMER STAPLES ONLY (LAGGED)
# ============================================================================
print("\n" + "="*70)
print("MODEL 1A: 2-YEAR MOVING AVERAGES OF LAGGED ESG - CONSUMER STAPLES ONLY")
print("="*70)

# Create 2-year moving averages of LAGGED ESG
for component in ['E_lag', 'S_lag', 'G_lag']:
    ma2_values = df_cs.groupby(level=0)[component].rolling(window=2, min_periods=2).mean().reset_index(level=0, drop=True)
    df_cs = safe_assign(df_cs, f'{component}_ma2', ma2_values)

# Clean data for analysis
df_clean_ma2 = df_cs.dropna(subset=['E_lag_ma2', 'S_lag_ma2', 'G_lag_ma2', 'ROA', 'Size', 'Leverage', 'Tobin_Q'])

# Model with 2-year moving averages of lagged ESG
y_ma2 = df_clean_ma2['Tobin_Q']
X_ma2 = df_clean_ma2[['E_lag_ma2', 'S_lag_ma2', 'G_lag_ma2', 'ROA', 'Size', 'Leverage']]
X_ma2 = pd.concat([pd.Series(1, index=X_ma2.index, name='const'), X_ma2], axis=1)

model_ma2 = PanelOLS(y_ma2, X_ma2, entity_effects=False, time_effects=False)
results_ma2 = model_ma2.fit(cov_type='robust')
y_pred_ma2 = results_ma2.predict().fitted_values
residuals_ma2 = y_ma2 - y_pred_ma2

print("2-Year Moving Average of LAGGED ESG (Consumer Staples Only):")
print(results_ma2)

# Generate equation
generate_regression_equation(
    "MODEL 1A: 2-Year Moving Average of Lagged ESG (CS Only)",
    results_ma2,
    ['const', 'E_lag_ma2', 'S_lag_ma2', 'G_lag_ma2', 'ROA', 'Size', 'Leverage'],
    special_notes=[
        "E_lag_ma2 = 2-year moving average of LAGGED Environmental score",
        "S_lag_ma2 = 2-year moving average of LAGGED Social score",
        "G_lag_ma2 = 2-year moving average of LAGGED Governance score",
        "ESG(t-1, t-2) → Tobin's Q(t)",
        "Consumer Staples only"
    ]
)

# Plot diagnostics
fig_ma2 = plot_model_diagnostics("2-Year MA of Lagged ESG (CS Only)", y_ma2, y_pred_ma2, residuals_ma2, results_ma2)
plt.show()

# Store results
original_model_results['2-Year MA of Lagged ESG (CS)'] = {
    'results': results_ma2,
    'y_pred': y_pred_ma2,
    'residuals': residuals_ma2,
    'nobs': results_ma2.nobs,
    'type': 'Consumer Staples (Lagged)'
}

# ============================================================================
# MODEL 1B: 3-YEAR MOVING AVERAGES - CONSUMER STAPLES ONLY (LAGGED)
# ============================================================================
print("\n" + "="*70)
print("MODEL 1B: 3-YEAR MOVING AVERAGES OF LAGGED ESG - CONSUMER STAPLES ONLY")
print("="*70)

# Create 3-year moving averages of LAGGED ESG
for component in ['E_lag', 'S_lag', 'G_lag']:
    ma3_values = df_cs.groupby(level=0)[component].rolling(window=3, min_periods=3).mean().reset_index(level=0, drop=True)
    df_cs = safe_assign(df_cs, f'{component}_ma3', ma3_values)

# Clean data for analysis
df_clean_ma3 = df_cs.dropna(subset=['E_lag_ma3', 'S_lag_ma3', 'G_lag_ma3', 'ROA', 'Size', 'Leverage', 'Tobin_Q'])

# Model with 3-year moving averages of lagged ESG
y_ma3 = df_clean_ma3['Tobin_Q']
X_ma3 = df_clean_ma3[['E_lag_ma3', 'S_lag_ma3', 'G_lag_ma3', 'ROA', 'Size', 'Leverage']]
X_ma3 = pd.concat([pd.Series(1, index=X_ma3.index, name='const'), X_ma3], axis=1)

model_ma3 = PanelOLS(y_ma3, X_ma3, entity_effects=False, time_effects=False)
results_ma3 = model_ma3.fit(cov_type='robust')
y_pred_ma3 = results_ma3.predict().fitted_values
residuals_ma3 = y_ma3 - y_pred_ma3

print("3-Year Moving Average of LAGGED ESG (Consumer Staples Only):")
print(results_ma3)

# Generate equation
generate_regression_equation(
    "MODEL 1B: 3-Year Moving Average of Lagged ESG (CS Only)",
    results_ma3,
    ['const', 'E_lag_ma3', 'S_lag_ma3', 'G_lag_ma3', 'ROA', 'Size', 'Leverage'],
    special_notes=[
        "E_lag_ma3 = 3-year moving average of LAGGED Environmental score",
        "S_lag_ma3 = 3-year moving average of LAGGED Social score",
        "G_lag_ma3 = 3-year moving average of LAGGED Governance score",
        "ESG(t-1, t-2, t-3) → Tobin's Q(t)",
        "X̄₃ = (X_{t-1} + X_{t-2} + X_{t-3})/3",
        "Consumer Staples only"
    ]
)

# Plot diagnostics
fig_ma3 = plot_model_diagnostics("3-Year MA of Lagged ESG (CS Only)", y_ma3, y_pred_ma3, residuals_ma3, results_ma3)
plt.show()

# Store results
original_model_results['3-Year MA of Lagged ESG (CS)'] = {
    'results': results_ma3,
    'y_pred': y_pred_ma3,
    'residuals': residuals_ma3,
    'nobs': results_ma3.nobs,
    'type': 'Consumer Staples (Lagged)'
}

# ============================================================================
# MODEL 2: DISTRIBUTED LAG MODEL (2-5 YEAR LAGS) - CONSUMER STAPLES ONLY
# ============================================================================
print("\n" + "="*70)
print("MODEL 2: DISTRIBUTED LAG MODEL (2-5 Year Lags) - CONSUMER STAPLES ONLY")
print("="*70)

# For each firm-year, we need multiple lags of original ESG (not already lagged)
df_lags = df_cs.copy()
for component in ['E', 'S', 'G']:
    for lag in range(1, 6):  # Lags 1-5 years (t-1 to t-5)
        df_lags[f'{component}_lag{lag}'] = df_lags.groupby(level=0)[component].shift(lag)

# Clean data with multiple lags
lag_columns = [f'{c}_lag{l}' for c in ['E', 'S', 'G'] for l in range(1, 6)]
df_lags_clean = df_lags.dropna(subset=lag_columns + ['ROA', 'Size', 'Leverage', 'Tobin_Q'])

if len(df_lags_clean) >= 10:
    # Estimate distributed lag model (unrestricted)
    X_lags = lag_columns + ['ROA', 'Size', 'Leverage']
    X_distributed = df_lags_clean[X_lags]
    X_distributed = pd.concat([pd.Series(1, index=X_distributed.index, name='const'), X_distributed], axis=1)
    y_distributed = df_lags_clean['Tobin_Q']

    model_distributed = PanelOLS(y_distributed, X_distributed, entity_effects=False, time_effects=False)
    results_distributed = model_distributed.fit(cov_type='robust')
    y_pred_distributed = results_distributed.predict().fitted_values
    residuals_distributed = y_distributed - y_pred_distributed

    print("Distributed Lag Model (1-5 year lags, CS Only):")
    print(results_distributed)

    # Generate simplified equation (only significant lags)
    print("\n" + "-"*70)
    print("REGRESSION EQUATION - MODEL 2: Distributed Lag (CS Only, Simplified):")
    print("-"*70)

    eq_parts = []
    const = results_distributed.params['const']
    eq_parts.append(f"{const:.4f}" if const >= 0 else f"- {abs(const):.4f}")

    # Add only significant lagged variables (p < 0.10)
    for var in results_distributed.params.index:
        if var != 'const':
            pval = results_distributed.pvalues[var] if var in results_distributed.pvalues else 1.0
            if pval < 0.10:
                coef = results_distributed.params[var]
                stars = "***" if pval < 0.01 else "**" if pval < 0.05 else "*"

                if coef >= 0:
                    eq_parts.append(f"+ {abs(coef):.4f}{stars}·{var}")
                else:
                    eq_parts.append(f"- {abs(coef):.4f}{stars}·{var}")

    equation = "Tobin's Q(t) = " + " ".join(eq_parts) + " + ε"
    print(equation)
    print(f"\nR² = {results_distributed.rsquared:.4f}, N = {results_distributed.nobs}")
    print("Showing only variables with p < 0.10")
    print("ESG(t-1 to t-5) → Tobin's Q(t)")
    print("Consumer Staples only")

    # Plot diagnostics
    fig_dist = plot_model_diagnostics("Distributed Lag (1-5 years, CS Only)", y_distributed,
                                      y_pred_distributed, residuals_distributed, results_distributed)
    plt.show()

    # Lag structure plot
    fig_lags, axes = plt.subplots(1, 3, figsize=(15, 4))
    fig_lags.suptitle('Lag Structure by ESG Component (1-5 year lags, CS Only)', fontsize=14, fontweight='bold')

    for idx, component in enumerate(['E', 'S', 'G']):
        lags = list(range(1, 6))
        coeffs = []
        errors = []

        for lag in lags:
            var = f'{component}_lag{lag}'
            if var in results_distributed.params:
                coeffs.append(results_distributed.params[var])
                if hasattr(results_distributed, 'std_errors') and var in results_distributed.std_errors:
                    errors.append(results_distributed.std_errors[var])
                else:
                    errors.append(0)
            else:
                coeffs.append(0)
                errors.append(0)

        axes[idx].errorbar(lags, coeffs, yerr=errors, fmt='o-', capsize=5, linewidth=2, markersize=8)
        axes[idx].axhline(y=0, color='r', linestyle='--', alpha=0.5)
        axes[idx].set_xlabel('Years Lag')
        axes[idx].set_ylabel('Coefficient')
        axes[idx].set_title(f'{component} Component')
        axes[idx].set_xticks(lags)
        axes[idx].grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

    # Store results
    original_model_results['Distributed Lag 1-5 years (CS)'] = {
        'results': results_distributed,
        'y_pred': y_pred_distributed,
        'residuals': residuals_distributed,
        'nobs': results_distributed.nobs,
        'type': 'Consumer Staples (Lagged)'
    }
else:
    print("Insufficient data for distributed lag model (needs at least 10 observations)")

# ============================================================================
# MODEL 3: NON-LINEAR TRANSFORMATIONS - CONSUMER STAPLES ONLY (LAGGED)
# ============================================================================
print("\n" + "="*70)
print("MODEL 3: NON-LINEAR EFFECTS OF LAGGED ESG - CONSUMER STAPLES ONLY")
print("="*70)

# Create non-linear transformations of LAGGED ESG
df_nonlinear = df_clean_ma3.copy()

# 1. Logarithmic transformation (diminishing returns) of LAGGED ESG
for component in ['E_lag', 'S_lag', 'G_lag']:
    df_nonlinear[f'log_{component}'] = np.log(df_nonlinear[component])

# 2. Square root transformation of LAGGED ESG
for component in ['E_lag', 'S_lag', 'G_lag']:
    df_nonlinear[f'sqrt_{component}'] = np.sqrt(df_nonlinear[component])

# 3. Quadratic terms (U-shaped or inverted U) of LAGGED ESG
for component in ['E_lag', 'S_lag', 'G_lag']:
    df_nonlinear[f'{component}_sq'] = df_nonlinear[component] ** 2

# 4. Threshold effects (dummy for high LAGGED ESG)
for component in ['E_lag', 'S_lag', 'G_lag']:
    threshold = df_nonlinear[component].quantile(0.75)  # Top quartile
    df_nonlinear[f'high_{component}'] = (df_nonlinear[component] > threshold).astype(int)

# Test different non-linear specifications
nonlinear_specs = [
    (['log_E_lag', 'log_S_lag', 'log_G_lag'], "Logarithmic (diminishing returns)"),
    (['sqrt_E_lag', 'sqrt_S_lag', 'sqrt_G_lag'], "Square Root"),
    (['E_lag', 'E_lag_sq', 'S_lag', 'S_lag_sq', 'G_lag', 'G_lag_sq'], "Quadratic (U-shaped possible)"),
    (['high_E_lag', 'high_S_lag', 'high_G_lag'], "Threshold (top quartile)")
]

print("\nNon-linear Model Comparison (LAGGED ESG, CS Only):")
print("-"*70)
print(f"{'Specification':<40} {'R-squared':<10} {'E Effect':<10} {'S Effect':<10} {'G Effect':<10}")
print("-"*70)

best_nonlinear = None
best_r2 = -1
best_results = None

for vars_list, desc in nonlinear_specs:
    temp = df_nonlinear.dropna(subset=vars_list + ['Tobin_Q', 'ROA', 'Size', 'Leverage'])

    if len(temp) < 5:
        continue

    y_temp = temp['Tobin_Q']
    X_temp = temp[vars_list + ['ROA', 'Size', 'Leverage']]
    X_temp = pd.concat([pd.Series(1, index=X_temp.index, name='const'), X_temp], axis=1)

    model_temp = PanelOLS(y_temp, X_temp, entity_effects=False, time_effects=False)
    results_temp = model_temp.fit(cov_type='robust')
    y_pred_temp = results_temp.predict().fitted_values
    residuals_temp = y_temp - y_pred_temp

    # Extract key coefficients
    e_effect = np.nan
    s_effect = np.nan
    g_effect = np.nan

    for var in vars_list:
        if 'E_lag' in var and var in results_temp.params:
            e_effect = results_temp.params[var]
        elif 'S_lag' in var and var in results_temp.params:
            s_effect = results_temp.params[var]
        elif 'G_lag' in var and var in results_temp.params:
            g_effect = results_temp.params[var]

    print(f"{desc:<40} {results_temp.rsquared:<10.4f} {e_effect:<10.4f} {s_effect:<10.4f} {g_effect:<10.4f}")

    # Track best model
    if results_temp.rsquared > best_r2:
        best_r2 = results_temp.rsquared
        best_nonlinear = (desc, vars_list, results_temp, y_temp, y_pred_temp, residuals_temp)

# Generate equation and plot diagnostics for best non-linear model
if best_nonlinear:
    desc, vars_list, results_temp, y_temp, y_pred_temp, residuals_temp = best_nonlinear
    print(f"\n{'='*70}")
    print(f"BEST NON-LINEAR MODEL (LAGGED ESG, CS Only): {desc}")
    print(f"{'='*70}")

    # Generate equation
    eq_vars = ['const'] + vars_list + ['ROA', 'Size', 'Leverage']
    generate_regression_equation(
        f"MODEL 3: {desc} of Lagged ESG (CS Only)",
        results_temp,
        eq_vars,
        special_notes=[f"Best non-linear specification (R² = {best_r2:.4f})",
                       "ESG(t-1) → Tobin's Q(t)",
                       "Consumer Staples only"]
    )

    # Plot diagnostics
    fig_nonlinear = plot_model_diagnostics(f"Non-linear of Lagged ESG: {desc} (CS Only)", y_temp, y_pred_temp, residuals_temp, results_temp)
    plt.show()

    # Store results
    original_model_results[f'Non-linear of Lagged ESG {desc[:15]}... (CS)'] = {
        'results': results_temp,
        'y_pred': y_pred_temp,
        'residuals': residuals_temp,
        'nobs': results_temp.nobs,
        'type': 'Consumer Staples (Lagged)'
    }

# ============================================================================
# MODEL 4: OPTIMAL DECAY RATE OF LAGGED ESG - CONSUMER STAPLES ONLY
# ============================================================================
print("\n" + "="*70)
print("MODEL 4: OPTIMAL DECAY RATE OF LAGGED ESG - CONSUMER STAPLES ONLY")
print("="*70)

# Estimate optimal decay rate by grid search using LAGGED ESG
def objective_function_lagged(decay_rate, component='S_lag', df_data=None):
    """Minimize negative R-squared for different decay rates on LAGGED ESG"""
    if df_data is None:
        return 1e10

    df_test = df_data.copy()

    # Create decaying average for each firm
    decaying_avgs = []
    for firm in df_test.index.get_level_values(0).unique():
        firm_data = df_test.xs(firm, level=0)
        if len(firm_data) > 0:
            decaying_avg = create_decaying_avg(firm_data[component], decay_rate, max_lag=3)
            decaying_avgs.extend(decaying_avg)
        else:
            decaying_avgs.extend([np.nan] * len(firm_data))

    df_test[f'{component}_decay'] = decaying_avgs

    # Drop NA
    temp = df_test.dropna(subset=[f'{component}_decay', 'Tobin_Q', 'ROA', 'Size', 'Leverage'])

    if len(temp) < 5:
        return 1e10  # Return large value if insufficient data

    # Run regression
    y_temp = temp['Tobin_Q']
    X_temp = temp[[f'{component}_decay', 'ROA', 'Size', 'Leverage']]
    X_temp = pd.concat([pd.Series(1, index=X_temp.index, name='const'), X_temp], axis=1)

    model_temp = PanelOLS(y_temp, X_temp, entity_effects=False, time_effects=False)
    results_temp = model_temp.fit(cov_type='robust')

    # Return negative R-squared (we want to maximize R²)
    return -results_temp.rsquared

# Find optimal decay rate for each LAGGED component using available data
optimal_decays_lagged = {}
if 'df_lags_clean' in locals() and len(df_lags_clean) > 0:
    for component in ['E_lag', 'S_lag', 'G_lag']:
        # Grid search from 0.1 to 0.9
        decay_grid = np.linspace(0.1, 0.9, 9)
        r2_values = []

        for decay in decay_grid:
            r2 = -objective_function_lagged(decay, component, df_lags_clean)
            r2_values.append(r2)

        # Find valid R² values
        valid_r2 = [r for r in r2_values if r != -1e10]
        if valid_r2:
            optimal_decay = decay_grid[np.argmax(valid_r2)]
            optimal_decays_lagged[component] = optimal_decay

            print(f"\n{component} - Optimal decay rate search (CS Only):")
            for decay, r2 in zip(decay_grid, r2_values):
                if r2 != -1e10:
                    print(f"  Decay {decay:.1f}: R² = {r2:.4f}")
            print(f"  → Optimal: {optimal_decay:.3f}")
        else:
            optimal_decays_lagged[component] = 0.5
            print(f"\n{component}: Using default decay rate 0.5 (insufficient data for optimization)")
else:
    print("Using default decay rates (insufficient data for optimization)")
    optimal_decays_lagged = {'E_lag': 0.5, 'S_lag': 0.5, 'G_lag': 0.5}

# Apply optimal decay rates to LAGGED ESG
df_optimal = df_lags_clean.copy() if 'df_lags_clean' in locals() else df_cs.copy()
for component, decay_rate in optimal_decays_lagged.items():
    decaying_avgs = []
    for firm in df_optimal.index.get_level_values(0).unique():
        firm_data = df_optimal.xs(firm, level=0)
        if len(firm_data) > 0:
            decaying_avg = create_decaying_avg(firm_data[component], decay_rate, max_lag=3)
            decaying_avgs.extend(decaying_avg)
        else:
            decaying_avgs.extend([np.nan] * len(firm_data))

    df_optimal[f'{component}_optdecay'] = decaying_avgs

# Estimate final model with optimal decay rates of LAGGED ESG
df_optimal_clean = df_optimal.dropna(subset=[f'{c}_optdecay' for c in ['E_lag', 'S_lag', 'G_lag']] +
                                     ['ROA', 'Size', 'Leverage', 'Tobin_Q'])

if len(df_optimal_clean) > 0:
    y_opt = df_optimal_clean['Tobin_Q']
    X_opt = df_optimal_clean[[f'{c}_optdecay' for c in ['E_lag', 'S_lag', 'G_lag']] + ['ROA', 'Size', 'Leverage']]
    X_opt = pd.concat([pd.Series(1, index=X_opt.index, name='const'), X_opt], axis=1)

    model_optimal = PanelOLS(y_opt, X_opt, entity_effects=False, time_effects=False)
    results_optimal = model_optimal.fit(cov_type='robust')
    y_pred_opt = results_optimal.predict().fitted_values
    residuals_opt = y_opt - y_pred_opt

    print("\nOptimal Decay Rate of LAGGED ESG (CS Only):")
    print(results_optimal)

    # Generate equation
    eq_vars = ['const'] + [f'{c}_optdecay' for c in ['E_lag', 'S_lag', 'G_lag']] + ['ROA', 'Size', 'Leverage']
    eq_notes = [
        f"Optimal decay rates: E_lag: δ={optimal_decays_lagged.get('E_lag', 0.5):.1f}, S_lag: δ={optimal_decays_lagged.get('S_lag', 0.5):.1f}, G_lag: δ={optimal_decays_lagged.get('G_lag', 0.5):.1f}",
        f"{component}_decay(δ) = Σ δ^k·{component}_t-k / Σ δ^k, for k = 0 to 3",
        "ESG(t-1) with decay → Tobin's Q(t)",
        "Consumer Staples only"
    ]

    generate_regression_equation(
        "MODEL 4: Optimal Decay Rate of Lagged ESG (CS Only)",
        results_optimal,
        eq_vars,
        special_notes=eq_notes
    )

    # Plot diagnostics
    fig_opt = plot_model_diagnostics("Optimal Decay Rate of Lagged ESG (CS Only)", y_opt, y_pred_opt, residuals_opt, results_optimal)
    plt.show()

    # Store results
    original_model_results['Optimal Decay of Lagged ESG (CS)'] = {
        'results': results_optimal,
        'y_pred': y_pred_opt,
        'residuals': residuals_opt,
        'nobs': results_optimal.nobs,
        'type': 'Consumer Staples (Lagged)'
    }
else:
    print("\nInsufficient data for optimal decay rate model of lagged ESG")

# ============================================================================
# MODEL 5: ESG CAPITAL STOCK FROM LAGGED ESG - CONSUMER STAPLES ONLY
# ============================================================================
print("\n" + "="*70)
print("MODEL 5: ESG CAPITAL STOCK FROM LAGGED ESG - CONSUMER STAPLES ONLY")
print("="*70)

# Calculate ESG capital for each firm-component using LAGGED ESG
df_capital = df_cs.copy()
for component in ['E_lag', 'S_lag', 'G_lag']:
    capital_series = []
    for firm in df_capital.index.get_level_values(0).unique():
        firm_data = df_capital.xs(firm, level=0)
        capital = create_cumulative(firm_data[component], depreciation=0.3)
        capital_series.extend(capital)

    df_capital[f'{component}_capital'] = capital_series

# Model with ESG capital from LAGGED ESG
df_capital_clean = df_capital.dropna(subset=[f'{c}_capital' for c in ['E_lag', 'S_lag', 'G_lag']] +
                                     ['ROA', 'Size', 'Leverage', 'Tobin_Q'])

if len(df_capital_clean) > 0:
    y_capital = df_capital_clean['Tobin_Q']
    X_capital = df_capital_clean[[f'{c}_capital' for c in ['E_lag', 'S_lag', 'G_lag']] + ['ROA', 'Size', 'Leverage']]
    X_capital = pd.concat([pd.Series(1, index=X_capital.index, name='const'), X_capital], axis=1)

    model_capital = PanelOLS(y_capital, X_capital, entity_effects=False, time_effects=False)
    results_capital = model_capital.fit(cov_type='robust')
    y_pred_capital = results_capital.predict().fitted_values
    residuals_capital = y_capital - y_pred_capital

    print("ESG Capital Stock from LAGGED ESG (30% annual depreciation, CS Only):")
    print(results_capital)

    # Generate equation
    generate_regression_equation(
        "MODEL 5: ESG Capital Stock from Lagged ESG (CS Only)",
        results_capital,
        ['const', 'E_lag_capital', 'S_lag_capital', 'G_lag_capital', 'ROA', 'Size', 'Leverage'],
        special_notes=[
            "X_capital(t) = (1-δ)·X_capital(t-1) + X_lag(t), where δ = 0.3",
            "X_capital(0) = X_lag(0)",
            "ESG(t-1) capital → Tobin's Q(t)",
            "Consumer Staples only"
        ]
    )

    # Plot diagnostics
    fig_capital = plot_model_diagnostics("ESG Capital Stock from Lagged ESG (CS Only)", y_capital, y_pred_capital,
                                         residuals_capital, results_capital)
    plt.show()

    # Store results
    original_model_results['ESG Capital from Lagged ESG (CS)'] = {
        'results': results_capital,
        'y_pred': y_pred_capital,
        'residuals': residuals_capital,
        'nobs': results_capital.nobs,
        'type': 'Consumer Staples (Lagged)'
    }
else:
    print("Insufficient data for ESG capital stock model from lagged ESG")

# ============================================================================
# MODEL 6: CHANGE IN LAGGED ESG - CONSUMER STAPLES ONLY
# ============================================================================
print("\n" + "="*70)
print("MODEL 6: CHANGE IN LAGGED ESG - CONSUMER STAPLES ONLY")
print("="*70)

# Create change variables for LAGGED ESG
df_change = df_cs.copy()
for component in ['E_lag', 'S_lag', 'G_lag']:
    df_change[f'{component}_chg1'] = df_change.groupby(level=0)[component].diff(1)

df_change_clean = df_change.dropna(subset=['E_lag_chg1', 'S_lag_chg1', 'G_lag_chg1', 'ROA', 'Size', 'Leverage', 'Tobin_Q'])

if len(df_change_clean) > 0:
    y_change = df_change_clean['Tobin_Q']
    X_change = df_change_clean[['E_lag_chg1', 'S_lag_chg1', 'G_lag_chg1', 'ROA', 'Size', 'Leverage']]
    X_change = pd.concat([pd.Series(1, index=X_change.index, name='const'), X_change], axis=1)

    model_change = PanelOLS(y_change, X_change, entity_effects=False, time_effects=False)
    results_change = model_change.fit(cov_type='robust')
    y_pred_change = results_change.predict().fitted_values
    residuals_change = y_change - y_pred_change

    print("1-Year Change in LAGGED ESG (CS Only):")
    print(results_change)

    # Generate equation
    generate_regression_equation(
        "MODEL 6: Change in Lagged ESG (CS Only)",
        results_change,
        ['const', 'E_lag_chg1', 'S_lag_chg1', 'G_lag_chg1', 'ROA', 'Size', 'Leverage'],
        special_notes=[
            "X_lag_chg1 = X_lag(t) - X_lag(t-1)",
            "Measures impact of changes in past ESG performance",
            "ΔESG(t-1) → Tobin's Q(t)",
            "Consumer Staples only"
        ]
    )

    # Plot diagnostics
    fig_change = plot_model_diagnostics("Change in Lagged ESG (CS Only)", y_change, y_pred_change,
                                        residuals_change, results_change)
    plt.show()

    # Store results
    original_model_results['Change in Lagged ESG (CS)'] = {
        'results': results_change,
        'y_pred': y_pred_change,
        'residuals': residuals_change,
        'nobs': results_change.nobs,
        'type': 'Consumer Staples (Lagged)'
    }
else:
    print("Insufficient data for change in lagged ESG model")

# ============================================================================
# ALL SECTORS VERSIONS OF THESE MODELS
# ============================================================================
print("\n" + "="*70)
print("PHASE 3: ALL SECTORS VERSIONS OF LAGGED ESG MODELS")
print("="*70)

# Create transformations for all data
print("\nCreating lagged transformations for all firms...")

# Create quadratic terms for lagged ESG for all data
for component in ['E_lag', 'S_lag', 'G_lag']:
    df[f'{component}_sq'] = df[component] ** 2

# Create capital stock for lagged ESG for all data
for component in ['E_lag', 'S_lag', 'G_lag']:
    capital_series = []
    for firm in df.index.get_level_values(0).unique():
        firm_data = df.xs(firm, level=0)
        capital = create_cumulative(firm_data[component], depreciation=0.3)
        capital_series.extend(capital)
    df[f'{component}_capital'] = capital_series

# Create change variables for lagged ESG for all data
for component in ['E_lag', 'S_lag', 'G_lag']:
    df[f'{component}_chg1'] = df.groupby(level=0)[component].diff(1)

print("✓ Created all lagged transformations for all data")

# Function to run All Sectors models with lagged ESG
def run_all_sectors_lagged_model_parallel(df, model_name, esg_vars, control_vars, special_notes=None):
    """Run a model with all sectors data using LAGGED ESG - PARALLEL WITH TRANSFORMED"""
    print(f"\n{'='*70}")
    print(f"{model_name.upper()} - ALL SECTORS (LAGGED ESG)")
    print("PARALLEL ANALYSIS: Original & Log-Transformed Tobin's Q")
    print(f"{'='*70}")

    # Clean data - need all variables
    all_vars = esg_vars + control_vars + ['Tobin_Q', 'Tobin_Q_log', 'ConsumerStaples']
    df_clean = df.dropna(subset=all_vars)

    if len(df_clean) < 10:
        print(f"Insufficient data: Only {len(df_clean)} observations")
        return None, None

    results_dict = {}

    # Run ORIGINAL Tobin's Q model
    y_original = df_clean['Tobin_Q']
    X = df_clean[esg_vars + control_vars + ['ConsumerStaples']]
    X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

    model_original = PanelOLS(y_original, X, entity_effects=False, time_effects=False)
    results_original = model_original.fit(cov_type='robust')
    results_dict['original'] = results_original

    # Run LOG-TRANSFORMED Tobin's Q model
    y_log = df_clean['Tobin_Q_log']
    model_log = PanelOLS(y_log, X, entity_effects=False, time_effects=False)
    results_log = model_log.fit(cov_type='robust')
    results_dict['log'] = results_log

    print(f"Observations: {results_original.nobs}")
    print(f"Firms: {df_clean.index.get_level_values(0).nunique()}")
    print(f"Consumer Staples firms: {df_clean[df_clean['ConsumerStaples']==1].index.get_level_values(0).nunique()}")
    print(f"Other firms: {df_clean[df_clean['ConsumerStaples']==0].index.get_level_values(0).nunique()}")

    # Print comparison table
    print("\n" + "="*60)
    print(f"{'VARIABLE':<15} {'ORIGINAL':<20} {'LOG-TRANSFORMED':<20} {'ROBUST?'}")
    print("="*60)

    # Check key variables
    key_vars = esg_vars + control_vars + ['ConsumerStaples']

    for var in key_vars:
        if var in results_original.params:
            # Original results
            orig_coeff = results_original.params[var]
            orig_pval = results_original.pvalues.get(var, 1.0)
            orig_sig = "***" if orig_pval < 0.01 else "**" if orig_pval < 0.05 else "*" if orig_pval < 0.10 else ""
            orig_str = f"{orig_coeff:.4f}{orig_sig} (p={orig_pval:.3f})"

            # Log results
            log_coeff = results_log.params[var]
            log_pval = results_log.pvalues.get(var, 1.0)
            log_sig = "***" if log_pval < 0.01 else "**" if log_pval < 0.05 else "*" if log_pval < 0.10 else ""
            log_str = f"{log_coeff:.4f}{log_sig} (p={log_pval:.3f})"

            # Check robustness
            same_sign = (orig_coeff * log_coeff > 0)  # Both positive or both negative
            same_sig = ((orig_pval < 0.10) == (log_pval < 0.10))  # Both significant or both not

            if same_sign and same_sig:
                robust = "✓"
            elif same_sign:
                robust = "⚠ (sign)"  # Same sign but different significance
            elif same_sig:
                robust = "⚠ (sig)"   # Same significance but different sign
            else:
                robust = "✗"

            print(f"{var:<15} {orig_str:<20} {log_str:<20} {robust}")

    print(f"\nR-squared comparison:")
    print(f"  Original Tobin's Q: {results_original.rsquared:.4f}")
    print(f"  Log Tobin's Q: {results_log.rsquared:.4f}")

    # Generate equation for original
    eq_notes = special_notes + ["ESG(t-1) → Tobin's Q(t)", "All sectors with Consumer Staples dummy"]
    print("\n" + "-"*70)
    print("REGRESSION EQUATION - ORIGINAL Tobin's Q:")
    print("-"*70)
    generate_regression_equation(
        f"{model_name} (Original Tobin's Q)",
        results_original,
        ['const'] + esg_vars + control_vars + ['ConsumerStaples'],
        special_notes=eq_notes
    )

    print("\n" + "-"*70)
    print("REGRESSION EQUATION - LOG-TRANSFORMED Tobin's Q:")
    print("-"*70)
    generate_regression_equation(
        f"{model_name} (Log Tobin's Q)",
        results_log,
        ['const'] + esg_vars + control_vars + ['ConsumerStaples'],
        special_notes=eq_notes + ["Interpretation: 1 unit ESG → exp(coeff)-1 percent change in Tobin's Q"]
    )

    return results_dict

def run_cs_model_parallel(df, model_name, esg_vars, control_vars, special_notes=None):
    """Run Consumer Staples model in parallel with original and transformed Tobin's Q"""
    print(f"\n{'='*70}")
    print(f"{model_name.upper()} - CONSUMER STAPLES")
    print("PARALLEL ANALYSIS: Original & Log-Transformed Tobin's Q")
    print(f"{'='*70}")

    # Filter Consumer Staples data
    cs_data = df[df['ConsumerStaples'] == 1].copy()

    # Clean data
    all_vars = esg_vars + control_vars + ['Tobin_Q', 'Tobin_Q_log']
    cs_clean = cs_data.dropna(subset=all_vars)

    if len(cs_clean) < 5:
        print(f"Insufficient data: Only {len(cs_clean)} observations")
        return None, None

    results_dict = {}

    # Run ORIGINAL Tobin's Q model
    y_original = cs_clean['Tobin_Q']
    X = cs_clean[esg_vars + control_vars]
    X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

    model_original = PanelOLS(y_original, X, entity_effects=False, time_effects=False)
    results_original = model_original.fit(cov_type='robust')
    results_dict['original'] = results_original

    # Run LOG-TRANSFORMED Tobin's Q model
    y_log = cs_clean['Tobin_Q_log']
    model_log = PanelOLS(y_log, X, entity_effects=False, time_effects=False)
    results_log = model_log.fit(cov_type='robust')
    results_dict['log'] = results_log

    print(f"Observations: {results_original.nobs}")
    print(f"Firms: {cs_clean.index.get_level_values(0).nunique()}")

    # Print comparison table
    print("\n" + "="*60)
    print(f"{'VARIABLE':<15} {'ORIGINAL':<20} {'LOG-TRANSFORMED':<20} {'ROBUST?'}")
    print("="*60)

    for var in esg_vars + control_vars:
        if var in results_original.params:
            # Original results
            orig_coeff = results_original.params[var]
            orig_pval = results_original.pvalues.get(var, 1.0)
            orig_sig = "***" if orig_pval < 0.01 else "**" if orig_pval < 0.05 else "*" if orig_pval < 0.10 else ""
            orig_str = f"{orig_coeff:.4f}{orig_sig} (p={orig_pval:.3f})"

            # Log results
            log_coeff = results_log.params[var]
            log_pval = results_log.pvalues.get(var, 1.0)
            log_sig = "***" if log_pval < 0.01 else "**" if log_pval < 0.05 else "*" if log_pval < 0.10 else ""
            log_str = f"{log_coeff:.4f}{log_sig} (p={log_pval:.3f})"

            # Check robustness
            same_sign = (orig_coeff * log_coeff > 0)
            same_sig = ((orig_pval < 0.10) == (log_pval < 0.10))

            if same_sign and same_sig:
                robust = "✓"
            elif same_sign:
                robust = "⚠ (sign)"
            elif same_sig:
                robust = "⚠ (sig)"
            else:
                robust = "✗"

            print(f"{var:<15} {orig_str:<20} {log_str:<20} {robust}")

    print(f"\nR-squared comparison:")
    print(f"  Original Tobin's Q: {results_original.rsquared:.4f}")
    print(f"  Log Tobin's Q: {results_log.rsquared:.4f}")

    # Generate equations
    eq_notes = special_notes + ["ESG(t-1) → Tobin's Q(t)", "Consumer Staples only"]

    print("\n" + "-"*70)
    print("REGRESSION EQUATION - ORIGINAL Tobin's Q:")
    print("-"*70)
    generate_regression_equation(
        f"{model_name} (Original Tobin's Q, CS)",
        results_original,
        ['const'] + esg_vars + control_vars,
        special_notes=eq_notes
    )

    print("\n" + "-"*70)
    print("REGRESSION EQUATION - LOG-TRANSFORMED Tobin's Q:")
    print("-"*70)
    generate_regression_equation(
        f"{model_name} (Log Tobin's Q, CS)",
        results_log,
        ['const'] + esg_vars + control_vars,
        special_notes=eq_notes + ["Interpretation: 1 unit ESG → exp(coeff)-1 percent change in Tobin's Q"]
    )

    return results_dict
# Run All Sectors versions of key models
print("\nRunning All Sectors versions of key models...")

# Model 1A All Sectors: 2-Year MA of Lagged ESG
results_ma2_all = run_all_sectors_lagged_model_parallel(
    df,
    "2-Year MA of Lagged ESG",
    ['E_lag_ma2', 'S_lag_ma2', 'G_lag_ma2'],
    ['ROA', 'Size', 'Leverage'],
    ["2-year moving average of LAGGED ESG scores"]
)

if results_ma2_all:
    original_model_results['2-Year MA of Lagged ESG (All)'] = {
        'results': results_ma2_all,
        'type': 'All Sectors (Lagged)'
    }

# Model 1B All Sectors: 3-Year MA of Lagged ESG
results_ma3_all = run_all_sectors_lagged_model_parallel(
    df,
    "3-Year MA of Lagged ESG",
    ['E_lag_ma3', 'S_lag_ma3', 'G_lag_ma3'],
    ['ROA', 'Size', 'Leverage'],
    ["3-year moving average of LAGGED ESG scores"]
)

if results_ma3_all:
    original_model_results['3-Year MA of Lagged ESG (All)'] = {
        'results': results_ma3_all,
        'type': 'All Sectors (Lagged)'
    }

# Simple lagged model (t-1 only) for all sectors
results_simple_all = run_all_sectors_lagged_model_parallel(
    df,
    "Simple Lagged ESG",
    ['E_lag', 'S_lag', 'G_lag'],
    ['ROA', 'Size', 'Leverage'],
    ["Simple 1-year lag of ESG scores"]
)

if results_simple_all:
    original_model_results['Simple Lagged ESG (All)'] = {
        'results': results_simple_all,
        'type': 'All Sectors (Lagged)'
    }

# ============================================================================
# COMPREHENSIVE COMPARISON
# ============================================================================
print("\n" + "="*70)
print("COMPREHENSIVE MODEL COMPARISON WITH LAGGED ESG")
print("="*70)

# Combine enhanced results with original model results
all_model_results = {}
if 'enhanced_results_cs' in locals() and enhanced_results_cs:
    all_model_results.update({'Enhanced_CS': enhanced_results_cs})
if 'enhanced_results_all' in locals() and enhanced_results_all:
    all_model_results.update({'Enhanced_All': enhanced_results_all})

all_model_results.update(original_model_results)

# Print comprehensive summary table
print("\nCOMPREHENSIVE MODEL PERFORMANCE SUMMARY (LAGGED ESG):")
print("-" * 120)
print(f"{'Model':<35} {'Type':<20} {'R-squared':<12} {'Social Coeff':<15} {'N':<10} {'Timing'}")
print("-" * 120)

for model_name, data in all_model_results.items():
    # Handle different data structures
    if isinstance(data, dict):
        if 'results' in data:  # Original model results
            results = data['results']
            model_type = data.get('type', 'Unknown')

            # SAFETY CHECK: Make sure results is not a dict
            if isinstance(results, dict):
                print(f"⚠ Warning: Results for {model_name} is a dict, not results object")
                continue  # Skip this model

            r2 = results.rsquared if hasattr(results, 'rsquared') else 0
            nobs = data.get('nobs', results.nobs if hasattr(results, 'nobs') else 0)

            # Find Social coefficient
            social_coeff = 0
            social_sig = ""
            social_vars = ['S_lag', 'S_lag_ma2', 'S_lag_ma3', 'S_lag_capital', 'S_lag_chg1', 'S']
            for var in social_vars:
                if hasattr(results, 'params') and var in results.params:
                    social_coeff = results.params[var]
                    if hasattr(results, 'pvalues') and var in results.pvalues:
                        pval = results.pvalues[var]
                        social_sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                    break

        elif 'main' in data:  # Enhanced results structure
            # Get the first result from main - but be careful!
            if data['main']:
                first_key = list(data['main'].keys())[0]
                possible_results = data['main'][first_key]

                # Check if it's actually a results object
                if hasattr(possible_results, 'rsquared'):
                    results = possible_results
                    model_type = 'Enhanced Analysis'
                    r2 = results.rsquared
                    nobs = results.nobs if hasattr(results, 'nobs') else 0
                    social_coeff = results.params.get('S_lag', 0) if hasattr(results, 'params') else 0
                    social_sig = ""
                else:
                    # Skip this enhanced model
                    print(f"⚠ Skipping enhanced model {model_name}: data structure issue")
                    continue
            else:
                continue
    elif hasattr(data, 'rsquared'):  # Direct results object
        results = data
        model_type = 'Direct Results'
        r2 = results.rsquared
        nobs = results.nobs if hasattr(results, 'nobs') else 0
        social_coeff = results.params.get('S_lag', 0) if hasattr(results, 'params') else 0
        social_sig = ""
    else:
        continue  # Skip unknown data types

    # Determine timing specification
    timing = "ESG(t-1)→Q(t)" if any(x in str(model_name).lower() for x in ['lag', 'lagged']) else "ESG(t)→Q(t)"

    # Find key findings
    key_findings = []
    if hasattr(results, 'params'):
        for var in results.params.index if hasattr(results.params, 'index') else results.params.keys():
            if any(c in str(var) for c in ['E', 'S', 'G']) and var != 'const':
                if hasattr(results, 'pvalues') and var in results.pvalues and results.pvalues[var] < 0.10:
                    coeff = results.params[var]
                    sig = "***" if results.pvalues[var] < 0.01 else "**" if results.pvalues[var] < 0.05 else "*"
                    # Shorten variable name for display
                    if len(var) > 8:
                        var_display = var[:8] + "..."
                    else:
                        var_display = var
                    key_findings.append(f"{var_display}:{coeff:.2f}{sig}")

    findings_str = "; ".join(key_findings[:2]) if key_findings else ""

    print(f"{model_name:<35} {model_type:<20} {r2:<12.4f} {social_coeff:<8.4f}{social_sig:<7} {nobs:<10} {timing:<15} {findings_str}")

    # Determine timing specification
    timing = "ESG(t-1)→Q(t)" if any(x in str(model_name).lower() for x in ['lag', 'lagged']) else "ESG(t)→Q(t)"

    # Find key findings
    key_findings = []
    if hasattr(results, 'params'):
        for var in results.params.index:
            if any(c in str(var) for c in ['E', 'S', 'G']) and var != 'const':
                if hasattr(results, 'pvalues') and var in results.pvalues and results.pvalues[var] < 0.10:
                    coeff = results.params[var]
                    sig = "***" if results.pvalues[var] < 0.01 else "**" if results.pvalues[var] < 0.05 else "*"
                    # Shorten variable name for display
                    if len(var) > 8:
                        var_display = var[:8] + "..."
                    else:
                        var_display = var
                    key_findings.append(f"{var_display}:{coeff:.2f}{sig}")

    findings_str = "; ".join(key_findings[:2]) if key_findings else ""

    print(f"{model_name:<35} {model_type:<20} {r2:<12.4f} {social_coeff:<8.4f}{social_sig:<7} {nobs:<10} {timing:<15} {findings_str}")

print("\n" + "="*70)
print("ANALYSIS COMPLETE")
print("="*70)
print("Summary:")
print("1. Enhanced Analysis: Credibility improvements, FE testing, transformations")
print("2. Original Models: 2-year MA, 3-year MA, distributed lags, non-linear, optimal decay")
print("3. All Sectors Versions: Sector comparisons with Consumer Staples dummy")
print("4. Comprehensive Comparison: All models compared in single table")
print("="*70)

# ============================================================================
# SAVE RESULTS
# ============================================================================
try:
    output_path = os.path.join(OUT_DIR, 'enhanced_results.xlsx')

    # Create DataFrames for saving
    cs_summary = create_summary_table(enhanced_results_cs) if enhanced_results_cs else pd.DataFrame()
    all_summary = create_summary_table(enhanced_results_all) if enhanced_results_all else pd.DataFrame()

    with pd.ExcelWriter(output_path) as writer:
        if not cs_summary.empty:
            cs_summary.to_excel(writer, sheet_name='Consumer_Staples', index=False)
        if not all_summary.empty:
            all_summary.to_excel(writer, sheet_name='All_Sectors', index=False)

    print(f"\n✓ Results saved to: {output_path}")
except Exception as e:
    print(f"\n✗ Could not save results: {e}")

print("\n" + "="*80)
print("ENHANCED ANALYSIS COMPLETE")
print("="*80)
print("\nKey Enhancements Implemented:")
print("1. ✅ Multiple standard error choices reported")
print("2. ✅ FE structure justified with F-tests")
print("3. ✅ Tobin's Q transformation analysis")
print("4. ✅ Binned scatterplots with FE partialling")
print("5. ✅ Coefficient plots across model variants")
print("6. ✅ High-impact robustness specifications")
print("7. ✅ Two-way FE with firm-clustered SE")
print("8. ✅ Publication-quality summary tables")
print("="*80)

# ============================================================================
# NOW YOU CAN RUN YOUR ORIGINAL SECTOR ANALYSIS
# Initialize the results dictionary here before using it
all_model_results = {}

# Now you can run your original sector analysis functions
# ============================================================================
# SECTOR ANALYSIS FUNCTIONS (FROM YOUR ORIGINAL CODE)
# ============================================================================
def run_sector_interaction_models_lagged(df):
    """Run models with sector interactions using LAGGED ESG"""

    results_dict = {}

    # MODEL A: Consumer Staples Only (Baseline) with LAGGED ESG
    cs_data = df[df['ConsumerStaples'] == 1]
    if len(cs_data) >= 10:
        print(f"\n{'='*70}")
        print("CONSUMER STAPLES ONLY (Baseline Model) - LAGGED ESG")
        print(f"{'='*70}")
        print("Using ESG(t-1) to predict Tobin's Q(t)")

        # Simple model with LAGGED E, S, G
        cs_clean = cs_data.dropna(subset=['E_lag', 'S_lag', 'G_lag', 'ROA', 'Size', 'Leverage', 'Tobin_Q'])
        if len(cs_clean) > 0:
            y_cs = cs_clean['Tobin_Q']
            X_cs = cs_clean[['E_lag', 'S_lag', 'G_lag', 'ROA', 'Size', 'Leverage']]
            X_cs = pd.concat([pd.Series(1, index=X_cs.index, name='const'), X_cs], axis=1)

            model_cs = PanelOLS(y_cs, X_cs, entity_effects=False, time_effects=False)
            results_cs = model_cs.fit(cov_type='robust')

            results_dict['CS Only (Lagged ESG)'] = results_cs

            print(f"Observations: {len(cs_clean)}")
            print(f"Firms: {cs_clean.index.get_level_values(0).nunique()}")
            print(f"R-squared: {results_cs.rsquared:.4f}")

            # Print key coefficients
            print("\nKey coefficients (ESG(t-1) → Tobin's Q(t)):")
            for var in ['E_lag', 'S_lag', 'G_lag']:
                if var in results_cs.params:
                    pval = results_cs.pvalues.get(var, 1.0)
                    sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                    print(f"  {var}: {results_cs.params[var]:.4f}{sig} (p={pval:.3f})")

    # MODEL B: All Sectors with Sector Interaction (LAGGED)
    print(f"\n{'='*70}")
    print("ALL SECTORS WITH SECTOR INTERACTIONS - LAGGED ESG")
    print(f"{'='*70}")
    print("Using ESG(t-1) to predict Tobin's Q(t)")

    # Create interaction terms with LAGGED ESG
    df_interaction = df.copy()
    for var in ['E_lag', 'S_lag', 'G_lag']:
        base_var = var.replace('_lag', '')
        df_interaction[f'{var}_CS'] = df_interaction[var] * df_interaction['ConsumerStaples']

    # Clean data
    interaction_vars = []
    for var in ['E_lag', 'S_lag', 'G_lag']:
        if var in df_interaction.columns:
            interaction_vars.extend([var, f'{var}_CS'])

    df_interaction_clean = df_interaction.dropna(
        subset=interaction_vars + ['ROA', 'Size', 'Leverage', 'Tobin_Q']
    )

    if len(df_interaction_clean) > 0:
        y_int = df_interaction_clean['Tobin_Q']
        X_int = df_interaction_clean[interaction_vars + ['ROA', 'Size', 'Leverage']]
        X_int = pd.concat([pd.Series(1, index=X_int.index, name='const'), X_int], axis=1)

        model_int = PanelOLS(y_int, X_int, entity_effects=False, time_effects=False)
        results_int = model_int.fit(cov_type='robust')

        results_dict['Sector Interaction (Lagged)'] = results_int

        print(f"Total observations: {len(df_interaction_clean)}")
        print(f"Total firms: {df_interaction_clean.index.get_level_values(0).nunique()}")
        print(f"Consumer Staples firms: {df_interaction_clean[df_interaction_clean['ConsumerStaples']==1].index.get_level_values(0).nunique()}")
        print(f"R-squared: {results_int.rsquared:.4f}")

        # Calculate Consumer Staples-specific effects
        print("\nCONSUMER STAPLES-SPECIFIC EFFECTS (LAGGED ESG):")
        for base_var in ['E', 'S', 'G']:
            var = f'{base_var}_lag'
            var_cs = f'{var}_CS'

            if var in results_int.params and var_cs in results_int.params:
                base_effect = results_int.params[var]
                interaction_effect = results_int.params[var_cs]
                cs_effect = base_effect + interaction_effect

                # Get p-values
                p_base = results_int.pvalues.get(var, 1.0)
                p_int = results_int.pvalues.get(var_cs, 1.0)

                sig_base = "***" if p_base < 0.01 else "**" if p_base < 0.05 else "*" if p_base < 0.10 else ""
                sig_int = "***" if p_int < 0.01 else "**" if p_int < 0.05 else "*" if p_int < 0.10 else ""

                print(f"{base_var}:")
                print(f"  Base (all firms): {base_effect:.4f}{sig_base}")
                print(f"  CS interaction: {interaction_effect:.4f}{sig_int}")
                print(f"  CS total effect: {cs_effect:.4f}")
                if p_int < 0.10:
                    print(f"  → CS effect is SIGNIFICANTLY different from other sectors")
                else:
                    print(f"  → CS effect NOT significantly different from other sectors")

    # MODEL C: All Sectors with Sector Dummy (LAGGED)
    print(f"\n{'='*70}")
    print("ALL SECTORS WITH SECTOR DUMMY - LAGGED ESG")
    print(f"{'='*70}")
    print("Using ESG(t-1) to predict Tobin's Q(t)")

    df_dummy_clean = df.dropna(subset=['E_lag', 'S_lag', 'G_lag', 'ROA', 'Size', 'Leverage', 'Tobin_Q'])

    if len(df_dummy_clean) > 0:
        y_dummy = df_dummy_clean['Tobin_Q']
        X_dummy = df_dummy_clean[['E_lag', 'S_lag', 'G_lag', 'ROA', 'Size', 'Leverage', 'ConsumerStaples']]
        X_dummy = pd.concat([pd.Series(1, index=X_dummy.index, name='const'), X_dummy], axis=1)

        model_dummy = PanelOLS(y_dummy, X_dummy, entity_effects=False, time_effects=False)
        results_dummy = model_dummy.fit(cov_type='robust')

        results_dict['Sector Dummy (Lagged)'] = results_dummy

        print(f"Observations: {len(df_dummy_clean)}")
        print(f"Firms: {df_dummy_clean.index.get_level_values(0).nunique()}")
        print(f"R-squared: {results_dummy.rsquared:.4f}")
        cs_coeff = results_dummy.params.get('ConsumerStaples', 0)
        cs_pval = results_dummy.pvalues.get('ConsumerStaples', 1.0)
        sig = "***" if cs_pval < 0.01 else "**" if cs_pval < 0.05 else "*" if cs_pval < 0.10 else ""
        print(f"Consumer Staples dummy: {cs_coeff:.4f}{sig} (p={cs_pval:.3f})")
        if cs_pval < 0.10:
            print(f"→ Being a Consumer Staples firm has a {'positive' if cs_coeff > 0 else 'negative'} effect on Tobin's Q")

    return results_dict

# ============================================================================
# RUN YOUR ORIGINAL SECTOR ANALYSIS
# ============================================================================
print("\n" + "="*70)
print("ORIGINAL SECTOR ANALYSIS WITH LAGGED ESG")
print("="*70)

# Run sector interaction models with lagged ESG
sector_results = run_sector_interaction_models_lagged(df)

# Store sector results
for key, value in sector_results.items():
    all_model_results[key] = value

# ============================================================================
# COMPREHENSIVE SENSITIVITY ANALYSIS FOR ALL MODELS
# ============================================================================
print("\n" + "="*70)
print("COMPREHENSIVE SENSITIVITY ANALYSIS: ALL MODELS WITH LOG-TRANSFORMED TOBIN'S Q")
print("="*70)

# Create log-transformed Tobin's Q (if not already done)
if 'Tobin_Q_log' not in df.columns:
    df['Tobin_Q_log'] = np.log(df['Tobin_Q'] + 0.001)

# Store log-transformed results for all models
log_results_all = {}

print("\nRUNNING ALL ORIGINAL MODELS WITH LOG-TRANSFORMED TOBIN'S Q:")
print("="*70)

# ----------------------------------------------------------------------------
# MODEL 1A: 2-YEAR MOVING AVERAGES - CONSUMER STAPLES (LOG TRANSFORMED)
# ----------------------------------------------------------------------------
print("\n1. MODEL 1A: 2-YEAR MA OF LAGGED ESG (CS Only, Log Tobin's Q)")
print("-" * 60)

# Use the same data cleaning as original Model 1A
cs_data_ma2 = df[df['ConsumerStaples'] == 1].copy()
cs_clean_ma2 = cs_data_ma2.dropna(subset=['E_lag_ma2', 'S_lag_ma2', 'G_lag_ma2',
                                           'ROA', 'Size', 'Leverage', 'Tobin_Q_log'])

if len(cs_clean_ma2) >= 5:
    y_log = cs_clean_ma2['Tobin_Q_log']
    X_log = cs_clean_ma2[['E_lag_ma2', 'S_lag_ma2', 'G_lag_ma2', 'ROA', 'Size', 'Leverage']]
    X_log = pd.concat([pd.Series(1, index=X_log.index, name='const'), X_log], axis=1)

    model_log = PanelOLS(y_log, X_log, entity_effects=False, time_effects=False)
    results_log = model_log.fit(cov_type='robust')

    log_results_all['Model_1A_2Y_MA_CS_Log'] = results_log

    print(f"Observations: {results_log.nobs}")
    print(f"R-squared: {results_log.rsquared:.4f}")
    print("Key coefficients (ESG(t-1) → log(Tobin's Q)):")
    for var in ['E_lag_ma2', 'S_lag_ma2', 'G_lag_ma2']:
        if var in results_log.params:
            coeff = results_log.params[var]
            pval = results_log.pvalues.get(var, 1.0)
            sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
            print(f"  {var}: {coeff:.4f}{sig} (p={pval:.4f})")

# ----------------------------------------------------------------------------
# MODEL 1B: 3-YEAR MOVING AVERAGES - CONSUMER STAPLES (LOG TRANSFORMED)
# ----------------------------------------------------------------------------
print("\n2. MODEL 1B: 3-YEAR MA OF LAGGED ESG (CS Only, Log Tobin's Q)")
print("-" * 60)

cs_data_ma3 = df[df['ConsumerStaples'] == 1].copy()
cs_clean_ma3 = cs_data_ma3.dropna(subset=['E_lag_ma3', 'S_lag_ma3', 'G_lag_ma3',
                                           'ROA', 'Size', 'Leverage', 'Tobin_Q_log'])

if len(cs_clean_ma3) >= 5:
    y_log = cs_clean_ma3['Tobin_Q_log']
    X_log = cs_clean_ma3[['E_lag_ma3', 'S_lag_ma3', 'G_lag_ma3', 'ROA', 'Size', 'Leverage']]
    X_log = pd.concat([pd.Series(1, index=X_log.index, name='const'), X_log], axis=1)

    model_log = PanelOLS(y_log, X_log, entity_effects=False, time_effects=False)
    results_log = model_log.fit(cov_type='robust')

    log_results_all['Model_1B_3Y_MA_CS_Log'] = results_log

    print(f"Observations: {results_log.nobs}")
    print(f"R-squared: {results_log.rsquared:.4f}")
    print("Key coefficients (ESG(t-1) → log(Tobin's Q)):")
    for var in ['E_lag_ma3', 'S_lag_ma3', 'G_lag_ma3']:
        if var in results_log.params:
            coeff = results_log.params[var]
            pval = results_log.pvalues.get(var, 1.0)
            sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
            print(f"  {var}: {coeff:.4f}{sig} (p={pval:.4f})")

# ----------------------------------------------------------------------------
# MODEL 2: DISTRIBUTED LAG MODEL (1-5 YEARS) - CONSUMER STAPLES (LOG TRANSFORMED)
# ----------------------------------------------------------------------------
print("\n3. MODEL 2: DISTRIBUTED LAG (1-5 YEARS) - CONSUMER STAPLES (Log Tobin's Q)")
print("-" * 60)

# Prepare distributed lag data (same as original Model 2)
df_lags = df[df['ConsumerStaples'] == 1].copy()
for component in ['E', 'S', 'G']:
    for lag in range(1, 6):
        df_lags[f'{component}_lag{lag}'] = df_lags.groupby(level=0)[component].shift(lag)

lag_columns = [f'{c}_lag{l}' for c in ['E', 'S', 'G'] for l in range(1, 6)]
cs_lags_clean = df_lags.dropna(subset=lag_columns + ['ROA', 'Size', 'Leverage', 'Tobin_Q_log'])

if len(cs_lags_clean) >= 10:
    X_lags = lag_columns + ['ROA', 'Size', 'Leverage']
    X_dist_log = cs_lags_clean[X_lags]
    X_dist_log = pd.concat([pd.Series(1, index=X_dist_log.index, name='const'), X_dist_log], axis=1)
    y_dist_log = cs_lags_clean['Tobin_Q_log']

    model_dist_log = PanelOLS(y_dist_log, X_dist_log, entity_effects=False, time_effects=False)
    results_dist_log = model_dist_log.fit(cov_type='robust')

    log_results_all['Model_2_Distributed_Lag_CS_Log'] = results_dist_log

    print(f"Observations: {results_dist_log.nobs}")
    print(f"R-squared: {results_dist_log.rsquared:.4f}")

    # Find significant lagged coefficients
    print("\nSignificant lagged coefficients (p < 0.10):")
    for var in lag_columns:
        if var in results_dist_log.params:
            pval = results_dist_log.pvalues.get(var, 1.0)
            if pval < 0.10:
                coeff = results_dist_log.params[var]
                sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*"
                print(f"  {var}: {coeff:.4f}{sig} (p={pval:.4f})")

# ----------------------------------------------------------------------------
# MODEL 4: OPTIMAL DECAY RATE - CONSUMER STAPLES (LOG TRANSFORMED)
# ----------------------------------------------------------------------------
print("\n4. MODEL 4: OPTIMAL DECAY RATE - CONSUMER STAPLES (Log Tobin's Q)")
print("-" * 60)

# Recreate optimal decay variables for log model
df_opt_decay = df[df['ConsumerStaples'] == 1].copy()

# Apply the optimal decay rates you found earlier (0.9 for all)
optimal_decays_lagged = {'E_lag': 0.9, 'S_lag': 0.9, 'G_lag': 0.9}

for component, decay_rate in optimal_decays_lagged.items():
    decaying_avgs = []
    for firm in df_opt_decay.index.get_level_values(0).unique():
        firm_data = df_opt_decay.xs(firm, level=0)
        if len(firm_data) > 0:
            decaying_avg = create_decaying_avg(firm_data[component], decay_rate, max_lag=3)
            decaying_avgs.extend(decaying_avg)
        else:
            decaying_avgs.extend([np.nan] * len(firm_data))
    df_opt_decay[f'{component}_optdecay'] = decaying_avgs

opt_clean = df_opt_decay.dropna(subset=[f'{c}_optdecay' for c in ['E_lag', 'S_lag', 'G_lag']] +
                                 ['ROA', 'Size', 'Leverage', 'Tobin_Q_log'])

if len(opt_clean) >= 5:
    y_opt_log = opt_clean['Tobin_Q_log']
    X_opt_log = opt_clean[[f'{c}_optdecay' for c in ['E_lag', 'S_lag', 'G_lag']] + ['ROA', 'Size', 'Leverage']]
    X_opt_log = pd.concat([pd.Series(1, index=X_opt_log.index, name='const'), X_opt_log], axis=1)

    model_opt_log = PanelOLS(y_opt_log, X_opt_log, entity_effects=False, time_effects=False)
    results_opt_log = model_opt_log.fit(cov_type='robust')

    log_results_all['Model_4_Optimal_Decay_CS_Log'] = results_opt_log

    print(f"Observations: {results_opt_log.nobs}")
    print(f"R-squared: {results_opt_log.rsquared:.4f}")
    print("Key coefficients (ESG(t-1) with decay → log(Tobin's Q)):")
    for var in [f'{c}_optdecay' for c in ['E_lag', 'S_lag', 'G_lag']]:
        if var in results_opt_log.params:
            coeff = results_opt_log.params[var]
            pval = results_opt_log.pvalues.get(var, 1.0)
            sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
            print(f"  {var}: {coeff:.4f}{sig} (p={pval:.4f})")

# ----------------------------------------------------------------------------
# MODEL 5: ESG CAPITAL STOCK - CONSUMER STAPLES (LOG TRANSFORMED)
# ----------------------------------------------------------------------------
print("\n5. MODEL 5: ESG CAPITAL STOCK - CONSUMER STAPLES (Log Tobin's Q)")
print("-" * 60)

# Recreate ESG capital stock for log model
df_capital = df[df['ConsumerStaples'] == 1].copy()

for component in ['E_lag', 'S_lag', 'G_lag']:
    capital_series = []
    for firm in df_capital.index.get_level_values(0).unique():
        firm_data = df_capital.xs(firm, level=0)
        capital = create_cumulative(firm_data[component], depreciation=0.3)
        capital_series.extend(capital)
    df_capital[f'{component}_capital'] = capital_series

capital_clean = df_capital.dropna(subset=[f'{c}_capital' for c in ['E_lag', 'S_lag', 'G_lag']] +
                                   ['ROA', 'Size', 'Leverage', 'Tobin_Q_log'])

if len(capital_clean) >= 5:
    y_cap_log = capital_clean['Tobin_Q_log']
    X_cap_log = capital_clean[[f'{c}_capital' for c in ['E_lag', 'S_lag', 'G_lag']] + ['ROA', 'Size', 'Leverage']]
    X_cap_log = pd.concat([pd.Series(1, index=X_cap_log.index, name='const'), X_cap_log], axis=1)

    model_cap_log = PanelOLS(y_cap_log, X_cap_log, entity_effects=False, time_effects=False)
    results_cap_log = model_cap_log.fit(cov_type='robust')

    log_results_all['Model_5_ESG_Capital_CS_Log'] = results_cap_log

    print(f"Observations: {results_cap_log.nobs}")
    print(f"R-squared: {results_cap_log.rsquared:.4f}")
    print("Key coefficients (ESG capital → log(Tobin's Q)):")
    for var in [f'{c}_capital' for c in ['E_lag', 'S_lag', 'G_lag']]:
        if var in results_cap_log.params:
            coeff = results_cap_log.params[var]
            pval = results_cap_log.pvalues.get(var, 1.0)
            sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
            print(f"  {var}: {coeff:.4f}{sig} (p={pval:.4f})")

# ----------------------------------------------------------------------------
# MODEL 6: CHANGE IN LAGGED ESG - CONSUMER STAPLES (LOG TRANSFORMED)
# ----------------------------------------------------------------------------
print("\n6. MODEL 6: CHANGE IN LAGGED ESG - CONSUMER STAPLES (Log Tobin's Q)")
print("-" * 60)

# Recreate change variables for log model
df_change = df[df['ConsumerStaples'] == 1].copy()
for component in ['E_lag', 'S_lag', 'G_lag']:
    df_change[f'{component}_chg1'] = df_change.groupby(level=0)[component].diff(1)

change_clean = df_change.dropna(subset=['E_lag_chg1', 'S_lag_chg1', 'G_lag_chg1',
                                        'ROA', 'Size', 'Leverage', 'Tobin_Q_log'])

if len(change_clean) >= 5:
    y_chg_log = change_clean['Tobin_Q_log']
    X_chg_log = change_clean[['E_lag_chg1', 'S_lag_chg1', 'G_lag_chg1', 'ROA', 'Size', 'Leverage']]
    X_chg_log = pd.concat([pd.Series(1, index=X_chg_log.index, name='const'), X_chg_log], axis=1)

    model_chg_log = PanelOLS(y_chg_log, X_chg_log, entity_effects=False, time_effects=False)
    results_chg_log = model_chg_log.fit(cov_type='robust')

    log_results_all['Model_6_Change_ESG_CS_Log'] = results_chg_log

    print(f"Observations: {results_chg_log.nobs}")
    print(f"R-squared: {results_chg_log.rsquared:.4f}")
    print("Key coefficients (ΔESG(t-1) → log(Tobin's Q)):")
    for var in ['E_lag_chg1', 'S_lag_chg1', 'G_lag_chg1']:
        if var in results_chg_log.params:
            coeff = results_chg_log.params[var]
            pval = results_chg_log.pvalues.get(var, 1.0)
            sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
            print(f"  {var}: {coeff:.4f}{sig} (p={pval:.4f})")

# ----------------------------------------------------------------------------
# ALL SECTORS MODELS (LOG TRANSFORMED)
# ----------------------------------------------------------------------------
print("\n" + "="*70)
print("ALL SECTORS MODELS WITH LOG-TRANSFORMED TOBIN'S Q")
print("="*70)

# Simple lagged ESG for all sectors
print("\n7. ALL SECTORS: SIMPLE LAGGED ESG (Log Tobin's Q)")
print("-" * 60)

all_simple_clean = df.dropna(subset=['E_lag', 'S_lag', 'G_lag', 'ROA', 'Size',
                                      'Leverage', 'ConsumerStaples', 'Tobin_Q_log'])

if len(all_simple_clean) >= 10:
    y_all_log = all_simple_clean['Tobin_Q_log']
    X_all_log = all_simple_clean[['E_lag', 'S_lag', 'G_lag', 'ROA', 'Size', 'Leverage', 'ConsumerStaples']]
    X_all_log = pd.concat([pd.Series(1, index=X_all_log.index, name='const'), X_all_log], axis=1)

    model_all_log = PanelOLS(y_all_log, X_all_log, entity_effects=False, time_effects=False)
    results_all_log = model_all_log.fit(cov_type='robust')

    log_results_all['All_Sectors_Simple_Log'] = results_all_log

    print(f"Observations: {results_all_log.nobs}")
    print(f"R-squared: {results_all_log.rsquared:.4f}")
    print("Key coefficients (ESG(t-1) → log(Tobin's Q)):")
    for var in ['E_lag', 'S_lag', 'G_lag']:
        if var in results_all_log.params:
            coeff = results_all_log.params[var]
            pval = results_all_log.pvalues.get(var, 1.0)
            sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
            print(f"  {var}: {coeff:.4f}{sig} (p={pval:.4f})")

# 2-Year MA for all sectors
print("\n8. ALL SECTORS: 2-YEAR MA OF LAGGED ESG (Log Tobin's Q)")
print("-" * 60)

all_2y_clean = df.dropna(subset=['E_lag_ma2', 'S_lag_ma2', 'G_lag_ma2', 'ROA', 'Size',
                                  'Leverage', 'ConsumerStaples', 'Tobin_Q_log'])

if len(all_2y_clean) >= 10:
    y_all_log = all_2y_clean['Tobin_Q_log']
    X_all_log = all_2y_clean[['E_lag_ma2', 'S_lag_ma2', 'G_lag_ma2', 'ROA', 'Size', 'Leverage', 'ConsumerStaples']]
    X_all_log = pd.concat([pd.Series(1, index=X_all_log.index, name='const'), X_all_log], axis=1)

    model_all_log = PanelOLS(y_all_log, X_all_log, entity_effects=False, time_effects=False)
    results_all_log = model_all_log.fit(cov_type='robust')

    log_results_all['All_Sectors_2Y_MA_Log'] = results_all_log

    print(f"Observations: {results_all_log.nobs}")
    print(f"R-squared: {results_all_log.rsquared:.4f}")
    print("Key coefficients (ESG(t-1) → log(Tobin's Q)):")
    for var in ['E_lag_ma2', 'S_lag_ma2', 'G_lag_ma2']:
        if var in results_all_log.params:
            coeff = results_all_log.params[var]
            pval = results_all_log.pvalues.get(var, 1.0)
            sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
            print(f"  {var}: {coeff:.4f}{sig} (p={pval:.4f})")

# ============================================================================
# COMPLETE SENSITIVITY ANALYSIS FOR ALL MODELS
# ============================================================================
print("\n" + "="*70)
print("COMPLETE SENSITIVITY ANALYSIS: ALL MODELS WITH LOG-TRANSFORMED TOBIN'S Q")
print("="*70)

# Ensure log-transformed variable exists
if 'Tobin_Q_log' not in df.columns:
    df['Tobin_Q_log'] = np.log(df['Tobin_Q'] + 0.001)

# Store all sensitivity results
sensitivity_results = {}

print("\nRUNNING SENSITIVITY ANALYSIS FOR ALL MODELS...")
print("="*70)

# ============================================================================
# 1. ALL CONSUMER STAPLES MODELS
# ============================================================================
print("\n" + "="*70)
print("CONSUMER STAPLES MODELS SENSITIVITY ANALYSIS")
print("="*70)

# Filter Consumer Staples data
cs_data = df[df['ConsumerStaples'] == 1].copy()

# Model 1A: 2-Year Moving Average
print("\n1. MODEL 1A: 2-YEAR MOVING AVERAGE OF LAGGED ESG")
print("-" * 60)

cs_clean_ma2 = cs_data.dropna(subset=['E_lag_ma2', 'S_lag_ma2', 'G_lag_ma2',
                                       'ROA', 'Size', 'Leverage', 'Tobin_Q', 'Tobin_Q_log'])

if len(cs_clean_ma2) >= 5:
    # Original model
    y_orig = cs_clean_ma2['Tobin_Q']
    X = cs_clean_ma2[['E_lag_ma2', 'S_lag_ma2', 'G_lag_ma2', 'ROA', 'Size', 'Leverage']]
    X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

    model_orig = PanelOLS(y_orig, X, entity_effects=False, time_effects=False)
    results_orig = model_orig.fit(cov_type='robust')

    # Log-transformed model
    y_log = cs_clean_ma2['Tobin_Q_log']
    model_log = PanelOLS(y_log, X, entity_effects=False, time_effects=False)
    results_log = model_log.fit(cov_type='robust')

    sensitivity_results['CS_2Y_MA'] = {
        'original': results_orig,
        'log': results_log
    }

    print(f"Observations: {results_orig.nobs}")
    print(f"R² Original: {results_orig.rsquared:.4f}")
    print(f"R² Log: {results_log.rsquared:.4f}")

    # Check key coefficients
    social_robust = (results_orig.pvalues.get('S_lag_ma2', 1.0) < 0.10 and
                     results_log.pvalues.get('S_lag_ma2', 1.0) < 0.10 and
                     results_orig.params.get('S_lag_ma2', 0) * results_log.params.get('S_lag_ma2', 0) > 0)

    print(f"Social coefficient robust: {'✓' if social_robust else '✗'}")

# Model 1B: 3-Year Moving Average
print("\n2. MODEL 1B: 3-YEAR MOVING AVERAGE OF LAGGED ESG")
print("-" * 60)

cs_clean_ma3 = cs_data.dropna(subset=['E_lag_ma3', 'S_lag_ma3', 'G_lag_ma3',
                                       'ROA', 'Size', 'Leverage', 'Tobin_Q', 'Tobin_Q_log'])

if len(cs_clean_ma3) >= 5:
    # Original model
    y_orig = cs_clean_ma3['Tobin_Q']
    X = cs_clean_ma3[['E_lag_ma3', 'S_lag_ma3', 'G_lag_ma3', 'ROA', 'Size', 'Leverage']]
    X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

    model_orig = PanelOLS(y_orig, X, entity_effects=False, time_effects=False)
    results_orig = model_orig.fit(cov_type='robust')

    # Log-transformed model
    y_log = cs_clean_ma3['Tobin_Q_log']
    model_log = PanelOLS(y_log, X, entity_effects=False, time_effects=False)
    results_log = model_log.fit(cov_type='robust')

    sensitivity_results['CS_3Y_MA'] = {
        'original': results_orig,
        'log': results_log
    }

    print(f"Observations: {results_orig.nobs}")
    print(f"R² Original: {results_orig.rsquared:.4f}")
    print(f"R² Log: {results_log.rsquared:.4f}")

    social_robust = (results_orig.pvalues.get('S_lag_ma3', 1.0) < 0.10 and
                     results_log.pvalues.get('S_lag_ma3', 1.0) < 0.10 and
                     results_orig.params.get('S_lag_ma3', 0) * results_log.params.get('S_lag_ma3', 0) > 0)

    print(f"Social coefficient robust: {'✓' if social_robust else '✗'}")

# Model 2: Distributed Lag (1-5 years) - Simplified version
print("\n3. MODEL 2: DISTRIBUTED LAG (1-5 YEARS) - SIMPLIFIED")
print("-" * 60)

# Create lagged variables for distributed lag model
cs_lag_data = cs_data.copy()
lag_columns = []
for component in ['E', 'S', 'G']:
    for lag in range(1, 6):  # Lags 1-5
        cs_lag_data[f'{component}_lag{lag}'] = cs_lag_data.groupby(level=0)[component].shift(lag)
        lag_columns.append(f'{component}_lag{lag}')

cs_lag_clean = cs_lag_data.dropna(subset=lag_columns + ['ROA', 'Size', 'Leverage', 'Tobin_Q', 'Tobin_Q_log'])

if len(cs_lag_clean) >= 10:
    # Use only lag 1, 3, and 5 to avoid overfitting
    selected_lags = ['E_lag1', 'E_lag3', 'E_lag5', 'S_lag1', 'S_lag3', 'S_lag5', 'G_lag1', 'G_lag3', 'G_lag5']

    # Original model
    y_orig = cs_lag_clean['Tobin_Q']
    X = cs_lag_clean[selected_lags + ['ROA', 'Size', 'Leverage']]
    X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

    model_orig = PanelOLS(y_orig, X, entity_effects=False, time_effects=False)
    results_orig = model_orig.fit(cov_type='robust')

    # Log-transformed model
    y_log = cs_lag_clean['Tobin_Q_log']
    model_log = PanelOLS(y_log, X, entity_effects=False, time_effects=False)
    results_log = model_log.fit(cov_type='robust')

    sensitivity_results['CS_Distributed_Lag'] = {
        'original': results_orig,
        'log': results_log
    }

    print(f"Observations: {results_orig.nobs}")
    print(f"R² Original: {results_orig.rsquared:.4f}")
    print(f"R² Log: {results_log.rsquared:.4f}")

# Model 4: Optimal Decay Rate
print("\n4. MODEL 4: OPTIMAL DECAY RATE OF LAGGED ESG")
print("-" * 60)

# Recreate optimal decay variables
cs_decay_data = cs_data.copy()
optimal_decays = {'E_lag': 0.9, 'S_lag': 0.9, 'G_lag': 0.9}

for component, decay_rate in optimal_decays.items():
    decaying_avgs = []
    for firm in cs_decay_data.index.get_level_values(0).unique():
        firm_data = cs_decay_data.xs(firm, level=0)
        if len(firm_data) > 0:
            decaying_avg = create_decaying_avg(firm_data[component], decay_rate, max_lag=3)
            decaying_avgs.extend(decaying_avg)
        else:
            decaying_avgs.extend([np.nan] * len(firm_data))
    cs_decay_data[f'{component}_optdecay'] = decaying_avgs

cs_decay_clean = cs_decay_data.dropna(subset=['E_lag_optdecay', 'S_lag_optdecay', 'G_lag_optdecay',
                                              'ROA', 'Size', 'Leverage', 'Tobin_Q', 'Tobin_Q_log'])

if len(cs_decay_clean) >= 5:
    # Original model
    y_orig = cs_decay_clean['Tobin_Q']
    X = cs_decay_clean[['E_lag_optdecay', 'S_lag_optdecay', 'G_lag_optdecay', 'ROA', 'Size', 'Leverage']]
    X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

    model_orig = PanelOLS(y_orig, X, entity_effects=False, time_effects=False)
    results_orig = model_orig.fit(cov_type='robust')

    # Log-transformed model
    y_log = cs_decay_clean['Tobin_Q_log']
    model_log = PanelOLS(y_log, X, entity_effects=False, time_effects=False)
    results_log = model_log.fit(cov_type='robust')

    sensitivity_results['CS_Optimal_Decay'] = {
        'original': results_orig,
        'log': results_log
    }

    print(f"Observations: {results_orig.nobs}")
    print(f"R² Original: {results_orig.rsquared:.4f}")
    print(f"R² Log: {results_log.rsquared:.4f}")

# Model 5: ESG Capital Stock
print("\n5. MODEL 5: ESG CAPITAL STOCK FROM LAGGED ESG")
print("-" * 60)

# Recreate capital stock
cs_capital_data = cs_data.copy()
for component in ['E_lag', 'S_lag', 'G_lag']:
    capital_series = []
    for firm in cs_capital_data.index.get_level_values(0).unique():
        firm_data = cs_capital_data.xs(firm, level=0)
        capital = create_cumulative(firm_data[component], depreciation=0.3)
        capital_series.extend(capital)
    cs_capital_data[f'{component}_capital'] = capital_series

cs_capital_clean = cs_capital_data.dropna(subset=['E_lag_capital', 'S_lag_capital', 'G_lag_capital',
                                                  'ROA', 'Size', 'Leverage', 'Tobin_Q', 'Tobin_Q_log'])

if len(cs_capital_clean) >= 5:
    # Original model
    y_orig = cs_capital_clean['Tobin_Q']
    X = cs_capital_clean[['E_lag_capital', 'S_lag_capital', 'G_lag_capital', 'ROA', 'Size', 'Leverage']]
    X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

    model_orig = PanelOLS(y_orig, X, entity_effects=False, time_effects=False)
    results_orig = model_orig.fit(cov_type='robust')

    # Log-transformed model
    y_log = cs_capital_clean['Tobin_Q_log']
    model_log = PanelOLS(y_log, X, entity_effects=False, time_effects=False)
    results_log = model_log.fit(cov_type='robust')

    sensitivity_results['CS_ESG_Capital'] = {
        'original': results_orig,
        'log': results_log
    }

    print(f"Observations: {results_orig.nobs}")
    print(f"R² Original: {results_orig.rsquared:.4f}")
    print(f"R² Log: {results_log.rsquared:.4f}")

# Model 6: Change in Lagged ESG
print("\n6. MODEL 6: CHANGE IN LAGGED ESG")
print("-" * 60)

# Create change variables
cs_change_data = cs_data.copy()
for component in ['E_lag', 'S_lag', 'G_lag']:
    cs_change_data[f'{component}_chg1'] = cs_change_data.groupby(level=0)[component].diff(1)

cs_change_clean = cs_change_data.dropna(subset=['E_lag_chg1', 'S_lag_chg1', 'G_lag_chg1',
                                                'ROA', 'Size', 'Leverage', 'Tobin_Q', 'Tobin_Q_log'])

if len(cs_change_clean) >= 5:
    # Original model
    y_orig = cs_change_clean['Tobin_Q']
    X = cs_change_clean[['E_lag_chg1', 'S_lag_chg1', 'G_lag_chg1', 'ROA', 'Size', 'Leverage']]
    X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

    model_orig = PanelOLS(y_orig, X, entity_effects=False, time_effects=False)
    results_orig = model_orig.fit(cov_type='robust')

    # Log-transformed model
    y_log = cs_change_clean['Tobin_Q_log']
    model_log = PanelOLS(y_log, X, entity_effects=False, time_effects=False)
    results_log = model_log.fit(cov_type='robust')

    sensitivity_results['CS_Change_ESG'] = {
        'original': results_orig,
        'log': results_log
    }

    print(f"Observations: {results_orig.nobs}")
    print(f"R² Original: {results_orig.rsquared:.4f}")
    print(f"R² Log: {results_log.rsquared:.4f}")

# ============================================================================
# ALL 6 MODELS FOR ALL SECTORS (COMPLETE ANALYSIS)
# ============================================================================
print("\n" + "="*70)
print("COMPLETE ANALYSIS: ALL 6 MODELS FOR ALL SECTORS")
print("="*70)

# Store all sectors results for all 6 models
all_sectors_complete_results = {}

print("\nRunning all 6 model specifications for All Sectors...")
print("="*70)

# Helper function to run model for all sectors
def run_all_sectors_model(model_name, esg_vars, control_vars, special_notes=None):
    """Run a model specification for all sectors"""
    print(f"\n{model_name.upper()} - ALL SECTORS")
    print("-" * 60)

    # Clean data
    all_vars = esg_vars + control_vars + ['Tobin_Q', 'ConsumerStaples']
    df_clean = df.dropna(subset=all_vars)

    if len(df_clean) < 10:
        print(f"Insufficient data: {len(df_clean)} observations")
        return None

    # Run model
    y = df_clean['Tobin_Q']
    X = df_clean[esg_vars + control_vars + ['ConsumerStaples']]
    X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

    model = PanelOLS(y, X, entity_effects=False, time_effects=False)
    results = model.fit(cov_type='robust')

    print(f"Observations: {results.nobs}")
    print(f"Firms: {df_clean.index.get_level_values(0).nunique()}")
    print(f"Consumer Staples firms: {df_clean[df_clean['ConsumerStaples']==1].index.get_level_values(0).nunique()}")
    print(f"R-squared: {results.rsquared:.4f}")

    # Print key coefficients
    print("\nKey coefficients (ESG(t-1) → Tobin's Q(t)):")
    for var in esg_vars:
        if var in results.params:
            coeff = results.params[var]
            pval = results.pvalues.get(var, 1.0)
            sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
            print(f"  {var}: {coeff:.4f}{sig} (p={pval:.3f})")

    # CS Dummy significance
    if 'ConsumerStaples' in results.params:
        cs_coeff = results.params['ConsumerStaples']
        cs_pval = results.pvalues.get('ConsumerStaples', 1.0)
        sig = "***" if cs_pval < 0.01 else "**" if cs_pval < 0.05 else "*" if cs_pval < 0.10 else ""
        print(f"  ConsumerStaples: {cs_coeff:.4f}{sig} (p={cs_pval:.3f})")

    # Generate equation
    if special_notes:
        print(f"\nNotes: {special_notes}")

    return results

# ----------------------------------------------------------------------------
# 1. SIMPLE LAGGED ESG (All Sectors)
# ----------------------------------------------------------------------------
results_simple = run_all_sectors_model(
    "1. SIMPLE LAGGED ESG",
    ['E_lag', 'S_lag', 'G_lag'],
    ['ROA', 'Size', 'Leverage'],
    "Simple 1-year lag of ESG scores"
)
if results_simple:
    all_sectors_complete_results['Simple_Lagged_ESG'] = results_simple

# ----------------------------------------------------------------------------
# 2. 2-YEAR MOVING AVERAGE (All Sectors)
# ----------------------------------------------------------------------------
results_2y_ma = run_all_sectors_model(
    "2. 2-YEAR MOVING AVERAGE",
    ['E_lag_ma2', 'S_lag_ma2', 'G_lag_ma2'],
    ['ROA', 'Size', 'Leverage'],
    "2-year moving average of LAGGED ESG scores"
)
if results_2y_ma:
    all_sectors_complete_results['2Y_MA_Lagged_ESG'] = results_2y_ma

# ----------------------------------------------------------------------------
# 3. 3-YEAR MOVING AVERAGE (All Sectors)
# ----------------------------------------------------------------------------
results_3y_ma = run_all_sectors_model(
    "3. 3-YEAR MOVING AVERAGE",
    ['E_lag_ma3', 'S_lag_ma3', 'G_lag_ma3'],
    ['ROA', 'Size', 'Leverage'],
    "3-year moving average of LAGGED ESG scores"
)
if results_3y_ma:
    all_sectors_complete_results['3Y_MA_Lagged_ESG'] = results_3y_ma

# ----------------------------------------------------------------------------
# 4. DISTRIBUTED LAG (1-5 years) - All Sectors (Simplified)
# ----------------------------------------------------------------------------
print("\n4. DISTRIBUTED LAG (1-5 YEARS) - ALL SECTORS")
print("-" * 60)

# Create distributed lag variables for all sectors
df_all_lags = df.copy()
lag_columns_all = []

# Create lagged variables for all sectors
for component in ['E', 'S', 'G']:
    for lag in range(1, 6):  # Lags 1-5
        df_all_lags[f'{component}_lag{lag}'] = df_all_lags.groupby(level=0)[component].shift(lag)
        lag_columns_all.append(f'{component}_lag{lag}')

# Use only key lags to avoid overfitting
selected_lags_all = ['E_lag1', 'E_lag3', 'E_lag5', 'S_lag1', 'S_lag3', 'S_lag5', 'G_lag1', 'G_lag3', 'G_lag5']

# Clean data
all_lags_clean = df_all_lags.dropna(subset=selected_lags_all + ['ROA', 'Size', 'Leverage', 'ConsumerStaples', 'Tobin_Q'])

if len(all_lags_clean) >= 20:
    y_lag_all = all_lags_clean['Tobin_Q']
    X_lag_all = all_lags_clean[selected_lags_all + ['ROA', 'Size', 'Leverage', 'ConsumerStaples']]
    X_lag_all = pd.concat([pd.Series(1, index=X_lag_all.index, name='const'), X_lag_all], axis=1)

    model_lag_all = PanelOLS(y_lag_all, X_lag_all, entity_effects=False, time_effects=False)
    results_lag_all = model_lag_all.fit(cov_type='robust')

    print(f"Observations: {results_lag_all.nobs}")
    print(f"R-squared: {results_lag_all.rsquared:.4f}")

    # Find significant coefficients
    print("\nSignificant ESG coefficients (p < 0.10):")
    for var in selected_lags_all:
        if var in results_lag_all.params:
            pval = results_lag_all.pvalues.get(var, 1.0)
            if pval < 0.10:
                coeff = results_lag_all.params[var]
                sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*"
                print(f"  {var}: {coeff:.4f}{sig} (p={pval:.3f})")

    all_sectors_complete_results['Distributed_Lag'] = results_lag_all
else:
    print("Insufficient data for distributed lag model")

# ----------------------------------------------------------------------------
# 5. OPTIMAL DECAY RATE - All Sectors
# ----------------------------------------------------------------------------
print("\n5. OPTIMAL DECAY RATE - ALL SECTORS")
print("-" * 60)

# Apply optimal decay rates to all sectors
df_all_decay = df.copy()
optimal_decays = {'E_lag': 0.9, 'S_lag': 0.9, 'G_lag': 0.9}  # Using your optimal rates

for component, decay_rate in optimal_decays.items():
    decaying_avgs = []
    for firm in df_all_decay.index.get_level_values(0).unique():
        firm_data = df_all_decay.xs(firm, level=0)
        if len(firm_data) > 0:
            decaying_avg = create_decaying_avg(firm_data[component], decay_rate, max_lag=3)
            decaying_avgs.extend(decaying_avg)
        else:
            decaying_avgs.extend([np.nan] * len(firm_data))
    df_all_decay[f'{component}_optdecay'] = decaying_avgs

all_decay_clean = df_all_decay.dropna(subset=['E_lag_optdecay', 'S_lag_optdecay', 'G_lag_optdecay',
                                               'ROA', 'Size', 'Leverage', 'ConsumerStaples', 'Tobin_Q'])

if len(all_decay_clean) >= 20:
    y_decay_all = all_decay_clean['Tobin_Q']
    X_decay_all = all_decay_clean[['E_lag_optdecay', 'S_lag_optdecay', 'G_lag_optdecay',
                                   'ROA', 'Size', 'Leverage', 'ConsumerStaples']]
    X_decay_all = pd.concat([pd.Series(1, index=X_decay_all.index, name='const'), X_decay_all], axis=1)

    model_decay_all = PanelOLS(y_decay_all, X_decay_all, entity_effects=False, time_effects=False)
    results_decay_all = model_decay_all.fit(cov_type='robust')

    print(f"Observations: {results_decay_all.nobs}")
    print(f"R-squared: {results_decay_all.rsquared:.4f}")

    print("\nKey coefficients (ESG with decay → Tobin's Q):")
    for var in ['E_lag_optdecay', 'S_lag_optdecay', 'G_lag_optdecay']:
        if var in results_decay_all.params:
            coeff = results_decay_all.params[var]
            pval = results_decay_all.pvalues.get(var, 1.0)
            sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
            print(f"  {var}: {coeff:.4f}{sig} (p={pval:.3f})")

    all_sectors_complete_results['Optimal_Decay'] = results_decay_all
else:
    print("Insufficient data for optimal decay model")

# ----------------------------------------------------------------------------
# 6. ESG CAPITAL STOCK - All Sectors
# ----------------------------------------------------------------------------
print("\n6. ESG CAPITAL STOCK - ALL SECTORS")
print("-" * 60)

# Create ESG capital for all sectors
df_all_capital = df.copy()

for component in ['E_lag', 'S_lag', 'G_lag']:
    capital_series = []
    for firm in df_all_capital.index.get_level_values(0).unique():
        firm_data = df_all_capital.xs(firm, level=0)
        capital = create_cumulative(firm_data[component], depreciation=0.3)
        capital_series.extend(capital)
    df_all_capital[f'{component}_capital'] = capital_series

all_capital_clean = df_all_capital.dropna(subset=['E_lag_capital', 'S_lag_capital', 'G_lag_capital',
                                                   'ROA', 'Size', 'Leverage', 'ConsumerStaples', 'Tobin_Q'])

if len(all_capital_clean) >= 20:
    y_capital_all = all_capital_clean['Tobin_Q']
    X_capital_all = all_capital_clean[['E_lag_capital', 'S_lag_capital', 'G_lag_capital',
                                        'ROA', 'Size', 'Leverage', 'ConsumerStaples']]
    X_capital_all = pd.concat([pd.Series(1, index=X_capital_all.index, name='const'), X_capital_all], axis=1)

    model_capital_all = PanelOLS(y_capital_all, X_capital_all, entity_effects=False, time_effects=False)
    results_capital_all = model_capital_all.fit(cov_type='robust')

    print(f"Observations: {results_capital_all.nobs}")
    print(f"R-squared: {results_capital_all.rsquared:.4f}")

    print("\nKey coefficients (ESG capital → Tobin's Q):")
    for var in ['E_lag_capital', 'S_lag_capital', 'G_lag_capital']:
        if var in results_capital_all.params:
            coeff = results_capital_all.params[var]
            pval = results_capital_all.pvalues.get(var, 1.0)
            sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
            print(f"  {var}: {coeff:.4f}{sig} (p={pval:.3f})")

    all_sectors_complete_results['ESG_Capital'] = results_capital_all
else:
    print("Insufficient data for ESG capital model")

# ----------------------------------------------------------------------------
# 7. CHANGE IN ESG - All Sectors (Bonus)
# ----------------------------------------------------------------------------
print("\n7. CHANGE IN LAGGED ESG - ALL SECTORS")
print("-" * 60)

# Create change variables for all sectors
df_all_change = df.copy()
for component in ['E_lag', 'S_lag', 'G_lag']:
    df_all_change[f'{component}_chg1'] = df_all_change.groupby(level=0)[component].diff(1)

all_change_clean = df_all_change.dropna(subset=['E_lag_chg1', 'S_lag_chg1', 'G_lag_chg1',
                                                 'ROA', 'Size', 'Leverage', 'ConsumerStaples', 'Tobin_Q'])

if len(all_change_clean) >= 20:
    y_change_all = all_change_clean['Tobin_Q']
    X_change_all = all_change_clean[['E_lag_chg1', 'S_lag_chg1', 'G_lag_chg1',
                                      'ROA', 'Size', 'Leverage', 'ConsumerStaples']]
    X_change_all = pd.concat([pd.Series(1, index=X_change_all.index, name='const'), X_change_all], axis=1)

    model_change_all = PanelOLS(y_change_all, X_change_all, entity_effects=False, time_effects=False)
    results_change_all = model_change_all.fit(cov_type='robust')

    print(f"Observations: {results_change_all.nobs}")
    print(f"R-squared: {results_change_all.rsquared:.4f}")

    print("\nKey coefficients (ΔESG → Tobin's Q):")
    for var in ['E_lag_chg1', 'S_lag_chg1', 'G_lag_chg1']:
        if var in results_change_all.params:
            coeff = results_change_all.params[var]
            pval = results_change_all.pvalues.get(var, 1.0)
            sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
            print(f"  {var}: {coeff:.4f}{sig} (p={pval:.3f})")

    all_sectors_complete_results['Change_ESG'] = results_change_all
else:
    print("Insufficient data for change model")

# ============================================================================
# SUMMARY TABLE FOR ALL SECTORS MODELS
# ============================================================================
print("\n" + "="*80)
print("SUMMARY: ALL 6 MODELS FOR ALL SECTORS")
print("="*80)

print("\n" + "-"*120)
print(f"{'MODEL':<30} {'SOCIAL (S)':<15} {'ENVIRONMENTAL (E)':<20} {'GOVERNANCE (G)':<20} {'R²':<10} {'N':<10}")
print("-"*120)

for model_name, results in all_sectors_complete_results.items():
    # Extract key coefficients
    social_coeff = results.params.get('S_lag',
                     results.params.get('S_lag_ma2',
                     results.params.get('S_lag_ma3',
                     results.params.get('S_lag_optdecay',
                     results.params.get('S_lag_capital',
                     results.params.get('S_lag_chg1', 0))))))

    env_coeff = results.params.get('E_lag',
                  results.params.get('E_lag_ma2',
                  results.params.get('E_lag_ma3',
                  results.params.get('E_lag_optdecay',
                  results.params.get('E_lag_capital',
                  results.params.get('E_lag_chg1', 0))))))

    gov_coeff = results.params.get('G_lag',
                  results.params.get('G_lag_ma2',
                  results.params.get('G_lag_ma3',
                  results.params.get('G_lag_optdecay',
                  results.params.get('G_lag_capital',
                  results.params.get('G_lag_chg1', 0))))))

    # Format with significance
    def format_coeff(coeff, results, var_patterns):
        for pattern in var_patterns:
            for key in results.params.keys():
                if pattern in str(key):
                    pval = results.pvalues.get(key, 1.0)
                    sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                    return f"{coeff:.4f}{sig}"
        return f"{coeff:.4f}"

    social_str = format_coeff(social_coeff, results, ['S_lag', 'S_lag_ma', 'S_lag_opt', 'S_lag_cap', 'S_lag_chg'])
    env_str = format_coeff(env_coeff, results, ['E_lag', 'E_lag_ma', 'E_lag_opt', 'E_lag_cap', 'E_lag_chg'])
    gov_str = format_coeff(gov_coeff, results, ['G_lag', 'G_lag_ma', 'G_lag_opt', 'G_lag_cap', 'G_lag_chg'])

    print(f"{model_name:<30} {social_str:<15} {env_str:<20} {gov_str:<20} {results.rsquared:<10.4f} {results.nobs:<10}")

print("\n" + "="*80)
print("KEY FINDINGS FOR ALL SECTORS:")
print("="*80)
print("1. Social (S): Consistently positive across ALL models and sectors")
print("2. Environmental (E): Generally negative or neutral")
print("3. Governance (G): Neutral for most sectors, but NEGATIVE for Consumer Staples")
print("4. Consumer Staples dummy: Shows sector-specific effects")
print("5. Moving averages (2-year, 3-year) perform best")
print("="*80)

# ============================================================================
# COMPREHENSIVE ROBUSTNESS SUMMARY
# ============================================================================
print("\n" + "="*80)
print("COMPREHENSIVE ROBUSTNESS SUMMARY")
print("="*80)

print("\n" + "-"*100)
print(f"{'MODEL':<30} {'SOCIAL COEFFICIENT':<25} {'R² (Orig/Log)':<20} {'N':<8} {'ROBUSTNESS'}")
print(f"{'':<30} {'Original vs Log':<25} {'':<20} {'':<8} {'':<10}")
print("-"*100)

robust_models = []
partially_robust_models = []
not_robust_models = []

for model_name, results_dict in sensitivity_results.items():
    if 'original' in results_dict and 'log' in results_dict:
        orig = results_dict['original']
        log = results_dict['log']

        # Find social coefficient variable
        social_var = None
        social_vars = ['S_lag', 'S_lag_ma2', 'S_lag_ma3', 'S_lag_optdecay',
                      'S_lag_capital', 'S_lag_chg1']

        for var in social_vars:
            if var in orig.params:
                social_var = var
                break

        if social_var:
            # Original results
            orig_coeff = orig.params[social_var]
            orig_pval = orig.pvalues.get(social_var, 1.0)
            orig_sig = "***" if orig_pval < 0.01 else "**" if orig_pval < 0.05 else "*" if orig_pval < 0.10 else ""

            # Log results
            log_coeff = log.params[social_var]
            log_pval = log.pvalues.get(social_var, 1.0)
            log_sig = "***" if log_pval < 0.01 else "**" if log_pval < 0.05 else "*" if log_pval < 0.10 else ""

            # Format comparison
            orig_str = f"{orig_coeff:.4f}{orig_sig}"
            log_str = f"{log_coeff:.4f}{log_sig}"
            comparison = f"{orig_str} vs {log_str}"

            # Check robustness
            same_sign = (orig_coeff * log_coeff > 0)
            both_sig = (orig_pval < 0.10 and log_pval < 0.10)

            if both_sig and same_sign:
                robustness = "✓✓ FULLY ROBUST"
                robust_models.append(model_name)
            elif same_sign:
                robustness = "✓ PARTIALLY ROBUST (sign)"
                partially_robust_models.append(model_name)
            elif both_sig:
                robustness = "⚠ PARTIALLY ROBUST (sig)"
                partially_robust_models.append(model_name)
            else:
                robustness = "✗ NOT ROBUST"
                not_robust_models.append(model_name)

            # R² comparison
            r2_comparison = f"{orig.rsquared:.4f} / {log.rsquared:.4f}"

            print(f"{model_name:<30} {comparison:<25} {r2_comparison:<20} {orig.nobs:<8} {robustness}")

print("\n" + "="*80)
print("ROBUSTNESS ASSESSMENT")
print("="*80)

print(f"\n✓ FULLY ROBUST MODELS ({len(robust_models)}):")
for model in robust_models:
    print(f"  • {model}")

print(f"\n⚠ PARTIALLY ROBUST MODELS ({len(partially_robust_models)}):")
for model in partially_robust_models:
    print(f"  • {model}")

print(f"\n✗ NOT ROBUST MODELS ({len(not_robust_models)}):")
for model in not_robust_models:
    print(f"  • {model}")

# Calculate overall robustness
total_models = len(robust_models) + len(partially_robust_models) + len(not_robust_models)
if total_models > 0:
    robust_percentage = (len(robust_models) / total_models) * 100
    print(f"\nOVERALL ROBUSTNESS: {robust_percentage:.1f}% of models fully robust")

    if robust_percentage >= 70:
        print("✓ CONCLUSION: Findings are HIGHLY ROBUST to log transformation")
    elif robust_percentage >= 50:
        print("✓ CONCLUSION: Findings are REASONABLY ROBUST to log transformation")
    elif robust_percentage >= 30:
        print("⚠ CONCLUSION: Findings show MODERATE ROBUSTNESS to log transformation")
    else:
        print("⚠ CONCLUSION: Findings are SENSITIVE to log transformation - interpret with caution")

print("\n" + "="*80)
print("SENSITIVITY ANALYSIS COMPLETE")
print("="*80)
print("All 9 models analyzed with both original and log-transformed Tobin's Q")
print("="*80)

print("\n" + "="*70)
print("ANALYSIS COMPLETE")
print("="*70)
print("Enhanced analysis and original sector analysis both completed successfully!")
```
