```python
# standalone_lag_model_individual_only.py
"""
Enhanced Standalone Lag Model Analysis - Individual Lags Only
With comprehensive statistical tests and robustness checks
"""

# ============================================================================
# SETUP AND IMPORTS
# ============================================================================
from google.colab import drive
drive.mount('/content/drive')

!pip install -q linearmodels pandas numpy matplotlib seaborn statsmodels scipy openpyxl

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from linearmodels.panel import PanelOLS, RandomEffects
from scipy import stats
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant
import warnings
warnings.filterwarnings('ignore')

# ============================================================================
# CONFIGURATION
#============================================================================
class Config:
    """Configuration parameters with enhanced settings"""
    PROJECT_DIR = '/content/drive/MyDrive/Colab Notebooks/FTSE Data'
    DATA_PATH = os.path.join(PROJECT_DIR, 'FTSE data.xlsx')
    OUTPUT_DIR = os.path.join(PROJECT_DIR, 'lag_model_outputs_individual')

    # Analysis parameters
    ESG_COMPONENTS = ['E', 'S', 'G']
    CONTROL_VARS = ['ROA', 'Size', 'Leverage']
    WINSORIZE_PERCENTILES = [0.01, 0.99]  # Winsorize at 1% and 99%

    # Consumer Staples firms
    CONSUMER_STAPLES_FIRMS = [
        'Associated British Foods', 'British American Tobacco', 'Coca Cola HBC AG',
        'Diageo', 'J. Sainsbury', 'Reckitt Benckiser', 'Tesco', 'Unilever', 'Haleon Plc'
    ]

    # Statistical thresholds
    SIGNIFICANCE_LEVELS = {'***': 0.01, '**': 0.05, '*': 0.10}

# ============================================================================
# DATA LOADER - ENHANCED
# ============================================================================
class DataLoader:
    """Enhanced data loader with quality checks and multiple variable specifications"""

    def __init__(self, config):
        self.config = config
        self.df = None
        self.df_cs = None
        self.df_all = None
        self.data_quality_report = {}

    def load_and_prepare_data(self):
        """Load data with comprehensive preparation and quality checks"""
        print("Loading and preparing data...")

        # 1. Load data
        self.df = pd.read_excel(self.config.DATA_PATH)
        self.df = self.df.set_index(['Firm', 'Year']).sort_index()
        self.df = self.df.rename(columns={"Tobin's Q": "Tobin_Q"})

        # 2. Create variables
        self.df['Size'] = np.log(self.df['Total Assets'])
        self.df['Leverage'] = self.df['Total Liabilities'] / self.df['Total Assets']
        self.df['Tobin_Q_log'] = np.log(self.df['Tobin_Q'] + 0.001)

        # 3. Winsorize extreme values
        self.df = self.winsorize_data(self.df)

        # 4. Create all lagged variables
        self._create_all_lagged_variables()

        # 5. Create sector variable
        sectors = []
        for firm in self.df.index.get_level_values(0):
            if firm in self.config.CONSUMER_STAPLES_FIRMS:
                sectors.append('Consumer Staples')
            else:
                sectors.append('Other')

        self.df['Sector'] = sectors
        self.df['ConsumerStaples'] = (self.df['Sector'] == 'Consumer Staples').astype(int)

        # 6. Create datasets
        self.df_cs = self.df[self.df['ConsumerStaples'] == 1].copy()
        self.df_all = self.df.copy()

        # 7. Run quality checks
        self.run_quality_checks()

        print(f"\n✓ Total observations: {len(self.df)}")
        print(f"✓ Consumer Staples: {len(self.df_cs)} observations")
        print(f"✓ All sectors: {len(self.df_all)} observations")

        return self.df, self.df_cs, self.df_all

    def winsorize_data(self, df):
        """Winsorize continuous variables to handle outliers"""
        df_winsorized = df.copy()
        variables_to_winsorize = ['Tobin_Q', 'ROA', 'Leverage', 'E', 'S', 'G']

        for var in variables_to_winsorize:
            if var in df.columns:
                lower = df[var].quantile(self.config.WINSORIZE_PERCENTILES[0])
                upper = df[var].quantile(self.config.WINSORIZE_PERCENTILES[1])
                df_winsorized[var] = df[var].clip(lower, upper)
                print(f"  Winsorized {var}: [{lower:.4f}, {upper:.4f}]")

        return df_winsorized

    def _create_all_lagged_variables(self):
        """Create all lagged variables - individual lags only"""
        # First create simple one-period lags (for comparison models)
        for component in self.config.ESG_COMPONENTS:
            self.df[f'{component}_lag'] = self.df.groupby(level=0)[component].shift(1)

        # Create lagged controls
        for control in self.config.CONTROL_VARS:
            self.df[f'{control}_lag'] = self.df.groupby(level=0)[control].shift(1)

        # Create individual lags up to 5 periods
        for component in self.config.ESG_COMPONENTS:
            for lag in range(1, 6):  # Create up to 5 lags
                self.df[f'{component}_lag{lag}'] = self.df.groupby(level=0)[component].shift(lag)

    def run_quality_checks(self):
        """Comprehensive data quality checks"""
        print("\n" + "="*60)
        print("DATA QUALITY CHECKS")
        print("="*60)

        # Check missing values
        missing_pct = (self.df.isnull().sum() / len(self.df)) * 100
        self.data_quality_report['missing_pct'] = missing_pct[missing_pct > 0].to_dict()

        print("\n1. Missing Values (%):")
        for var, pct in self.data_quality_report['missing_pct'].items():
            if pct > 0:
                print(f"   {var}: {pct:.2f}%")

        # Check summary statistics
        print("\n2. Summary Statistics (after winsorization):")
        key_vars = ['Tobin_Q', 'ROA', 'Size', 'Leverage', 'E', 'S', 'G']
        summary = self.df[key_vars].describe().T[['mean', 'std', 'min', 'max']]
        print(summary.round(4))

        # Check for outliers
        print("\n3. Outlier Detection (Z-score > 3):")
        for var in key_vars:
            z_scores = np.abs((self.df[var] - self.df[var].mean()) / self.df[var].std())
            outliers = (z_scores > 3).sum()
            if outliers > 0:
                print(f"   {var}: {outliers} outliers ({outliers/len(self.df)*100:.2f}%)")

    def create_lagged_variables_for_model(self, df, min_lag=1, max_lag=5, use_original_controls=True):
        """
        Create lagged variables for specific model specification
        """
        df_copy = df.copy()

        # Ensure all lag variables exist
        for component in self.config.ESG_COMPONENTS:
            for lag in range(min_lag, max_lag + 1):
                if f'{component}_lag{lag}' not in df_copy.columns:
                    df_copy[f'{component}_lag{lag}'] = df_copy.groupby(level=0)[component].shift(lag)

        # Determine control variables
        if use_original_controls:
            control_vars = self.config.CONTROL_VARS
        else:
            control_vars = [f'{var}_lag' for var in self.config.CONTROL_VARS]

        return df_copy, control_vars

# ============================================================================
# LAG MODEL ANALYZER - INDIVIDUAL LAGS ONLY
# ============================================================================
class LagModelAnalyzer:
    """Enhanced analyzer with comprehensive statistical tests - Individual lags only"""

    def __init__(self, config):
        self.config = config
        self.loader = DataLoader(config)
        self.results = {}
        self.test_results = {}

    def run_analysis(self, max_lag=5):
        """Run comprehensive analysis with all specifications - individual lags only"""
        print("\n" + "="*80)
        print(f"ENHANCED LAG MODEL ANALYSIS (Individual Lags 1-{max_lag})")
        print("="*80)

        # Load data
        df, df_cs, df_all = self.loader.load_and_prepare_data()

        # Run models for different sectors
        sectors = [
            ('cs', df_cs, 'Consumer Staples Only'),
            ('all', df_all, 'All Sectors')
        ]

        for sector_code, sector_df, sector_name in sectors:
            print(f"\n{'='*60}")
            print(f"ANALYZING: {sector_name}")
            print(f"{'='*60}")

            # Create lagged variables
            df_lagged, control_vars = self.loader.create_lagged_variables_for_model(
                sector_df, min_lag=1, max_lag=max_lag, use_original_controls=True
            )

            # Run individual lag models (no summed models)
            self._run_pooled_ols_model(df_lagged, max_lag, sector_code, sector_name, control_vars)
            self._run_fixed_effects_model(df_lagged, max_lag, sector_code, sector_name, control_vars)
            self._run_random_effects_model(df_lagged, max_lag, sector_code, sector_name, control_vars)

    def _run_fixed_effects_model(self, df, max_lag, sector_code, sector_name, control_vars):
        """Run fixed effects panel model (entity effects) with individual lags"""
        print(f"\nFIXED EFFECTS MODEL (with entity effects)")

        # Prepare variables - individual lags only
        lag_vars = []
        for component in ['E', 'S', 'G']:
            for lag in range(1, max_lag + 1):
                lag_vars.append(f'{component}_lag{lag}')

        # Add ConsumerStaples dummy for all sectors (note: time-invariant will be dropped with FE)
        if sector_code == 'all':
            all_vars = lag_vars + control_vars + ['ConsumerStaples', 'Tobin_Q']
        else:
            all_vars = lag_vars + control_vars + ['Tobin_Q']

        # Clean data
        df_clean = df.dropna(subset=all_vars)

        if len(df_clean) < 10:
            print(f"⚠ Insufficient data: {len(df_clean)} observations")
            return None

        print(f"Observations: {len(df_clean)}")
        print(f"Firms: {df_clean.index.get_level_values(0).nunique()}")

        # Prepare regression
        y = df_clean['Tobin_Q']
        X_vars = lag_vars + control_vars
        if sector_code == 'all':
            X_vars = X_vars + ['ConsumerStaples']

        X = df_clean[X_vars]
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

        # Run fixed effects regression
        try:
            model = PanelOLS(y, X, entity_effects=True, time_effects=False)
            results = model.fit(cov_type='clustered', cluster_entity=True)

            # Store results
            key = f"fe_individual_{max_lag}_{sector_code}"
            self.results[key] = {
                'results': results,
                'type': 'fixed_effects_individual',
                'sector': sector_code,
                'max_lag': max_lag,
                'nobs': results.nobs,
                'r2': results.rsquared,
                'r2_within': results.rsquared_within if hasattr(results, 'rsquared_within') else None,
                'data': df_clean
            }

            # Print results
            self._print_results(results, f"Fixed Effects Model - {sector_name}")

            # Run additional tests
            self._run_diagnostic_tests(results, df_clean, X_vars, key)

            # Create plots
            self._create_lag_structure_plot(results, max_lag, f"Fixed Effects - {sector_name}")
            self._create_cumulative_effects_plot(results, max_lag, f"Fixed Effects - {sector_name}")

            return results

        except Exception as e:
            print(f"Error running fixed effects model: {e}")
            return None

    def _run_random_effects_model(self, df, max_lag, sector_code, sector_name, control_vars):
        """Run random effects panel model with individual lags"""
        print(f"\nRANDOM EFFECTS MODEL")

        # Prepare variables - individual lags only
        lag_vars = []
        for component in ['E', 'S', 'G']:
            for lag in range(1, max_lag + 1):
                lag_vars.append(f'{component}_lag{lag}')

        if sector_code == 'all':
            all_vars = lag_vars + control_vars + ['ConsumerStaples', 'Tobin_Q']
        else:
            all_vars = lag_vars + control_vars + ['Tobin_Q']

        df_clean = df.dropna(subset=all_vars)

        if len(df_clean) < 10:
            return None

        y = df_clean['Tobin_Q']
        X_vars = lag_vars + control_vars
        if sector_code == 'all':
            X_vars = X_vars + ['ConsumerStaples']

        X = df_clean[X_vars]
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

        try:
            model = RandomEffects(y, X)
            results = model.fit(cov_type='robust')

            key = f"re_individual_{max_lag}_{sector_code}"
            self.results[key] = {
                'results': results,
                'type': 'random_effects_individual',
                'sector': sector_code,
                'max_lag': max_lag,
                'nobs': results.nobs,
                'r2': results.rsquared,
                'data': df_clean
            }

            self._print_results(results, f"Random Effects Model - {sector_name}")

            # Hausman test (compare with FE)
            self._run_hausman_test(sector_code, max_lag)

            return results

        except Exception as e:
            print(f"Error running random effects model: {e}")
            return None

    def _run_pooled_ols_model(self, df, max_lag, sector_code, sector_name, control_vars):
        """Run pooled OLS model with individual lags"""
        print(f"\nPOOLED OLS MODEL (Individual Lags 1-{max_lag})")

        lag_vars = []
        for component in ['E', 'S', 'G']:
            for lag in range(1, max_lag + 1):
                lag_vars.append(f'{component}_lag{lag}')

        if sector_code == 'all':
            all_vars = lag_vars + control_vars + ['ConsumerStaples', 'Tobin_Q']
        else:
            all_vars = lag_vars + control_vars + ['Tobin_Q']

        df_clean = df.dropna(subset=all_vars)

        if len(df_clean) < 10:
            print(f"⚠ Insufficient data: {len(df_clean)} observations")
            return None

        print(f"Observations: {len(df_clean)}")
        print(f"Firms: {df_clean.index.get_level_values(0).nunique()}")

        y = df_clean['Tobin_Q']
        X_vars = lag_vars + control_vars
        if sector_code == 'all':
            X_vars = X_vars + ['ConsumerStaples']

        X = df_clean[X_vars]
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

        try:
            model = PanelOLS(y, X, entity_effects=False, time_effects=False)
            results = model.fit(cov_type='robust')

            key = f"pooled_individual_{max_lag}_{sector_code}"
            self.results[key] = {
                'results': results,
                'type': 'pooled_ols_individual',
                'sector': sector_code,
                'max_lag': max_lag,
                'nobs': results.nobs,
                'r2': results.rsquared,
                'data': df_clean
            }

            self._print_results(results, f"Pooled OLS - {sector_name}")
            self._check_multicollinearity(df_clean, X_vars, key)
            self._create_lag_structure_plot(results, max_lag, f"Pooled OLS - {sector_name}")
            self._create_cumulative_effects_plot(results, max_lag, f"Pooled OLS - {sector_name}")

            return results

        except Exception as e:
            print(f"Error running model: {e}")
            return None

    def _run_diagnostic_tests(self, results, df, X_vars, key):
        """Run comprehensive diagnostic tests"""
        print("\n" + "-"*40)
        print("DIAGNOSTIC TESTS")
        print("-"*40)

        tests = {}

        # 1. Joint significance test for all lags
        lag_vars = [v for v in X_vars if any(c in v for c in ['E_lag', 'S_lag', 'G_lag'])]
        if lag_vars:
            # F-test for joint significance
            f_stat = self._joint_significance_test(results, lag_vars)
            tests['joint_significance'] = f_stat
            print(f"Joint significance of all lags: F={f_stat['stat']:.4f}, p={f_stat['pval']:.4f}")

        # 2. Test for linear trend in lag coefficients
        for component in ['E', 'S', 'G']:
            comp_lags = [v for v in lag_vars if component in v]
            if len(comp_lags) >= 3:
                trend_test = self._test_lag_trend(results, comp_lags)
                tests[f'{component}_trend'] = trend_test
                print(f"{component} lag trend test: p={trend_test['pval']:.4f}")

        # 3. Normality of residuals
        residuals = results.resids
        jb_stat, jb_pval = stats.jarque_bera(residuals)
        tests['jarque_bera'] = {'stat': jb_stat, 'pval': jb_pval}
        print(f"Jarque-Bera normality test: stat={jb_stat:.4f}, p={jb_pval:.4f}")

        # 4. Heteroskedasticity test (simplified)
        if hasattr(results, 'wald_test'):
            # Breusch-Pagan style test
            squared_resid = residuals**2
            bp_stat = len(residuals) * (squared_resid.corr(df[X_vars].mean(axis=1))**2)
            tests['heteroskedasticity'] = {'stat': bp_stat, 'pval': 1 - stats.chi2.cdf(bp_stat, 1)}
            print(f"Heteroskedasticity test: stat={bp_stat:.4f}, p={tests['heteroskedasticity']['pval']:.4f}")

        self.test_results[key] = tests

    def _run_hausman_test(self, sector_code, max_lag):
        """Run Hausman test to compare FE vs RE"""
        fe_key = f"fe_individual_{max_lag}_{sector_code}"
        re_key = f"re_individual_{max_lag}_{sector_code}"

        if fe_key in self.results and re_key in self.results:
            fe_results = self.results[fe_key]['results']
            re_results = self.results[re_key]['results']

            # Get common variables
            common_vars = set(fe_results.params.index) & set(re_results.params.index)
            common_vars = [v for v in common_vars if v != 'const']

            if len(common_vars) > 0:
                # Calculate Hausman statistic
                b_diff = fe_results.params[common_vars] - re_results.params[common_vars]

                # Get covariance matrices
                try:
                    var_diff = fe_results.cov[common_vars].loc[common_vars, common_vars] - \
                              re_results.cov[common_vars].loc[common_vars, common_vars]

                    # Handle non-positive definite matrix
                    eigvals = np.linalg.eigvals(var_diff)
                    if np.any(eigvals < 0):
                        print("\nHausman test: Variance matrix not positive definite")
                        return

                    # Calculate statistic
                    h_stat = b_diff @ np.linalg.inv(var_diff) @ b_diff
                    h_pval = 1 - stats.chi2.cdf(h_stat, len(common_vars))

                    print("\n" + "-"*40)
                    print("HAUSMAN TEST (FE vs RE)")
                    print("-"*40)
                    print(f"Chi-square: {h_stat:.4f}")
                    print(f"df: {len(common_vars)}")
                    print(f"p-value: {h_pval:.4f}")

                    if h_pval < 0.05:
                        print("Conclusion: Reject RE - FE is preferred")
                    else:
                        print("Conclusion: Cannot reject RE - both are consistent")

                except Exception as e:
                    print(f"Hausman test failed: {e}")

    def _check_multicollinearity(self, df, X_vars, key):
        """Check for multicollinearity using VIF"""
        print("\n" + "-"*40)
        print("MULTICOLLINEARITY CHECK (VIF)")
        print("-"*40)

        X = df[X_vars].copy()
        X = add_constant(X)

        vif_data = pd.DataFrame()
        vif_data["Variable"] = X.columns
        vif_data["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]

        print(vif_data.round(2))

        # Flag high VIF
        high_vif = vif_data[vif_data["VIF"] > 10]
        if len(high_vif) > 0:
            print("\n⚠ High multicollinearity detected (VIF > 10):")
            print(high_vif)

    def _joint_significance_test(self, results, lag_vars):
        """Test joint significance of lag variables"""
        # Simplified F-test
        r2_full = results.rsquared

        # Run restricted model (without lags)
        # This is a simplification - in practice you'd need to re-run the model
        f_stat = (r2_full / len(lag_vars)) / ((1 - r2_full) / (results.nobs - len(lag_vars) - 1))
        p_val = 1 - stats.f.cdf(f_stat, len(lag_vars), results.nobs - len(lag_vars) - 1)

        return {'stat': f_stat, 'pval': p_val}

    def _test_lag_trend(self, results, lag_vars):
        """Test for linear trend in lag coefficients"""
        if len(lag_vars) < 3:
            return {'pval': 1.0}

        # Get coefficients in order
        coeffs = [results.params[v] for v in lag_vars]

        # Simple linear regression of coefficients on lag number
        lags = list(range(1, len(coeffs) + 1))
        slope, intercept, r_value, p_value, std_err = stats.linregress(lags, coeffs)

        return {'slope': slope, 'pval': p_value}

    def _print_results(self, results, title):
        """Print formatted results with significance stars"""
        print(f"\n{title}")
        print("=" * 60)
        print(f"R-squared: {results.rsquared:.4f}")
        if hasattr(results, 'rsquared_within') and results.rsquared_within:
            print(f"R-squared (within): {results.rsquared_within:.4f}")
        print(f"Observations: {results.nobs}")
        if hasattr(results, 'entity_info'):
            print(f"Firms: {results.entity_info.get('n_entities', 'N/A')}")

        if hasattr(results, 'f_statistic'):
            print(f"F-statistic: {results.f_statistic.stat:.4f} (p={results.f_statistic.pval:.4f})")

        # Print coefficients with significance
        print("\nCoefficients:")
        print("-" * 60)

        # Group by component
        all_vars = results.params.index
        const_vars = [v for v in all_vars if v == 'const']
        e_vars = [v for v in all_vars if 'E' in str(v) and v != 'const']
        s_vars = [v for v in all_vars if 'S' in str(v) and v != 'const']
        g_vars = [v for v in all_vars if 'G' in str(v) and v != 'const']
        control_vars = [v for v in all_vars if v not in const_vars + e_vars + s_vars + g_vars]

        # Print constant
        if const_vars:
            for var in const_vars:
                self._print_coef(results, var)

        # Print ESG variables
        if e_vars:
            print("\nE Component:")
            for var in sorted(e_vars):
                self._print_coef(results, var)

        if s_vars:
            print("\nS Component:")
            for var in sorted(s_vars):
                self._print_coef(results, var)

        if g_vars:
            print("\nG Component:")
            for var in sorted(g_vars):
                self._print_coef(results, var)

        # Print controls
        if control_vars:
            print("\nControl Variables:")
            for var in sorted(control_vars):
                self._print_coef(results, var)

    def _print_coef(self, results, var):
        """Print a single coefficient with significance"""
        coeff = results.params[var]
        pval = results.pvalues.get(var, 1.0)
        std_err = results.std_errors.get(var, np.nan) if hasattr(results, 'std_errors') else np.nan

        # Determine significance
        sig = ""
        for stars, level in self.config.SIGNIFICANCE_LEVELS.items():
            if pval < level:
                sig = stars
                break

        print(f"  {var:<20} {coeff:>10.4f}{sig:<3} (p={pval:.4f})" +
              (f" [SE={std_err:.4f}]" if not np.isnan(std_err) else ""))

    def _create_lag_structure_plot(self, results, max_lag, title):
        """Create lag structure plot with confidence intervals"""
        fig, axes = plt.subplots(1, 3, figsize=(15, 5))
        fig.suptitle(f'Lag Structure: {title}', fontsize=14, fontweight='bold')

        for idx, component in enumerate(['E', 'S', 'G']):
            lags = list(range(1, max_lag + 1))
            coeffs = []
            cis = []  # Confidence intervals

            for lag in lags:
                var = f'{component}_lag{lag}'
                if var in results.params:
                    coeff = results.params[var]
                    coeffs.append(coeff)

                    # Get confidence interval
                    if hasattr(results, 'conf_int'):
                        ci = results.conf_int().loc[var]
                        cis.append((ci['lower'], ci['upper']))
                    else:
                        cis.append((coeff - 1.96*0.1, coeff + 1.96*0.1))  # Approximate
                else:
                    coeffs.append(0)
                    cis.append((0, 0))

            # Plot with confidence bands
            lags_array = np.array(lags)
            coeffs_array = np.array(coeffs)
            ci_lower = np.array([ci[0] for ci in cis])
            ci_upper = np.array([ci[1] for ci in cis])

            axes[idx].fill_between(lags_array, ci_lower, ci_upper, alpha=0.2, color='blue')
            axes[idx].plot(lags_array, coeffs_array, 'o-', linewidth=2, markersize=8, color='blue')
            axes[idx].axhline(y=0, color='red', linestyle='--', alpha=0.5)
            axes[idx].axhline(y=coeffs_array.mean(), color='green', linestyle=':', alpha=0.5,
                            label=f'Mean: {coeffs_array.mean():.4f}')

            axes[idx].set_xlabel('Years Lag', fontsize=11)
            axes[idx].set_ylabel('Coefficient', fontsize=11)
            axes[idx].set_title(f'{component} Component', fontsize=12, fontweight='bold')
            axes[idx].set_xticks(lags)
            axes[idx].grid(True, alpha=0.3)
            axes[idx].legend(loc='best', fontsize=9)

        plt.tight_layout()

        # Save plot
        os.makedirs(self.config.OUTPUT_DIR, exist_ok=True)
        plot_filename = f"lag_structure_{title.replace(' ', '_').replace('-', '_').replace('>', '')}.png"
        plt.savefig(os.path.join(self.config.OUTPUT_DIR, plot_filename),
                   dpi=300, bbox_inches='tight')
        plt.show()

        return fig

    def _create_cumulative_effects_plot(self, results, max_lag, title):
        """Create cumulative effects plot"""
        fig, ax = plt.subplots(figsize=(10, 6))
        fig.suptitle(f'Cumulative ESG Effects: {title}', fontsize=14, fontweight='bold')

        for component in ['E', 'S', 'G']:
            cumulative = []
            cumulative_se = []

            for lag in range(1, max_lag + 1):
                # Sum coefficients up to current lag
                sum_coeff = 0
                sum_se = 0
                for l in range(1, lag + 1):
                    var = f'{component}_lag{l}'
                    if var in results.params:
                        sum_coeff += results.params[var]
                        if hasattr(results, 'std_errors') and var in results.std_errors:
                            sum_se += results.std_errors[var]**2

                cumulative.append(sum_coeff)
                cumulative_se.append(np.sqrt(sum_se) if sum_se > 0 else 0)

            # Plot cumulative effect with confidence bands
            lags = range(1, max_lag + 1)
            cumulative = np.array(cumulative)
            cumulative_se = np.array(cumulative_se)

            ax.fill_between(lags,
                          cumulative - 1.96*cumulative_se,
                          cumulative + 1.96*cumulative_se,
                          alpha=0.2)
            ax.plot(lags, cumulative, 'o-', linewidth=2, markersize=8,
                   label=f'{component} (final: {cumulative[-1]:.4f})')

        ax.axhline(y=0, color='red', linestyle='--', alpha=0.5)
        ax.set_xlabel('Lag Length', fontsize=12)
        ax.set_ylabel('Cumulative Effect', fontsize=12)
        ax.set_title('Cumulative Effect Over Time', fontsize=12, fontweight='bold')
        ax.legend(loc='best')
        ax.grid(True, alpha=0.3)

        plt.tight_layout()

        # Save plot
        plot_filename = f"cumulative_effects_{title.replace(' ', '_').replace('-', '_').replace('>', '')}.png"
        plt.savefig(os.path.join(self.config.OUTPUT_DIR, plot_filename),
                   dpi=300, bbox_inches='tight')
        plt.show()

        return fig

    def compare_with_original_simple(self, max_lag=5):
        """Run simple lag model for comparison"""
        print("\n" + "="*80)
        print("SIMPLE LAG MODEL (For Comparison)")
        print("="*80)

        # Load data
        df, df_cs, df_all = self.loader.load_and_prepare_data()

        # For Consumer Staples
        print("\nCONSUMER STAPLES ONLY (Simple t-1 model):")

        # Use simple approach: just E_lag, S_lag, G_lag
        simple_vars = ['E_lag', 'S_lag', 'G_lag'] + self.config.CONTROL_VARS + ['Tobin_Q']
        df_simple = df_cs.dropna(subset=simple_vars)

        if len(df_simple) > 0:
            y = df_simple['Tobin_Q']
            X = df_simple[['E_lag', 'S_lag', 'G_lag'] + self.config.CONTROL_VARS]
            X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

            # Try with and without fixed effects
            for fe in [False, True]:
                print(f"\n{'Fixed Effects' if fe else 'Pooled OLS'} Model:")
                model = PanelOLS(y, X, entity_effects=fe, time_effects=False)
                results = model.fit(cov_type='clustered' if fe else 'robust',
                                  cluster_entity=fe)

                print(f"Observations: {results.nobs}")
                print(f"R-squared: {results.rsquared:.4f}")
                if fe and hasattr(results, 'rsquared_within'):
                    print(f"R-squared (within): {results.rsquared_within:.4f}")

                for var in ['E_lag', 'S_lag', 'G_lag'] + self.config.CONTROL_VARS:
                    if var in results.params:
                        coeff = results.params[var]
                        pval = results.pvalues.get(var, 1.0)
                        sig = ""
                        for stars, level in self.config.SIGNIFICANCE_LEVELS.items():
                            if pval < level:
                                sig = stars
                                break
                        print(f"  {var}: {coeff:.4f}{sig} (p={pval:.4f})")

        return results

# ============================================================================
# RESULTS COMPARATOR - ENHANCED
# ============================================================================
class ResultsComparator:
    """Enhanced results comparator with statistical comparisons - Individual models only"""

    def __init__(self, analyzer):
        self.analyzer = analyzer
        self.comparison_df = None
        self.summary_stats = {}

    def create_comparison_table(self):
        """Create comprehensive comparison table - individual models only"""
        comparison_data = []

        for key, model_info in self.analyzer.results.items():
            results = model_info['results']

            row = {
                'Model_ID': key,
                'Model_Type': model_info['type'],
                'Sector': model_info['sector'],
                'Max_Lag': model_info['max_lag'],
                'N': results.nobs,
                'R2': results.rsquared,
                'R2_Adj': getattr(results, 'rsquared_adjusted', results.rsquared),
                'AIC': self._calculate_aic(results),
                'BIC': self._calculate_bic(results),
            }

            # Add within R2 for FE models
            if 'fixed_effects' in model_info['type'] and hasattr(results, 'rsquared_within'):
                row['R2_Within'] = results.rsquared_within

            # Extract key coefficients and their significance
            for component in ['E', 'S', 'G']:
                # Get all lag coefficients for this component
                coeffs = []
                pvals = []
                for var in results.params.index:
                    if component in str(var) and 'lag' in str(var):
                        coeffs.append(results.params[var])
                        if var in results.pvalues:
                            pvals.append(results.pvalues[var])

                if coeffs:
                    row[f'{component}_avg'] = np.mean(coeffs)
                    row[f'{component}_sum'] = np.sum(coeffs)
                    row[f'{component}_min'] = np.min(coeffs)
                    row[f'{component}_max'] = np.max(coeffs)
                    row[f'{component}_n_sig'] = sum(1 for p in pvals if p < 0.05)

            # Add diagnostic test results
            if key in self.analyzer.test_results:
                for test_name, test_result in self.analyzer.test_results[key].items():
                    if isinstance(test_result, dict) and 'pval' in test_result:
                        row[f'test_{test_name}_pval'] = test_result['pval']

            comparison_data.append(row)

        self.comparison_df = pd.DataFrame(comparison_data)
        return self.comparison_df

    def _calculate_aic(self, results):
        """Calculate AIC"""
        try:
            k = len(results.params)
            return -2 * results.loglik + 2 * k
        except:
            return np.nan

    def _calculate_bic(self, results):
        """Calculate BIC"""
        try:
            k = len(results.params)
            return -2 * results.loglik + k * np.log(results.nobs)
        except:
            return np.nan

    def print_comparison(self):
        """Print enhanced comparison of results"""
        if self.comparison_df is None:
            self.create_comparison_table()

        print("\n" + "="*80)
        print("MODEL COMPARISON SUMMARY - INDIVIDUAL LAGS ONLY")
        print("="*80)

        # Sort by R2
        comparison_sorted = self.comparison_df.sort_values('R2', ascending=False)

        # Print best models
        print("\nTop 5 Models by R²:")
        print("-" * 60)
        top_models = comparison_sorted.head(5)
        for _, row in top_models.iterrows():
            print(f"{row['Model_Type']:<30} | R²={row['R2']:.4f} | N={row['N']} | " +
                  f"AIC={row['AIC']:.1f}" if not pd.isna(row['AIC']) else "")

        # Sector comparison
        print("\n" + "="*80)
        print("SECTOR COMPARISON (Fixed Effects Models)")
        print("="*80)

        fe_models = self.comparison_df[self.comparison_df['Model_Type'].str.contains('fixed_effects')]
        if len(fe_models) > 0:
            for sector in ['cs', 'all']:
                sector_models = fe_models[fe_models['Sector'] == sector]
                if len(sector_models) > 0:
                    print(f"\n{sector.upper()}:")
                    best = sector_models.loc[sector_models['R2'].idxmax()]
                    print(f"  Best R²: {best['R2']:.4f} (Model: {best['Model_Type']})")

                    # Average ESG effects
                    print("  Average ESG effects:")
                    for comp in ['E', 'S', 'G']:
                        if f'{comp}_avg' in best and not pd.isna(best[f'{comp}_avg']):
                            print(f"    {comp}: {best[f'{comp}_avg']:.4f} " +
                                  f"[{best[f'{comp}_min']:.4f}, {best[f'{comp}_max']:.4f}]")

        # Model type comparison
        print("\n" + "="*80)
        print("MODEL TYPE COMPARISON")
        print("="*80)

        model_types = self.comparison_df.groupby('Model_Type').agg({
            'R2': ['mean', 'max', 'min'],
            'N': 'mean',
            'AIC': 'mean'
        }).round(4)
        print(model_types)

    def export_results(self, filename='lag_model_results_individual.xlsx'):
        """Export all results to Excel with multiple sheets"""
        if self.comparison_df is None:
            self.create_comparison_table()

        os.makedirs(self.analyzer.config.OUTPUT_DIR, exist_ok=True)
        filepath = os.path.join(self.analyzer.config.OUTPUT_DIR, filename)

        with pd.ExcelWriter(filepath, engine='openpyxl') as writer:
            # Main comparison
            self.comparison_df.to_excel(writer, sheet_name='Model_Comparison', index=False)

            # Detailed results for each model
            for key, model_info in self.analyzer.results.items():
                results = model_info['results']

                # Create detailed results DataFrame
                params_df = pd.DataFrame({
                    'Coefficient': results.params,
                    'Std.Error': results.std_errors if hasattr(results, 'std_errors') else np.nan,
                    't-stat': results.params / results.std_errors if hasattr(results, 'std_errors') else np.nan,
                    'P-value': results.pvalues,
                    'CI_lower': results.conf_int()['lower'] if hasattr(results, 'conf_int') else np.nan,
                    'CI_upper': results.conf_int()['upper'] if hasattr(results, 'conf_int') else np.nan
                })

                # Add significance stars
                params_df['Significance'] = ''
                for idx, row in params_df.iterrows():
                    pval = row['P-value']
                    if pval < 0.01:
                        params_df.loc[idx, 'Significance'] = '***'
                    elif pval < 0.05:
                        params_df.loc[idx, 'Significance'] = '**'
                    elif pval < 0.10:
                        params_df.loc[idx, 'Significance'] = '*'

                sheet_name = key[:30]  # Excel sheet name limit
                params_df.to_excel(writer, sheet_name=sheet_name)

            # Diagnostic tests
            if self.analyzer.test_results:
                tests_data = []
                for key, tests in self.analyzer.test_results.items():
                    for test_name, test_result in tests.items():
                        if isinstance(test_result, dict):
                            row = {'Model': key, 'Test': test_name}
                            row.update(test_result)
                            tests_data.append(row)

                if tests_data:
                    tests_df = pd.DataFrame(tests_data)
                    tests_df.to_excel(writer, sheet_name='Diagnostic_Tests', index=False)

            # Summary statistics
            summary = self.comparison_df.describe()
            summary.to_excel(writer, sheet_name='Summary_Stats')

        print(f"\n✓ Results exported to: {filepath}")
        return filepath

# ============================================================================
# MAIN EXECUTION - INDIVIDUAL LAGS ONLY
# ============================================================================
def main():
    """Enhanced main execution function - Individual lags only"""
    print("\n" + "="*80)
    print("ENHANCED LAG MODEL ANALYSIS - INDIVIDUAL LAGS ONLY")
    print("="*80)
    print("\nThis analysis examines the distributed lag relationship between")
    print("ESG scores and financial performance (Tobin's Q) using individual lags")

    # Initialize
    config = Config()
    os.makedirs(config.OUTPUT_DIR, exist_ok=True)

    analyzer = LagModelAnalyzer(config)

    # Run analysis with different lag specifications
    lag_ranges = [2, 3, 4, 5]  # Test different lag lengths

    for max_lag in lag_ranges:
        print(f"\n{'#'*80}")
        print(f"ANALYZING LAG RANGE: 1-{max_lag} years")
        print(f"{'#'*80}")

        analyzer.run_analysis(max_lag=max_lag)

    # Compare results
    comparator = ResultsComparator(analyzer)
    comparator.create_comparison_table()
    comparator.print_comparison()

    # Export results
    comparator.export_results()

    # Run simple model for comparison
    analyzer.compare_with_original_simple(max_lag=5)

    print("\n" + "="*80)
    print("ANALYSIS COMPLETE")
    print("="*80)
    print(f"\nAll outputs saved to: {config.OUTPUT_DIR}")

    return analyzer, comparator

# ============================================================================
# EXECUTION
# ============================================================================
if __name__ == "__main__":
    analyzer, comparator = main()
```
