```python
# This cell installs and imports required libraries for network design and mapping.
# We separate installs to keep things clean and fast.

try:
    import pandas as pd
    import numpy as np
    import matplotlib.pyplot as plt
    import seaborn as sns
    from tqdm import tqdm
    import geopandas as gpd
    from shapely.geometry import Point
    from sklearn.neighbors import BallTree
    import contextily as ctx
except Exception as e:
    pass

# Try importing again and install missing packages if needed
missing = []
try:
    import pandas as pd
except Exception as e:
    missing.append("pandas")
try:
    import numpy as np
except Exception as e:
    missing.append("numpy")
try:
    import matplotlib.pyplot as plt
except Exception as e:
    missing.append("matplotlib")
try:
    import seaborn as sns
except Exception as e:
    missing.append("seaborn")
try:
    from tqdm import tqdm
except Exception as e:
    missing.append("tqdm")
try:
    import geopandas as gpd
except Exception as e:
    missing.append("geopandas")
try:
    from shapely.geometry import Point
except Exception as e:
    missing.append("shapely")
try:
    from sklearn.neighbors import BallTree
except Exception as e:
    missing.append("scikit-learn")
try:
    import contextily as ctx
except Exception as e:
    missing.append("contextily")

if len(missing) > 0:
    import sys
    for pkg in missing:
        print("Installing package " + pkg)
        if pkg == "shapely":
            # shapely comes with geopandas install often, but ensure
            pass
        %pip install --quiet --no-input  pandas numpy matplotlib seaborn tqdm geopandas shapely scikit-learn contextily
        break

# Final imports after install
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
import geopandas as gpd
from shapely.geometry import Point
from sklearn.neighbors import BallTree
import contextily as ctx

# Basic plotting style
sns.set(style="whitegrid")
print(pd.__version__)
print(np.__version__)
print(gpd.__version__)
print("Imports ready")

# Install OR-Tools for optimization if not present
try:
    import ortools
except Exception as e:
    %pip install ortools --quiet --no-input

from ortools.linear_solver import pywraplp
print("ortools ready")

# Create a comprehensive supply chain optimization platform for African health facilities
# This will be a modular framework that can handle facility data, city/warehouse locations, and optimization

import pandas as pd
import numpy as np
from scipy.spatial.distance import cdist
from scipy.optimize import linear_sum_assignment
import matplotlib.pyplot as plt

print("Building Supply Chain Optimization Platform for African Health Facilities")
print("=" * 70)
print("")
print("Platform Components:")
print("1. Data Management Module - Load and validate facility and city data")
print("2. Distance Calculation Module - Compute distances between locations")
print("3. Cost Estimation Module - Calculate transportation and operational costs")
print("4. Optimization Module - Find optimal warehouse locations and assignments")
print("5. Visualization Module - Map and analyze the supply chain network")
print("")
print("Ready to build the platform...")

# Create the main SupplyChainOptimizer class
# This will be the core platform for designing optimal health supply chains

class SupplyChainOptimizer:
    """
    A comprehensive platform for optimizing public health supply chains in Africa.
    Handles facility data, warehouse location optimization, and cost analysis.
    """

    def __init__(self):
        self.facilities = None
        self.cities = None
        self.capital = None
        self.warehouses = None
        self.assignments = None
        self.cost_params = self._default_cost_params()

    def _default_cost_params(self):
        """Default cost parameters based on African logistics (USD)"""
        return {
            'fuel_per_km': 0.15,  # USD per km (diesel)
            'truck_hire_per_day': 80,  # USD per day
            'driver_per_day': 25,  # USD per day
            'maintenance_per_km': 0.05,  # USD per km
            'warehouse_fixed_annual': 12000,  # USD per year (rent, utilities)
            'warehouse_variable_per_facility': 50,  # USD per facility served annually
            'avg_speed_kmh': 50,  # Average speed in km/h
            'deliveries_per_month': 2,  # Delivery frequency
            'overhead_multiplier': 1.2  # 20% overhead on transport costs
        }

    def load_facilities(self, df):
        """Load health facility data with lat/lon coordinates"""
        required_cols = ['latitude', 'longitude']
        if not all(col in df.columns for col in required_cols):
            raise ValueError("Facility data must have latitude and longitude columns")
        self.facilities = df.copy()
        return len(self.facilities)

    def load_cities(self, df, capital_name=None):
        """Load city/town data for potential warehouse locations"""
        required_cols = ['latitude', 'longitude']
        if not all(col in df.columns for col in required_cols):
            raise ValueError("City data must have latitude and longitude columns")
        self.cities = df.copy()
        if capital_name and 'name' in df.columns:
            self.capital = df[df['name'] == capital_name].iloc[0] if len(df[df['name'] == capital_name]) > 0 else None
        return len(self.cities)

    def haversine_distance(self, lat1, lon1, lat2, lon2):
        """Calculate haversine distance between coordinates in km"""
        R = 6371  # Earth radius in km
        lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
        dlat = lat2 - lat1
        dlon = lon2 - lon1
        a = np.sin(dlat/2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon/2)**2
        c = 2 * np.arcsin(np.sqrt(a))
        return R * c

    def calculate_distance_matrix(self, from_locs, to_locs):
        """Calculate distance matrix between two sets of locations"""
        from_coords = np.array([[loc['latitude'], loc['longitude']] for _, loc in from_locs.iterrows()])
        to_coords = np.array([[loc['latitude'], loc['longitude']] for _, loc in to_locs.iterrows()])

        distances = np.zeros((len(from_coords), len(to_coords)))
        for i in range(len(from_coords)):
            for j in range(len(to_coords)):
                distances[i, j] = self.haversine_distance(
                    from_coords[i, 0], from_coords[i, 1],
                    to_coords[j, 0], to_coords[j, 1]
                )
        return distances

print("SupplyChainOptimizer class created successfully")
print("Key methods available:")
print("- load_facilities(): Load health facility geolocation data")
print("- load_cities(): Load potential warehouse city locations")
print("- calculate_distance_matrix(): Compute distances between locations")
print("- Default cost parameters loaded for African logistics")

# Load and inspect the health facilities data
# The file has headers starting at row 1 based on the preview

import pandas as pd
import numpy as np
from google.colab import drive
drive.mount('/content/drive')

# Load the Excel file, skipping the first empty row
path = '/content/drive/MyDrive/Colab Notebooks/health facilities list.xlsx'
df_facilities = pd.read_excel(path, sheet_name='Sheet1', header=1)

# Display basic info
print("Health Facilities Data Loaded")
print("=" * 60)
print("Shape: " + str(df_facilities.shape[0]) + " rows x " + str(df_facilities.shape[1]) + " columns")
print("")
print(df_facilities.head(10))

# Clean up the dataframe - the actual headers are in the first row
# Remove the first row and set proper column names

# The first row contains the actual headers
df_facilities.columns = df_facilities.iloc[0]
df_facilities = df_facilities.drop(0).reset_index(drop=True)

# Drop the unnamed columns (first 3 columns are empty)
df_facilities = df_facilities.iloc[:, 3:]

# Clean column names
df_facilities.columns = [str(col).strip() for col in df_facilities.columns]

# Convert numeric columns
df_facilities['Longitude'] = pd.to_numeric(df_facilities['Longitude'], errors='coerce')
df_facilities['Latitude'] = pd.to_numeric(df_facilities['Latitude'], errors='coerce')
df_facilities['Frequency'] = pd.to_numeric(df_facilities['Frequency'], errors='coerce')

# Remove any rows with missing coordinates
df_facilities = df_facilities.dropna(subset=['Longitude', 'Latitude'])

# Rename columns to match our platform's expected format
df_facilities = df_facilities.rename(columns={
    'Longitude': 'longitude',
    'Latitude': 'latitude',
    'State': 'state',
    'Facility name': 'facility_name',
    'Facility type': 'facility_type',
    'Frequency': 'frequency'
})

print("Cleaned Health Facilities Dataset")
print("=" * 60)
print("Total facilities: " + str(len(df_facilities)))
print("States covered: " + str(df_facilities['state'].nunique()))
print("")
print(df_facilities.head(15))


# Analyze the distribution of facilities by state and type
import matplotlib.pyplot as plt
import seaborn as sns

# Facility distribution by state
state_counts = df_facilities['state'].value_counts().head(15)

# Facility type distribution
type_counts = df_facilities['facility_type'].value_counts()

print("Top 15 States by Number of Facilities:")
print(state_counts)
print("")
print("Facility Types Distribution:")
print(type_counts)

# Load the city/capital data provided by the user
import pandas as pd
import numpy as np

# Create the cities dataframe from the provided data
cities_data = {
    'State': ['Abia', 'Adamawa', 'Akwa-Ibom', 'Anambra', 'Bauchi', 'Bayelsa', 'Benue', 'Borno',
              'Cross River', 'Delta', 'Ebonyi', 'Edo', 'Ekiti', 'Enugu', 'Gombe', 'Imo', 'Jigawa',
              'Kaduna', 'Kano', 'Katsina', 'Kebbi', 'Kogi', 'Kwara', 'Lagos', 'Nasarawa', 'Niger',
              'Ogun', 'Ondo', 'Osun', 'Oyo', 'Plateau', 'Rivers', 'Sokoto', 'Taraba', 'Yobe',
              'Zamfara', 'FCT'],
    'Capital': ['Umuahia', 'Yola', 'Uyo', 'Awka', 'Bauchi', 'Yenagoa', 'Makurdi', 'Maiduguri',
                'Calabar', 'Asaba', 'Abakaliki', 'Benin City', 'Ado-Ekiti', 'Enugu', 'Gombe',
                'Owerri', 'Dutse', 'Kaduna', 'Kano', 'Katsina', 'Birnin Kebbi', 'Lokoja', 'Ilorin',
                'Ikeja', 'Lafia', 'Minna', 'Abeokuta', 'Akure', 'Osogbo', 'Ibadan', 'Jos',
                'Port Harcourt', 'Sokoto', 'Jalingo', 'Damaturu', 'Gusau', 'Abuja'],
    'Lat': [5.52491, 9.20839, 5.05127, 6.21269, 10.31032, 4.92675, 7.73375, 11.84692, 4.95893,
            6.19824, 6.32485, 6.33815, 7.62329, 6.44132, 10.28969, 5.48363, 11.75618, 10.52641,
            12.00012, 12.99082, 12.45389, 7.79688, 8.49664, 6.59651, 8.4939, 9.61524, 7.15571,
            7.25256, 7.77104, 7.37756, 9.92849, 4.77742, 13.06269, 8.89367, 11.74697, 12.17024,
            9.05785],
    'Long': [7.49461, 12.48146, 7.9335, 7.07199, 9.84388, 6.26764, 8.52139, 13.15712, 8.32695,
             6.73187, 8.11368, 5.62575, 5.22087, 7.49883, 11.16729, 7.03325, 9.33896, 7.43879,
             8.51672, 7.60177, 4.1975, 6.74048, 4.54214, 3.34205, 8.51532, 6.54776, 3.34509,
             5.19312, 4.55698, 3.90591, 8.89212, 7.0134, 5.24322, 11.3596, 11.96083, 6.66412,
             7.49508]
}

df_cities = pd.DataFrame(cities_data)

# Rename columns to match our platform format
df_cities = df_cities.rename(columns={
    'Lat': 'latitude',
    'Long': 'longitude',
    'Capital': 'city',
    'State': 'state'
})

print("State Capitals Dataset Loaded")
print("=" * 60)
print("Total potential warehouse locations: " + str(len(df_cities)))
print("")
print(df_cities.head(10))

# Build the supply chain optimization platform
# Step 1: Calculate haversine distance between two coordinates

def haversine_distance(lat1, lon1, lat2, lon2):
    """Calculate distance in km between two points using haversine formula"""
    R = 6371  # Earth's radius in km

    lat1_rad = np.radians(lat1)
    lat2_rad = np.radians(lat2)
    delta_lat = np.radians(lat2 - lat1)
    delta_lon = np.radians(lon2 - lon1)

    a = np.sin(delta_lat/2)**2 + np.cos(lat1_rad) * np.cos(lat2_rad) * np.sin(delta_lon/2)**2
    c = 2 * np.arctan2(np.sqrt(a), np.sqrt(1-a))

    return R * c

# Step 2: Calculate transportation cost
def calculate_transport_cost(distance_km, cost_per_km=0.5):
    """
    Calculate transportation cost based on distance
    Default: $0.50 per km (adjustable based on fuel, vehicle type, etc.)
    """
    return distance_km * cost_per_km

# Step 3: Find nearest warehouse for each facility
def assign_facilities_to_warehouses(facilities_df, warehouses_df):
    """
    Assign each facility to its nearest warehouse
    Returns facilities_df with warehouse assignments and distances
    """
    assignments = []

    for idx, facility in facilities_df.iterrows():
        min_distance = float('inf')
        nearest_warehouse = None
        warehouse_city = None

        for w_idx, warehouse in warehouses_df.iterrows():
            dist = haversine_distance(
                facility['latitude'], facility['longitude'],
                warehouse['latitude'], warehouse['longitude']
            )

            if dist < min_distance:
                min_distance = dist
                nearest_warehouse = w_idx
                warehouse_city = warehouse['city']

        assignments.append({
            'warehouse_id': nearest_warehouse,
            'warehouse_city': warehouse_city,
            'distance_km': min_distance,
            'transport_cost': calculate_transport_cost(min_distance)
        })

    assignment_df = pd.DataFrame(assignments)
    result_df = pd.concat([facilities_df.reset_index(drop=True), assignment_df], axis=1)

    return result_df

print("Supply Chain Optimization Functions Loaded")
print("=" * 60)
print("Functions available:")
print("- haversine_distance(): Calculate distance between coordinates")
print("- calculate_transport_cost(): Estimate transportation costs")
print("- assign_facilities_to_warehouses(): Optimize facility-warehouse assignments")

# Run optimization scenarios with different numbers of warehouses
# We'll test 1, 3, 5, 7, and 10 regional warehouses plus Abuja as main hub

from sklearn.cluster import KMeans
from tqdm import tqdm

# Scenario 1: Single central warehouse (Abuja only)
print("SCENARIO 1: Single Central Warehouse (Abuja)")
print("=" * 60)

abuja = df_cities[df_cities['city'] == 'Abuja']
scenario_1 = assign_facilities_to_warehouses(df_facilities, abuja)

total_distance_1 = scenario_1['distance_km'].sum()
avg_distance_1 = scenario_1['distance_km'].mean()
total_cost_1 = scenario_1['transport_cost'].sum()

print("Total distance to facilities: " + str(round(total_distance_1, 2)) + " km")
print("Average distance per facility: " + str(round(avg_distance_1, 2)) + " km")
print("Total transportation cost: $" + str(round(total_cost_1, 2)))
print("Max distance to any facility: " + str(round(scenario_1['distance_km'].max(), 2)) + " km")
print("")


# Scenario 2-5: Multiple regional warehouses using K-means clustering
# We'll cluster facilities and find nearest capital city to each cluster centroid

scenarios = {}

for n_warehouses in [3, 5, 7, 10]:
    print("SCENARIO: " + str(n_warehouses) + " Regional Warehouses + Abuja Hub")
    print("=" * 60)

    # Use K-means to cluster facilities
    facility_coords = df_facilities[['latitude', 'longitude']].values
    kmeans = KMeans(n_clusters=n_warehouses, random_state=42, n_init=10)
    clusters = kmeans.fit_predict(facility_coords)
    centroids = kmeans.cluster_centers_

    # Find nearest capital city to each cluster centroid
    selected_warehouses = []

    for centroid in centroids:
        min_dist = float('inf')
        nearest_city = None

        for idx, city in df_cities.iterrows():
            dist = haversine_distance(centroid[0], centroid[1],
                                     city['latitude'], city['longitude'])
            if dist < min_dist:
                min_dist = dist
                nearest_city = idx

        if nearest_city not in selected_warehouses:
            selected_warehouses.append(nearest_city)

    # Get warehouse dataframe
    warehouses = df_cities.iloc[selected_warehouses]

    # Assign facilities to warehouses
    result = assign_facilities_to_warehouses(df_facilities, warehouses)

    total_distance = result['distance_km'].sum()
    avg_distance = result['distance_km'].mean()
    total_cost = result['transport_cost'].sum()
    max_distance = result['distance_km'].max()

    scenarios[n_warehouses] = {
        'warehouses': warehouses['city'].tolist(),
        'total_distance': total_distance,
        'avg_distance': avg_distance,
        'total_cost': total_cost,
        'max_distance': max_distance,
        'assignments': result
    }

    print("Selected warehouse cities: " + ", ".join(warehouses['city'].tolist()))
    print("Total distance: " + str(round(total_distance, 2)) + " km")
    print("Average distance: " + str(round(avg_distance, 2)) + " km")
    print("Total transport cost: $" + str(round(total_cost, 2)))
    print("Max distance: " + str(round(max_distance, 2)) + " km")
    print("")

    # Step 1: Import and load the health facilities Excel file (fixing the header row)
# We use header=1 because the preview shows the labels on the second row (0-indexed)
import pandas as pd
import numpy as np

excel_path = '/content/drive/MyDrive/Colab Notebooks/health facilities list.xlsx'
raw_df = pd.read_excel(excel_path, sheet_name='Sheet1', header=1)

print(raw_df.head(8))
print(list(raw_df.columns))

# Visualize the 7-warehouse scenario with Plotly
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from datetime import datetime
import random

# Retrieve the 7-warehouse scenario assignments from the scenarios dict created earlier
assignments_7 = scenarios[7]['assignments'].copy()

# Ensure expected columns exist
expected_cols = ['latitude','longitude','warehouse_city','distance_km','transport_cost']
missing = [c for c in expected_cols if c not in assignments_7.columns]
if missing:
    raise ValueError(f"Missing columns in assignments for plotting: {missing}")

# Build color labels
assignments_7['warehouse_city'] = assignments_7['warehouse_city'].astype(str)

# Create a scatter geo plot of facilities colored by assigned warehouse city
fig = px.scatter_geo(
    assignments_7,
    lat='latitude',
    lon='longitude',
    color='warehouse_city',
    hover_name='warehouse_city',
    hover_data={'distance_km':':.1f', 'transport_cost':':.2f', 'latitude':False, 'longitude':False},
    title='Nigeria Health Facilities Assigned to 7 Regional Warehouses',
    opacity=0.6,
    height=720
)

# Add warehouse city markers: take one point per warehouse (nearest facility centroid approx.)
warehouse_markers = assignments_7.groupby('warehouse_city', as_index=False)[['latitude','longitude']].median()
fig.add_trace(go.Scattergeo(
    lat=warehouse_markers['latitude'],
    lon=warehouse_markers['longitude'],
    mode='markers+text',
    text=warehouse_markers['warehouse_city'],
    textposition='top center',
    marker=dict(size=12, color='black', symbol='x'),
    name='Warehouse (label at cluster median)'
))

fig.update_geos(
    showcountries=True, countrycolor='LightGray',
    showland=True, landcolor='rgb(240,240,240)',
    showcoastlines=False, fitbounds="locations"
)
fig.update_layout(legend_title_text='Assigned Warehouse City')

# Save to file
rand_id = random.randint(1000,9999)
html_file = f"seven_warehouses_map_{rand_id}.html"
fig.write_html(html_file, include_plotlyjs='cdn', full_html=True)

# Also prepare a small summary table (head)
assignments_head = assignments_7[['warehouse_city','distance_km','transport_cost','latitude','longitude']].head(10)
print(assignments_head)
print(f"Saved visualization to {html_file}")
```
