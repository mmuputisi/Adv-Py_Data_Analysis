```python
# esg_capital_stock_optimal_rates_enhanced.py
"""
ESG-Tobin's Q analysis that finds optimal depreciation rates:
1. SEPARATE optimal rates for All Sectors vs Consumer Staples
2. Different optimal rates for E, S, and G components separately
3. Compare optimal rates across different sectors
4. Test stability of optimal rates over time
"""

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from linearmodels.panel import PanelOLS
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

from google.colab import drive
drive.mount('/content/drive')

!pip install -q linearmodels pandas numpy matplotlib seaborn statsmodels


class ESGCapitalStockOptimalRatesEnhanced:
    """
    Enhanced ESG Capital Stock analysis that finds optimal depreciation rates for:
    - Different samples (All sectors, Consumer Staples, other sectors)
    - Different ESG components (E, S, G separately)
    - Tests stability over time
    """

    def __init__(self, data_path, output_dir):
        """
        Initialize the enhanced analysis.
        """
        self.data_path = data_path
        self.output_dir = output_dir
        self.df = None
        self.df_cs = None
        self.capital_stock_data = {}
        self.results = {}

        # Optimal rates for different samples
        self.optimal_rates = {
            'all_sectors': {},
            'consumer_staples': {},
            'other_sectors': {},
            'by_sector': {}
        }

        # Time period splits for stability testing
        self.time_periods = {}

        # Create output directory
        os.makedirs(output_dir, exist_ok=True)

        # Setup plotting
        plt.style.use('seaborn-v0_8-whitegrid')
        sns.set_palette("husl")

        # Define Consumer Staples firms
        self.consumer_staples_firms = [
            'Associated British Foods', 'British American Tobacco', 'Coca Cola HBC AG',
            'Diageo', 'J. Sainsbury', 'Reckitt Benckiser', 'Tesco', 'Unilever', 'Haleon Plc'
        ]

        # Define other sectors for comparison
        self.other_sectors = {
            'Industrials': [],
            'Financials': [],
            'Healthcare': [],
            'Technology': [],
            'Energy': []
        }

    def load_and_prepare_data(self):
        """
        Load data and prepare basic transformations.
        """
        print("=" * 70)
        print("LOADING AND PREPARING DATA")
        print("=" * 70)

        try:
            # Load data
            self.df = pd.read_excel(self.data_path)
            print(f"✓ Loaded data from: {self.data_path}")

            # Set multi-index
            if 'Firm' in self.df.columns and 'Year' in self.df.columns:
                self.df = self.df.set_index(['Firm', 'Year']).sort_index()
                print("✓ Set multi-index (Firm, Year)")
            else:
                self.df = self.df.set_index([self.df.columns[0], self.df.columns[1]]).sort_index()
                print(f"✓ Using {self.df.index.names} as index")

            # Rename columns
            column_renames = {}
            if "Tobin's Q" in self.df.columns:
                column_renames["Tobin's Q"] = "Tobin_Q"

            if column_renames:
                self.df = self.df.rename(columns=column_renames)
                print(f"✓ Renamed columns: {column_renames}")

            # Create control variables
            if 'Total Assets' in self.df.columns:
                self.df['Size'] = np.log(self.df['Total Assets'].replace(0, np.nan))
                print("✓ Created Size variable")

            if 'Total Liabilities' in self.df.columns and 'Total Assets' in self.df.columns:
                self.df['Leverage'] = self.df['Total Liabilities'] / self.df['Total Assets'].replace(0, np.nan)
                print("✓ Created Leverage variable")

            # Create Consumer Staples dummy
            self.df['ConsumerStaples'] = self.df.index.get_level_values(0).isin(self.consumer_staples_firms).astype(int)

            # Create Consumer Staples only dataset
            self.df_cs = self.df[self.df['ConsumerStaples'] == 1].copy()

            # Create Other Sectors dataset (exclude Consumer Staples)
            self.df_other = self.df[self.df['ConsumerStaples'] == 0].copy()

            print(f"\n✓ Total observations: {len(self.df)}")
            print(f"✓ Total firms: {self.df.index.get_level_values(0).nunique()}")
            print(f"✓ Consumer Staples observations: {len(self.df_cs)}")
            print(f"✓ Consumer Staples firms: {self.df_cs.index.get_level_values(0).nunique() if len(self.df_cs) > 0 else 0}")
            print(f"✓ Other sectors observations: {len(self.df_other)}")
            print(f"✓ Other sectors firms: {self.df_other.index.get_level_values(0).nunique() if len(self.df_other) > 0 else 0}")
            print(f"✓ Years: {sorted(self.df.index.get_level_values(1).unique())}")

            # Create time periods for stability testing
            years = sorted(self.df.index.get_level_values(1).unique())
            mid_point = len(years) // 2
            self.time_periods = {
                'full': years,
                'early': years[:mid_point],
                'late': years[mid_point:],
                'pre_2015': [y for y in years if y < 2015],
                'post_2015': [y for y in years if y >= 2015]
            }

            print(f"\n✓ Created time periods for stability testing:")
            for period, period_years in self.time_periods.items():
                if period_years:
                    print(f"  {period}: {period_years[0]}-{period_years[-1]} ({len(period_years)} years)")

            return self.df

        except Exception as e:
            print(f"✗ Error loading data: {e}")
            import traceback
            traceback.print_exc()
            return None

    def create_lagged_esg_variables(self, dataset):
        """
        Create lagged ESG variables.
        """
        if dataset is None or len(dataset) == 0:
            return dataset

        dataset_copy = dataset.copy()

        for component in ['E', 'S', 'G']:
            if component in dataset_copy.columns:
                dataset_copy[f'{component}_lag'] = dataset_copy.groupby(level=0)[component].shift(1)

        return dataset_copy

    def create_esg_capital_stock_single_component(self, dataset, component,
                                                  depreciation_rate=0.3, use_lagged_esg=True):
        """
        Create ESG capital stock for a SINGLE component.
        This allows finding optimal rates separately for E, S, and G.
        """
        if dataset is None or len(dataset) == 0:
            return dataset

        dataset_copy = dataset.copy()

        # Determine which variable to use
        if use_lagged_esg:
            component_var = f'{component}_lag'
            suffix = '_lag'
        else:
            component_var = component
            suffix = ''

        if component_var not in dataset_copy.columns:
            print(f"  Warning: {component_var} not found")
            return dataset_copy

        col_name = f'{component}_capital_{int(depreciation_rate*100)}{suffix}'

        # Process each firm separately
        capital_series = []

        for firm in dataset_copy.index.get_level_values(0).unique():
            firm_data = dataset_copy.xs(firm, level=0).copy()

            if len(firm_data) > 0 and component_var in firm_data.columns:
                esg_series = firm_data[component_var].fillna(0).values

                # Calculate capital stock
                capital = np.zeros(len(esg_series))

                if len(esg_series) > 0:
                    capital[0] = esg_series[0]

                    for t in range(1, len(esg_series)):
                        capital[t] = (1 - depreciation_rate) * capital[t-1] + esg_series[t]

                capital_series.extend(capital)
            else:
                capital_series.extend([np.nan] * len(firm_data))

        dataset_copy[col_name] = capital_series

        return dataset_copy

    def create_esg_capital_stock_all_components(self, dataset, depreciation_rate=0.3,
                                                use_lagged_esg=True):
        """
        Create ESG capital stock for all components together.
        """
        if dataset is None or len(dataset) == 0:
            return dataset

        for component in ['E', 'S', 'G']:
            dataset = self.create_esg_capital_stock_single_component(
                dataset, component, depreciation_rate, use_lagged_esg
            )

        return dataset

    def run_regression_safe(self, df, y_var, x_vars, control_vars=None,
                           add_consumer_staples_dummy=False):
        """
        Run a safe regression that won't produce R² = 1.
        """
        if df is None or len(df) == 0:
            return None

        # Combine x variables
        all_x_vars = x_vars.copy()
        if control_vars:
            all_x_vars.extend(control_vars)
        if add_consumer_staples_dummy and 'ConsumerStaples' in df.columns:
            all_x_vars.append('ConsumerStaples')

        # Check if variables exist
        available_vars = [v for v in all_x_vars if v in df.columns]

        if y_var not in df.columns:
            return None

        if len(available_vars) == 0:
            return None

        # Get clean data
        df_clean = df.dropna(subset=[y_var] + available_vars).copy()

        if len(df_clean) < 10:
            return None

        # Check for constant variables
        valid_vars = []
        for var in available_vars:
            if df_clean[var].nunique() > 1:
                valid_vars.append(var)

        if len(valid_vars) == 0:
            return None

        # Prepare X and y
        y = df_clean[y_var]
        X = df_clean[valid_vars].copy()

        # Add constant
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)
        used_constant = True

        # Check for perfect multicollinearity
        try:
            from numpy.linalg import cond
            X_matrix = X.values.astype(float)
            cond_num = cond(X_matrix)

            if cond_num > 1e10:
                X_no_const = df_clean[valid_vars].copy()
                used_constant = False
                X = X_no_const
        except:
            pass

        try:
            import statsmodels.api as sm

            model = sm.OLS(y, X.astype(float))
            results = model.fit(cov_type='HC3')

            return {
                'results': results,
                'nobs': int(results.nobs),
                'r2': results.rsquared,
                'r2_adj': results.rsquared_adj,
                'valid_vars': valid_vars,
                'used_constant': used_constant
            }

        except Exception as e:
            return None

    def find_optimal_rate_for_component(self, dataset, dataset_name, component,
                                        depreciation_rates=None, use_lagged_esg=True,
                                        add_consumer_staples_dummy=False):
        """
        Find optimal depreciation rate for a SINGLE ESG component.
        """
        if depreciation_rates is None:
            depreciation_rates = [0.10, 0.15, 0.20, 0.25, 0.30, 0.35, 0.40, 0.45, 0.50]

        print(f"\n  Testing {component} capital - {len(depreciation_rates)} rates")

        # Create lagged variables
        dataset = self.create_lagged_esg_variables(dataset)

        results_summary = []

        for rate in depreciation_rates:
            # Create capital stock for single component
            df_with_capital = self.create_esg_capital_stock_single_component(
                dataset.copy(),
                component=component,
                depreciation_rate=rate,
                use_lagged_esg=use_lagged_esg
            )

            # Define capital variable
            suffix = '_lag' if use_lagged_esg else ''
            capital_var = f'{component}_capital_{int(rate*100)}{suffix}'

            if capital_var not in df_with_capital.columns:
                continue

            # Define control variables
            control_vars = ['ROA', 'Size', 'Leverage']
            control_vars = [v for v in control_vars if v in df_with_capital.columns]

            # Run regression with just this component
            result = self.run_regression_safe(
                df=df_with_capital,
                y_var='Tobin_Q',
                x_vars=[capital_var],
                control_vars=control_vars,
                add_consumer_staples_dummy=add_consumer_staples_dummy
            )

            if result is not None:
                coeff = result['results'].params[capital_var] if capital_var in result['results'].params else np.nan
                pval = result['results'].pvalues[capital_var] if capital_var in result['results'].pvalues else 1.0
                se = result['results'].bse[capital_var] if capital_var in result['results'].bse else np.nan

                results_summary.append({
                    'Component': component,
                    'Depreciation_Rate': rate,
                    'R_Squared': result['r2'],
                    'R_Squared_Adj': result['r2_adj'],
                    'Coefficient': coeff,
                    'PValue': pval,
                    'StdError': se,
                    'Observations': result['nobs'],
                    'Significant_10': pval < 0.10,
                    'Significant_05': pval < 0.05,
                    'Significant_01': pval < 0.01
                })

        if results_summary:
            results_df = pd.DataFrame(results_summary)

            # Find best rate based on R-squared
            best_idx = results_df['R_Squared'].idxmax()
            best_rate = results_df.loc[best_idx, 'Depreciation_Rate']

            print(f"    ✓ {component}: optimal δ = {best_rate:.2f}, R² = {results_df.loc[best_idx, 'R_Squared']:.4f}")

            return best_rate, results_df
        else:
            print(f"    ✗ {component}: no successful models")
            return None, None

    def find_optimal_rates_all_components(self, dataset, dataset_name,
                                          depreciation_rates=None, use_lagged_esg=True,
                                          add_consumer_staples_dummy=False):
        """
        Find optimal depreciation rates for E, S, and G separately.
        """
        print("\n" + "=" * 70)
        print(f"FINDING COMPONENT-SPECIFIC OPTIMAL RATES FOR {dataset_name}")
        print("=" * 70)

        components = ['E', 'S', 'G']
        results = {}

        for component in components:
            best_rate, comp_df = self.find_optimal_rate_for_component(
                dataset=dataset,
                dataset_name=dataset_name,
                component=component,
                depreciation_rates=depreciation_rates,
                use_lagged_esg=use_lagged_esg,
                add_consumer_staples_dummy=add_consumer_staples_dummy
            )

            if best_rate is not None:
                results[component] = {
                    'optimal_rate': best_rate,
                    'results': comp_df
                }

                # Save component results
                comp_path = os.path.join(self.output_dir,
                                        f'{dataset_name.lower().replace(" ", "_")}_{component}_optimal.csv')
                comp_df.to_csv(comp_path, index=False)

        return results

    def find_optimal_rate_combined_components(self, dataset, dataset_name,
                                              depreciation_rates=None, use_lagged_esg=True,
                                              add_consumer_staples_dummy=False):
        """
        Find optimal depreciation rate when using all components together.
        """
        if depreciation_rates is None:
            depreciation_rates = [0.10, 0.15, 0.20, 0.25, 0.30, 0.35, 0.40, 0.45, 0.50]

        print("\n" + "=" * 70)
        print(f"FINDING COMBINED OPTIMAL RATE FOR {dataset_name}")
        print("=" * 70)
        print(f"Testing {len(depreciation_rates)} depreciation rates")

        # Create lagged variables
        dataset = self.create_lagged_esg_variables(dataset)

        results_summary = []

        for rate in depreciation_rates:
            print(f"\n  Testing δ = {rate:.2f}")

            # Create capital stock for all components
            df_with_capital = self.create_esg_capital_stock_all_components(
                dataset.copy(),
                depreciation_rate=rate,
                use_lagged_esg=use_lagged_esg
            )

            # Define capital variables
            suffix = '_lag' if use_lagged_esg else ''
            capital_vars = [f'{c}_capital_{int(rate*100)}{suffix}' for c in ['E', 'S', 'G']]
            capital_vars = [v for v in capital_vars if v in df_with_capital.columns]

            if not capital_vars:
                print(f"    No capital variables created")
                continue

            # Define control variables
            control_vars = ['ROA', 'Size', 'Leverage']
            control_vars = [v for v in control_vars if v in df_with_capital.columns]

            # Run regression
            result = self.run_regression_safe(
                df=df_with_capital,
                y_var='Tobin_Q',
                x_vars=capital_vars,
                control_vars=control_vars,
                add_consumer_staples_dummy=add_consumer_staples_dummy
            )

            if result is not None:
                # Extract coefficients for each component
                row_data = {
                    'Depreciation_Rate': rate,
                    'R_Squared': result['r2'],
                    'R_Squared_Adj': result['r2_adj'],
                    'Observations': result['nobs'],
                    'N_Firms': df_with_capital.index.get_level_values(0).nunique()
                }

                for component in ['E', 'S', 'G']:
                    var = f'{component}_capital_{int(rate*100)}{suffix}'
                    if var in result['results'].params:
                        row_data[f'{component}_Coefficient'] = result['results'].params[var]
                        row_data[f'{component}_PValue'] = result['results'].pvalues[var]
                        row_data[f'{component}_StdError'] = result['results'].bse[var]

                results_summary.append(row_data)
                print(f"    ✓ R² = {result['r2']:.4f}")
            else:
                print(f"    ✗ Model failed")

        if results_summary:
            results_df = pd.DataFrame(results_summary)

            # Find best rate based on R-squared
            best_idx = results_df['R_Squared'].idxmax()
            best_rate = results_df.loc[best_idx, 'Depreciation_Rate']

            print("\n" + "=" * 70)
            print(f"OPTIMAL COMBINED RATE FOR {dataset_name}: δ = {best_rate:.2f}")
            print("=" * 70)
            print(f"R² = {results_df.loc[best_idx, 'R_Squared']:.4f}")

            # Display coefficients at optimal rate
            print(f"\nCoefficients at optimal rate:")
            for component in ['E', 'S', 'G']:
                coeff_col = f'{component}_Coefficient'
                pval_col = f'{component}_PValue'
                if coeff_col in results_df.columns:
                    coeff = results_df.loc[best_idx, coeff_col]
                    pval = results_df.loc[best_idx, pval_col]
                    stars = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                    print(f"  {component}: {coeff:.4f}{stars} (p={pval:.4f})")

            # Save results
            comp_path = os.path.join(self.output_dir,
                                    f'{dataset_name.lower().replace(" ", "_")}_combined_optimal.csv')
            results_df.to_csv(comp_path, index=False)
            print(f"\n✓ Saved combined optimal results to: {comp_path}")

            # Plot comparison
            self.plot_combined_comparison(results_df, best_rate, dataset_name)

            return best_rate, results_df
        else:
            print(f"\n✗ No successful models for {dataset_name}")
            return None, None

    def plot_combined_comparison(self, results_df, best_rate, dataset_name):
        """
        Plot combined model comparison.
        """
        fig, axes = plt.subplots(1, 3, figsize=(18, 5))

        # Plot 1: R-squared vs Rate
        ax1 = axes[0]
        ax1.plot(results_df['Depreciation_Rate'], results_df['R_Squared'],
                marker='o', linewidth=2, markersize=8)
        ax1.axvline(x=best_rate, color='red', linestyle='--',
                   label=f'Optimal: δ={best_rate:.2f}')
        ax1.set_xlabel('Depreciation Rate (δ)')
        ax1.set_ylabel('R-squared')
        ax1.set_title(f'{dataset_name}: Combined Model Fit')
        ax1.legend()
        ax1.grid(True, alpha=0.3)

        # Plot 2: Component Coefficients vs Rate
        ax2 = axes[1]
        colors = {'E': 'blue', 'S': 'green', 'G': 'orange'}

        for component in ['E', 'S', 'G']:
            coeff_col = f'{component}_Coefficient'
            if coeff_col in results_df.columns:
                ax2.plot(results_df['Depreciation_Rate'], results_df[coeff_col],
                        marker='o', linewidth=2, markersize=6,
                        color=colors[component], label=f'{component}')

        ax2.axhline(y=0, color='gray', linestyle='-', alpha=0.5)
        ax2.axvline(x=best_rate, color='red', linestyle='--', alpha=0.5)
        ax2.set_xlabel('Depreciation Rate (δ)')
        ax2.set_ylabel('Coefficient')
        ax2.set_title(f'{dataset_name}: Component Coefficients')
        ax2.legend()
        ax2.grid(True, alpha=0.3)

        # Plot 3: Significance Heatmap
        ax3 = axes[2]

        significance_data = []
        for component in ['E', 'S', 'G']:
            pval_col = f'{component}_PValue'
            if pval_col in results_df.columns:
                sig_levels = []
                for pval in results_df[pval_col]:
                    if pval < 0.01:
                        sig_levels.append(3)  # ***
                    elif pval < 0.05:
                        sig_levels.append(2)  # **
                    elif pval < 0.10:
                        sig_levels.append(1)  # *
                    else:
                        sig_levels.append(0)  # Not significant
                significance_data.append(sig_levels)

        if significance_data:
            im = ax3.imshow(significance_data, aspect='auto', cmap='RdYlGn_r',
                          vmin=0, vmax=3)
            ax3.set_yticks(range(len(['E', 'S', 'G'])))
            ax3.set_yticklabels(['E', 'S', 'G'])
            ax3.set_xticks(range(len(results_df['Depreciation_Rate'])))
            ax3.set_xticklabels([f'{r:.2f}' for r in results_df['Depreciation_Rate']],
                              rotation=45, ha='right')
            ax3.set_xlabel('Depreciation Rate (δ)')
            ax3.set_title(f'{dataset_name}: Significance Levels\n(*** p<0.01, ** p<0.05, * p<0.10)')
            plt.colorbar(im, ax=ax3, ticks=[0, 1, 2, 3])

        plt.tight_layout()
        plot_path = os.path.join(self.output_dir,
                                f'{dataset_name.lower().replace(" ", "_")}_combined_comparison.png')
        plt.savefig(plot_path, dpi=150, bbox_inches='tight')
        plt.show()
        print(f"✓ Saved combined comparison plot to: {plot_path}")

    def analyze_sector_differences(self):
        """
        Compare optimal depreciation rates across different sectors.
        """
        print("\n" + "=" * 70)
        print("ANALYZING SECTOR DIFFERENCES IN OPTIMAL DEPRECIATION RATES")
        print("=" * 70)

        # Get all unique firms
        all_firms = self.df.index.get_level_values(0).unique()

        # Try to identify sectors based on firm names
        # This is a simplified approach - ideally you'd have actual sector classification
        sector_keywords = {
            'Industrials': ['Industrial', 'Engineering', 'Construction', 'Manufacturing',
                           'Transport', 'Logistics', 'BAE', 'Rolls-Royce', 'GKN'],
            'Financials': ['Bank', 'Financial', 'Insurance', 'Lloyds', 'Barclays',
                          'HSBC', 'Standard Chartered', 'Prudential', 'Aviva'],
            'Healthcare': ['Health', 'Pharma', 'Medical', 'GSK', 'AstraZeneca',
                          'Smith & Nephew', 'Haleon'],
            'Technology': ['Tech', 'Software', 'Hardware', 'Sage', 'Micro Focus'],
            'Energy': ['Energy', 'Oil', 'Gas', 'Shell', 'BP', 'Centrica', 'SSE'],
            'Consumer Staples': self.consumer_staples_firms,
            'Consumer Discretionary': ['Retail', 'Leisure', 'Hotel', 'Restaurant',
                                      'Marks & Spencer', 'Next', 'Burberry']
        }

        sector_datasets = {}

        for sector, keywords in sector_keywords.items():
            sector_firms = []
            for firm in all_firms:
                if any(keyword.lower() in str(firm).lower() for keyword in keywords):
                    sector_firms.append(firm)

            if sector_firms:
                sector_datasets[sector] = self.df[
                    self.df.index.get_level_values(0).isin(sector_firms)
                ].copy()
                print(f"\n{sector}:")
                print(f"  Firms: {len(sector_firms)}")
                print(f"  Observations: {len(sector_datasets[sector])}")

        # Find optimal rates for each sector
        sector_results = {}

        for sector, sector_df in sector_datasets.items():
            if len(sector_df) >= 50:  # Need minimum observations
                print(f"\n" + "-" * 50)
                print(f"Finding optimal rate for {sector}")
                print("-" * 50)

                best_rate, results_df = self.find_optimal_rate_combined_components(
                    dataset=sector_df,
                    dataset_name=sector,
                    add_consumer_staples_dummy=False
                )

                if best_rate is not None:
                    sector_results[sector] = {
                        'optimal_rate': best_rate,
                        'observations': len(sector_df),
                        'firms': sector_df.index.get_level_values(0).nunique(),
                        'r2': results_df.loc[results_df['R_Squared'].idxmax(), 'R_Squared'] if results_df is not None else None
                    }

        # Save sector comparison
        if sector_results:
            sector_df = pd.DataFrame.from_dict(sector_results, orient='index')
            sector_df.index.name = 'Sector'
            sector_df = sector_df.reset_index()

            sector_path = os.path.join(self.output_dir, 'sector_comparison.csv')
            sector_df.to_csv(sector_path, index=False)
            print(f"\n✓ Saved sector comparison to: {sector_path}")

            # Plot sector comparison
            self.plot_sector_comparison(sector_df)

            self.optimal_rates['by_sector'] = sector_results

        return sector_results

    def plot_sector_comparison(self, sector_df):
        """
        Plot comparison of optimal rates across sectors.
        """
        fig, axes = plt.subplots(1, 2, figsize=(14, 6))

        # Plot 1: Optimal rates by sector
        ax1 = axes[0]
        bars = ax1.bar(sector_df['Sector'], sector_df['optimal_rate'],
                      alpha=0.8, color=sns.color_palette("husl", len(sector_df)))
        ax1.axhline(y=self.optimal_rates['all_sectors']['combined']['optimal_rate']
                   if 'combined' in self.optimal_rates['all_sectors'] else 0.3,
                   color='red', linestyle='--', label='All Sectors Avg')
        ax1.set_xlabel('Sector')
        ax1.set_ylabel('Optimal Depreciation Rate (δ)')
        ax1.set_title('Optimal Depreciation Rates by Sector')
        ax1.tick_params(axis='x', rotation=45)
        ax1.legend()
        ax1.grid(True, alpha=0.3)

        # Add value labels on bars
        for bar, rate in zip(bars, sector_df['optimal_rate']):
            height = bar.get_height()
            ax1.text(bar.get_x() + bar.get_width()/2., height,
                    f'{rate:.2f}', ha='center', va='bottom')

        # Plot 2: R-squared by sector
        ax2 = axes[1]
        if 'r2' in sector_df.columns:
            bars2 = ax2.bar(sector_df['Sector'], sector_df['r2'],
                          alpha=0.8, color=sns.color_palette("husl", len(sector_df)))
            ax2.set_xlabel('Sector')
            ax2.set_ylabel('R-squared at Optimal Rate')
            ax2.set_title('Model Fit by Sector')
            ax2.tick_params(axis='x', rotation=45)
            ax2.grid(True, alpha=0.3)

            for bar, r2 in zip(bars2, sector_df['r2']):
                height = bar.get_height()
                ax2.text(bar.get_x() + bar.get_width()/2., height,
                        f'{r2:.3f}', ha='center', va='bottom')

        plt.tight_layout()
        plot_path = os.path.join(self.output_dir, 'sector_comparison.png')
        plt.savefig(plot_path, dpi=150, bbox_inches='tight')
        plt.show()
        print(f"✓ Saved sector comparison plot to: {plot_path}")

    def test_stability_over_time(self, dataset, dataset_name,
                                optimal_rate, use_lagged_esg=True,
                                add_consumer_staples_dummy=False):
        """
        Test stability of optimal rate over different time periods.
        """
        print("\n" + "=" * 70)
        print(f"TESTING STABILITY OF δ={optimal_rate:.2f} OVER TIME FOR {dataset_name}")
        print("=" * 70)

        stability_results = []

        for period_name, period_years in self.time_periods.items():
            if not period_years:
                continue

            print(f"\n  Testing period: {period_name}")

            # Filter data for this time period
            period_data = dataset[dataset.index.get_level_values(1).isin(period_years)].copy()

            if len(period_data) < 30:
                print(f"    Insufficient data: {len(period_data)} observations")
                continue

            # Create capital stock with the optimal rate
            period_data = self.create_lagged_esg_variables(period_data)
            period_data = self.create_esg_capital_stock_all_components(
                period_data,
                depreciation_rate=optimal_rate,
                use_lagged_esg=use_lagged_esg
            )

            # Define capital variables
            suffix = '_lag' if use_lagged_esg else ''
            capital_vars = [f'{c}_capital_{int(optimal_rate*100)}{suffix}' for c in ['E', 'S', 'G']]
            capital_vars = [v for v in capital_vars if v in period_data.columns]

            # Define control variables
            control_vars = ['ROA', 'Size', 'Leverage']
            control_vars = [v for v in control_vars if v in period_data.columns]

            # Run regression
            result = self.run_regression_safe(
                df=period_data,
                y_var='Tobin_Q',
                x_vars=capital_vars,
                control_vars=control_vars,
                add_consumer_staples_dummy=add_consumer_staples_dummy
            )

            if result is not None:
                row_data = {
                    'Period': period_name,
                    'Years': f"{period_years[0]}-{period_years[-1]}",
                    'N_Years': len(period_years),
                    'Observations': result['nobs'],
                    'R_Squared': result['r2'],
                    'R_Squared_Adj': result['r2_adj']
                }

                for component in ['E', 'S', 'G']:
                    var = f'{component}_capital_{int(optimal_rate*100)}{suffix}'
                    if var in result['results'].params:
                        row_data[f'{component}_Coefficient'] = result['results'].params[var]
                        row_data[f'{component}_PValue'] = result['results'].pvalues[var]
                        row_data[f'{component}_StdError'] = result['results'].bse[var]

                stability_results.append(row_data)
                print(f"    ✓ R² = {result['r2']:.4f}, N = {result['nobs']}")
            else:
                print(f"    ✗ Model failed")

        if stability_results:
            stability_df = pd.DataFrame(stability_results)

            # Save stability results
            stability_path = os.path.join(self.output_dir,
                                         f'{dataset_name.lower().replace(" ", "_")}_stability_test.csv')
            stability_df.to_csv(stability_path, index=False)
            print(f"\n✓ Saved stability test results to: {stability_path}")

            # Plot stability
            self.plot_stability_results(stability_df, optimal_rate, dataset_name)

            return stability_df
        else:
            print(f"\n✗ No stability test results for {dataset_name}")
            return None

    def plot_stability_results(self, stability_df, optimal_rate, dataset_name):
        """
        Plot stability test results.
        """
        fig, axes = plt.subplots(2, 2, figsize=(14, 10))

        # Plot 1: R-squared by period
        ax1 = axes[0, 0]
        bars = ax1.bar(stability_df['Period'], stability_df['R_Squared'],
                      alpha=0.8, color='steelblue')
        ax1.axhline(y=stability_df[stability_df['Period'] == 'full']['R_Squared'].values[0]
                   if 'full' in stability_df['Period'].values else None,
                   color='red', linestyle='--', label='Full Period')
        ax1.set_xlabel('Time Period')
        ax1.set_ylabel('R-squared')
        ax1.set_title(f'{dataset_name}: Model Stability Over Time\nδ = {optimal_rate:.2f}')
        ax1.tick_params(axis='x', rotation=45)
        ax1.legend()
        ax1.grid(True, alpha=0.3)

        # Plot 2: Component coefficients by period
        ax2 = axes[0, 1]
        colors = {'E': 'blue', 'S': 'green', 'G': 'orange'}

        x_pos = np.arange(len(stability_df))
        width = 0.25

        for i, component in enumerate(['E', 'S', 'G']):
            coeff_col = f'{component}_Coefficient'
            if coeff_col in stability_df.columns:
                coeffs = stability_df[coeff_col].values
                ax2.bar(x_pos + i*width, coeffs, width,
                       label=component, color=colors[component], alpha=0.8)

                # Add significance stars
                pval_col = f'{component}_PValue'
                for j, (coeff, pval) in enumerate(zip(coeffs, stability_df[pval_col])):
                    stars = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                    ax2.text(x_pos[j] + i*width, coeff + (0.01 if coeff >= 0 else -0.02),
                            stars, ha='center', va='bottom' if coeff >= 0 else 'top')

        ax2.axhline(y=0, color='gray', linestyle='-', alpha=0.5)
        ax2.set_xlabel('Time Period')
        ax2.set_ylabel('Coefficient')
        ax2.set_title(f'{dataset_name}: ESG Coefficients Over Time')
        ax2.set_xticks(x_pos + width)
        ax2.set_xticklabels(stability_df['Period'], rotation=45)
        ax2.legend()
        ax2.grid(True, alpha=0.3)

        # Plot 3: Observations by period
        ax3 = axes[1, 0]
        bars3 = ax3.bar(stability_df['Period'], stability_df['Observations'],
                       alpha=0.8, color='coral')
        ax3.set_xlabel('Time Period')
        ax3.set_ylabel('Observations')
        ax3.set_title(f'{dataset_name}: Sample Size by Period')
        ax3.tick_params(axis='x', rotation=45)
        ax3.grid(True, alpha=0.3)

        for bar, obs in zip(bars3, stability_df['Observations']):
            height = bar.get_height()
            ax3.text(bar.get_x() + bar.get_width()/2., height,
                    f'{int(obs)}', ha='center', va='bottom')

        # Plot 4: Coefficient stability heatmap
        ax4 = axes[1, 1]

        # Create matrix of coefficients
        coeff_matrix = []
        for component in ['E', 'S', 'G']:
            coeff_col = f'{component}_Coefficient'
            if coeff_col in stability_df.columns:
                coeff_matrix.append(stability_df[coeff_col].values)

        if coeff_matrix:
            im = ax4.imshow(coeff_matrix, aspect='auto', cmap='RdBu_r')
            ax4.set_yticks(range(len(['E', 'S', 'G'])))
            ax4.set_yticklabels(['E', 'S', 'G'])
            ax4.set_xticks(range(len(stability_df)))
            ax4.set_xticklabels(stability_df['Period'], rotation=45, ha='right')
            ax4.set_xlabel('Time Period')
            ax4.set_title('Coefficient Stability Heatmap')
            plt.colorbar(im, ax=ax4, label='Coefficient')

        plt.tight_layout()
        plot_path = os.path.join(self.output_dir,
                                f'{dataset_name.lower().replace(" ", "_")}_stability_plot.png')
        plt.savefig(plot_path, dpi=150, bbox_inches='tight')
        plt.show()
        print(f"✓ Saved stability plot to: {plot_path}")

    def run_complete_analysis(self):
        """
        Run complete enhanced analysis with all four extensions.
        """
        print("=" * 70)
        print("ENHANCED ESG CAPITAL STOCK ANALYSIS")
        print("=" * 70)
        print("\nThis enhanced analysis includes:")
        print("1. SEPARATE optimal rates for All Sectors vs Consumer Staples")
        print("2. Component-specific optimal rates for E, S, and G")
        print("3. Sector comparison of optimal rates")
        print("4. Stability testing over time")
        print("=" * 70)

        # Step 1: Load data
        print("\n" + "=" * 70)
        print("STEP 1: LOADING DATA")
        print("=" * 70)
        self.load_and_prepare_data()

        if self.df is None or len(self.df) == 0:
            print("✗ No data loaded. Exiting.")
            return

        # =========================================================
        # PART 1: ALL SECTORS ANALYSIS
        # =========================================================
        print("\n" + "=" * 70)
        print("PART 1: ALL SECTORS ANALYSIS")
        print("=" * 70)

        # 1A: Combined optimal rate for All Sectors
        print("\n" + "-" * 50)
        print("1A: Combined Model - All Sectors")
        print("-" * 50)

        best_rate_all, results_all = self.find_optimal_rate_combined_components(
            dataset=self.df,
            dataset_name="ALL SECTORS",
            add_consumer_staples_dummy=True
        )

        self.optimal_rates['all_sectors']['combined'] = {
            'optimal_rate': best_rate_all,
            'results': results_all
        }

        # 1B: Component-specific rates for All Sectors
        print("\n" + "-" * 50)
        print("1B: Component-Specific Models - All Sectors")
        print("-" * 50)

        component_rates_all = self.find_optimal_rates_all_components(
            dataset=self.df,
            dataset_name="ALL SECTORS",
            add_consumer_staples_dummy=True
        )

        self.optimal_rates['all_sectors']['components'] = component_rates_all

        # 1C: Stability test for All Sectors
        if best_rate_all is not None:
            print("\n" + "-" * 50)
            print("1C: Stability Test - All Sectors")
            print("-" * 50)

            stability_all = self.test_stability_over_time(
                dataset=self.df,
                dataset_name="ALL SECTORS",
                optimal_rate=best_rate_all,
                add_consumer_staples_dummy=True
            )

            self.optimal_rates['all_sectors']['stability'] = stability_all

        # =========================================================
        # PART 2: CONSUMER STAPLES ANALYSIS
        # =========================================================
        print("\n" + "=" * 70)
        print("PART 2: CONSUMER STAPLES ANALYSIS")
        print("=" * 70)

        if self.df_cs is not None and len(self.df_cs) > 0:
            # 2A: Combined optimal rate for Consumer Staples
            print("\n" + "-" * 50)
            print("2A: Combined Model - Consumer Staples")
            print("-" * 50)

            best_rate_cs, results_cs = self.find_optimal_rate_combined_components(
                dataset=self.df_cs,
                dataset_name="CONSUMER STAPLES",
                add_consumer_staples_dummy=False
            )

            self.optimal_rates['consumer_staples']['combined'] = {
                'optimal_rate': best_rate_cs,
                'results': results_cs
            }

            # 2B: Component-specific rates for Consumer Staples
            print("\n" + "-" * 50)
            print("2B: Component-Specific Models - Consumer Staples")
            print("-" * 50)

            component_rates_cs = self.find_optimal_rates_all_components(
                dataset=self.df_cs,
                dataset_name="CONSUMER STAPLES",
                add_consumer_staples_dummy=False
            )

            self.optimal_rates['consumer_staples']['components'] = component_rates_cs

            # 2C: Stability test for Consumer Staples
            if best_rate_cs is not None:
                print("\n" + "-" * 50)
                print("2C: Stability Test - Consumer Staples")
                print("-" * 50)

                stability_cs = self.test_stability_over_time(
                    dataset=self.df_cs,
                    dataset_name="CONSUMER STAPLES",
                    optimal_rate=best_rate_cs,
                    add_consumer_staples_dummy=False
                )

                self.optimal_rates['consumer_staples']['stability'] = stability_cs
        else:
            print("✗ No Consumer Staples data available. Skipping Part 2.")

        # =========================================================
        # PART 3: SECTOR COMPARISON
        # =========================================================
        print("\n" + "=" * 70)
        print("PART 3: SECTOR COMPARISON ANALYSIS")
        print("=" * 70)

        sector_results = self.analyze_sector_differences()

        # =========================================================
        # PART 4: SUMMARY AND SAVE RESULTS
        # =========================================================
        print("\n" + "=" * 70)
        print("PART 4: SUMMARY AND SAVING RESULTS")
        print("=" * 70)

        self.save_enhanced_results()

        print("\n" + "=" * 70)
        print("ENHANCED ANALYSIS COMPLETE")
        print("=" * 70)

        # Print comprehensive summary
        self.print_comprehensive_summary()

    def save_enhanced_results(self):
        """
        Save all enhanced analysis results.
        """
        print("\nSaving all enhanced results...")

        # 1. Save optimal rates summary
        summary_path = os.path.join(self.output_dir, 'enhanced_optimal_rates_summary.csv')

        summary_data = []

        # All sectors - combined
        if 'combined' in self.optimal_rates['all_sectors']:
            summary_data.append({
                'Sample': 'All Sectors',
                'Model': 'Combined',
                'Optimal_Rate': self.optimal_rates['all_sectors']['combined']['optimal_rate'],
                'R_Squared': self.optimal_rates['all_sectors']['combined']['results'].loc[
                    self.optimal_rates['all_sectors']['combined']['results']['R_Squared'].idxmax()
                ]['R_Squared'] if self.optimal_rates['all_sectors']['combined']['results'] is not None else None
            })

        # All sectors - components
        if 'components' in self.optimal_rates['all_sectors']:
            for component, comp_data in self.optimal_rates['all_sectors']['components'].items():
                summary_data.append({
                    'Sample': 'All Sectors',
                    'Model': f'{component} Only',
                    'Optimal_Rate': comp_data['optimal_rate'],
                    'R_Squared': comp_data['results'].loc[
                        comp_data['results']['R_Squared'].idxmax()
                    ]['R_Squared'] if comp_data['results'] is not None else None
                })

        # Consumer Staples - combined
        if 'combined' in self.optimal_rates['consumer_staples']:
            summary_data.append({
                'Sample': 'Consumer Staples',
                'Model': 'Combined',
                'Optimal_Rate': self.optimal_rates['consumer_staples']['combined']['optimal_rate'],
                'R_Squared': self.optimal_rates['consumer_staples']['combined']['results'].loc[
                    self.optimal_rates['consumer_staples']['combined']['results']['R_Squared'].idxmax()
                ]['R_Squared'] if self.optimal_rates['consumer_staples']['combined']['results'] is not None else None
            })

        # Consumer Staples - components
        if 'components' in self.optimal_rates['consumer_staples']:
            for component, comp_data in self.optimal_rates['consumer_staples']['components'].items():
                summary_data.append({
                    'Sample': 'Consumer Staples',
                    'Model': f'{component} Only',
                    'Optimal_Rate': comp_data['optimal_rate'],
                    'R_Squared': comp_data['results'].loc[
                        comp_data['results']['R_Squared'].idxmax()
                    ]['R_Squared'] if comp_data['results'] is not None else None
                })

        # Sector comparison
        if 'by_sector' in self.optimal_rates:
            for sector, sector_data in self.optimal_rates['by_sector'].items():
                summary_data.append({
                    'Sample': sector,
                    'Model': 'Combined',
                    'Optimal_Rate': sector_data['optimal_rate'],
                    'R_Squared': sector_data['r2']
                })

        if summary_data:
            summary_df = pd.DataFrame(summary_data)
            summary_df.to_csv(summary_path, index=False)
            print(f"✓ Saved enhanced optimal rates summary to: {summary_path}")

        # 2. Create comparison visualization
        self.create_comprehensive_comparison_plot()

    def create_comprehensive_comparison_plot(self):
        """
        Create comprehensive visualization comparing all optimal rates.
        """
        fig, axes = plt.subplots(2, 2, figsize=(16, 12))

        # Plot 1: All Sectors - Combined vs Component
        ax1 = axes[0, 0]

        all_sectors_data = []
        all_sectors_labels = []

        if 'combined' in self.optimal_rates['all_sectors']:
            all_sectors_data.append(self.optimal_rates['all_sectors']['combined']['optimal_rate'])
            all_sectors_labels.append('Combined')

        if 'components' in self.optimal_rates['all_sectors']:
            for component in ['E', 'S', 'G']:
                if component in self.optimal_rates['all_sectors']['components']:
                    all_sectors_data.append(self.optimal_rates['all_sectors']['components'][component]['optimal_rate'])
                    all_sectors_labels.append(f'{component} Only')

        if all_sectors_data:
            bars1 = ax1.bar(all_sectors_labels, all_sectors_data, color=['navy', 'blue', 'green', 'orange'])
            ax1.set_ylabel('Optimal Depreciation Rate (δ)')
            ax1.set_title('All Sectors: Optimal Rates by Model')
            ax1.grid(True, alpha=0.3)

            for bar, rate in zip(bars1, all_sectors_data):
                height = bar.get_height()
                ax1.text(bar.get_x() + bar.get_width()/2., height,
                        f'{rate:.2f}', ha='center', va='bottom')

        # Plot 2: Consumer Staples - Combined vs Component
        ax2 = axes[0, 1]

        cs_data = []
        cs_labels = []

        if 'combined' in self.optimal_rates['consumer_staples']:
            cs_data.append(self.optimal_rates['consumer_staples']['combined']['optimal_rate'])
            cs_labels.append('Combined')

        if 'components' in self.optimal_rates['consumer_staples']:
            for component in ['E', 'S', 'G']:
                if component in self.optimal_rates['consumer_staples']['components']:
                    cs_data.append(self.optimal_rates['consumer_staples']['components'][component]['optimal_rate'])
                    cs_labels.append(f'{component} Only')

        if cs_data:
            bars2 = ax2.bar(cs_labels, cs_data, color=['darkred', 'red', 'lightcoral', 'salmon'])
            ax2.set_ylabel('Optimal Depreciation Rate (δ)')
            ax2.set_title('Consumer Staples: Optimal Rates by Model')
            ax2.grid(True, alpha=0.3)

            for bar, rate in zip(bars2, cs_data):
                height = bar.get_height()
                ax2.text(bar.get_x() + bar.get_width()/2., height,
                        f'{rate:.2f}', ha='center', va='bottom')

        # Plot 3: Sector Comparison
        ax3 = axes[1, 0]

        if 'by_sector' in self.optimal_rates:
            sectors = list(self.optimal_rates['by_sector'].keys())
            rates = [self.optimal_rates['by_sector'][s]['optimal_rate'] for s in sectors]

            colors = sns.color_palette("husl", len(sectors))
            bars3 = ax3.bar(sectors, rates, color=colors)
            ax3.set_xlabel('Sector')
            ax3.set_ylabel('Optimal Depreciation Rate (δ)')
            ax3.set_title('Optimal Rates by Sector')
            ax3.tick_params(axis='x', rotation=45)
            ax3.grid(True, alpha=0.3)

            # Add Consumer Staples benchmark
            if 'combined' in self.optimal_rates['consumer_staples']:
                cs_rate = self.optimal_rates['consumer_staples']['combined']['optimal_rate']
                ax3.axhline(y=cs_rate, color='red', linestyle='--',
                          label=f'Consumer Staples: {cs_rate:.2f}')
                ax3.legend()

            for bar, rate in zip(bars3, rates):
                height = bar.get_height()
                ax3.text(bar.get_x() + bar.get_width()/2., height,
                        f'{rate:.2f}', ha='center', va='bottom')

        # Plot 4: Stability Summary
        ax4 = axes[1, 1]

        stability_data = []
        stability_labels = []

        if 'stability' in self.optimal_rates['all_sectors']:
            stability_all = self.optimal_rates['all_sectors']['stability']
            if stability_all is not None and 'full' in stability_all['Period'].values:
                full_r2 = stability_all[stability_all['Period'] == 'full']['R_Squared'].values[0]
                early_r2 = stability_all[stability_all['Period'] == 'early']['R_Squared'].values[0] if 'early' in stability_all['Period'].values else None
                late_r2 = stability_all[stability_all['Period'] == 'late']['R_Squared'].values[0] if 'late' in stability_all['Period'].values else None

                stability_data.extend([full_r2, early_r2, late_r2])
                stability_labels.extend(['Full', 'Early', 'Late'])

        if stability_data:
            bars4 = ax4.bar(stability_labels, stability_data, color=['purple', 'violet', 'plum'])
            ax4.set_ylabel('R-squared')
            ax4.set_title('All Sectors: Stability Over Time')
            ax4.grid(True, alpha=0.3)

            for bar, r2 in zip(bars4, stability_data):
                if r2 is not None:
                    height = bar.get_height()
                    ax4.text(bar.get_x() + bar.get_width()/2., height,
                            f'{r2:.3f}', ha='center', va='bottom')

        plt.tight_layout()
        plot_path = os.path.join(self.output_dir, 'comprehensive_comparison.png')
        plt.savefig(plot_path, dpi=150, bbox_inches='tight')
        plt.show()
        print(f"✓ Saved comprehensive comparison plot to: {plot_path}")

    def print_comprehensive_summary(self):
        """
        Print comprehensive summary of all findings.
        """
        print("\n" + "=" * 70)
        print("COMPREHENSIVE SUMMARY OF FINDINGS")
        print("=" * 70)

        print("\n1. OPTIMAL DEPRECIATION RATES - ALL SECTORS")
        print("-" * 50)
        if 'combined' in self.optimal_rates['all_sectors']:
            rate = self.optimal_rates['all_sectors']['combined']['optimal_rate']
            print(f"   Combined Model: δ = {rate:.2f}")

        if 'components' in self.optimal_rates['all_sectors']:
            for component in ['E', 'S', 'G']:
                if component in self.optimal_rates['all_sectors']['components']:
                    rate = self.optimal_rates['all_sectors']['components'][component]['optimal_rate']
                    print(f"   {component} Only: δ = {rate:.2f}")

        print("\n2. OPTIMAL DEPRECIATION RATES - CONSUMER STAPLES")
        print("-" * 50)
        if 'combined' in self.optimal_rates['consumer_staples']:
            rate = self.optimal_rates['consumer_staples']['combined']['optimal_rate']
            print(f"   Combined Model: δ = {rate:.2f}")

        if 'components' in self.optimal_rates['consumer_staples']:
            for component in ['E', 'S', 'G']:
                if component in self.optimal_rates['consumer_staples']['components']:
                    rate = self.optimal_rates['consumer_staples']['components'][component]['optimal_rate']
                    print(f"   {component} Only: δ = {rate:.2f}")

        print("\n3. COMPARISON: ALL SECTORS vs CONSUMER STAPLES")
        print("-" * 50)
        if ('combined' in self.optimal_rates['all_sectors'] and
            'combined' in self.optimal_rates['consumer_staples']):
            rate_all = self.optimal_rates['all_sectors']['combined']['optimal_rate']
            rate_cs = self.optimal_rates['consumer_staples']['combined']['optimal_rate']
            diff = rate_cs - rate_all
            print(f"   All Sectors: δ = {rate_all:.2f}")
            print(f"   Consumer Staples: δ = {rate_cs:.2f}")
            print(f"   Difference: {diff:+.2f}")
            if diff > 0:
                print(f"   → Consumer Staples has FASTER depreciation (higher δ)")
            else:
                print(f"   → Consumer Staples has SLOWER depreciation (lower δ)")

        print("\n4. COMPONENT-SPECIFIC DIFFERENCES")
        print("-" * 50)
        if ('components' in self.optimal_rates['all_sectors'] and
            'components' in self.optimal_rates['consumer_staples']):
            for component in ['E', 'S', 'G']:
                if (component in self.optimal_rates['all_sectors']['components'] and
                    component in self.optimal_rates['consumer_staples']['components']):
                    rate_all = self.optimal_rates['all_sectors']['components'][component]['optimal_rate']
                    rate_cs = self.optimal_rates['consumer_staples']['components'][component]['optimal_rate']
                    diff = rate_cs - rate_all
                    print(f"   {component}: All={rate_all:.2f}, CS={rate_cs:.2f}, Diff={diff:+.2f}")

        print("\n5. SECTOR COMPARISON")
        print("-" * 50)
        if 'by_sector' in self.optimal_rates:
            # Sort sectors by optimal rate
            sorted_sectors = sorted(self.optimal_rates['by_sector'].items(),
                                  key=lambda x: x[1]['optimal_rate'])

            print("   Optimal rates by sector (lowest to highest):")
            for sector, data in sorted_sectors:
                print(f"   {sector}: δ = {data['optimal_rate']:.2f}")

            # Find highest and lowest
            if sorted_sectors:
                lowest_sector, lowest_data = sorted_sectors[0]
                highest_sector, highest_data = sorted_sectors[-1]
                print(f"\n   Lowest: {lowest_sector} (δ = {lowest_data['optimal_rate']:.2f})")
                print(f"   Highest: {highest_sector} (δ = {highest_data['optimal_rate']:.2f})")

        print("\n6. STABILITY OVER TIME")
        print("-" * 50)
        if 'stability' in self.optimal_rates['all_sectors']:
            stability = self.optimal_rates['all_sectors']['stability']
            if stability is not None:
                full_r2 = stability[stability['Period'] == 'full']['R_Squared'].values[0] if 'full' in stability['Period'].values else None
                early_r2 = stability[stability['Period'] == 'early']['R_Squared'].values[0] if 'early' in stability['Period'].values else None
                late_r2 = stability[stability['Period'] == 'late']['R_Squared'].values[0] if 'late' in stability['Period'].values else None

                print(f"   All Sectors - Full period: R² = {full_r2:.4f}" if full_r2 else "")
                print(f"   All Sectors - Early period: R² = {early_r2:.4f}" if early_r2 else "")
                print(f"   All Sectors - Late period: R² = {late_r2:.4f}" if late_r2 else "")

                if early_r2 and late_r2:
                    change = late_r2 - early_r2
                    print(f"   R² change: {change:+.4f}")

        print("\n" + "=" * 70)
        print("All results saved to: {}".format(self.output_dir))
        print("=" * 70)


def main():
    """
    Main function to run the enhanced analysis.
    """
    # Configuration
    DATA_PATH = '/content/drive/MyDrive/Colab Notebooks/FTSE Data/FTSE data.xlsx'
    OUTPUT_DIR = '/content/drive/MyDrive/Colab Notebooks/FTSE Data/esg_capital_stock_enhanced'

    # Initialize and run enhanced analysis
    analyzer = ESGCapitalStockOptimalRatesEnhanced(DATA_PATH, OUTPUT_DIR)
    analyzer.run_complete_analysis()


if __name__ == "__main__":
    main()
```
