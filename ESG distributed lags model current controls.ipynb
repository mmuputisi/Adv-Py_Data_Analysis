```python
# standalone_lag_model_corrected.py
"""
Standalone Lag Model Analysis - Corrected to match original structured analysis
"""

# ============================================================================
# SETUP AND IMPORTS
# ============================================================================
from google.colab import drive
drive.mount('/content/drive')

!pip install -q linearmodels pandas numpy matplotlib seaborn statsmodels

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from linearmodels.panel import PanelOLS
import warnings
warnings.filterwarnings('ignore')

# ============================================================================
# CONFIGURATION
# ============================================================================
class Config:
    """Configuration parameters matching original structured analysis"""
    PROJECT_DIR = '/content/drive/MyDrive/Colab Notebooks/FTSE Data'
    DATA_PATH = os.path.join(PROJECT_DIR, 'FTSE data.xlsx')
    OUTPUT_DIR = os.path.join(PROJECT_DIR, 'lag_model_outputs_corrected')

    # Match original structured analysis settings
    ESG_COMPONENTS = ['E', 'S', 'G']
    CONTROL_VARS = ['ROA', 'Size', 'Leverage']

    # Consumer Staples firms
    CONSUMER_STAPLES_FIRMS = [
        'Associated British Foods', 'British American Tobacco', 'Coca Cola HBC AG',
        'Diageo', 'J. Sainsbury', 'Reckitt Benckiser', 'Tesco', 'Unilever', 'Haleon Plc'
    ]

# ============================================================================
# DATA LOADER - MATCHING ORIGINAL
# ============================================================================
class DataLoader:
    """Load and prepare data exactly as in structured analysis"""

    def __init__(self, config):
        self.config = config
        self.df = None
        self.df_cs = None  # Consumer Staples only
        self.df_all = None  # All sectors

    def load_and_prepare_data(self):
        """Load data exactly as in original structured analysis"""
        print("Loading and preparing data...")

        # 1. Load data
        self.df = pd.read_excel(self.config.DATA_PATH)
        self.df = self.df.set_index(['Firm', 'Year']).sort_index()
        self.df = self.df.rename(columns={"Tobin's Q": "Tobin_Q"})

        # 2. Create variables exactly as in original
        self.df['Size'] = np.log(self.df['Total Assets'])
        self.df['Leverage'] = self.df['Total Liabilities'] / self.df['Total Assets']
        self.df['Tobin_Q_log'] = np.log(self.df['Tobin_Q'] + 0.001)

        # 3. Create lagged ESG variables (t-1)
        self.df['E_lag'] = self.df.groupby(level=0)['E'].shift(1)
        self.df['S_lag'] = self.df.groupby(level=0)['S'].shift(1)
        self.df['G_lag'] = self.df.groupby(level=0)['G'].shift(1)

        # 4. Create lagged control variables
        self.df['ROA_lag'] = self.df.groupby(level=0)['ROA'].shift(1)
        self.df['Size_lag'] = self.df.groupby(level=0)['Size'].shift(1)
        self.df['Leverage_lag'] = self.df.groupby(level=0)['Leverage'].shift(1)

        # 5. Create sector variable
        sectors = []
        for firm in self.df.index.get_level_values(0):
            if firm in self.config.CONSUMER_STAPLES_FIRMS:
                sectors.append('Consumer Staples')
            else:
                sectors.append('Other')

        self.df['Sector'] = sectors
        self.df['ConsumerStaples'] = (self.df['Sector'] == 'Consumer Staples').astype(int)

        # 6. Create datasets
        self.df_cs = self.df[self.df['ConsumerStaples'] == 1].copy()
        self.df_all = self.df.copy()

        print(f"✓ Total observations: {len(self.df)}")
        print(f"✓ Consumer Staples: {len(self.df_cs)} observations")
        print(f"✓ All sectors: {len(self.df_all)} observations")

        return self.df, self.df_cs, self.df_all

    def create_lagged_variables_for_model(self, df, min_lag=1, max_lag=5, use_original_controls=True):
        """
        Create lagged variables exactly as in original distributed lag model

        Parameters:
        -----------
        df : DataFrame
            Input dataframe
        min_lag : int
            Minimum lag (1)
        max_lag : int
            Maximum lag (1-5)
        use_original_controls : bool
            If True, use original controls (not lagged) as in structured analysis
        """
        df_copy = df.copy()

        # Create distributed lags for ESG - INDIVIDUAL LAGS
        print(f"Creating distributed lags {min_lag}-{max_lag} for ESG...")
        for component in ['E', 'S', 'G']:
            for lag in range(min_lag, max_lag + 1):
                df_copy[f'{component}_lag{lag}'] = df_copy.groupby(level=0)[component].shift(lag)

        # Also create summed versions as in original
        for component in ['E', 'S', 'G']:
            # Initialize with first lag
            if max_lag >= 1:
                lag1 = df_copy.groupby(level=0)[component].shift(1)
                df_copy[f'{component}_lag1'] = lag1

            # Create summed versions
            if max_lag >= 2:
                lag2 = df_copy.groupby(level=0)[component].shift(2)
                df_copy[f'{component}_lag12'] = lag1 + lag2

            if max_lag >= 3:
                lag3 = df_copy.groupby(level=0)[component].shift(3)
                df_copy[f'{component}_lag13'] = lag1 + lag2 + lag3

            if max_lag >= 4:
                lag4 = df_copy.groupby(level=0)[component].shift(4)
                df_copy[f'{component}_lag14'] = lag1 + lag2 + lag3 + lag4

            if max_lag >= 5:
                lag5 = df_copy.groupby(level=0)[component].shift(5)
                df_copy[f'{component}_lag15'] = lag1 + lag2 + lag3 + lag4 + lag5

        # Use control variables as in original structured analysis
        if use_original_controls:
            control_vars = ['ROA', 'Size', 'Leverage']
        else:
            control_vars = ['ROA_lag', 'Size_lag', 'Leverage_lag']

        return df_copy, control_vars

# ============================================================================
# LAG MODEL ANALYZER - MATCHING ORIGINAL
# ============================================================================
class LagModelAnalyzer:
    """Analyzer that matches original structured analysis"""

    def __init__(self, config):
        self.config = config
        self.loader = DataLoader(config)
        self.results = {}

    def run_analysis(self, max_lag=5):
        """Run analysis matching original structured analysis"""
        print("\n" + "="*80)
        print(f"RUNNING LAG MODEL ANALYSIS (1-{max_lag} years)")
        print("="*80)

        # Load data
        df, df_cs, df_all = self.loader.load_and_prepare_data()

        # Run models for different sectors
        sectors = [
            ('cs', df_cs, 'Consumer Staples Only'),
            ('all', df_all, 'All Sectors')
        ]

        for sector_code, sector_df, sector_name in sectors:
            print(f"\n{'='*60}")
            print(f"ANALYZING: {sector_name}")
            print(f"{'='*60}")

            # Create lagged variables
            df_lagged, control_vars = self.loader.create_lagged_variables_for_model(
                sector_df, min_lag=1, max_lag=max_lag, use_original_controls=True
            )

            # Run individual distributed lag model (Model 4 in original)
            print(f"\n1. DISTRIBUTED LAG MODEL (Individual Lags 1-{max_lag})")
            self._run_distributed_lag_individual(df_lagged, max_lag, sector_code,
                                                sector_name, control_vars)

            # Run summed distributed lag model (Model 5 in original)
            print(f"\n2. DISTRIBUTED LAG MODEL (Summed Lags 1-{max_lag})")
            self._run_distributed_lag_summed(df_lagged, max_lag, sector_code,
                                           sector_name, control_vars)

    def _run_distributed_lag_individual(self, df, max_lag, sector_code,
                                       sector_name, control_vars):
        """Run distributed lag model with individual lags (Model 4 in original)"""
        # Prepare lag variables
        lag_vars = []
        for component in ['E', 'S', 'G']:
            for lag in range(1, max_lag + 1):
                lag_vars.append(f'{component}_lag{lag}')

        # Add ConsumerStaples dummy for all sectors
        if sector_code == 'all':
            all_vars = lag_vars + control_vars + ['ConsumerStaples', 'Tobin_Q']
        else:
            all_vars = lag_vars + control_vars + ['Tobin_Q']

        # Clean data
        df_clean = df.dropna(subset=all_vars)

        if len(df_clean) < 10:
            print(f"⚠ Insufficient data: {len(df_clean)} observations")
            return None

        print(f"Observations: {len(df_clean)}")
        print(f"Firms: {df_clean.index.get_level_values(0).nunique()}")

        # Prepare regression
        y = df_clean['Tobin_Q']
        X_vars = lag_vars + control_vars
        if sector_code == 'all':
            X_vars = X_vars + ['ConsumerStaples']

        X = df_clean[X_vars]
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

        # Run regression
        try:
            model = PanelOLS(y, X, entity_effects=False, time_effects=False)
            results = model.fit(cov_type='robust')

            # Store results
            key = f"distributed_individual_{max_lag}_{sector_code}"
            self.results[key] = {
                'results': results,
                'type': 'distributed_individual',
                'sector': sector_code,
                'max_lag': max_lag,
                'nobs': results.nobs,
                'r2': results.rsquared,
                'data': df_clean
            }

            # Print results
            self._print_results(results, f"Distributed Lag (Individual) - {sector_name}")

            # Create lag structure plot
            self._create_lag_structure_plot(results, max_lag,
                                          f"Distributed Lag Individual - {sector_name}")

            return results

        except Exception as e:
            print(f"Error running model: {e}")
            return None

    def _run_distributed_lag_summed(self, df, max_lag, sector_code,
                                   sector_name, control_vars):
        """Run distributed lag model with summed lags (Model 5 in original)"""
        # Use summed variables (e.g., E_lag15 for sum of lags 1-5)
        summed_vars = [f'{component}_lag1{max_lag}' for component in ['E', 'S', 'G']]

        # Add ConsumerStaples dummy for all sectors
        if sector_code == 'all':
            all_vars = summed_vars + control_vars + ['ConsumerStaples', 'Tobin_Q']
        else:
            all_vars = summed_vars + control_vars + ['Tobin_Q']

        # Clean data
        df_clean = df.dropna(subset=all_vars)

        if len(df_clean) < 10:
            print(f"⚠ Insufficient data: {len(df_clean)} observations")
            return None

        print(f"Observations: {len(df_clean)}")

        # Prepare regression
        y = df_clean['Tobin_Q']
        X_vars = summed_vars + control_vars
        if sector_code == 'all':
            X_vars = X_vars + ['ConsumerStaples']

        X = df_clean[X_vars]
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

        # Run regression
        try:
            model = PanelOLS(y, X, entity_effects=False, time_effects=False)
            results = model.fit(cov_type='robust')

            # Store results
            key = f"distributed_summed_{max_lag}_{sector_code}"
            self.results[key] = {
                'results': results,
                'type': 'distributed_summed',
                'sector': sector_code,
                'max_lag': max_lag,
                'nobs': results.nobs,
                'r2': results.rsquared,
                'data': df_clean
            }

            # Print results
            self._print_results(results, f"Distributed Lag (Summed) - {sector_name}")

            return results

        except Exception as e:
            print(f"Error running model: {e}")
            return None

    def _print_results(self, results, title):
        """Print formatted results matching original style"""
        print(f"\n{title}")
        print("-" * 60)
        print(f"R-squared: {results.rsquared:.4f}")
        print(f"Observations: {results.nobs}")

        if hasattr(results, 'f_statistic'):
            print(f"F-statistic: {results.f_statistic.stat:.4f} (p={results.f_statistic.pval:.4f})")

        # Print ESG coefficients with significance
        print("\nESG Coefficients:")
        print("-" * 40)

        # Group by component
        for component in ['E', 'S', 'G']:
            component_vars = [v for v in results.params.index if component in str(v)]
            if component_vars:
                print(f"\n{component} component:")
                for var in component_vars:
                    if var in results.params and var != 'const':
                        coeff = results.params[var]
                        pval = results.pvalues.get(var, 1.0)
                        sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                        print(f"  {var:<15} {coeff:>10.4f}{sig:<3} (p={pval:.4f})")

    def _create_lag_structure_plot(self, results, max_lag, title):
        """Create lag structure plot"""
        fig, axes = plt.subplots(1, 3, figsize=(15, 4))
        fig.suptitle(title, fontsize=14, fontweight='bold')

        for idx, component in enumerate(['E', 'S', 'G']):
            lags = list(range(1, max_lag + 1))
            coeffs = []
            errors = []

            for lag in lags:
                var = f'{component}_lag{lag}'
                if var in results.params:
                    coeffs.append(results.params[var])
                    if hasattr(results, 'std_errors') and var in results.std_errors:
                        errors.append(results.std_errors[var])
                    else:
                        errors.append(0)
                else:
                    coeffs.append(0)
                    errors.append(0)

            axes[idx].errorbar(lags, coeffs, yerr=errors, fmt='o-',
                             capsize=5, linewidth=2, markersize=8)
            axes[idx].axhline(y=0, color='r', linestyle='--', alpha=0.5)
            axes[idx].set_xlabel('Years Lag')
            axes[idx].set_ylabel('Coefficient')
            axes[idx].set_title(f'{component} Component')
            axes[idx].set_xticks(lags)
            axes[idx].grid(True, alpha=0.3)

        plt.tight_layout()

        # Save plot
        os.makedirs(self.config.OUTPUT_DIR, exist_ok=True)
        plot_filename = f"lag_structure_{title.replace(' ', '_').replace('-', '_')}.png"
        plt.savefig(os.path.join(self.config.OUTPUT_DIR, plot_filename),
                   dpi=300, bbox_inches='tight')
        plt.show()

        return fig

    def compare_with_original_simple(self, max_lag=5):
        """
        Run simple lag model matching the ORIGINAL simple approach
        (not the structured analysis, but the simple lag model)
        """
        print("\n" + "="*80)
        print("SIMPLE LAG MODEL (Matching Original Simple Approach)")
        print("="*80)

        # Load data
        df, df_cs, df_all = self.loader.load_and_prepare_data()

        # For Consumer Staples
        print("\nCONSUMER STAPLES ONLY:")
        df_cs_lagged, control_vars = self.loader.create_lagged_variables_for_model(
            df_cs, min_lag=1, max_lag=max_lag, use_original_controls=True
        )

        # Use simple approach: just E_lag, S_lag, G_lag
        simple_vars = ['E_lag', 'S_lag', 'G_lag'] + control_vars + ['Tobin_Q']
        df_simple = df_cs_lagged.dropna(subset=simple_vars)

        if len(df_simple) > 0:
            y = df_simple['Tobin_Q']
            X = df_simple[['E_lag', 'S_lag', 'G_lag'] + control_vars]
            X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

            model = PanelOLS(y, X, entity_effects=False, time_effects=False)
            results = model.fit(cov_type='robust')

            print(f"\nSimple Lag Model (t-1 only):")
            print(f"Observations: {results.nobs}")
            print(f"R-squared: {results.rsquared:.4f}")

            for var in ['E_lag', 'S_lag', 'G_lag']:
                if var in results.params:
                    coeff = results.params[var]
                    pval = results.pvalues.get(var, 1.0)
                    sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                    print(f"  {var}: {coeff:.4f}{sig} (p={pval:.4f})")

        return results

# ============================================================================
# RESULTS COMPARISON
# ============================================================================
class ResultsComparator:
    """Compare results from different model specifications"""

    def __init__(self, analyzer):
        self.analyzer = analyzer
        self.comparison_df = None

    def create_comparison_table(self):
        """Create comparison table of all results"""
        comparison_data = []

        for key, model_info in self.analyzer.results.items():
            results = model_info['results']

            row = {
                'Model_Type': model_info['type'],
                'Sector': model_info['sector'],
                'Max_Lag': model_info['max_lag'],
                'N': results.nobs,
                'R2': results.rsquared,
            }

            # Extract key coefficients
            for component in ['E', 'S', 'G']:
                # For individual models, get average across lags
                if 'individual' in model_info['type']:
                    coeffs = []
                    for var in results.params.index:
                        if component in str(var) and 'lag' in str(var):
                            coeffs.append(results.params[var])
                    if coeffs:
                        row[f'{component}_avg'] = np.mean(coeffs)
                        row[f'{component}_min'] = min(coeffs)
                        row[f'{component}_max'] = max(coeffs)

                # For summed models
                elif 'summed' in model_info['type']:
                    var = f'{component}_lag1{model_info["max_lag"]}'
                    if var in results.params:
                        row[f'{component}_sum'] = results.params[var]

            comparison_data.append(row)

        self.comparison_df = pd.DataFrame(comparison_data)
        return self.comparison_df

    def print_comparison(self):
        """Print comparison of results"""
        if self.comparison_df is None:
            self.create_comparison_table()

        print("\n" + "="*80)
        print("MODEL COMPARISON")
        print("="*80)

        # Sort by R2
        self.comparison_df = self.comparison_df.sort_values('R2', ascending=False)

        for _, row in self.comparison_df.iterrows():
            print(f"\n{row['Model_Type'].replace('_', ' ').title()}")
            print(f"Sector: {row['Sector'].upper()}, Max Lag: {row['Max_Lag']}")
            print(f"R²: {row['R2']:.4f}, N: {row['N']}")

            if 'individual' in row['Model_Type']:
                print("Average ESG effects:")
                for component in ['E', 'S', 'G']:
                    if f'{component}_avg' in row and not pd.isna(row[f'{component}_avg']):
                        print(f"  {component}: {row[f'{component}_avg']:.4f} " +
                              f"[{row[f'{component}_min']:.4f}, {row[f'{component}_max']:.4f}]")

# ============================================================================
# MAIN EXECUTION
# ============================================================================
def main():
    """Main execution function"""
    print("\n" + "="*80)
    print("LAG MODEL ANALYSIS - CORRECTED TO MATCH ORIGINAL")
    print("="*80)

    # Initialize
    config = Config()
    analyzer = LagModelAnalyzer(config)

    # Run analysis with different lag specifications
    lag_ranges = [2, 3, 4, 5]  # Test different ranges

    for max_lag in lag_ranges:
        print(f"\n{'='*80}")
        print(f"ANALYZING LAG RANGE: 1-{max_lag} years")
        print(f"{'='*80}")

        analyzer.run_analysis(max_lag=max_lag)

    # Compare results
    comparator = ResultsComparator(analyzer)
    comparator.create_comparison_table()
    comparator.print_comparison()

    # Also run simple model for comparison
    analyzer.compare_with_original_simple(max_lag=5)

    print("\n" + "="*80)
    print("ANALYSIS COMPLETE")
    print("="*80)

    return analyzer, comparator

# ============================================================================
# EXECUTION
# ============================================================================
if __name__ == "__main__":
    analyzer, comparator = main()
```
