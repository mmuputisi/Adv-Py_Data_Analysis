```python
# esg_capital_stock_optimal_rates_separate.py
"""
ESG-Tobin's Q analysis that finds:
1. Optimal depreciation rate for ALL sectors (applied to all sectors)
2. Optimal depreciation rate for CONSUMER STAPLES (applied to consumer staples)
Both rates are found independently on their respective samples.
"""

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from linearmodels.panel import PanelOLS
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

from google.colab import drive
drive.mount('/content/drive')

!pip install -q linearmodels pandas numpy matplotlib seaborn statsmodels


class ESGCapitalStockOptimalRates:
    """
    ESG Capital Stock analysis that finds optimal depreciation rates SEPARATELY for:
    1. All sectors
    2. Consumer Staples
    """

    def __init__(self, data_path, output_dir):
        """
        Initialize the analysis.
        """
        self.data_path = data_path
        self.output_dir = output_dir
        self.df = None
        self.df_cs = None
        self.capital_stock_data = {}
        self.results = {}

        # Separate optimal rates for different samples
        self.best_rate_all_sectors = None
        self.best_rate_consumer_staples = None

        # Separate comparison dataframes
        self.comparison_all_sectors = None
        self.comparison_consumer_staples = None

        # Create output directory
        os.makedirs(output_dir, exist_ok=True)

        # Setup plotting
        plt.style.use('seaborn-v0_8-whitegrid')
        sns.set_palette("husl")

        # Define Consumer Staples firms
        self.consumer_staples_firms = [
            'Associated British Foods', 'British American Tobacco', 'Coca Cola HBC AG',
            'Diageo', 'J. Sainsbury', 'Reckitt Benckiser', 'Tesco', 'Unilever', 'Haleon Plc'
        ]

    def load_and_prepare_data(self):
        """
        Load data and prepare basic transformations.
        """
        print("=" * 70)
        print("LOADING AND PREPARING DATA")
        print("=" * 70)

        try:
            # Load data
            self.df = pd.read_excel(self.data_path)
            print(f"✓ Loaded data from: {self.data_path}")

            # Set multi-index
            if 'Firm' in self.df.columns and 'Year' in self.df.columns:
                self.df = self.df.set_index(['Firm', 'Year']).sort_index()
                print("✓ Set multi-index (Firm, Year)")
            else:
                # Try first two columns
                self.df = self.df.set_index([self.df.columns[0], self.df.columns[1]]).sort_index()
                print(f"✓ Using {self.df.index.names} as index")

            # Rename columns
            column_renames = {}
            if "Tobin's Q" in self.df.columns:
                column_renames["Tobin's Q"] = "Tobin_Q"

            if column_renames:
                self.df = self.df.rename(columns=column_renames)
                print(f"✓ Renamed columns: {column_renames}")

            # Create control variables
            if 'Total Assets' in self.df.columns:
                self.df['Size'] = np.log(self.df['Total Assets'].replace(0, np.nan))
                print("✓ Created Size variable")

            if 'Total Liabilities' in self.df.columns and 'Total Assets' in self.df.columns:
                self.df['Leverage'] = self.df['Total Liabilities'] / self.df['Total Assets'].replace(0, np.nan)
                print("✓ Created Leverage variable")

            # Create Consumer Staples dummy
            self.df['ConsumerStaples'] = self.df.index.get_level_values(0).isin(self.consumer_staples_firms).astype(int)

            # Create Consumer Staples only dataset
            self.df_cs = self.df[self.df['ConsumerStaples'] == 1].copy()

            print(f"\n✓ Total observations: {len(self.df)}")
            print(f"✓ Total firms: {self.df.index.get_level_values(0).nunique()}")
            print(f"✓ Consumer Staples observations: {len(self.df_cs)}")
            print(f"✓ Consumer Staples firms: {self.df_cs.index.get_level_values(0).nunique() if len(self.df_cs) > 0 else 0}")
            print(f"✓ Years: {sorted(self.df.index.get_level_values(1).unique())}")

            return self.df

        except Exception as e:
            print(f"✗ Error loading data: {e}")
            import traceback
            traceback.print_exc()
            return None

    def create_lagged_esg_variables(self, dataset):
        """
        Create lagged ESG variables.
        """
        if dataset is None or len(dataset) == 0:
            return dataset

        dataset_copy = dataset.copy()

        for component in ['E', 'S', 'G']:
            if component in dataset_copy.columns:
                dataset_copy[f'{component}_lag'] = dataset_copy.groupby(level=0)[component].shift(1)

        return dataset_copy

    def create_esg_capital_stock(self, dataset, depreciation_rate=0.3, use_lagged_esg=True):
        """
        Create ESG capital stock variables.
        """
        if dataset is None or len(dataset) == 0:
            return dataset

        dataset_copy = dataset.copy()

        # Determine which ESG variables to use
        if use_lagged_esg:
            esg_components = ['E_lag', 'S_lag', 'G_lag']
            suffix = '_lag'
        else:
            esg_components = ['E', 'S', 'G']
            suffix = ''

        # Check which components are available
        available_components = [c for c in esg_components if c in dataset_copy.columns]

        if not available_components:
            return dataset_copy

        for component in available_components:
            base_component = component.replace('_lag', '')
            col_name = f'{base_component}_capital_{int(depreciation_rate*100)}{suffix}'

            # Process each firm separately
            capital_series = []

            for firm in dataset_copy.index.get_level_values(0).unique():
                firm_data = dataset_copy.xs(firm, level=0).copy()

                if len(firm_data) > 0 and component in firm_data.columns:
                    esg_series = firm_data[component].fillna(0).values

                    # Calculate capital stock
                    capital = np.zeros(len(esg_series))

                    if len(esg_series) > 0:
                        capital[0] = esg_series[0]

                        for t in range(1, len(esg_series)):
                            capital[t] = (1 - depreciation_rate) * capital[t-1] + esg_series[t]

                    capital_series.extend(capital)
                else:
                    capital_series.extend([np.nan] * len(firm_data))

            dataset_copy[col_name] = capital_series

        return dataset_copy

    def run_regression_safe(self, df, y_var, x_vars, control_vars=None, add_consumer_staples_dummy=False):
        """
        Run a safe regression that won't produce R² = 1.
        """
        if df is None or len(df) == 0:
            return None

        # Combine x variables
        all_x_vars = x_vars.copy()
        if control_vars:
            all_x_vars.extend(control_vars)
        if add_consumer_staples_dummy and 'ConsumerStaples' in df.columns:
            all_x_vars.append('ConsumerStaples')

        # Check if variables exist
        available_vars = [v for v in all_x_vars if v in df.columns]

        if y_var not in df.columns:
            print(f"  Dependent variable {y_var} not found")
            return None

        if len(available_vars) == 0:
            print(f"  No independent variables available")
            return None

        # Get clean data
        df_clean = df.dropna(subset=[y_var] + available_vars).copy()

        if len(df_clean) < 10:
            print(f"  Insufficient observations: {len(df_clean)}")
            return None

        # Check for constant or near-constant variables
        valid_vars = []
        for var in available_vars:
            if df_clean[var].nunique() > 1:
                valid_vars.append(var)
            else:
                print(f"  Dropping constant variable: {var}")

        if len(valid_vars) == 0:
            print(f"  No valid independent variables with variation")
            return None

        # Prepare X and y
        y = df_clean[y_var]
        X = df_clean[valid_vars].copy()

        # Add constant
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)
        used_constant = True

        # Check for perfect multicollinearity
        try:
            from numpy.linalg import cond
            X_matrix = X.values.astype(float)
            cond_num = cond(X_matrix)

            if cond_num > 1e10:
                print(f"  Warning: Extreme multicollinearity (condition number: {cond_num:.2e})")
                # Try without constant
                X_no_const = df_clean[valid_vars].copy()
                used_constant = False

                X_no_const_matrix = X_no_const.values.astype(float)
                cond_num_no_const = cond(X_no_const_matrix)

                if cond_num_no_const < cond_num:
                    X = X_no_const
                    print(f"  Using model without constant term")
        except:
            pass

        try:
            import statsmodels.api as sm

            model = sm.OLS(y, X.astype(float))
            results = model.fit(cov_type='HC3')

            # Verify R² is not 1
            if results.rsquared > 0.999:
                print(f"  Warning: R² = {results.rsquared:.6f} (near perfect fit)")

                if used_constant:
                    X_no_const = df_clean[valid_vars].copy()
                    model_no_const = sm.OLS(y, X_no_const.astype(float))
                    results_no_const = model_no_const.fit(cov_type='HC3')

                    if results_no_const.rsquared < results.rsquared:
                        print(f"  Using model without constant: R² = {results_no_const.rsquared:.4f}")
                        return {
                            'results': results_no_const,
                            'nobs': int(results_no_const.nobs),
                            'r2': results_no_const.rsquared,
                            'r2_adj': results_no_const.rsquared_adj,
                            'valid_vars': valid_vars,
                            'used_constant': False
                        }

            return {
                'results': results,
                'nobs': int(results.nobs),
                'r2': results.rsquared,
                'r2_adj': results.rsquared_adj,
                'valid_vars': valid_vars,
                'used_constant': used_constant
            }

        except Exception as e:
            print(f"  Error in regression: {e}")
            return None

    def find_optimal_rate_for_dataset(self, dataset, dataset_name,
                                       depreciation_rates=None,
                                       use_lagged_esg=True,
                                       add_consumer_staples_dummy=False):
        """
        Find optimal depreciation rate for a specific dataset.
        This is a generic function that works for any dataset.
        """
        if depreciation_rates is None:
            depreciation_rates = [0.10, 0.15, 0.20, 0.25, 0.30, 0.35, 0.40, 0.45, 0.50]

        print("\n" + "=" * 70)
        print(f"FINDING OPTIMAL DEPRECIATION RATE FOR {dataset_name}")
        print("=" * 70)
        print(f"Testing {len(depreciation_rates)} depreciation rates")

        # Create lagged variables
        dataset = self.create_lagged_esg_variables(dataset)

        results_summary = []

        for i, rate in enumerate(depreciation_rates, 1):
            print(f"\n[{i}/{len(depreciation_rates)}] Testing δ = {rate:.2f}")

            # Create capital stock
            df_with_capital = self.create_esg_capital_stock(
                dataset.copy(),
                depreciation_rate=rate,
                use_lagged_esg=use_lagged_esg
            )

            # Define capital variables
            suffix = '_lag' if use_lagged_esg else ''
            capital_vars = [f'{c}_capital_{int(rate*100)}{suffix}' for c in ['E', 'S', 'G']]
            capital_vars = [v for v in capital_vars if v in df_with_capital.columns]

            if not capital_vars:
                print(f"  No capital variables created")
                continue

            # Define control variables
            control_vars = ['ROA', 'Size', 'Leverage']
            control_vars = [v for v in control_vars if v in df_with_capital.columns]

            # Run regression
            result = self.run_regression_safe(
                df=df_with_capital,
                y_var='Tobin_Q',
                x_vars=capital_vars,
                control_vars=control_vars,
                add_consumer_staples_dummy=add_consumer_staples_dummy
            )

            if result is not None:
                # Extract Social coefficient
                social_var = f'S_capital_{int(rate*100)}{suffix}'
                social_coeff = np.nan
                social_pval = 1.0
                social_se = np.nan

                if social_var in result['results'].params:
                    social_coeff = result['results'].params[social_var]
                    social_pval = result['results'].pvalues[social_var]
                    social_se = result['results'].bse[social_var]

                results_summary.append({
                    'Depreciation_Rate': rate,
                    'R_Squared': result['r2'],
                    'R_Squared_Adj': result['r2_adj'],
                    'Observations': result['nobs'],
                    'Social_Coefficient': social_coeff,
                    'Social_PValue': social_pval,
                    'Social_SE': social_se,
                    'Significant_10': social_pval < 0.10,
                    'Significant_05': social_pval < 0.05,
                    'Significant_01': social_pval < 0.01,
                    'N_Firms': df_with_capital.index.get_level_values(0).nunique()
                })

                print(f"  ✓ R² = {result['r2']:.4f}, Social coeff = {social_coeff:.4f} (p={social_pval:.4f})")
            else:
                print(f"  ✗ Model failed")

        if results_summary:
            comparison_df = pd.DataFrame(results_summary)

            # Find best rate based on R-squared
            best_idx = comparison_df['R_Squared'].idxmax()
            best_rate = comparison_df.loc[best_idx, 'Depreciation_Rate']

            print("\n" + "=" * 70)
            print(f"OPTIMAL DEPRECIATION RATE FOR {dataset_name}: δ = {best_rate:.2f}")
            print("=" * 70)
            print(f"R² = {comparison_df.loc[best_idx, 'R_Squared']:.4f}")
            print(f"Social coefficient = {comparison_df.loc[best_idx, 'Social_Coefficient']:.4f}")
            print(f"P-value = {comparison_df.loc[best_idx, 'Social_PValue']:.4f}")
            print(f"Observations = {comparison_df.loc[best_idx, 'Observations']}")
            print(f"Firms = {comparison_df.loc[best_idx, 'N_Firms']}")

            # Display all results
            print("\n" + "-" * 70)
            print(f"All depreciation rates tested for {dataset_name}:")
            print("-" * 70)
            display_cols = ['Depreciation_Rate', 'R_Squared', 'Social_Coefficient',
                           'Social_PValue', 'Observations', 'N_Firms']
            print(comparison_df[display_cols].to_string())

            # Save comparison
            comp_path = os.path.join(self.output_dir, f'{dataset_name.lower().replace(" ", "_")}_depreciation_comparison.csv')
            comparison_df.to_csv(comp_path, index=False)
            print(f"\n✓ Saved depreciation comparison to: {comp_path}")

            # Plot
            self.plot_depreciation_comparison(comparison_df, best_rate, dataset_name)

            return best_rate, comparison_df
        else:
            print(f"\n✗ No successful models for {dataset_name}. Using default rate of 0.30")
            return 0.30, None

    def plot_depreciation_comparison(self, comparison_df, best_rate, dataset_name):
        """
        Plot depreciation rate comparison.
        """
        if comparison_df is None or len(comparison_df) == 0:
            return

        fig, axes = plt.subplots(1, 2, figsize=(14, 5))

        # Plot 1: R-squared vs Depreciation Rate
        ax1 = axes[0]
        ax1.plot(comparison_df['Depreciation_Rate'],
                comparison_df['R_Squared'],
                marker='o', linewidth=2, markersize=8)
        ax1.axvline(x=best_rate, color='red', linestyle='--',
                   label=f'Optimal: δ={best_rate:.2f}')
        ax1.set_xlabel('Depreciation Rate (δ)')
        ax1.set_ylabel('R-squared')
        ax1.set_title(f'{dataset_name}: Model Fit by Depreciation Rate')
        ax1.legend()
        ax1.grid(True, alpha=0.3)
        ax1.set_ylim([0, 1])

        # Plot 2: Social Coefficient vs Depreciation Rate
        ax2 = axes[1]

        # Filter out rows with NaN social coefficient
        plot_data = comparison_df.dropna(subset=['Social_Coefficient'])

        if len(plot_data) > 0:
            ax2.errorbar(plot_data['Depreciation_Rate'],
                        plot_data['Social_Coefficient'],
                        yerr=plot_data['Social_SE'],
                        marker='o', linewidth=2, markersize=8, capsize=4)
            ax2.axhline(y=0, color='gray', linestyle='-', alpha=0.5)
            ax2.axvline(x=best_rate, color='red', linestyle='--',
                       label=f'Optimal: δ={best_rate:.2f}')

            # Add significance stars
            for _, row in plot_data.iterrows():
                if row['Significant_10']:
                    stars = '***' if row['Significant_01'] else '**' if row['Significant_05'] else '*'
                    ax2.annotate(stars,
                               (row['Depreciation_Rate'], row['Social_Coefficient']),
                               textcoords="offset points", xytext=(0, 10), ha='center', fontsize=12)

        ax2.set_xlabel('Depreciation Rate (δ)')
        ax2.set_ylabel('Social Capital Coefficient')
        ax2.set_title(f'{dataset_name}: Social Capital Coefficient by Depreciation Rate')
        ax2.legend()
        ax2.grid(True, alpha=0.3)

        plt.tight_layout()
        plot_path = os.path.join(self.output_dir, f'{dataset_name.lower().replace(" ", "_")}_depreciation_comparison.png')
        plt.savefig(plot_path, dpi=150, bbox_inches='tight')
        plt.show()
        print(f"✓ Saved comparison plot to: {plot_path}")

    def run_final_model_with_optimal_rate(self, dataset, dataset_name,
                                          optimal_rate, use_lagged_esg=True,
                                          add_consumer_staples_dummy=False):
        """
        Run final model with the optimal depreciation rate for a dataset.
        """
        print("\n" + "=" * 70)
        print(f"FINAL MODEL FOR {dataset_name} WITH OPTIMAL δ = {optimal_rate:.2f}")
        print("=" * 70)

        # Create lagged variables
        dataset = self.create_lagged_esg_variables(dataset)

        # Create capital stock with optimal rate
        df_with_capital = self.create_esg_capital_stock(
            dataset.copy(),
            depreciation_rate=optimal_rate,
            use_lagged_esg=use_lagged_esg
        )

        # Define capital variables
        suffix = '_lag' if use_lagged_esg else ''
        capital_vars = [f'{c}_capital_{int(optimal_rate*100)}{suffix}' for c in ['E', 'S', 'G']]
        capital_vars = [v for v in capital_vars if v in df_with_capital.columns]

        # Define control variables
        control_vars = ['ROA', 'Size', 'Leverage']
        control_vars = [v for v in control_vars if v in df_with_capital.columns]

        print(f"\nData Summary:")
        print(f"  Observations: {len(df_with_capital)}")
        print(f"  Firms: {df_with_capital.index.get_level_values(0).nunique()}")
        print(f"  Years: {sorted(df_with_capital.index.get_level_values(1).unique())}")
        print(f"  Capital variables: {capital_vars}")
        print(f"  Control variables: {control_vars}")

        # Run regression
        result = self.run_regression_safe(
            df=df_with_capital,
            y_var='Tobin_Q',
            x_vars=capital_vars,
            control_vars=control_vars,
            add_consumer_staples_dummy=add_consumer_staples_dummy
        )

        if result is None:
            print(f"✗ Regression failed for {dataset_name}")
            return None

        # Store results
        result_key = dataset_name.upper().replace(" ", "_")
        self.results[result_key] = {
            'results': result['results'],
            'optimal_rate': optimal_rate,
            'use_lagged_esg': use_lagged_esg,
            'nobs': result['nobs'],
            'r2': result['r2'],
            'r2_adj': result['r2_adj'],
            'capital_vars': capital_vars,
            'control_vars': control_vars,
            'df_with_capital': df_with_capital,
            'used_constant': result['used_constant'],
            'dataset_name': dataset_name
        }

        # Print detailed results
        print("\n" + "=" * 70)
        print(f"{dataset_name} REGRESSION RESULTS")
        print("=" * 70)
        print(f"Optimal Depreciation Rate: δ = {optimal_rate:.2f}")
        print(f"R-squared: {result['r2']:.4f}")
        print(f"Adjusted R-squared: {result['r2_adj']:.4f}")
        print(f"Observations: {result['nobs']}")
        print(f"Firms: {df_with_capital.index.get_level_values(0).nunique()}")
        print(f"Included constant: {result['used_constant']}")

        print("\n" + "-" * 50)
        print("ESG CAPITAL STOCK COEFFICIENTS")
        print("-" * 50)
        for var in capital_vars:
            if var in result['results'].params:
                coeff = result['results'].params[var]
                pval = result['results'].pvalues[var]
                se = result['results'].bse[var]
                stars = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                component = var.split('_')[0]
                t_stat = coeff / se if se != 0 else np.nan
                print(f"{component} Capital: {coeff:.4f}{stars}")
                print(f"  Std.Err: {se:.4f}, t-stat: {t_stat:.4f}, p-value: {pval:.4f}")

        print("\n" + "-" * 50)
        print("CONTROL VARIABLES")
        print("-" * 50)

        vars_to_show = control_vars.copy()
        if result['used_constant']:
            vars_to_show.append('const')

        # Add Consumer Staples dummy if included
        if add_consumer_staples_dummy and 'ConsumerStaples' in result['results'].params:
            vars_to_show.append('ConsumerStaples')

        for var in vars_to_show:
            if var in result['results'].params:
                coeff = result['results'].params[var]
                pval = result['results'].pvalues[var]
                se = result['results'].bse[var]
                stars = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                t_stat = coeff / se if se != 0 else np.nan
                print(f"{var}: {coeff:.4f}{stars}")
                print(f"  Std.Err: {se:.4f}, t-stat: {t_stat:.4f}, p-value: {pval:.4f}")

        # Plot results
        self.plot_coefficients(result, capital_vars, control_vars, optimal_rate, dataset_name,
                              add_consumer_staples_dummy)

        return result

    def plot_coefficients(self, result, capital_vars, control_vars, optimal_rate,
                         dataset_name, add_consumer_staples_dummy=False):
        """
        Plot coefficient estimates.
        """
        fig, axes = plt.subplots(1, 2, figsize=(14, 5))

        # Plot ESG capital coefficients
        ax1 = axes[0]
        esg_coeffs = []
        esg_errors = []
        esg_names = []
        esg_pvalues = []

        for var in capital_vars:
            if var in result['results'].params:
                esg_coeffs.append(result['results'].params[var])
                esg_errors.append(result['results'].bse[var] * 1.96)
                esg_names.append(var.split('_')[0])
                esg_pvalues.append(result['results'].pvalues[var])

        if esg_coeffs:
            y_pos = np.arange(len(esg_names))
            ax1.errorbar(esg_coeffs, y_pos, xerr=esg_errors, fmt='o', capsize=5, capthick=2)
            ax1.axvline(x=0, color='gray', linestyle='--', alpha=0.7)
            ax1.set_yticks(y_pos)
            ax1.set_yticklabels(esg_names)
            ax1.set_xlabel('Coefficient Estimate')
            ax1.set_title(f'{dataset_name}: ESG Capital Coefficients\nδ = {optimal_rate:.2f}')

            for i, (coeff, pval) in enumerate(zip(esg_coeffs, esg_pvalues)):
                stars = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                ax1.annotate(stars, (coeff, i), xytext=(5, -2), textcoords='offset points', fontsize=12)

        # Plot control variable coefficients
        ax2 = axes[1]
        control_coeffs = []
        control_errors = []
        control_names = []
        control_pvalues = []

        vars_to_show = control_vars.copy()
        if result['used_constant']:
            vars_to_show.append('const')
        if add_consumer_staples_dummy and 'ConsumerStaples' in result['results'].params:
            vars_to_show.append('ConsumerStaples')

        for var in vars_to_show:
            if var in result['results'].params:
                control_coeffs.append(result['results'].params[var])
                control_errors.append(result['results'].bse[var] * 1.96)
                control_names.append(var)
                control_pvalues.append(result['results'].pvalues[var])

        if control_coeffs:
            y_pos2 = np.arange(len(control_names))
            ax2.errorbar(control_coeffs, y_pos2, xerr=control_errors, fmt='o', capsize=5, capthick=2)
            ax2.axvline(x=0, color='gray', linestyle='--', alpha=0.7)
            ax2.set_yticks(y_pos2)
            ax2.set_yticklabels(control_names)
            ax2.set_xlabel('Coefficient Estimate')
            ax2.set_title(f'{dataset_name}: Control Variables')

            for i, (coeff, pval) in enumerate(zip(control_coeffs, control_pvalues)):
                stars = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                ax2.annotate(stars, (coeff, i), xytext=(5, -2), textcoords='offset points', fontsize=12)

        plt.tight_layout()
        plot_path = os.path.join(self.output_dir, f'{dataset_name.lower().replace(" ", "_")}_coefficients_delta_{int(optimal_rate*100)}.png')
        plt.savefig(plot_path, dpi=150, bbox_inches='tight')
        plt.show()
        print(f"✓ Saved coefficient plot to: {plot_path}")

    def save_all_results(self):
        """
        Save all analysis results.
        """
        print("\n" + "=" * 70)
        print("SAVING RESULTS")
        print("=" * 70)

        # 1. Save summary of optimal rates
        summary_path = os.path.join(self.output_dir, 'optimal_depreciation_rates_summary.txt')
        with open(summary_path, 'w') as f:
            f.write("ESG CAPITAL STOCK ANALYSIS - OPTIMAL DEPRECIATION RATES\n")
            f.write("=" * 60 + "\n\n")

            f.write(f"ALL SECTORS:\n")
            f.write(f"  Optimal δ = {self.best_rate_all_sectors:.2f}\n")
            if self.comparison_all_sectors is not None:
                best_row = self.comparison_all_sectors.loc[
                    self.comparison_all_sectors['R_Squared'].idxmax()
                ]
                f.write(f"  R² = {best_row['R_Squared']:.4f}\n")
                f.write(f"  Social Coefficient = {best_row['Social_Coefficient']:.4f}\n")
                f.write(f"  P-value = {best_row['Social_PValue']:.4f}\n")
                f.write(f"  Observations = {best_row['Observations']}\n\n")

            f.write(f"CONSUMER STAPLES:\n")
            f.write(f"  Optimal δ = {self.best_rate_consumer_staples:.2f}\n")
            if self.comparison_consumer_staples is not None:
                best_row = self.comparison_consumer_staples.loc[
                    self.comparison_consumer_staples['R_Squared'].idxmax()
                ]
                f.write(f"  R² = {best_row['R_Squared']:.4f}\n")
                f.write(f"  Social Coefficient = {best_row['Social_Coefficient']:.4f}\n")
                f.write(f"  P-value = {best_row['Social_PValue']:.4f}\n")
                f.write(f"  Observations = {best_row['Observations']}\n")

        print(f"✓ Saved optimal rates summary to: {summary_path}")

        # 2. Save all regression results
        for result_key, result_data in self.results.items():
            # Create summary DataFrame
            summary_data = []
            summary_data.append(['Metric', 'Value'])
            summary_data.append(['Dataset', result_data['dataset_name']])
            summary_data.append(['Optimal_Depreciation_Rate', result_data['optimal_rate']])
            summary_data.append(['R_Squared', result_data['r2']])
            summary_data.append(['R_Squared_Adjusted', result_data['r2_adj']])
            summary_data.append(['Observations', result_data['nobs']])
            summary_data.append(['Firms', result_data['df_with_capital'].index.get_level_values(0).nunique()])
            summary_data.append(['Used_Constant', result_data['used_constant']])

            # Add coefficients
            for var in result_data['capital_vars'] + result_data['control_vars']:
                if var in result_data['results'].params:
                    summary_data.append([f'{var}_Coefficient', result_data['results'].params[var]])
                    summary_data.append([f'{var}_PValue', result_data['results'].pvalues[var]])
                    summary_data.append([f'{var}_StdError', result_data['results'].bse[var]])

            # Add constant if used
            if result_data['used_constant'] and 'const' in result_data['results'].params:
                summary_data.append(['const_Coefficient', result_data['results'].params['const']])
                summary_data.append(['const_PValue', result_data['results'].pvalues['const']])
                summary_data.append(['const_StdError', result_data['results'].bse['const']])

            # Add Consumer Staples dummy if present
            if 'ConsumerStaples' in result_data['results'].params:
                summary_data.append(['ConsumerStaples_Coefficient', result_data['results'].params['ConsumerStaples']])
                summary_data.append(['ConsumerStaples_PValue', result_data['results'].pvalues['ConsumerStaples']])
                summary_data.append(['ConsumerStaples_StdError', result_data['results'].bse['ConsumerStaples']])

            result_df = pd.DataFrame(summary_data[1:], columns=summary_data[0])
            result_path = os.path.join(self.output_dir, f'{result_key.lower()}_final_results.csv')
            result_df.to_csv(result_path, index=False)
            print(f"✓ Saved {result_data['dataset_name']} results to: {result_path}")

            # Save data with capital stock
            data_path = os.path.join(self.output_dir, f'{result_key.lower()}_with_capital.csv')
            result_data['df_with_capital'].reset_index().to_csv(data_path, index=False)
            print(f"✓ Saved {result_data['dataset_name']} data with capital stock to: {data_path}")

        print(f"\n✓ All results saved to: {self.output_dir}")

    def run_complete_analysis(self):
        """
        Run complete analysis workflow with SEPARATE optimal rates for:
        1. All sectors
        2. Consumer Staples
        """
        print("=" * 70)
        print("COMPLETE ESG CAPITAL STOCK ANALYSIS")
        print("=" * 70)
        print("\nThis analysis will find SEPARATE optimal depreciation rates for:")
        print("1. ALL SECTORS - using all firms in the dataset")
        print("2. CONSUMER STAPLES - using only Consumer Staples firms")
        print("\nEach rate is optimized independently on its respective sample.")
        print("=" * 70)

        # Step 1: Load data
        print("\n" + "=" * 70)
        print("STEP 1: LOADING DATA")
        print("=" * 70)
        self.load_and_prepare_data()

        if self.df is None or len(self.df) == 0:
            print("✗ No data loaded. Exiting.")
            return

        # Step 2: Find optimal rate for ALL SECTORS
        print("\n" + "=" * 70)
        print("STEP 2: FINDING OPTIMAL RATE FOR ALL SECTORS")
        print("=" * 70)

        self.best_rate_all_sectors, self.comparison_all_sectors = self.find_optimal_rate_for_dataset(
            dataset=self.df,
            dataset_name="ALL SECTORS",
            add_consumer_staples_dummy=True
        )

        # Step 3: Run final model for ALL SECTORS with its optimal rate
        print("\n" + "=" * 70)
        print("STEP 3: RUNNING FINAL MODEL FOR ALL SECTORS")
        print("=" * 70)

        if self.best_rate_all_sectors is not None:
            self.run_final_model_with_optimal_rate(
                dataset=self.df,
                dataset_name="ALL SECTORS",
                optimal_rate=self.best_rate_all_sectors,
                add_consumer_staples_dummy=True
            )

        # Step 4: Find optimal rate for CONSUMER STAPLES
        print("\n" + "=" * 70)
        print("STEP 4: FINDING OPTIMAL RATE FOR CONSUMER STAPLES")
        print("=" * 70)

        if self.df_cs is not None and len(self.df_cs) > 0:
            self.best_rate_consumer_staples, self.comparison_consumer_staples = self.find_optimal_rate_for_dataset(
                dataset=self.df_cs,
                dataset_name="CONSUMER STAPLES",
                add_consumer_staples_dummy=False
            )

            # Step 5: Run final model for CONSUMER STAPLES with its optimal rate
            print("\n" + "=" * 70)
            print("STEP 5: RUNNING FINAL MODEL FOR CONSUMER STAPLES")
            print("=" * 70)

            if self.best_rate_consumer_staples is not None:
                self.run_final_model_with_optimal_rate(
                    dataset=self.df_cs,
                    dataset_name="CONSUMER STAPLES",
                    optimal_rate=self.best_rate_consumer_staples,
                    add_consumer_staples_dummy=False
                )
        else:
            print("✗ No Consumer Staples data available. Skipping steps 4-5.")

        # Step 6: Save all results
        print("\n" + "=" * 70)
        print("STEP 6: SAVING RESULTS")
        print("=" * 70)
        self.save_all_results()

        print("\n" + "=" * 70)
        print("ANALYSIS COMPLETE")
        print("=" * 70)
        print(f"\n✓ Optimal depreciation rate for ALL SECTORS: δ = {self.best_rate_all_sectors:.2f}")

        if self.best_rate_consumer_staples is not None:
            print(f"✓ Optimal depreciation rate for CONSUMER STAPLES: δ = {self.best_rate_consumer_staples:.2f}")
            print(f"  (Optimized independently on Consumer Staples sample)")

        print(f"\n✓ All results saved to: {self.output_dir}")


def main():
    """
    Main function to run the analysis.
    """
    # Configuration
    DATA_PATH = '/content/drive/MyDrive/Colab Notebooks/FTSE Data/FTSE data.xlsx'
    OUTPUT_DIR = '/content/drive/MyDrive/Colab Notebooks/FTSE Data/esg_capital_stock_optimal_rates'

    # Initialize and run analysis
    analyzer = ESGCapitalStockOptimalRates(DATA_PATH, OUTPUT_DIR)
    analyzer.run_complete_analysis()


if __name__ == "__main__":
    main()
```
