```python
# esg_capital_stock_standalone.py
"""
Standalone program for ESG-Tobin's Q analysis using ESG Capital Stock models.
Focuses on creating and analyzing ESG capital stock with different depreciation rates.
"""
from google.colab import drive
drive.mount('/content/drive')

!pip install -q linearmodels pandas numpy matplotlib seaborn statsmodels
# change_esg_standalone.py
"""
Standalone program for ESG-Tobin's Q analysis using Change in Lagged ESG models.
Analyzes how changes in ESG scores affect Tobin's Q.
"""

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from linearmodels.panel import PanelOLS
import warnings
warnings.filterwarnings('ignore')

class ChangeESGAnalysis:
    """
    Standalone class for running Change in Lagged ESG models on Tobin's Q.
    Analyzes how changes in ESG scores over time affect financial performance.
    """

    def __init__(self, data_path, output_dir):
        """
        Initialize the Change in ESG analysis.

        Parameters:
        -----------
        data_path : str
            Path to the Excel data file
        output_dir : str
            Directory to save output files
        """
        self.data_path = data_path
        self.output_dir = output_dir
        self.df = None
        self.change_data = {}
        self.results = {}

        # Create output directory if it doesn't exist
        os.makedirs(output_dir, exist_ok=True)

        # Setup plotting style
        plt.style.use('seaborn-v0_8-whitegrid')
        sns.set_palette("husl")

    def load_and_prepare_data(self):
        """
        Load data and prepare basic transformations.
        """
        print("=" * 70)
        print("LOADING AND PREPARING DATA FOR CHANGE IN ESG ANALYSIS")
        print("=" * 70)

        try:
            # Load data
            self.df = pd.read_excel(self.data_path)
            print(f"✓ Loaded data from: {self.data_path}")

            # Set multi-index
            if 'Firm' in self.df.columns and 'Year' in self.df.columns:
                self.df = self.df.set_index(['Firm', 'Year']).sort_index()
                print("✓ Set multi-index (Firm, Year)")
            else:
                print("⚠ Warning: 'Firm' and/or 'Year' columns not found")
                print(f"  Available columns: {list(self.df.columns)}")
                # Try to use first two columns as index
                self.df = self.df.set_index([self.df.columns[0], self.df.columns[1]]).sort_index()
                print(f"  Using {self.df.index.names} as index")

            # Rename columns if needed
            column_renames = {}
            if "Tobin's Q" in self.df.columns:
                column_renames["Tobin's Q"] = "Tobin_Q"

            # Apply renames if any
            if column_renames:
                self.df = self.df.rename(columns=column_renames)
                print(f"✓ Renamed columns: {column_renames}")

            # Create basic control variables
            if 'Total Assets' in self.df.columns:
                self.df['Size'] = np.log(self.df['Total Assets'].replace(0, np.nan))
                print("✓ Created Size variable (log of Total Assets)")
            else:
                print("⚠ Warning: 'Total Assets' column not found")

            if 'Total Liabilities' in self.df.columns and 'Total Assets' in self.df.columns:
                self.df['Leverage'] = self.df['Total Liabilities'] / self.df['Total Assets'].replace(0, np.nan)
                print("✓ Created Leverage variable")
            else:
                print("⚠ Warning: Could not create Leverage variable")

            # Create log transformation of Tobin's Q
            if 'Tobin_Q' in self.df.columns:
                self.df['Tobin_Q_log'] = np.log(self.df['Tobin_Q'].replace(0, np.nan) + 0.001)
                print("✓ Created Tobin_Q_log variable")
            else:
                print("⚠ Warning: 'Tobin_Q' column not found")

            # Check for ESG columns
            esg_columns = [col for col in ['E', 'S', 'G'] if col in self.df.columns]
            if len(esg_columns) == 3:
                print(f"✓ Found ESG columns: {esg_columns}")
            else:
                print(f"⚠ Missing ESG columns. Found: {esg_columns}")
                print(f"  All columns: {list(self.df.columns)}")

            print(f"\n✓ Loaded {len(self.df)} observations")
            print(f"✓ Number of firms: {self.df.index.get_level_values(0).nunique()}")
            print(f"✓ Years: {sorted(self.df.index.get_level_values(1).unique())}")

            # Print column summary
            print(f"\nAvailable columns ({len(self.df.columns)}):")
            for i, col in enumerate(sorted(self.df.columns)):
                if i < 20:  # Show first 20 columns
                    non_null = self.df[col].notna().sum()
                    print(f"  {col:<20} ({non_null} non-null)")

            return self.df

        except Exception as e:
            print(f"✗ Error loading data: {e}")
            import traceback
            traceback.print_exc()
            return None

    def create_sector_variable(self):
        """
        Create Consumer Staples sector variable.
        """
        print("\n" + "-" * 50)
        print("CREATING SECTOR VARIABLES")
        print("-" * 50)

        # Define Consumer Staples firms
        consumer_staples_firms = [
            'Associated British Foods', 'British American Tobacco', 'Coca Cola HBC AG',
            'Diageo', 'J. Sainsbury', 'Reckitt Benckiser', 'Tesco', 'Unilever', 'Haleon Plc'
        ]

        # Get unique firm names
        firm_names = self.df.index.get_level_values(0).unique()
        print(f"Found {len(firm_names)} unique firms")

        # Check which Consumer Staples firms are in the data
        found_cs_firms = [firm for firm in consumer_staples_firms if firm in firm_names]
        print(f"Found {len(found_cs_firms)} Consumer Staples firms in data: {found_cs_firms}")

        # Create sector variable
        sectors = []
        consumer_staples_dummy = []

        for firm in self.df.index.get_level_values(0):
            if firm in consumer_staples_firms:
                sectors.append('Consumer Staples')
                consumer_staples_dummy.append(1)
            else:
                sectors.append('Other')
                consumer_staples_dummy.append(0)

        self.df['Sector'] = sectors
        self.df['ConsumerStaples'] = consumer_staples_dummy

        print(f"\n✓ Created sector variables:")
        print(f"  Consumer Staples firms: {self.df[self.df['ConsumerStaples']==1].index.get_level_values(0).nunique()}")
        print(f"  Other sectors firms: {self.df[self.df['ConsumerStaples']==0].index.get_level_values(0).nunique()}")
        print(f"  Consumer Staples observations: {self.df['ConsumerStaples'].sum()}")

        return self.df

    def create_change_variables(self, lag_period=1, use_absolute=True, use_percentage=False):
        """
        Create change in ESG variables.

        Parameters:
        -----------
        lag_period : int
            Period over which to calculate change (e.g., 1 for year-to-year)
        use_absolute : bool
            Whether to calculate absolute change
        use_percentage : bool
            Whether to calculate percentage change
        """
        print(f"\n{'='*70}")
        print(f"CREATING CHANGE IN ESG VARIABLES")
        print(f"Lag period: {lag_period} year(s)")
        print(f"Absolute change: {use_absolute}, Percentage change: {use_percentage}")
        print(f"{'='*70}")

        # Check which ESG components are available
        esg_components = [c for c in ['E', 'S', 'G'] if c in self.df.columns]
        print(f"Available ESG components: {esg_components}")

        if not esg_components:
            print("⚠ No ESG components found!")
            return self.df

        # Create lagged ESG variables first if they don't exist
        for component in esg_components:
            lag_col = f'{component}_lag{lag_period}'
            if lag_col not in self.df.columns:
                self.df[lag_col] = self.df.groupby(level=0)[component].shift(lag_period)
                print(f"✓ Created {lag_col}")

        for component in esg_components:
            print(f"\nProcessing {component}:")

            current_col = component
            lag_col = f'{component}_lag{lag_period}'

            if current_col in self.df.columns and lag_col in self.df.columns:

                # 1. Absolute change (ΔESG = ESG_t - ESG_{t-lag})
                if use_absolute:
                    change_col = f'{component}_chg{lag_period}'
                    self.df[change_col] = self.df[current_col] - self.df[lag_col]

                    # Store metadata
                    self.change_data[change_col] = {
                        'component': component,
                        'lag_period': lag_period,
                        'type': 'absolute',
                        'formula': f'{current_col} - {lag_col}'
                    }

                    non_null = self.df[change_col].notna().sum()
                    mean_val = self.df[change_col].mean()
                    print(f"  ✓ {change_col}: {non_null} non-null, mean={mean_val:.4f}")

                # 2. Percentage change (%ΔESG = (ESG_t - ESG_{t-lag}) / ESG_{t-lag})
                if use_percentage:
                    pct_col = f'{component}_pctchg{lag_period}'

                    # Avoid division by zero
                    denominator = self.df[lag_col].replace(0, np.nan)
                    self.df[pct_col] = (self.df[current_col] - self.df[lag_col]) / denominator

                    # Store metadata
                    self.change_data[pct_col] = {
                        'component': component,
                        'lag_period': lag_period,
                        'type': 'percentage',
                        'formula': f'({current_col} - {lag_col}) / {lag_col}'
                    }

                    non_null = self.df[pct_col].notna().sum()
                    mean_val = self.df[pct_col].mean()
                    print(f"  ✓ {pct_col}: {non_null} non-null, mean={mean_val:.4%}")

            else:
                print(f"  ⚠ Missing columns for {component}: {current_col} or {lag_col}")

        print(f"\n✓ Created change variables")

        # Summary of created variables
        change_cols = [col for col in self.df.columns if '_chg' in col or '_pctchg' in col]
        print(f"\nChange variables created ({len(change_cols)}):")
        for col in sorted(change_cols):
            non_null = self.df[col].notna().sum()
            print(f"  {col:<20} ({non_null} non-null)")

        return self.df

    def create_multiple_change_periods(self, periods=None):
        """
        Create change variables for multiple lag periods.

        Parameters:
        -----------
        periods : list
            List of lag periods to create (e.g., [1, 2, 3] for 1, 2, 3-year changes)
        """
        if periods is None:
            periods = [1, 2, 3]  # 1-year, 2-year, and 3-year changes

        print(f"\n{'='*70}")
        print(f"CREATING CHANGE VARIABLES FOR {len(periods)} PERIODS")
        print(f"Periods: {periods}")
        print(f"{'='*70}")

        for period in periods:
            print(f"\nCreating {period}-year change variables:")
            self.create_change_variables(
                lag_period=period,
                use_absolute=True,
                use_percentage=True
            )

        # Summary of all change variables
        change_cols = [col for col in self.df.columns if '_chg' in col or '_pctchg' in col]
        print(f"\n✓ Total change variables created: {len(change_cols)}")

        return self.df

    def analyze_change_distribution(self):
        """
        Analyze the distribution of ESG changes.
        """
        print(f"\n{'='*70}")
        print("ANALYZING ESG CHANGE DISTRIBUTIONS")
        print(f"{'='*70}")

        # Get all change variables
        change_cols = [col for col in self.df.columns if '_chg' in col and '_pctchg' not in col]
        pct_change_cols = [col for col in self.df.columns if '_pctchg' in col]

        if not change_cols and not pct_change_cols:
            print("No change variables found. Creating 1-year changes...")
            self.create_change_variables(lag_period=1, use_absolute=True, use_percentage=True)
            change_cols = [col for col in self.df.columns if '_chg' in col and '_pctchg' not in col]
            pct_change_cols = [col for col in self.df.columns if '_pctchg' in col]

        # Create summary statistics
        summary_data = []

        for col in change_cols + pct_change_cols:
            if col in self.df.columns:
                data = self.df[col].dropna()
                if len(data) > 0:
                    component = col.split('_')[0]
                    change_type = 'absolute' if '_chg' in col and '_pctchg' not in col else 'percentage'
                    period = ''.join(filter(str.isdigit, col.split('_')[-1])) or '1'

                    summary_data.append({
                        'Variable': col,
                        'Component': component,
                        'Type': change_type,
                        'Period': f'{period} year(s)',
                        'N': len(data),
                        'Mean': data.mean(),
                        'Std': data.std(),
                        'Min': data.min(),
                        'Max': data.max(),
                        'Positive': (data > 0).sum(),
                        'Negative': (data < 0).sum(),
                        'Zero': (data == 0).sum()
                    })

        if summary_data:
            summary_df = pd.DataFrame(summary_data)

            print("\nSummary Statistics:")
            print("-" * 100)
            print(f"{'Variable':<20} {'Type':<12} {'Period':<10} {'N':<6} {'Mean':<10} {'Std':<10} {'Min':<10} {'Max':<10}")
            print("-" * 100)

            for _, row in summary_df.iterrows():
                if row['Type'] == 'absolute':
                    mean_str = f"{row['Mean']:.4f}"
                    std_str = f"{row['Std']:.4f}"
                    min_str = f"{row['Min']:.4f}"
                    max_str = f"{row['Max']:.4f}"
                else:  # percentage
                    mean_str = f"{row['Mean']:.2%}"
                    std_str = f"{row['Std']:.2%}"
                    min_str = f"{row['Min']:.2%}"
                    max_str = f"{row['Max']:.2%}"

                print(f"{row['Variable']:<20} {row['Type']:<12} {row['Period']:<10} "
                      f"{row['N']:<6} {mean_str:<10} {std_str:<10} {min_str:<10} {max_str:<10}")

            # Create distribution plots
            self.plot_change_distributions(change_cols, pct_change_cols)

            return summary_df
        else:
            print("No change data available for analysis")
            return None

    def plot_change_distributions(self, change_cols, pct_change_cols):
        """
        Plot distributions of ESG changes.
        """
        # Filter to most common period (1-year changes)
        change_cols_1yr = [col for col in change_cols if col.endswith('chg1') or ('chg' in col and not any(char.isdigit() for char in col.split('_')[-1]))]
        pct_change_cols_1yr = [col for col in pct_change_cols if col.endswith('pctchg1')]

        if not change_cols_1yr and not pct_change_cols_1yr:
            print("No 1-year change variables found for plotting")
            return

        # Create plots
        fig, axes = plt.subplots(2, 3, figsize=(15, 10))
        fig.suptitle('Distribution of 1-Year ESG Changes', fontsize=16, fontweight='bold')

        # Plot absolute changes
        if change_cols_1yr:
            for idx, col in enumerate(change_cols_1yr[:3]):  # Plot first 3 components
                if idx < 3:
                    row, col_idx = divmod(idx, 3)
                    data = self.df[col].dropna()

                    if len(data) > 0:
                        axes[0, col_idx].hist(data, bins=30, edgecolor='black', alpha=0.7)
                        axes[0, col_idx].axvline(x=data.mean(), color='red', linestyle='--',
                                                label=f'Mean: {data.mean():.3f}')
                        axes[0, col_idx].axvline(x=0, color='green', linestyle=':', alpha=0.5)
                        axes[0, col_idx].set_xlabel('Absolute Change')
                        axes[0, col_idx].set_ylabel('Frequency')
                        axes[0, col_idx].set_title(f'{col.split("_")[0]} Component')
                        axes[0, col_idx].legend()
                        axes[0, col_idx].grid(True, alpha=0.3)

        # Plot percentage changes
        if pct_change_cols_1yr:
            for idx, col in enumerate(pct_change_cols_1yr[:3]):  # Plot first 3 components
                if idx < 3:
                    row, col_idx = divmod(idx, 3)
                    data = self.df[col].dropna() * 100  # Convert to percentage

                    if len(data) > 0:
                        axes[1, col_idx].hist(data, bins=30, edgecolor='black', alpha=0.7)
                        axes[1, col_idx].axvline(x=data.mean(), color='red', linestyle='--',
                                                label=f'Mean: {data.mean():.2f}%')
                        axes[1, col_idx].axvline(x=0, color='green', linestyle=':', alpha=0.5)
                        axes[1, col_idx].set_xlabel('Percentage Change (%)')
                        axes[1, col_idx].set_ylabel('Frequency')
                        axes[1, col_idx].set_title(f'{col.split("_")[0]} Component (% Change)')
                        axes[1, col_idx].legend()
                        axes[1, col_idx].grid(True, alpha=0.3)

        plt.tight_layout()

        # Save plot
        plot_path = os.path.join(self.output_dir, 'esg_change_distributions.png')
        plt.savefig(plot_path, dpi=300, bbox_inches='tight')
        plt.show()

        print(f"✓ Change distribution plots saved to: {plot_path}")

    def run_change_model(self, change_period=1, change_type='absolute',
                        y_var='Tobin_Q', sector='All', include_levels=True):
        """
        Run a change in ESG model.

        Parameters:
        -----------
        change_period : int
            Period for change calculation (e.g., 1 for 1-year change)
        change_type : str
            Type of change ('absolute' or 'percentage')
        y_var : str
            Dependent variable
        sector : str
            Sector to analyze
        include_levels : bool
            Whether to include lagged ESG levels in the model
        """
        print(f"\n{'='*70}")
        print(f"CHANGE IN ESG MODEL")
        print(f"Period: {change_period} year(s), Type: {change_type}")
        print(f"Sector: {sector}, Y: {y_var}, Include levels: {include_levels}")
        print(f"{'='*70}")

        # Filter data by sector
        if sector == 'CS':
            if 'ConsumerStaples' not in self.df.columns:
                self.create_sector_variable()
            df_filtered = self.df[self.df['ConsumerStaples'] == 1].copy()
        else:
            df_filtered = self.df.copy()
            if 'ConsumerStaples' not in self.df.columns:
                self.create_sector_variable()

        # Create change variables if they don't exist
        suffix = 'chg' if change_type == 'absolute' else 'pctchg'
        change_vars = [f'{c}_{suffix}{change_period}' for c in ['E', 'S', 'G']]

        # Check which variables exist
        existing_change_vars = [var for var in change_vars if var in df_filtered.columns]
        if len(existing_change_vars) < len(change_vars):
            print(f"Creating missing change variables...")
            self.create_change_variables(
                lag_period=change_period,
                use_absolute=(change_type == 'absolute'),
                use_percentage=(change_type == 'percentage')
            )
            # Refresh filtered data
            if sector == 'CS':
                df_filtered = self.df[self.df['ConsumerStaples'] == 1].copy()
            else:
                df_filtered = self.df.copy()

        # Define model variables
        model_vars = []

        # Add change variables
        for component in ['E', 'S', 'G']:
            change_var = f'{component}_{suffix}{change_period}'
            if change_var in df_filtered.columns:
                model_vars.append(change_var)

        # Add lagged ESG levels if requested
        if include_levels:
            for component in ['E', 'S', 'G']:
                level_var = f'{component}_lag{change_period}'
                if level_var not in df_filtered.columns:
                    # Create lagged variable if it doesn't exist
                    df_filtered[level_var] = df_filtered.groupby(level=0)[component].shift(change_period)
                model_vars.append(level_var)

        # Add control variables
        control_vars = []
        for var in ['ROA', 'Size', 'Leverage']:
            if var in df_filtered.columns:
                control_vars.append(var)
            else:
                print(f"⚠ Control variable '{var}' not found")

        if sector == 'All':
            control_vars.append('ConsumerStaples')

        model_vars.extend(control_vars)

        # Clean data
        all_vars = model_vars + [y_var]
        df_clean = df_filtered.dropna(subset=all_vars)

        if len(df_clean) < 10:
            print(f"✗ Insufficient data: Only {len(df_clean)} observations")
            return None

        # Print data summary
        print(f"\nData Summary:")
        print(f"  Observations: {len(df_clean)}")
        print(f"  Firms: {df_clean.index.get_level_values(0).nunique()}")
        print(f"  Years: {sorted(df_clean.index.get_level_values(1).unique())}")

        if sector == 'All':
            cs_firms = df_clean[df_clean['ConsumerStaples']==1].index.get_level_values(0).nunique()
            other_firms = df_clean[df_clean['ConsumerStaples']==0].index.get_level_values(0).nunique()
            print(f"  Consumer Staples firms: {cs_firms}")
            print(f"  Other firms: {other_firms}")

        # Prepare data for regression
        y = df_clean[y_var]
        X = df_clean[model_vars]
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

        print(f"\nRegression variables ({len(model_vars)}):")
        print(f"  Y: {y_var}")
        print(f"  X: {model_vars}")

        try:
            # Run regression
            model = PanelOLS(y, X, entity_effects=False, time_effects=False)
            results = model.fit(cov_type='robust')

            # Store results
            level_suffix = '_withlevels' if include_levels else ''
            model_key = f'Change_{change_period}yr_{change_type}{level_suffix}_{sector}_{y_var}'
            self.results[model_key] = {
                'results': results,
                'model_name': f'Change in ESG ({change_period}yr, {change_type})',
                'change_period': change_period,
                'change_type': change_type,
                'include_levels': include_levels,
                'y_var': y_var,
                'sector': sector,
                'nobs': results.nobs,
                'r2': results.rsquared,
                'change_vars': [v for v in model_vars if suffix in v],
                'level_vars': [v for v in model_vars if '_lag' in v] if include_levels else [],
                'control_vars': control_vars
            }

            # Print results
            print(f"\n✓ Regression successful!")
            print(f"  R-squared: {results.rsquared:.4f}")

            if hasattr(results, 'f_statistic') and results.f_statistic:
                print(f"  F-statistic: {results.f_statistic.stat:.4f} (p={results.f_statistic.pval:.4f})")

            # Print change coefficients
            print(f"\nChange in ESG Coefficients ({change_type}):")
            for component in ['E', 'S', 'G']:
                change_var = f'{component}_{suffix}{change_period}'
                if change_var in results.params:
                    coeff = results.params[change_var]
                    pval = results.pvalues.get(change_var, 1.0)
                    sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""

                    if change_type == 'percentage':
                        interpretation = f"(1% Δ{component} → {coeff:.4f} ΔTobin's Q)"
                    else:
                        interpretation = f"(1 unit Δ{component} → {coeff:.4f} ΔTobin's Q)"

                    print(f"  Δ{component}: {coeff:.4f}{sig} (p={pval:.4f}) {interpretation}")

            # Print level coefficients if included
            if include_levels:
                print(f"\nLagged ESG Level Coefficients:")
                for component in ['E', 'S', 'G']:
                    level_var = f'{component}_lag{change_period}'
                    if level_var in results.params:
                        coeff = results.params[level_var]
                        pval = results.pvalues.get(level_var, 1.0)
                        sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                        print(f"  {component}(t-{change_period}): {coeff:.4f}{sig} (p={pval:.4f})")

            # Print control variables
            print(f"\nControl Variables:")
            for var in control_vars:
                if var in results.params:
                    coeff = results.params[var]
                    pval = results.pvalues.get(var, 1.0)
                    sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                    print(f"  {var}: {coeff:.4f}{sig} (p={pval:.4f})")

            # Generate equation
            self._generate_equation(results, model_vars, change_period, change_type,
                                  include_levels, sector, y_var)

            return results

        except Exception as e:
            print(f"\n✗ Error running regression: {e}")
            import traceback
            traceback.print_exc()
            return None

    def _generate_equation(self, results, model_vars, change_period, change_type,
                          include_levels, sector, y_var):
        """Generate a formatted regression equation."""
        print(f"\n{'='*70}")
        print("REGRESSION EQUATION")
        print(f"{'='*70}")

        # Start with constant
        eq_parts = []
        if 'const' in results.params:
            const = results.params['const']
            eq_parts.append(f"{const:.4f}")

        # Add variables
        for var in model_vars:
            if var in results.params:
                coeff = results.params[var]
                pval = results.pvalues.get(var, 1.0)

                # Determine significance stars
                stars = ""
                if pval < 0.01:
                    stars = "***"
                elif pval < 0.05:
                    stars = "**"
                elif pval < 0.10:
                    stars = "*"

                # Format variable name
                if '_chg' in var or '_pctchg' in var:
                    component = var.split('_')[0]
                    change_sym = 'Δ' if '_chg' in var else '%Δ'
                    display_var = f"{change_sym}{component}"
                elif '_lag' in var:
                    component = var.split('_')[0]
                    display_var = f"{component}(t-{change_period})"
                else:
                    display_var = var

                if coeff >= 0:
                    eq_parts.append(f"+ {abs(coeff):.4f}{stars}·{display_var}")
                else:
                    eq_parts.append(f"- {abs(coeff):.4f}{stars}·{display_var}")

        # Build equation
        equation = f"{y_var}(t) = " + " ".join(eq_parts) + " + ε"
        print(equation)

        print(f"\nModel Details:")
        print(f"  Change period: {change_period} year(s)")
        print(f"  Change type: {change_type}")
        print(f"  Include lagged levels: {include_levels}")
        print(f"  Sector: {sector}")
        print(f"  R² = {results.rsquared:.4f}, N = {results.nobs}")

        if change_type == 'absolute':
            print(f"  Change formula: Δ{change_period}X = X(t) - X(t-{change_period})")
        else:
            print(f"  Change formula: %Δ{change_period}X = [X(t) - X(t-{change_period})] / X(t-{change_period})")

        print("  Significance: *** p<0.01, ** p<0.05, * p<0.10")

    def compare_change_periods(self, periods=None, change_type='absolute',
                              y_var='Tobin_Q', sector='All'):
        """
        Compare models with different change periods.

        Parameters:
        -----------
        periods : list
            List of change periods to compare
        change_type : str
            Type of change ('absolute' or 'percentage')
        y_var : str
            Dependent variable
        sector : str
            Sector to analyze
        """
        if periods is None:
            periods = [1, 2, 3]  # Compare 1, 2, and 3-year changes

        print(f"\n{'='*70}")
        print(f"COMPARING CHANGE PERIODS")
        print(f"Type: {change_type}, Sector: {sector}, Y: {y_var}")
        print(f"{'='*70}")

        comparison_data = []

        for period in periods:
            print(f"\nTesting {period}-year change:")

            # Run model without including levels (pure change model)
            result = self.run_change_model(
                change_period=period,
                change_type=change_type,
                y_var=y_var,
                sector=sector,
                include_levels=False
            )

            if result is not None:
                # Extract Social change coefficient
                suffix = 'chg' if change_type == 'absolute' else 'pctchg'
                social_var = f'S_{suffix}{period}'

                if social_var in result.params:
                    social_coeff = result.params[social_var]
                    social_pval = result.pvalues.get(social_var, 1.0)
                else:
                    social_coeff = np.nan
                    social_pval = 1.0

                comparison_data.append({
                    'Change_Period': period,
                    'R_Squared': result.rsquared,
                    'Social_Coefficient': social_coeff,
                    'Social_PValue': social_pval,
                    'Observations': result.nobs,
                    'Significant': social_pval < 0.10
                })

                print(f"  R² = {result.rsquared:.4f}, Social coeff = {social_coeff:.4f}")
            else:
                print(f"  Model failed")

        if comparison_data:
            # Create comparison DataFrame
            comp_df = pd.DataFrame(comparison_data)

            print(f"\n{'='*70}")
            print("COMPARISON RESULTS")
            print(f"{'='*70}")

            print(f"\n{'Period':<8} {'R²':<8} {'Social Coeff':<15} {'P-Value':<10} {'Sig':<6} {'N':<8}")
            print("-" * 60)

            for _, row in comp_df.iterrows():
                sig = "✓" if row['Significant'] else "✗"
                sig_level = ""
                pval = row['Social_PValue']
                if pval < 0.01:
                    sig_level = "***"
                elif pval < 0.05:
                    sig_level = "**"
                elif pval < 0.10:
                    sig_level = "*"

                print(f"{row['Change_Period']:<8} {row['R_Squared']:<8.4f} "
                      f"{row['Social_Coefficient']:<15.4f} {pval:<10.4f} {sig}{sig_level:<5} {row['Observations']:<8}")

            # Create comparison plot
            self.plot_change_period_comparison(comp_df, change_type, sector, y_var)

            return comp_df
        else:
            print("No comparison data available")
            return None

    def plot_change_period_comparison(self, comp_df, change_type, sector, y_var):
        """
        Plot comparison of different change periods.
        """
        fig, axes = plt.subplots(1, 2, figsize=(12, 5))

        # Plot 1: R-squared vs change period
        axes[0].plot(comp_df['Change_Period'], comp_df['R_Squared'], 'o-',
                    linewidth=2, markersize=8, color='blue')
        axes[0].set_xlabel('Change Period (years)', fontsize=12)
        axes[0].set_ylabel('R-squared', fontsize=12)
        axes[0].set_title(f'Model Fit vs Change Period\n{change_type.capitalize()} Change, {sector}', fontsize=14)
        axes[0].grid(True, alpha=0.3)

        # Highlight best R-squared
        best_idx = comp_df['R_Squared'].idxmax()
        best_period = comp_df.loc[best_idx, 'Change_Period']
        best_r2 = comp_df.loc[best_idx, 'R_Squared']
        axes[0].plot(best_period, best_r2, 'o', markersize=10,
                    markeredgecolor='red', markeredgewidth=2)
        axes[0].annotate(f'Best: {best_period}yr',
                        xy=(best_period, best_r2),
                        xytext=(best_period + 0.1, best_r2 - 0.02),
                        arrowprops=dict(arrowstyle='->', color='red'))

        # Plot 2: Social coefficient vs change period
        axes[1].plot(comp_df['Change_Period'], comp_df['Social_Coefficient'], 's-',
                    linewidth=2, markersize=8, color='green')
        axes[1].axhline(y=0, color='r', linestyle='--', alpha=0.5)
        axes[1].set_xlabel('Change Period (years)', fontsize=12)
        axes[1].set_ylabel('Social Coefficient', fontsize=12)
        axes[1].set_title('Social Change Effect vs Period', fontsize=14)
        axes[1].grid(True, alpha=0.3)

        # Add significance markers
        for _, row in comp_df.iterrows():
            if row['Significant']:
                axes[1].plot(row['Change_Period'], row['Social_Coefficient'], 's',
                           markersize=10, markeredgecolor='red', markeredgewidth=2)

        plt.tight_layout()

        # Save plot
        plot_path = os.path.join(self.output_dir, f'change_period_comparison_{change_type}_{sector}.png')
        plt.savefig(plot_path, dpi=300, bbox_inches='tight')
        plt.show()

        print(f"\n✓ Comparison plot saved to: {plot_path}")

    def run_interaction_model(self, change_period=1, change_type='absolute',
                            y_var='Tobin_Q'):
        """
        Run a model with interaction between ESG change and Consumer Staples dummy.

        Parameters:
        -----------
        change_period : int
            Period for change calculation
        change_type : str
            Type of change
        y_var : str
            Dependent variable
        """
        print(f"\n{'='*70}")
        print(f"INTERACTION MODEL: ESG Change × Consumer Staples")
        print(f"Period: {change_period} year(s), Type: {change_type}")
        print(f"Y: {y_var}")
        print(f"{'='*70}")

        # Ensure sector variable exists
        if 'ConsumerStaples' not in self.df.columns:
            self.create_sector_variable()

        # Create change variables if needed
        suffix = 'chg' if change_type == 'absolute' else 'pctchg'
        change_vars = [f'{c}_{suffix}{change_period}' for c in ['E', 'S', 'G']]

        existing_change_vars = [var for var in change_vars if var in self.df.columns]
        if len(existing_change_vars) < len(change_vars):
            print(f"Creating change variables...")
            self.create_change_variables(
                lag_period=change_period,
                use_absolute=(change_type == 'absolute'),
                use_percentage=(change_type == 'percentage')
            )

        # Create interaction terms
        for component in ['E', 'S', 'G']:
            change_var = f'{component}_{suffix}{change_period}'
            interaction_var = f'{change_var}_CS'
            self.df[interaction_var] = self.df[change_var] * self.df['ConsumerStaples']
            print(f"✓ Created interaction term: {interaction_var}")

        # Prepare data
        df_clean = self.df.dropna(subset=change_vars + ['ConsumerStaples', 'ROA', 'Size', 'Leverage', y_var])

        if len(df_clean) < 10:
            print(f"✗ Insufficient data: Only {len(df_clean)} observations")
            return None

        # Define model variables
        model_vars = []

        # Add change variables
        for component in ['E', 'S', 'G']:
            change_var = f'{component}_{suffix}{change_period}'
            model_vars.append(change_var)

        # Add Consumer Staples dummy
        model_vars.append('ConsumerStaples')

        # Add interaction terms
        for component in ['E', 'S', 'G']:
            interaction_var = f'{component}_{suffix}{change_period}_CS'
            model_vars.append(interaction_var)

        # Add control variables
        control_vars = ['ROA', 'Size', 'Leverage']
        model_vars.extend(control_vars)

        # Run regression
        y = df_clean[y_var]
        X = df_clean[model_vars]
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

        try:
            model = PanelOLS(y, X, entity_effects=False, time_effects=False)
            results = model.fit(cov_type='robust')

            # Store results
            model_key = f'Interaction_{change_period}yr_{change_type}_{y_var}'
            self.results[model_key] = {
                'results': results,
                'model_name': f'Interaction Model ({change_period}yr, {change_type})',
                'change_period': change_period,
                'change_type': change_type,
                'y_var': y_var,
                'sector': 'All',
                'nobs': results.nobs,
                'r2': results.rsquared,
                'change_vars': [v for v in model_vars if suffix in v and '_CS' not in v],
                'interaction_vars': [v for v in model_vars if '_CS' in v],
                'control_vars': control_vars
            }

            # Print results
            print(f"\n✓ Regression successful!")
            print(f"  R-squared: {results.rsquared:.4f}")

            print(f"\nMain Effects (Non-Consumer Staples):")
            for component in ['E', 'S', 'G']:
                change_var = f'{component}_{suffix}{change_period}'
                if change_var in results.params:
                    coeff = results.params[change_var]
                    pval = results.pvalues.get(change_var, 1.0)
                    sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                    print(f"  Δ{component}: {coeff:.4f}{sig} (p={pval:.4f})")

            print(f"\nInteraction Effects (Consumer Staples vs Others):")
            for component in ['E', 'S', 'G']:
                interaction_var = f'{component}_{suffix}{change_period}_CS'
                if interaction_var in results.params:
                    coeff = results.params[interaction_var]
                    pval = results.pvalues.get(interaction_var, 1.0)
                    sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""

                    # Calculate total effect for Consumer Staples
                    change_var = f'{component}_{suffix}{change_period}'
                    if change_var in results.params:
                        main_effect = results.params[change_var]
                        total_effect = main_effect + coeff
                        print(f"  Δ{component}×CS: {coeff:.4f}{sig} (p={pval:.4f})")
                        print(f"    Total for CS: {total_effect:.4f} (= {main_effect:.4f} + {coeff:.4f})")

            print(f"\nConsumer Staples Dummy:")
            if 'ConsumerStaples' in results.params:
                coeff = results.params['ConsumerStaples']
                pval = results.pvalues.get('ConsumerStaples', 1.0)
                sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                print(f"  CS: {coeff:.4f}{sig} (p={pval:.4f})")

            return results

        except Exception as e:
            print(f"\n✗ Error running regression: {e}")
            import traceback
            traceback.print_exc()
            return None

    def run_comprehensive_analysis(self, periods=None, change_types=None,
                                  y_vars=None, sectors=None):
        """
        Run comprehensive change in ESG analysis.

        Parameters:
        -----------
        periods : list
            List of change periods to analyze
        change_types : list
            List of change types to analyze
        y_vars : list
            List of dependent variables
        sectors : list
            List of sectors to analyze
        """
        if periods is None:
            periods = [1, 2, 3]

        if change_types is None:
            change_types = ['absolute', 'percentage']

        if y_vars is None:
            y_vars = ['Tobin_Q', 'Tobin_Q_log']

        if sectors is None:
            sectors = ['All', 'CS']

        print("=" * 70)
        print("COMPREHENSIVE CHANGE IN ESG ANALYSIS")
        print("=" * 70)

        results_summary = []

        # Create change variables for all periods
        print("\nCreating change variables...")
        self.create_multiple_change_periods(periods)

        # Run models for each combination
        print("\nRunning regression models...")
        for period in periods:
            for change_type in change_types:
                for y_var in y_vars:
                    for sector in sectors:
                        for include_levels in [False, True]:  # Test both with and without levels

                            print(f"\nAnalyzing: {period}yr {change_type}, "
                                  f"Levels={include_levels}, Y={y_var}, Sector={sector}")

                            # Run change model
                            result = self.run_change_model(
                                change_period=period,
                                change_type=change_type,
                                y_var=y_var,
                                sector=sector,
                                include_levels=include_levels
                            )

                            # Store summary if results exist
                            if result is not None:
                                # Extract Social coefficient
                                suffix = 'chg' if change_type == 'absolute' else 'pctchg'
                                social_var = f'S_{suffix}{period}'

                                if social_var in result.params:
                                    social_coeff = result.params[social_var]
                                    social_pval = result.pvalues.get(social_var, 1.0)
                                else:
                                    social_coeff = np.nan
                                    social_pval = 1.0

                                results_summary.append({
                                    'Change_Period': period,
                                    'Change_Type': change_type,
                                    'Include_Levels': include_levels,
                                    'Y_Variable': y_var,
                                    'Sector': sector,
                                    'Observations': result.nobs,
                                    'R_Squared': result.rsquared,
                                    'Social_Coefficient': social_coeff,
                                    'Social_PValue': social_pval,
                                    'Significant': social_pval < 0.10
                                })

        # Create summary DataFrame
        if results_summary:
            summary_df = pd.DataFrame(results_summary)

            # Save summary to CSV
            summary_path = os.path.join(self.output_dir, 'change_esg_summary.csv')
            summary_df.to_csv(summary_path, index=False)
            print(f"\n✓ Summary saved to: {summary_path}")

            # Display key findings
            self._display_key_findings(summary_df)

            return summary_df
        else:
            print("No models were successfully run.")
            return None

    def _display_key_findings(self, summary_df):
        """Display key findings from the analysis."""
        print(f"\n{'='*70}")
        print("KEY FINDINGS")
        print(f"{'='*70}")

        # Find best model by R-squared
        best_idx = summary_df['R_Squared'].idxmax()
        best_model = summary_df.loc[best_idx]

        print(f"\nBEST MODEL BY R-SQUARED:")
        print(f"  Change period: {best_model['Change_Period']} year(s)")
        print(f"  Change type: {best_model['Change_Type']}")
        print(f"  Include levels: {best_model['Include_Levels']}")
        print(f"  Y variable: {best_model['Y_Variable']}")
        print(f"  Sector: {best_model['Sector']}")
        print(f"  R-squared: {best_model['R_Squared']:.4f}")
        print(f"  Social coefficient: {best_model['Social_Coefficient']:.4f}")
        print(f"  Significance: {'Significant' if best_model['Significant'] else 'Not significant'}")

        # Summary by change period
        print(f"\nSUMMARY BY CHANGE PERIOD:")
        period_summary = summary_df.groupby('Change_Period').agg({
            'R_Squared': 'mean',
            'Social_Coefficient': 'mean',
            'Significant': 'mean'
        }).round(4)

        print(period_summary)

        # Summary by sector
        print(f"\nSUMMARY BY SECTOR:")
        sector_summary = summary_df.groupby('Sector').agg({
            'R_Squared': 'mean',
            'Social_Coefficient': 'mean',
            'Significant': 'mean'
        }).round(4)

        print(sector_summary)

    def save_results(self):
        """
        Save results to CSV files.
        """
        print("\n" + "=" * 70)
        print("SAVING RESULTS")
        print("=" * 70)

        # Save change variables summary
        if self.change_data:
            change_summary = []
            for col_name, data_info in self.change_data.items():
                if col_name in self.df.columns:
                    non_null = self.df[col_name].notna().sum()
                    mean_val = self.df[col_name].mean()
                    std_val = self.df[col_name].std()

                    change_summary.append({
                        'Variable': col_name,
                        'Component': data_info['component'],
                        'Lag_Period': data_info['lag_period'],
                        'Type': data_info['type'],
                        'Formula': data_info['formula'],
                        'Non_Null': non_null,
                        'Mean': mean_val,
                        'Std': std_val
                    })

            if change_summary:
                change_df = pd.DataFrame(change_summary)
                change_path = os.path.join(self.output_dir, 'change_variables_summary.csv')
                change_df.to_csv(change_path, index=False)
                print(f"✓ Saved change variables summary to: {change_path}")

        # Save model results
        if self.results:
            results_summary = []

            for model_key, model_info in self.results.items():
                results = model_info['results']

                # Extract coefficients for ESG changes
                for component in ['E', 'S', 'G']:
                    suffix = 'chg' if model_info['change_type'] == 'absolute' else 'pctchg'
                    change_var = f'{component}_{suffix}{model_info["change_period"]}'

                    if change_var in results.params:
                        coeff = results.params[change_var]
                        pval = results.pvalues.get(change_var, 1.0)

                        results_summary.append({
                            'Model_Key': model_key,
                            'Model_Name': model_info['model_name'],
                            'Change_Period': model_info['change_period'],
                            'Change_Type': model_info['change_type'],
                            'Include_Levels': model_info.get('include_levels', False),
                            'Component': component,
                            'Y_Variable': model_info['y_var'],
                            'Sector': model_info['sector'],
                            'Observations': results.nobs,
                            'R_Squared': results.rsquared,
                            'Coefficient': coeff,
                            'P_Value': pval,
                            'Significant': pval < 0.10
                        })

            if results_summary:
                results_df = pd.DataFrame(results_summary)
                results_path = os.path.join(self.output_dir, 'change_esg_results.csv')
                results_df.to_csv(results_path, index=False)
                print(f"✓ Saved model results to: {results_path}")

        # Save data sample with change variables
        change_cols = [col for col in self.df.columns if '_chg' in col or '_pctchg' in col]
        if change_cols:
            # Select a few important columns
            important_cols = ['Tobin_Q', 'Tobin_Q_log', 'ROA', 'Size', 'Leverage', 'ConsumerStaples'] + change_cols[:10]  # First 10 change variables
            available_cols = [col for col in important_cols if col in self.df.columns]

            if available_cols:
                sample_df = self.df[available_cols].reset_index()
                sample_path = os.path.join(self.output_dir, 'data_sample_with_changes.csv')
                sample_df.head(100).to_csv(sample_path, index=False)  # Save first 100 rows
                print(f"✓ Saved data sample to: {sample_path}")

        print(f"\n✓ All results saved to: {self.output_dir}")


def main():
    """
    Main function to run the Change in ESG analysis.
    """
    # Configuration
    DATA_PATH = '/content/drive/MyDrive/Colab Notebooks/FTSE Data/FTSE data.xlsx'
    OUTPUT_DIR = '/content/drive/MyDrive/Colab Notebooks/FTSE Data/change_esg_output'

    print("=" * 70)
    print("CHANGE IN LAGGED ESG ANALYSIS")
    print("=" * 70)

    # Initialize analysis
    analyzer = ChangeESGAnalysis(DATA_PATH, OUTPUT_DIR)

    # Step 1: Load and prepare data
    print("\n" + "=" * 70)
    print("STEP 1: LOADING DATA")
    print("=" * 70)

    df = analyzer.load_and_prepare_data()

    if df is None:
        print("✗ Failed to load data. Exiting.")
        return

    # Step 2: Create sector variable
    print("\n" + "=" * 70)
    print("STEP 2: CREATING SECTOR VARIABLES")
    print("=" * 70)

    analyzer.create_sector_variable()

    # Step 3: Analyze change distributions
    print("\n" + "=" * 70)
    print("STEP 3: ANALYZING ESG CHANGE DISTRIBUTIONS")
    print("=" * 70)

    change_summary = analyzer.analyze_change_distribution()

    # Step 4: Compare different change periods
    print("\n" + "=" * 70)
    print("STEP 4: COMPARING CHANGE PERIODS")
    print("=" * 70)

    # Compare absolute changes
    abs_comparison = analyzer.compare_change_periods(
        periods=[1, 2, 3],
        change_type='absolute',
        y_var='Tobin_Q',
        sector='All'
    )

    # Compare percentage changes
    pct_comparison = analyzer.compare_change_periods(
        periods=[1, 2, 3],
        change_type='percentage',
        y_var='Tobin_Q',
        sector='All'
    )

    # Step 5: Run interaction model
    print("\n" + "=" * 70)
    print("STEP 5: RUNNING INTERACTION MODEL")
    print("=" * 70)

    interaction_result = analyzer.run_interaction_model(
        change_period=1,
        change_type='absolute',
        y_var='Tobin_Q'
    )

    # Step 6: Comprehensive analysis
    print("\n" + "=" * 70)
    print("STEP 6: COMPREHENSIVE ANALYSIS")
    print("=" * 70)

    comprehensive_results = analyzer.run_comprehensive_analysis(
        periods=[1, 2],
        change_types=['absolute'],
        y_vars=['Tobin_Q'],
        sectors=['All', 'CS']
    )

    # Step 7: Save results
    print("\n" + "=" * 70)
    print("STEP 7: SAVING RESULTS")
    print("=" * 70)

    analyzer.save_results()

    print("\n" + "=" * 70)
    print("ANALYSIS COMPLETE")
    print("=" * 70)

    # Summary
    print("\nSUMMARY:")
    print(f"1. ✓ Data loaded: {len(analyzer.df)} observations")
    print(f"2. ✓ Change variables created: {len([c for c in analyzer.df.columns if '_chg' in c or '_pctchg' in c])}")
    print(f"3. ✓ Models run: {len(analyzer.results)}")

    if analyzer.results:
        # Show best model
        best_model_key = None
        best_r2 = -1

        for model_key, model_info in analyzer.results.items():
            if model_info['r2'] > best_r2:
                best_r2 = model_info['r2']
                best_model_key = model_key

        if best_model_key:
            best_model = analyzer.results[best_model_key]
            print(f"\nBEST MODEL:")
            print(f"  Name: {best_model['model_name']}")
            print(f"  Sector: {best_model['sector']}")
            print(f"  Y: {best_model['y_var']}")
            print(f"  R²: {best_model['r2']:.4f}")
            print(f"  N: {best_model['nobs']}")

        print("\nKEY INSIGHTS:")
        print("  1. Change models capture dynamic ESG effects")
        print("  2. Short-term (1-year) vs long-term (2-3 year) changes may differ")
        print("  3. Absolute vs percentage changes provide different perspectives")
        print("  4. Consumer Staples may respond differently to ESG changes")
        print("  5. Including lagged levels helps separate change from level effects")


if __name__ == "__main__":
    main()
```
