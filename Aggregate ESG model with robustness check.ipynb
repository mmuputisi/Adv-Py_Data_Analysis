```python
# standalone_lag_model_aggregate_esg_enhanced.py
"""
Standalone Lag Model Analysis - Using Aggregate ESG Score
With enhanced output including regression tables, equations, and Excel export
"""

# ============================================================================
# SETUP AND IMPORTS
# ============================================================================
from google.colab import drive
drive.mount('/content/drive')

!pip install -q linearmodels pandas numpy matplotlib seaborn statsmodels openpyxl

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from linearmodels.panel import PanelOLS
import warnings
from datetime import datetime
warnings.filterwarnings('ignore')

# ============================================================================
# CONFIGURATION
# ============================================================================
class Config:
    """Configuration parameters for aggregate ESG analysis"""
    PROJECT_DIR = '/content/drive/MyDrive/Colab Notebooks/FTSE Data'
    DATA_PATH = os.path.join(PROJECT_DIR, 'FTSE data.xlsx')
    OUTPUT_DIR = os.path.join(PROJECT_DIR, 'lag_model_outputs_aggregate_esg_enhanced')

    # Using aggregate ESG instead of individual components
    ESG_COMPONENTS = ['ESG']  # Single aggregate ESG score
    CONTROL_VARS = ['ROA', 'Size', 'Leverage']

    # Consumer Staples firms
    CONSUMER_STAPLES_FIRMS = [
        'Associated British Foods', 'British American Tobacco', 'Coca Cola HBC AG',
        'Diageo', 'J. Sainsbury', 'Reckitt Benckiser', 'Tesco', 'Unilever', 'Haleon Plc'
    ]

# ============================================================================
# DATA LOADER - AGGREGATE ESG VERSION
# ============================================================================
class DataLoader:
    """Load and prepare data with aggregate ESG score"""

    def __init__(self, config):
        self.config = config
        self.df = None
        self.df_cs = None  # Consumer Staples only
        self.df_all = None  # All sectors

    def load_and_prepare_data(self):
        """Load data and create aggregate ESG score"""
        print("Loading and preparing data...")

        # 1. Load data
        self.df = pd.read_excel(self.config.DATA_PATH)
        self.df = self.df.set_index(['Firm', 'Year']).sort_index()
        self.df = self.df.rename(columns={"Tobin's Q": "Tobin_Q"})

        # 2. Create aggregate ESG score (average of E, S, G if they exist)
        if all(col in self.df.columns for col in ['E', 'S', 'G']):
            self.df['ESG'] = self.df[['E', 'S', 'G']].mean(axis=1)
            print("✓ Created aggregate ESG score from E, S, G components")
        else:
            print("⚠ Individual ESG components not found - using existing ESG score if available")
            if 'ESG' not in self.df.columns:
                raise ValueError("No ESG score available in data")

        # 3. Create variables
        self.df['Size'] = np.log(self.df['Total Assets'])
        self.df['Leverage'] = self.df['Total Liabilities'] / self.df['Total Assets']
        self.df['Tobin_Q_log'] = np.log(self.df['Tobin_Q'] + 0.001)

        # 4. Create lagged ESG variables (t-1)
        self.df['ESG_lag'] = self.df.groupby(level=0)['ESG'].shift(1)

        # 5. Create lagged control variables
        self.df['ROA_lag'] = self.df.groupby(level=0)['ROA'].shift(1)
        self.df['Size_lag'] = self.df.groupby(level=0)['Size'].shift(1)
        self.df['Leverage_lag'] = self.df.groupby(level=0)['Leverage'].shift(1)

        # 6. Create sector variable
        sectors = []
        for firm in self.df.index.get_level_values(0):
            if firm in self.config.CONSUMER_STAPLES_FIRMS:
                sectors.append('Consumer Staples')
            else:
                sectors.append('Other')

        self.df['Sector'] = sectors
        self.df['ConsumerStaples'] = (self.df['Sector'] == 'Consumer Staples').astype(int)

        # 7. Create datasets
        self.df_cs = self.df[self.df['ConsumerStaples'] == 1].copy()
        self.df_all = self.df.copy()

        print(f"✓ Total observations: {len(self.df)}")
        print(f"✓ Consumer Staples: {len(self.df_cs)} observations")
        print(f"✓ All sectors: {len(self.df_all)} observations")

        return self.df, self.df_cs, self.df_all

    def create_lagged_variables_for_model(self, df, min_lag=1, max_lag=5, use_original_controls=True):
        """
        Create lagged variables for aggregate ESG

        Parameters:
        -----------
        df : DataFrame
            Input dataframe
        min_lag : int
            Minimum lag (1)
        max_lag : int
            Maximum lag (1-5)
        use_original_controls : bool
            If True, use original controls (not lagged)
        """
        df_copy = df.copy()

        # Create distributed lags for aggregate ESG - INDIVIDUAL LAGS
        print(f"Creating distributed lags {min_lag}-{max_lag} for aggregate ESG...")
        for lag in range(min_lag, max_lag + 1):
            df_copy[f'ESG_lag{lag}'] = df_copy.groupby(level=0)['ESG'].shift(lag)

        # Also create summed versions
        # Initialize with first lag
        if max_lag >= 1:
            lag1 = df_copy.groupby(level=0)['ESG'].shift(1)
            df_copy['ESG_lag1'] = lag1

        # Create summed versions
        if max_lag >= 2:
            lag2 = df_copy.groupby(level=0)['ESG'].shift(2)
            df_copy['ESG_lag12'] = lag1 + lag2

        if max_lag >= 3:
            lag3 = df_copy.groupby(level=0)['ESG'].shift(3)
            df_copy['ESG_lag13'] = lag1 + lag2 + lag3

        if max_lag >= 4:
            lag4 = df_copy.groupby(level=0)['ESG'].shift(4)
            df_copy['ESG_lag14'] = lag1 + lag2 + lag3 + lag4

        if max_lag >= 5:
            lag5 = df_copy.groupby(level=0)['ESG'].shift(5)
            df_copy['ESG_lag15'] = lag1 + lag2 + lag3 + lag4 + lag5

        # Use control variables as in original structured analysis
        if use_original_controls:
            control_vars = ['ROA', 'Size', 'Leverage']
        else:
            control_vars = ['ROA_lag', 'Size_lag', 'Leverage_lag']

        return df_copy, control_vars

# ============================================================================
# RESULTS EXPORTER
# ============================================================================
class ResultsExporter:
    """Export results to Excel and create formatted tables"""

    def __init__(self, config):
        self.config = config
        os.makedirs(config.OUTPUT_DIR, exist_ok=True)

    def create_regression_table(self, results, model_name, sector_name, max_lag, model_type):
        """Create a formatted regression table with coefficients, p-values, and significance"""
        table_data = []

        # Get all variables except const
        for var in results.params.index:
            if var != 'const':
                coeff = results.params[var]
                std_err = results.std_errors.get(var, np.nan) if hasattr(results, 'std_errors') else np.nan
                t_stat = results.tstats.get(var, np.nan) if hasattr(results, 'tstats') else np.nan
                pval = results.pvalues.get(var, np.nan)

                # Determine significance stars
                if pval < 0.01:
                    sig = '***'
                elif pval < 0.05:
                    sig = '**'
                elif pval < 0.10:
                    sig = '*'
                else:
                    sig = ''

                table_data.append({
                    'Variable': var,
                    'Coefficient': f"{coeff:.4f}{sig}",
                    'Coefficient_Value': coeff,
                    'Std_Error': f"{std_err:.4f}" if not np.isnan(std_err) else 'N/A',
                    't_stat': f"{t_stat:.4f}" if not np.isnan(t_stat) else 'N/A',
                    'P_value': f"{pval:.4f}",
                    'P_value_Num': pval,
                    'Significance': sig
                })

        # Add constant if present
        if 'const' in results.params:
            const_coeff = results.params['const']
            const_pval = results.pvalues.get('const', np.nan)
            if const_pval < 0.01:
                const_sig = '***'
            elif const_pval < 0.05:
                const_sig = '**'
            elif const_pval < 0.10:
                const_sig = '*'
            else:
                const_sig = ''

            table_data.insert(0, {
                'Variable': 'Constant',
                'Coefficient': f"{const_coeff:.4f}{const_sig}",
                'Coefficient_Value': const_coeff,
                'Std_Error': f"{results.std_errors.get('const', np.nan):.4f}" if hasattr(results, 'std_errors') else 'N/A',
                't_stat': f"{results.tstats.get('const', np.nan):.4f}" if hasattr(results, 'tstats') else 'N/A',
                'P_value': f"{const_pval:.4f}",
                'P_value_Num': const_pval,
                'Significance': const_sig
            })

        # Add model statistics
        table_data.append({
            'Variable': 'R-squared',
            'Coefficient': f"{results.rsquared:.4f}",
            'Coefficient_Value': results.rsquared,
            'Std_Error': '',
            't_stat': '',
            'P_value': '',
            'P_value_Num': np.nan,
            'Significance': ''
        })

        table_data.append({
            'Variable': 'Observations',
            'Coefficient': f"{results.nobs}",
            'Coefficient_Value': results.nobs,
            'Std_Error': '',
            't_stat': '',
            'P_value': '',
            'P_value_Num': np.nan,
            'Significance': ''
        })

        if hasattr(results, 'f_statistic'):
            table_data.append({
                'Variable': 'F-statistic',
                'Coefficient': f"{results.f_statistic.stat:.4f}",
                'Coefficient_Value': results.f_statistic.stat,
                'Std_Error': '',
                't_stat': '',
                'P_value': f"{results.f_statistic.pval:.4f}",
                'P_value_Num': results.f_statistic.pval,
                'Significance': ''
            })

        # Create DataFrame
        df_table = pd.DataFrame(table_data)

        # Add metadata
        df_table.attrs['model_name'] = model_name
        df_table.attrs['sector'] = sector_name
        df_table.attrs['max_lag'] = max_lag
        df_table.attrs['model_type'] = model_type

        return df_table

    def generate_model_equation(self, results, model_name, sector_name, max_lag, model_type):
        """Generate the regression equation as a string"""
        equation_parts = ["Tobin's Q = "]

        # Add constant
        if 'const' in results.params:
            const_val = results.params['const']
            equation_parts.append(f"{const_val:.4f}")

        # Add ESG variables
        esg_vars = [v for v in results.params.index if 'ESG' in str(v) and v != 'const']
        for var in sorted(esg_vars):
            coeff = results.params[var]
            sign = " + " if coeff >= 0 else " - "
            if coeff >= 0:
                equation_parts.append(f"+ {abs(coeff):.4f} × {var}")
            else:
                equation_parts.append(f"- {abs(coeff):.4f} × {var}")

        # Add control variables
        control_vars = [v for v in results.params.index if v not in esg_vars + ['const'] and v != 'ConsumerStaples']
        for var in control_vars:
            coeff = results.params[var]
            sign = " + " if coeff >= 0 else " - "
            if coeff >= 0:
                equation_parts.append(f"+ {abs(coeff):.4f} × {var}")
            else:
                equation_parts.append(f"- {abs(coeff):.4f} × {var}")

        # Add sector dummy if present
        if 'ConsumerStaples' in results.params:
            coeff = results.params['ConsumerStaples']
            sign = " + " if coeff >= 0 else " - "
            if coeff >= 0:
                equation_parts.append(f"+ {abs(coeff):.4f} × ConsumerStaples")
            else:
                equation_parts.append(f"- {abs(coeff):.4f} × ConsumerStaples")

        equation = " ".join(equation_parts)

        # Add model statistics
        equation += f"\n\nR² = {results.rsquared:.4f}"
        equation += f"\nN = {results.nobs}"
        if hasattr(results, 'f_statistic'):
            equation += f"\nF-statistic = {results.f_statistic.stat:.4f} (p={results.f_statistic.pval:.4f})"

        return equation

    def export_all_results(self, analyzer):
        """Export all results to Excel with multiple sheets"""
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
        excel_path = os.path.join(self.config.OUTPUT_DIR, f'aggregate_esg_results_{timestamp}.xlsx')

        with pd.ExcelWriter(excel_path, engine='openpyxl') as writer:
            # Sheet 1: Summary comparison
            summary_data = []

            for key, model_info in analyzer.results.items():
                results = model_info['results']

                # Get key ESG coefficients
                esg_coeffs = []
                esg_vars = [v for v in results.params.index if 'ESG' in str(v) and v != 'const']
                for var in esg_vars:
                    esg_coeffs.append({
                        'variable': var,
                        'coefficient': results.params[var],
                        'p_value': results.pvalues.get(var, np.nan)
                    })

                summary_row = {
                    'Model_Type': model_info['type'],
                    'Sector': 'Consumer Staples' if model_info['sector'] == 'cs' else 'All Sectors',
                    'Max_Lag': model_info['max_lag'],
                    'Observations': results.nobs,
                    'R_squared': results.rsquared,
                    'F_statistic': results.f_statistic.stat if hasattr(results, 'f_statistic') else np.nan,
                    'F_p_value': results.f_statistic.pval if hasattr(results, 'f_statistic') else np.nan,
                }

                # Add ESG coefficient info
                if esg_coeffs:
                    if model_info['type'] == 'distributed_individual':
                        # For individual lags, add average and each lag
                        avg_coeff = np.mean([c['coefficient'] for c in esg_coeffs])
                        summary_row['ESG_Avg_Coefficient'] = avg_coeff

                        for coeff_info in esg_coeffs:
                            summary_row[f"{coeff_info['variable']}_coeff"] = coeff_info['coefficient']
                            summary_row[f"{coeff_info['variable']}_pval"] = coeff_info['p_value']
                    else:
                        # For summed or simple models
                        summary_row['ESG_Coefficient'] = esg_coeffs[0]['coefficient'] if esg_coeffs else np.nan
                        summary_row['ESG_P_value'] = esg_coeffs[0]['p_value'] if esg_coeffs else np.nan

                summary_data.append(summary_row)

            pd.DataFrame(summary_data).to_excel(writer, sheet_name='Summary', index=False)

            # Sheet 2: Simple model results
            simple_results = []
            for key, model_info in analyzer.results.items():
                if model_info['type'] == 'simple':
                    df_table = self.create_regression_table(
                        model_info['results'],
                        f"Simple Model - {model_info['max_lag']} year lag",
                        'Consumer Staples' if model_info['sector'] == 'cs' else 'All Sectors',
                        model_info['max_lag'],
                        'simple'
                    )
                    df_table.to_excel(writer,
                                    sheet_name=f"Simple_{model_info['sector'].upper()}",
                                    index=False)

                    # Store equation
                    equation = self.generate_model_equation(
                        model_info['results'],
                        f"Simple Model",
                        'Consumer Staples' if model_info['sector'] == 'cs' else 'All Sectors',
                        model_info['max_lag'],
                        'simple'
                    )
                    simple_results.append({
                        'Sector': 'Consumer Staples' if model_info['sector'] == 'cs' else 'All Sectors',
                        'Equation': equation
                    })

            if simple_results:
                pd.DataFrame(simple_results).to_excel(writer, sheet_name='Simple_Equations', index=False)

            # Sheets for each lag range
            for max_lag in [2, 3, 4, 5]:
                lag_sheet_name = f"Lag_1_{max_lag}"

                for sector_code in ['cs', 'all']:
                    sector_name = 'Consumer_Staples' if sector_code == 'cs' else 'All_Sectors'

                    # Individual lags model
                    ind_key = f"distributed_individual_{max_lag}_{sector_code}"
                    if ind_key in analyzer.results:
                        df_ind = self.create_regression_table(
                            analyzer.results[ind_key]['results'],
                            f"Individual Lags 1-{max_lag}",
                            'Consumer Staples' if sector_code == 'cs' else 'All Sectors',
                            max_lag,
                            'distributed_individual'
                        )
                        df_ind.to_excel(writer,
                                      sheet_name=f"Ind_{max_lag}y_{sector_name[:10]}",
                                      index=False)

                    # Summed lags model
                    sum_key = f"distributed_summed_{max_lag}_{sector_code}"
                    if sum_key in analyzer.results:
                        df_sum = self.create_regression_table(
                            analyzer.results[sum_key]['results'],
                            f"Summed Lags 1-{max_lag}",
                            'Consumer Staples' if sector_code == 'cs' else 'All Sectors',
                            max_lag,
                            'distributed_summed'
                        )
                        df_sum.to_excel(writer,
                                      sheet_name=f"Sum_{max_lag}y_{sector_name[:10]}",
                                      index=False)

            # Sheet 3: All equations
            equations_data = []
            for key, model_info in analyzer.results.items():
                equation = self.generate_model_equation(
                    model_info['results'],
                    model_info['type'],
                    'Consumer Staples' if model_info['sector'] == 'cs' else 'All Sectors',
                    model_info['max_lag'],
                    model_info['type']
                )
                equations_data.append({
                    'Model_Type': model_info['type'],
                    'Sector': 'Consumer Staples' if model_info['sector'] == 'cs' else 'All Sectors',
                    'Max_Lag': model_info['max_lag'],
                    'Equation': equation
                })

            pd.DataFrame(equations_data).to_excel(writer, sheet_name='All_Equations', index=False)

        print(f"\n✓ Results exported to: {excel_path}")
        return excel_path

# ============================================================================
# LAG MODEL ANALYZER - AGGREGATE ESG VERSION
# ============================================================================
class LagModelAnalyzer:
    """Analyzer for aggregate ESG models"""

    def __init__(self, config):
        self.config = config
        self.loader = DataLoader(config)
        self.exporter = ResultsExporter(config)
        self.results = {}

    def run_analysis(self, max_lag=5):
        """Run analysis for aggregate ESG with different lag specifications"""
        print("\n" + "="*80)
        print(f"RUNNING AGGREGATE ESG LAG MODEL ANALYSIS (1-{max_lag} years)")
        print("="*80)

        # Load data
        df, df_cs, df_all = self.loader.load_and_prepare_data()

        # Run models for different sectors
        sectors = [
            ('cs', df_cs, 'Consumer Staples Only'),
            ('all', df_all, 'All Sectors')
        ]

        for sector_code, sector_df, sector_name in sectors:
            print(f"\n{'='*60}")
            print(f"ANALYZING: {sector_name}")
            print(f"{'='*60}")

            # Create lagged variables
            df_lagged, control_vars = self.loader.create_lagged_variables_for_model(
                sector_df, min_lag=1, max_lag=max_lag, use_original_controls=True
            )

            # Run distributed lag model with individual lags
            print(f"\n1. DISTRIBUTED LAG MODEL (Individual Lags 1-{max_lag})")
            self._run_distributed_lag_individual(df_lagged, max_lag, sector_code,
                                                sector_name, control_vars)

            # Run distributed lag model with summed lags
            print(f"\n2. DISTRIBUTED LAG MODEL (Summed Lags 1-{max_lag})")
            self._run_distributed_lag_summed(df_lagged, max_lag, sector_code,
                                           sector_name, control_vars)

    def _run_distributed_lag_individual(self, df, max_lag, sector_code,
                                       sector_name, control_vars):
        """Run distributed lag model with individual lags for aggregate ESG"""
        # Prepare lag variables
        lag_vars = [f'ESG_lag{lag}' for lag in range(1, max_lag + 1)]

        # Add ConsumerStaples dummy for all sectors
        if sector_code == 'all':
            all_vars = lag_vars + control_vars + ['ConsumerStaples', 'Tobin_Q']
        else:
            all_vars = lag_vars + control_vars + ['Tobin_Q']

        # Clean data
        df_clean = df.dropna(subset=all_vars)

        if len(df_clean) < 10:
            print(f"⚠ Insufficient data: {len(df_clean)} observations")
            return None

        print(f"Observations: {len(df_clean)}")
        print(f"Firms: {df_clean.index.get_level_values(0).nunique()}")

        # Prepare regression
        y = df_clean['Tobin_Q']
        X_vars = lag_vars + control_vars
        if sector_code == 'all':
            X_vars = X_vars + ['ConsumerStaples']

        X = df_clean[X_vars]
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

        # Run regression
        try:
            model = PanelOLS(y, X, entity_effects=False, time_effects=False)
            results = model.fit(cov_type='robust')

            # Store results
            key = f"distributed_individual_{max_lag}_{sector_code}"
            self.results[key] = {
                'results': results,
                'type': 'distributed_individual',
                'sector': sector_code,
                'max_lag': max_lag,
                'nobs': results.nobs,
                'r2': results.rsquared,
                'data': df_clean
            }

            # Print results
            self._print_results_aggregate(results, f"Distributed Lag (Individual) - {sector_name}")

            # Print regression table
            self._print_regression_table(results, "Individual Lags Model")

            # Print equation
            print("\n" + "="*60)
            print("REGRESSION EQUATION:")
            print("="*60)
            equation = self.exporter.generate_model_equation(
                results,
                "Individual Lags Model",
                sector_name,
                max_lag,
                'distributed_individual'
            )
            print(equation)

            # Create lag structure plot
            self._create_lag_structure_plot_aggregate(results, max_lag,
                                          f"Distributed Lag Individual - {sector_name}")

            return results

        except Exception as e:
            print(f"Error running model: {e}")
            return None

    def _run_distributed_lag_summed(self, df, max_lag, sector_code,
                                   sector_name, control_vars):
        """Run distributed lag model with summed lags for aggregate ESG"""
        # Use summed variable (e.g., ESG_lag15 for sum of lags 1-5)
        summed_var = f'ESG_lag1{max_lag}'

        # Add ConsumerStaples dummy for all sectors
        if sector_code == 'all':
            all_vars = [summed_var] + control_vars + ['ConsumerStaples', 'Tobin_Q']
        else:
            all_vars = [summed_var] + control_vars + ['Tobin_Q']

        # Clean data
        df_clean = df.dropna(subset=all_vars)

        if len(df_clean) < 10:
            print(f"⚠ Insufficient data: {len(df_clean)} observations")
            return None

        print(f"Observations: {len(df_clean)}")

        # Prepare regression
        y = df_clean['Tobin_Q']
        X_vars = [summed_var] + control_vars
        if sector_code == 'all':
            X_vars = X_vars + ['ConsumerStaples']

        X = df_clean[X_vars]
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

        # Run regression
        try:
            model = PanelOLS(y, X, entity_effects=False, time_effects=False)
            results = model.fit(cov_type='robust')

            # Store results
            key = f"distributed_summed_{max_lag}_{sector_code}"
            self.results[key] = {
                'results': results,
                'type': 'distributed_summed',
                'sector': sector_code,
                'max_lag': max_lag,
                'nobs': results.nobs,
                'r2': results.rsquared,
                'data': df_clean
            }

            # Print results
            self._print_results_aggregate(results, f"Distributed Lag (Summed) - {sector_name}")

            # Print regression table
            self._print_regression_table(results, "Summed Lags Model")

            # Print equation
            print("\n" + "="*60)
            print("REGRESSION EQUATION:")
            print("="*60)
            equation = self.exporter.generate_model_equation(
                results,
                "Summed Lags Model",
                sector_name,
                max_lag,
                'distributed_summed'
            )
            print(equation)

            return results

        except Exception as e:
            print(f"Error running model: {e}")
            return None

    def _print_results_aggregate(self, results, title):
        """Print formatted results for aggregate ESG"""
        print(f"\n{title}")
        print("-" * 60)
        print(f"R-squared: {results.rsquared:.4f}")
        print(f"Observations: {results.nobs}")

        if hasattr(results, 'f_statistic'):
            print(f"F-statistic: {results.f_statistic.stat:.4f} (p={results.f_statistic.pval:.4f})")

        # Print ESG coefficients with significance
        print("\nESG Coefficients:")
        print("-" * 40)

        esg_vars = [v for v in results.params.index if 'ESG' in str(v)]
        for var in esg_vars:
            if var in results.params and var != 'const':
                coeff = results.params[var]
                pval = results.pvalues.get(var, 1.0)
                sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                print(f"  {var:<15} {coeff:>10.4f}{sig:<3} (p={pval:.4f})")

    def _print_regression_table(self, results, title):
        """Print a formatted regression table"""
        print("\n" + "="*70)
        print(f"REGRESSION TABLE: {title}")
        print("="*70)
        print(f"{'Variable':<20} {'Coefficient':>12} {'Std.Error':>12} {'t-stat':>10} {'p-value':>10} {'Sig':>6}")
        print("-"*70)

        # Print all variables except const first
        for var in sorted(results.params.index):
            if var != 'const':
                coeff = results.params[var]
                std_err = results.std_errors.get(var, np.nan) if hasattr(results, 'std_errors') else np.nan
                t_stat = results.tstats.get(var, np.nan) if hasattr(results, 'tstats') else np.nan
                pval = results.pvalues.get(var, np.nan)

                # Determine significance
                if pval < 0.01:
                    sig = '***'
                elif pval < 0.05:
                    sig = '**'
                elif pval < 0.10:
                    sig = '*'
                else:
                    sig = ''

                print(f"{var:<20} {coeff:>12.4f} {std_err:>12.4f} {t_stat:>10.4f} {pval:>10.4f} {sig:>6}")

        # Print constant if present
        if 'const' in results.params:
            coeff = results.params['const']
            std_err = results.std_errors.get('const', np.nan) if hasattr(results, 'std_errors') else np.nan
            t_stat = results.tstats.get('const', np.nan) if hasattr(results, 'tstats') else np.nan
            pval = results.pvalues.get('const', np.nan)

            if pval < 0.01:
                sig = '***'
            elif pval < 0.05:
                sig = '**'
            elif pval < 0.10:
                sig = '*'
            else:
                sig = ''

            print(f"{'const':<20} {coeff:>12.4f} {std_err:>12.4f} {t_stat:>10.4f} {pval:>10.4f} {sig:>6}")

        print("-"*70)
        print(f"R-squared: {results.rsquared:.4f}")
        print(f"Observations: {results.nobs}")
        if hasattr(results, 'f_statistic'):
            print(f"F-statistic: {results.f_statistic.stat:.4f} (p={results.f_statistic.pval:.4f})")
        print("="*70)
        print("Significance codes: *** p<0.01, ** p<0.05, * p<0.10")

    def _create_lag_structure_plot_aggregate(self, results, max_lag, title):
        """Create lag structure plot for aggregate ESG"""
        fig, ax = plt.subplots(figsize=(10, 6))
        fig.suptitle(title, fontsize=14, fontweight='bold')

        lags = list(range(1, max_lag + 1))
        coeffs = []
        errors = []

        for lag in lags:
            var = f'ESG_lag{lag}'
            if var in results.params:
                coeffs.append(results.params[var])
                if hasattr(results, 'std_errors') and var in results.std_errors:
                    errors.append(results.std_errors[var])
                else:
                    errors.append(0)
            else:
                coeffs.append(0)
                errors.append(0)

        ax.errorbar(lags, coeffs, yerr=errors, fmt='o-',
                   capsize=5, linewidth=2, markersize=8, color='green')
        ax.axhline(y=0, color='r', linestyle='--', alpha=0.5)
        ax.set_xlabel('Years Lag')
        ax.set_ylabel('Coefficient')
        ax.set_title('Aggregate ESG Lag Structure')
        ax.set_xticks(lags)
        ax.grid(True, alpha=0.3)

        # Add value labels
        for i, (lag, coeff) in enumerate(zip(lags, coeffs)):
            ax.annotate(f'{coeff:.3f}', (lag, coeff),
                       textcoords="offset points", xytext=(0,10),
                       ha='center', fontsize=9)

        plt.tight_layout()

        # Save plot
        os.makedirs(self.config.OUTPUT_DIR, exist_ok=True)
        plot_filename = f"lag_structure_aggregate_{title.replace(' ', '_').replace('-', '_')}.png"
        plt.savefig(os.path.join(self.config.OUTPUT_DIR, plot_filename),
                   dpi=300, bbox_inches='tight')
        plt.show()

        return fig

    def run_simple_lag_model(self, max_lag=5):
        """
        Run simple lag model with aggregate ESG
        """
        print("\n" + "="*80)
        print("SIMPLE LAG MODEL (t-1 only) - Aggregate ESG")
        print("="*80)

        # Load data
        df, df_cs, df_all = self.loader.load_and_prepare_data()

        # For Consumer Staples
        print("\nCONSUMER STAPLES ONLY:")
        df_cs_lagged, control_vars = self.loader.create_lagged_variables_for_model(
            df_cs, min_lag=1, max_lag=max_lag, use_original_controls=True
        )

        # Use simple approach: just ESG_lag
        simple_vars = ['ESG_lag'] + control_vars + ['Tobin_Q']
        df_simple = df_cs_lagged.dropna(subset=simple_vars)

        if len(df_simple) > 0:
            y = df_simple['Tobin_Q']
            X = df_simple[['ESG_lag'] + control_vars]
            X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

            model = PanelOLS(y, X, entity_effects=False, time_effects=False)
            results = model.fit(cov_type='robust')

            print(f"\nSimple Lag Model (t-1 only) - Aggregate ESG:")
            print(f"Observations: {results.nobs}")
            print(f"R-squared: {results.rsquared:.4f}")

            # Print regression table
            self._print_regression_table(results, "Simple Lag Model (t-1)")

            # Print equation
            print("\n" + "="*60)
            print("REGRESSION EQUATION:")
            print("="*60)
            equation = self.exporter.generate_model_equation(
                results,
                "Simple Model",
                "Consumer Staples Only",
                1,
                'simple'
            )
            print(equation)

            # Store results
            key = f"simple_lag_{max_lag}_cs"
            self.results[key] = {
                'results': results,
                'type': 'simple',
                'sector': 'cs',
                'max_lag': 1,
                'nobs': results.nobs,
                'r2': results.rsquared,
                'data': df_simple
            }

        # For All Sectors
        print("\nALL SECTORS:")
        df_all_lagged, control_vars = self.loader.create_lagged_variables_for_model(
            df_all, min_lag=1, max_lag=max_lag, use_original_controls=True
        )

        simple_vars = ['ESG_lag'] + control_vars + ['ConsumerStaples', 'Tobin_Q']
        df_simple_all = df_all_lagged.dropna(subset=simple_vars)

        if len(df_simple_all) > 0:
            y = df_simple_all['Tobin_Q']
            X = df_simple_all[['ESG_lag'] + control_vars + ['ConsumerStaples']]
            X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

            model = PanelOLS(y, X, entity_effects=False, time_effects=False)
            results_all = model.fit(cov_type='robust')

            print(f"\nSimple Lag Model (t-1 only) - All Sectors:")
            print(f"Observations: {results_all.nobs}")
            print(f"R-squared: {results_all.rsquared:.4f}")

            # Print regression table
            self._print_regression_table(results_all, "Simple Lag Model (t-1) - All Sectors")

            # Print equation
            print("\n" + "="*60)
            print("REGRESSION EQUATION:")
            print("="*60)
            equation = self.exporter.generate_model_equation(
                results_all,
                "Simple Model",
                "All Sectors",
                1,
                'simple'
            )
            print(equation)

            # Store results
            key = f"simple_lag_{max_lag}_all"
            self.results[key] = {
                'results': results_all,
                'type': 'simple',
                'sector': 'all',
                'max_lag': 1,
                'nobs': results_all.nobs,
                'r2': results_all.rsquared,
                'data': df_simple_all
            }

        return results

# ============================================================================
# RESULTS COMPARISON
# ============================================================================
class ResultsComparator:
    """Compare results from different model specifications"""

    def __init__(self, analyzer):
        self.analyzer = analyzer
        self.comparison_df = None

    def create_comparison_table(self):
        """Create comparison table of all results"""
        comparison_data = []

        for key, model_info in self.analyzer.results.items():
            results = model_info['results']

            row = {
                'Model_Type': model_info['type'],
                'Sector': 'Consumer Staples' if model_info['sector'] == 'cs' else 'All Sectors',
                'Max_Lag': model_info['max_lag'],
                'N': results.nobs,
                'R2': results.rsquared,
                'R2_Adj': getattr(results, 'rsquared_adj', np.nan),
            }

            # Add F-statistic if available
            if hasattr(results, 'f_statistic'):
                row['F_stat'] = results.f_statistic.stat
                row['F_pval'] = results.f_statistic.pval

            # Extract ESG coefficients
            if 'individual' in model_info['type']:
                # For individual models, get average, min, max across lags
                esg_coeffs = []
                esg_pvals = []
                for var in results.params.index:
                    if 'ESG' in str(var) and 'lag' in str(var):
                        esg_coeffs.append(results.params[var])
                        esg_pvals.append(results.pvalues.get(var, np.nan))

                if esg_coeffs:
                    row['ESG_avg_coeff'] = np.mean(esg_coeffs)
                    row['ESG_min_coeff'] = min(esg_coeffs)
                    row['ESG_max_coeff'] = max(esg_coeffs)
                    row['ESG_avg_pval'] = np.mean(esg_pvals)
                    row['ESG_min_pval'] = min(esg_pvals)
                    row['ESG_max_pval'] = max(esg_pvals)

                    # Count significant lags
                    sig_count = sum(1 for p in esg_pvals if p < 0.10)
                    row['ESG_sig_lags_10pct'] = sig_count

            elif 'summed' in model_info['type']:
                # For summed models
                var = f'ESG_lag1{model_info["max_lag"]}'
                if var in results.params:
                    row['ESG_sum_coeff'] = results.params[var]
                    row['ESG_sum_pval'] = results.pvalues.get(var, np.nan)

            elif 'simple' in model_info['type']:
                # For simple model
                if 'ESG_lag' in results.params:
                    row['ESG_simple_coeff'] = results.params['ESG_lag']
                    row['ESG_simple_pval'] = results.pvalues.get('ESG_lag', np.nan)

            comparison_data.append(row)

        self.comparison_df = pd.DataFrame(comparison_data)
        return self.comparison_df

    def print_comparison(self):
        """Print comparison of results"""
        if self.comparison_df is None:
            self.create_comparison_table()

        print("\n" + "="*100)
        print("MODEL COMPARISON - AGGREGATE ESG")
        print("="*100)

        # Sort by R2
        comparison_sorted = self.comparison_df.sort_values('R2', ascending=False)

        # Print header
        print(f"\n{'Model Type':<25} {'Sector':<18} {'Lags':<6} {'N':<8} {'R²':<8} {'ESG Effect':<20}")
        print("-"*100)

        for _, row in comparison_sorted.iterrows():
            model_type = row['Model_Type'].replace('_', ' ').title()

            if 'Individual' in model_type:
                esg_effect = f"Avg: {row.get('ESG_avg_coeff', np.nan):.4f}"
                if 'ESG_sig_lags_10pct' in row:
                    esg_effect += f" ({int(row['ESG_sig_lags_10pct'])} sig)"
            elif 'Summed' in model_type:
                esg_effect = f"Sum: {row.get('ESG_sum_coeff', np.nan):.4f}"
            else:
                esg_effect = f"Simple: {row.get('ESG_simple_coeff', np.nan):.4f}"

            print(f"{model_type[:25]:<25} {row['Sector']:<18} {int(row['Max_Lag']):<6} "
                  f"{int(row['N']):<8} {row['R2']:<8.4f} {esg_effect:<20}")

        print("="*100)

# ============================================================================
# MAIN EXECUTION
# ============================================================================
def main():
    """Main execution function"""
    print("\n" + "="*80)
    print("LAG MODEL ANALYSIS - AGGREGATE ESG SCORE")
    print("="*80)

    # Initialize
    config = Config()
    analyzer = LagModelAnalyzer(config)

    # Run simple model first (t-1 only)
    analyzer.run_simple_lag_model(max_lag=5)

    # Run analysis with different lag specifications (1-2, 1-3, 1-4, 1-5)
    lag_ranges = [2, 3, 4, 5]

    for max_lag in lag_ranges:
        print(f"\n{'='*80}")
        print(f"ANALYZING LAG RANGE: 1-{max_lag} years")
        print(f"{'='*80}")

        analyzer.run_analysis(max_lag=max_lag)

    # Compare results
    comparator = ResultsComparator(analyzer)
    comparator.create_comparison_table()
    comparator.print_comparison()

    # Export all results to Excel
    print("\n" + "="*80)
    print("EXPORTING RESULTS TO EXCEL")
    print("="*80)
    excel_path = analyzer.exporter.export_all_results(analyzer)

    print("\n" + "="*80)
    print("ANALYSIS COMPLETE")
    print("="*80)
    print(f"\nAll results saved to: {config.OUTPUT_DIR}")
    print(f"Main Excel file: {excel_path}")

    return analyzer, comparator

# ============================================================================
# EXECUTION
# ============================================================================
if __name__ == "__main__":
    analyzer, comparator = main()

# ============================================================================
# ROBUSTNESS CHECK FOR AGGREGATE ESG LAG MODEL
# ============================================================================
print("\n" + "="*80)
print("ROBUSTNESS CHECK: AGGREGATE ESG LAG MODEL")
print("="*80)

# First, capture the original results from your model
print("\n" + "-"*40)
print("ORIGINAL MODEL RESULTS")
print("-"*40)

# Get the simple lag model results (Consumer Staples and All Sectors)
original_results = {}

# Consumer Staples Simple Model
if 'simple_lag_5_cs' in analyzer.results:
    res_cs = analyzer.results['simple_lag_5_cs']['results']
    original_results['Consumer Staples - Simple'] = {
        'R2': res_cs.rsquared,
        'ESG_Coeff': res_cs.params.get('ESG_lag', 0),
        'ESG_Pval': res_cs.pvalues.get('ESG_lag', 1.0),
        'N': res_cs.nobs,
        'Significant': res_cs.pvalues.get('ESG_lag', 1.0) < 0.10
    }
    print(f"\nConsumer Staples (Simple Model):")
    print(f"  R² = {res_cs.rsquared:.4f}")
    print(f"  ESG Coefficient = {res_cs.params.get('ESG_lag', 0):.4f}")
    print(f"  ESG p-value = {res_cs.pvalues.get('ESG_lag', 1.0):.4f}")
    print(f"  N = {res_cs.nobs}")

# All Sectors Simple Model
if 'simple_lag_5_all' in analyzer.results:
    res_all = analyzer.results['simple_lag_5_all']['results']
    original_results['All Sectors - Simple'] = {
        'R2': res_all.rsquared,
        'ESG_Coeff': res_all.params.get('ESG_lag', 0),
        'ESG_Pval': res_all.pvalues.get('ESG_lag', 1.0),
        'N': res_all.nobs,
        'Significant': res_all.pvalues.get('ESG_lag', 1.0) < 0.10
    }
    print(f"\nAll Sectors (Simple Model):")
    print(f"  R² = {res_all.rsquared:.4f}")
    print(f"  ESG Coefficient = {res_all.params.get('ESG_lag', 0):.4f}")
    print(f"  ESG p-value = {res_all.pvalues.get('ESG_lag', 1.0):.4f}")
    print(f"  N = {res_all.nobs}")

# Get distributed lag results
print(f"\n\nDistributed Lag Models (1-5 years):")
for max_lag in [2, 3, 4, 5]:
    for sector_code, sector_name in [('cs', 'Consumer Staples'), ('all', 'All Sectors')]:
        # Individual lags
        ind_key = f"distributed_individual_{max_lag}_{sector_code}"
        if ind_key in analyzer.results:
            res_ind = analyzer.results[ind_key]['results']
            key = f"{sector_name} - Individual Lags 1-{max_lag}"
            # Calculate average ESG coefficient
            esg_coeffs = []
            for i in range(1, max_lag+1):
                coeff = res_ind.params.get(f'ESG_lag{i}', 0)
                esg_coeffs.append(coeff)

            # Count significant lags
            sig_count = 0
            for i in range(1, max_lag+1):
                pval = res_ind.pvalues.get(f'ESG_lag{i}', 1.0)
                if pval < 0.10:
                    sig_count += 1

            original_results[key] = {
                'R2': res_ind.rsquared,
                'ESG_Avg_Coeff': np.mean(esg_coeffs) if esg_coeffs else 0,
                'ESG_Sig_Count': sig_count,
                'N': res_ind.nobs
            }
            print(f"\n  {sector_name} - Lags 1-{max_lag}:")
            print(f"    R² = {res_ind.rsquared:.4f}")
            print(f"    Avg ESG Coeff = {original_results[key]['ESG_Avg_Coeff']:.4f}")
            print(f"    Significant lags = {original_results[key]['ESG_Sig_Count']}/{max_lag}")

        # Summed lags
        sum_key = f"distributed_summed_{max_lag}_{sector_code}"
        if sum_key in analyzer.results:
            res_sum = analyzer.results[sum_key]['results']
            key = f"{sector_name} - Summed Lags 1-{max_lag}"
            original_results[key] = {
                'R2': res_sum.rsquared,
                'ESG_Coeff': res_sum.params.get(f'ESG_lag1{max_lag}', 0),
                'ESG_Pval': res_sum.pvalues.get(f'ESG_lag1{max_lag}', 1.0),
                'N': res_sum.nobs,
                'Significant': res_sum.pvalues.get(f'ESG_lag1{max_lag}', 1.0) < 0.10
            }
            print(f"    Summed Model R² = {res_sum.rsquared:.4f}")

# ============================================================================
# CREATE TRANSFORMATION COLUMNS FIRST
# ============================================================================
print("\n" + "="*80)
print("CREATING TRANSFORMATION COLUMNS")
print("="*80)

# Get the dataframe from analyzer
df, df_cs, df_all = analyzer.loader.load_and_prepare_data()

# Create transformation columns safely
print("\nCreating transformation columns...")

# Log transformation (add small constant to avoid log(0))
if 'Tobin_Q_log' not in df.columns:
    df['Tobin_Q_log'] = np.log(df['Tobin_Q'].clip(lower=0.001))
    print("✓ Created Tobin_Q_log")
else:
    print("✓ Tobin_Q_log already exists")

# Square root transformation (ensure non-negative)
if 'Tobin_Q_sqrt' not in df.columns:
    df['Tobin_Q_sqrt'] = np.sqrt(df['Tobin_Q'].clip(lower=0))
    print("✓ Created Tobin_Q_sqrt")
else:
    print("✓ Tobin_Q_sqrt already exists")

# Inverse transformation (handle division by zero)
if 'Tobin_Q_inv' not in df.columns:
    with np.errstate(divide='ignore', invalid='ignore'):
        df['Tobin_Q_inv'] = 1 / df['Tobin_Q']
        df['Tobin_Q_inv'] = df['Tobin_Q_inv'].replace([np.inf, -np.inf], np.nan)
    print("✓ Created Tobin_Q_inv")
else:
    print("✓ Tobin_Q_inv already exists")

# Update the dataframes
df_cs = df[df['ConsumerStaples'] == 1].copy() if 'ConsumerStaples' in df.columns else df[df.index.get_level_values(0).isin(Config.CONSUMER_STAPLES_FIRMS)].copy()
df_all = df.copy()

print(f"\nTransformation columns created. Data shape: {df.shape}")

# ============================================================================
# ROBUSTNESS CHECK 1: ALTERNATIVE DEPENDENT VARIABLE TRANSFORMATIONS
# ============================================================================
print("\n" + "="*80)
print("ROBUSTNESS CHECK 1: ALTERNATIVE DEPENDENT VARIABLE TRANSFORMATIONS")
print("="*80)

# Test different transformations for the simple model
print("\n" + "-"*60)
print("TESTING DIFFERENT TRANSFORMATIONS - SIMPLE MODEL (Consumer Staples)")
print("-"*60)

transformations = [
    ('Original', 'Tobin_Q'),
    ('Log', 'Tobin_Q_log'),
    ('Square Root', 'Tobin_Q_sqrt'),
    ('Inverse', 'Tobin_Q_inv')
]

transformation_results = {}

for trans_name, trans_var in transformations:
    print(f"\nTesting {trans_name} transformation...")

    # Prepare data for Consumer Staples
    df_cs_trans = df_cs.copy()

    # Create lagged ESG
    df_cs_trans['ESG_lag'] = df_cs_trans.groupby(level=0)['ESG'].shift(1)

    # Create control variables if they don't exist
    if 'Size' not in df_cs_trans.columns:
        df_cs_trans['Size'] = np.log(df_cs_trans['Total Assets'])
    if 'Leverage' not in df_cs_trans.columns:
        df_cs_trans['Leverage'] = df_cs_trans['Total Liabilities'] / df_cs_trans['Total Assets']

    # Check if transformation column exists and has data
    if trans_var not in df_cs_trans.columns:
        print(f"  ⚠ {trans_var} column not found - skipping")
        transformation_results[trans_name] = None
        continue

    # Check if transformation column has non-NaN values
    if df_cs_trans[trans_var].isna().all():
        print(f"  ⚠ {trans_var} has all NaN values - skipping")
        transformation_results[trans_name] = None
        continue

    # Drop NA
    model_vars = ['ESG_lag', 'ROA', 'Size', 'Leverage', trans_var]
    df_clean = df_cs_trans.dropna(subset=model_vars)

    print(f"  Observations after cleaning: {len(df_clean)}")

    if len(df_clean) >= 10:
        # Run regression
        y = df_clean[trans_var]
        X = df_clean[['ESG_lag', 'ROA', 'Size', 'Leverage']]
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

        try:
            model = PanelOLS(y, X, entity_effects=False, time_effects=False)
            results = model.fit(cov_type='robust')

            transformation_results[trans_name] = {
                'ESG_Coeff': results.params.get('ESG_lag', 0),
                'ESG_Pval': results.pvalues.get('ESG_lag', 1.0),
                'R2': results.rsquared,
                'N': results.nobs,
                'Significant': results.pvalues.get('ESG_lag', 1.0) < 0.10
            }

            sig = "***" if results.pvalues.get('ESG_lag', 1.0) < 0.01 else \
                  "**" if results.pvalues.get('ESG_lag', 1.0) < 0.05 else \
                  "*" if results.pvalues.get('ESG_lag', 1.0) < 0.10 else ""

            print(f"  ✓ Completed: ESG Coeff = {results.params.get('ESG_lag', 0):.4f}{sig}")
            print(f"    p-value = {results.pvalues.get('ESG_lag', 1.0):.4f}")
            print(f"    R² = {results.rsquared:.4f}")

        except Exception as e:
            print(f"  ✗ Failed: {str(e)[:100]}")
            transformation_results[trans_name] = None
    else:
        print(f"  ✗ Insufficient data: {len(df_clean)} observations (need at least 10)")
        transformation_results[trans_name] = None

# ============================================================================
# ROBUSTNESS CHECK 2: INDUSTRY FIXED EFFECTS
# ============================================================================
print("\n" + "="*80)
print("ROBUSTNESS CHECK 2: INDUSTRY FIXED EFFECTS")
print("="*80)

if 'Sector' in df.columns:
    # Create industry dummies
    industry_dummies = pd.get_dummies(df['Sector'], prefix='ind', drop_first=True)
    df_with_industry = pd.concat([df, industry_dummies], axis=1)
    industry_cols = [col for col in df_with_industry.columns if col.startswith('ind_')]

    print(f"\nIndustry sectors found: {df['Sector'].nunique()}")
    print(f"Industry dummies created: {len(industry_cols)}")

    # Test simple model with and without industry FE
    print("\n" + "-"*60)
    print("SIMPLE MODEL WITH INDUSTRY FIXED EFFECTS (All Sectors)")
    print("-"*60)

    # Create lagged variables
    df_all_lagged = df_with_industry.copy()
    df_all_lagged['ESG_lag'] = df_all_lagged.groupby(level=0)['ESG'].shift(1)

    if 'Size' not in df_all_lagged.columns:
        df_all_lagged['Size'] = np.log(df_all_lagged['Total Assets'])
    if 'Leverage' not in df_all_lagged.columns:
        df_all_lagged['Leverage'] = df_all_lagged['Total Liabilities'] / df_all_lagged['Total Assets']

    # Model WITHOUT industry FE
    vars_no_fe = ['ESG_lag', 'ROA', 'Size', 'Leverage', 'Tobin_Q']
    df_clean_no_fe = df_all_lagged.dropna(subset=vars_no_fe)

    if len(df_clean_no_fe) >= 10:
        y_no_fe = df_clean_no_fe['Tobin_Q']
        X_no_fe = df_clean_no_fe[['ESG_lag', 'ROA', 'Size', 'Leverage']]
        X_no_fe = pd.concat([pd.Series(1, index=X_no_fe.index, name='const'), X_no_fe], axis=1)

        model_no_fe = PanelOLS(y_no_fe, X_no_fe, entity_effects=False, time_effects=False)
        results_no_fe = model_no_fe.fit(cov_type='robust')

        print(f"\nWithout Industry FE:")
        print(f"  ESG Coeff = {results_no_fe.params.get('ESG_lag', 0):.4f}")
        print(f"  p-value = {results_no_fe.pvalues.get('ESG_lag', 1.0):.4f}")
        print(f"  R² = {results_no_fe.rsquared:.4f}")
        print(f"  N = {results_no_fe.nobs}")

        # Model WITH industry FE
        vars_with_fe = ['ESG_lag', 'ROA', 'Size', 'Leverage'] + industry_cols + ['Tobin_Q']
        df_clean_with_fe = df_all_lagged.dropna(subset=vars_with_fe)

        if len(df_clean_with_fe) >= 10:
            y_with_fe = df_clean_with_fe['Tobin_Q']
            X_with_fe = df_clean_with_fe[['ESG_lag', 'ROA', 'Size', 'Leverage'] + industry_cols]
            X_with_fe = pd.concat([pd.Series(1, index=X_with_fe.index, name='const'), X_with_fe], axis=1)

            model_with_fe = PanelOLS(y_with_fe, X_with_fe, entity_effects=False, time_effects=False)
            results_with_fe = model_with_fe.fit(cov_type='robust')

            print(f"\nWith Industry FE:")
            print(f"  ESG Coeff = {results_with_fe.params.get('ESG_lag', 0):.4f}")
            print(f"  p-value = {results_with_fe.pvalues.get('ESG_lag', 1.0):.4f}")
            print(f"  R² = {results_with_fe.rsquared:.4f}")
            print(f"  N = {results_with_fe.nobs}")

            # Calculate changes
            coeff_change = results_with_fe.params.get('ESG_lag', 0) - results_no_fe.params.get('ESG_lag', 0)
            coeff_change_pct = (coeff_change / abs(results_no_fe.params.get('ESG_lag', 0))) * 100 if results_no_fe.params.get('ESG_lag', 0) != 0 else 0

            print(f"\nImpact of Industry FE:")
            print(f"  Coefficient change: {coeff_change:+.4f} ({coeff_change_pct:+.1f}%)")
            print(f"  Significance maintained: {'YES' if (results_with_fe.pvalues.get('ESG_lag', 1.0) < 0.10) == (results_no_fe.pvalues.get('ESG_lag', 1.0) < 0.10) else 'NO'}")
        else:
            print(f"\nInsufficient data for model WITH industry FE: {len(df_clean_with_fe)} observations")
    else:
        print(f"\nInsufficient data for model WITHOUT industry FE: {len(df_clean_no_fe)} observations")
else:
    print("Sector variable not available for industry fixed effects")

# ============================================================================
# ROBUSTNESS CHECK 3: NON-LINEAR THRESHOLDS
# ============================================================================
print("\n" + "="*80)
print("ROBUSTNESS CHECK 3: NON-LINEAR THRESHOLDS")
print("="*80)

# Test different ESG thresholds for Consumer Staples
df_cs_thresh = df_cs.copy()
df_cs_thresh['ESG_lag'] = df_cs_thresh.groupby(level=0)['ESG'].shift(1)

if 'Size' not in df_cs_thresh.columns:
    df_cs_thresh['Size'] = np.log(df_cs_thresh['Total Assets'])
if 'Leverage' not in df_cs_thresh.columns:
    df_cs_thresh['Leverage'] = df_cs_thresh['Total Liabilities'] / df_cs_thresh['Total Assets']

thresholds = {
    'High ESG (Top 25%)': 0.75,
    'Medium ESG (Median)': 0.50,
    'Very High ESG (Top 10%)': 0.90
}

print("\n" + "-"*60)
print("THRESHOLD MODELS - Consumer Staples")
print("-"*60)

threshold_results = {}

for thresh_name, percentile in thresholds.items():
    print(f"\nTesting {thresh_name}...")

    # Create threshold dummy
    thresh_value = df_cs_thresh['ESG_lag'].quantile(percentile)
    df_cs_thresh['ESG_high'] = (df_cs_thresh['ESG_lag'] > thresh_value).astype(int)

    # Run model with threshold dummy
    model_vars = ['ESG_high', 'ROA', 'Size', 'Leverage', 'Tobin_Q']
    df_clean = df_cs_thresh.dropna(subset=model_vars)

    print(f"  Observations: {len(df_clean)}")
    print(f"  Threshold value: {thresh_value:.4f}")

    if len(df_clean) >= 10:
        y = df_clean['Tobin_Q']
        X = df_clean[['ESG_high', 'ROA', 'Size', 'Leverage']]
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

        try:
            model = PanelOLS(y, X, entity_effects=False, time_effects=False)
            results = model.fit(cov_type='robust')

            threshold_results[thresh_name] = {
                'ESG_High_Coeff': results.params.get('ESG_high', 0),
                'ESG_High_Pval': results.pvalues.get('ESG_high', 1.0),
                'R2': results.rsquared,
                'N': results.nobs,
                'Threshold_Value': thresh_value
            }

            sig = "***" if results.pvalues.get('ESG_high', 1.0) < 0.01 else \
                  "**" if results.pvalues.get('ESG_high', 1.0) < 0.05 else \
                  "*" if results.pvalues.get('ESG_high', 1.0) < 0.10 else ""

            print(f"  ✓ Completed: ESG_High Coeff = {results.params.get('ESG_high', 0):.4f}{sig}")
            print(f"    p-value = {results.pvalues.get('ESG_high', 1.0):.4f}")
            print(f"    R² = {results.rsquared:.4f}")

        except Exception as e:
            print(f"  ✗ Failed: {str(e)[:100]}")
            threshold_results[thresh_name] = None
    else:
        print(f"  ✗ Insufficient data: {len(df_clean)} observations")
        threshold_results[thresh_name] = None

# ============================================================================
# ROBUSTNESS CHECK 4: MODERATING EFFECTS
# ============================================================================
print("\n" + "="*80)
print("ROBUSTNESS CHECK 4: MODERATING EFFECTS")
print("="*80)

# Test moderators for Consumer Staples
df_cs_mod = df_cs.copy()
df_cs_mod['ESG_lag'] = df_cs_mod.groupby(level=0)['ESG'].shift(1)

if 'Size' not in df_cs_mod.columns:
    df_cs_mod['Size'] = np.log(df_cs_mod['Total Assets'])
if 'Leverage' not in df_cs_mod.columns:
    df_cs_mod['Leverage'] = df_cs_mod['Total Liabilities'] / df_cs_mod['Total Assets']

moderators = {
    'Size': 'Size',
    'ROA': 'ROA',
    'Leverage': 'Leverage'
}

print("\n" + "-"*60)
print("MODERATING EFFECTS - Consumer Staples")
print("-"*60)

moderation_results = {}

for mod_name, mod_var in moderators.items():
    print(f"\nTesting moderator: {mod_name}")

    # Create interaction term
    df_cs_mod[f'ESG_x_{mod_var}'] = df_cs_mod['ESG_lag'] * df_cs_mod[mod_var]

    # Run model with interaction
    model_vars = ['ESG_lag', mod_var, f'ESG_x_{mod_var}', 'ROA', 'Size', 'Leverage', 'Tobin_Q']
    # Remove duplicates if moderator is in controls
    model_vars = list(dict.fromkeys(model_vars))

    df_clean = df_cs_mod.dropna(subset=model_vars)

    print(f"  Observations: {len(df_clean)}")

    if len(df_clean) >= 10:
        y = df_clean['Tobin_Q']
        X = df_clean[[v for v in model_vars if v != 'Tobin_Q']]
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

        try:
            model = PanelOLS(y, X, entity_effects=False, time_effects=False)
            results = model.fit(cov_type='robust')

            inter_var = f'ESG_x_{mod_var}'
            if inter_var in results.params:
                moderation_results[mod_name] = {
                    'Interaction_Coeff': results.params.get(inter_var, 0),
                    'Interaction_Pval': results.pvalues.get(inter_var, 1.0),
                    'ESG_Direct_Coeff': results.params.get('ESG_lag', 0),
                    'ESG_Direct_Pval': results.pvalues.get('ESG_lag', 1.0),
                    'R2': results.rsquared,
                    'N': results.nobs
                }

                sig = "***" if results.pvalues.get(inter_var, 1.0) < 0.01 else \
                      "**" if results.pvalues.get(inter_var, 1.0) < 0.05 else \
                      "*" if results.pvalues.get(inter_var, 1.0) < 0.10 else ""

                print(f"  ✓ Completed: ESG × {mod_name} Coeff = {results.params.get(inter_var, 0):.4f}{sig}")
                print(f"    p-value = {results.pvalues.get(inter_var, 1.0):.4f}")
                print(f"    Direct ESG Coeff = {results.params.get('ESG_lag', 0):.4f}")
                print(f"    R² = {results.rsquared:.4f}")
            else:
                print(f"  ⚠ Interaction term not in model")
                moderation_results[mod_name] = None

        except Exception as e:
            print(f"  ✗ Failed: {str(e)[:100]}")
            moderation_results[mod_name] = None
    else:
        print(f"  ✗ Insufficient data: {len(df_clean)} observations")
        moderation_results[mod_name] = None

# ============================================================================
# SUMMARY COMPARISON TABLE
# ============================================================================
print("\n" + "="*100)
print("ROBUSTNESS CHECK SUMMARY - AGGREGATE ESG MODEL")
print("="*100)

# Create comparison table
comparison_data = []

# Original results
for model_name, results in original_results.items():
    if 'Simple' in model_name:
        comparison_data.append({
            'Model': model_name,
            'Type': 'Original',
            'R2': results['R2'],
            'ESG_Coeff': results.get('ESG_Coeff', results.get('ESG_Avg_Coeff', 0)),
            'ESG_Pval': results.get('ESG_Pval', 0.5),
            'N': results['N'],
            'Significant': results.get('Significant', results.get('ESG_Sig_Count', 0) > 0)
        })

# Transformation results
for trans_name, results in transformation_results.items():
    if results:
        comparison_data.append({
            'Model': f"Transformation: {trans_name}",
            'Type': 'Robustness',
            'R2': results['R2'],
            'ESG_Coeff': results['ESG_Coeff'],
            'ESG_Pval': results['ESG_Pval'],
            'N': results['N'],
            'Significant': results['Significant']
        })

# Industry FE results
if 'results_with_fe' in locals() and 'results_no_fe' in locals():
    comparison_data.append({
        'Model': "With Industry FE",
        'Type': 'Robustness',
        'R2': results_with_fe.rsquared,
        'ESG_Coeff': results_with_fe.params.get('ESG_lag', 0),
        'ESG_Pval': results_with_fe.pvalues.get('ESG_lag', 1.0),
        'N': results_with_fe.nobs,
        'Significant': results_with_fe.pvalues.get('ESG_lag', 1.0) < 0.10
    })

# Best threshold result
valid_thresholds = {k: v for k, v in threshold_results.items() if v is not None}
if valid_thresholds:
    best_threshold = max(valid_thresholds.items(), key=lambda x: x[1]['R2'])
    if best_threshold[1]:
        comparison_data.append({
            'Model': f"Threshold: {best_threshold[0]}",
            'Type': 'Robustness',
            'R2': best_threshold[1]['R2'],
            'ESG_Coeff': best_threshold[1]['ESG_High_Coeff'],
            'ESG_Pval': best_threshold[1]['ESG_High_Pval'],
            'N': best_threshold[1]['N'],
            'Significant': best_threshold[1]['ESG_High_Pval'] < 0.10
        })

# Best moderation result
valid_moderations = {k: v for k, v in moderation_results.items() if v is not None}
if valid_moderations:
    # Find moderation with significant interaction
    sig_mods = {k: v for k, v in valid_moderations.items() if v['Interaction_Pval'] < 0.10}
    if sig_mods:
        best_mod = max(sig_mods.items(), key=lambda x: abs(x[1]['Interaction_Coeff']))
    else:
        best_mod = max(valid_moderations.items(), key=lambda x: abs(x[1]['Interaction_Coeff']))

    if best_mod[1]:
        comparison_data.append({
            'Model': f"Moderation: {best_mod[0]}",
            'Type': 'Robustness',
            'R2': best_mod[1]['R2'],
            'ESG_Coeff': best_mod[1]['Interaction_Coeff'],
            'ESG_Pval': best_mod[1]['Interaction_Pval'],
            'N': best_mod[1]['N'],
            'Significant': best_mod[1]['Interaction_Pval'] < 0.10
        })

# Create DataFrame and print
if comparison_data:
    comparison_df = pd.DataFrame(comparison_data)
    print("\n" + comparison_df.to_string(index=False))

    # Calculate robustness metrics
    print("\n" + "="*60)
    print("ROBUSTNESS ASSESSMENT")
    print("="*60)

    # Check sign consistency
    original_models = comparison_df[comparison_df['Type'] == 'Original']
    if len(original_models) > 0:
        original_sign = np.sign(original_models['ESG_Coeff'].iloc[0])

        robustness_checks = comparison_df[comparison_df['Type'] == 'Robustness']
        if len(robustness_checks) > 0:
            sign_consistent = sum(np.sign(robustness_checks['ESG_Coeff']) == original_sign) / len(robustness_checks) * 100

            # Check significance consistency
            original_sig = original_models['Significant'].iloc[0]
            sig_consistent = sum(robustness_checks['Significant'] == original_sig) / len(robustness_checks) * 100

            print(f"\nOriginal Model: ESG Coeff = {original_models['ESG_Coeff'].iloc[0]:.4f}, "
                  f"Significant = {original_sig}")

            print(f"\nRobustness Check Results ({len(robustness_checks)} checks):")
            print(f"  Sign consistency: {sign_consistent:.1f}% of checks have same sign as original")
            print(f"  Significance consistency: {sig_consistent:.1f}% of checks have same significance as original")

            if sign_consistent >= 80 and sig_consistent >= 80:
                print("\n✅ MODEL IS HIGHLY ROBUST - Results consistent across all checks")
            elif sign_consistent >= 60 or sig_consistent >= 60:
                print("\n⚠️ MODEL IS MODERATELY ROBUST - Some variation in results")
            else:
                print("\n❌ MODEL IS NOT ROBUST - Results vary substantially across checks")
        else:
            print("\nNo robustness checks completed successfully")
    else:
        print("\nNo original models found for comparison")
else:
    print("\nNo comparison data available")

# Export to Excel
try:
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    robustness_path = os.path.join(Config.OUTPUT_DIR, f'aggregate_esg_robustness_{timestamp}.xlsx')

    with pd.ExcelWriter(robustness_path, engine='openpyxl') as writer:
        # Original results
        if original_results:
            original_df = pd.DataFrame([
                {'Model': k, **v} for k, v in original_results.items()
            ])
            original_df.to_excel(writer, sheet_name='Original_Results', index=False)

        # Transformation results
        if transformation_results:
            trans_data = []
            for k, v in transformation_results.items():
                if v:
                    row = {'Transformation': k}
                    row.update(v)
                    trans_data.append(row)
            if trans_data:
                trans_df = pd.DataFrame(trans_data)
                trans_df.to_excel(writer, sheet_name='Transformations', index=False)

        # Industry FE results
        if 'results_with_fe' in locals() and 'results_no_fe' in locals():
            industry_df = pd.DataFrame([
                {'Model': 'Without FE', 'ESG_Coeff': results_no_fe.params.get('ESG_lag', 0),
                 'ESG_Pval': results_no_fe.pvalues.get('ESG_lag', 1.0), 'R2': results_no_fe.rsquared},
                {'Model': 'With FE', 'ESG_Coeff': results_with_fe.params.get('ESG_lag', 0),
                 'ESG_Pval': results_with_fe.pvalues.get('ESG_lag', 1.0), 'R2': results_with_fe.rsquared}
            ])
            industry_df.to_excel(writer, sheet_name='Industry_FE', index=False)

        # Threshold results
        if threshold_results:
            thresh_data = []
            for k, v in threshold_results.items():
                if v:
                    row = {'Threshold': k}
                    row.update(v)
                    thresh_data.append(row)
            if thresh_data:
                thresh_df = pd.DataFrame(thresh_data)
                thresh_df.to_excel(writer, sheet_name='Thresholds', index=False)

        # Moderation results
        if moderation_results:
            mod_data = []
            for k, v in moderation_results.items():
                if v:
                    row = {'Moderator': k}
                    row.update(v)
                    mod_data.append(row)
            if mod_data:
                mod_df = pd.DataFrame(mod_data)
                mod_df.to_excel(writer, sheet_name='Moderation', index=False)

        # Summary comparison
        if comparison_data:
            comparison_df.to_excel(writer, sheet_name='Summary_Comparison', index=False)

    print(f"\n✓ Robustness check results exported to: {robustness_path}")
except Exception as e:
    print(f"\n⚠ Could not export to Excel: {str(e)}")

print("\n" + "="*80)
print("ROBUSTNESS CHECK COMPLETE")
print("="*80)
```
