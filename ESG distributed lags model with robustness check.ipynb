```python
# standalone_lag_model_corrected.py
"""
Standalone Lag Model Analysis - Corrected to match original structured analysis
"""

# ============================================================================
# SETUP AND IMPORTS
# ============================================================================
from google.colab import drive
drive.mount('/content/drive')

!pip install -q linearmodels pandas numpy matplotlib seaborn statsmodels

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from linearmodels.panel import PanelOLS
import warnings
warnings.filterwarnings('ignore')

# ============================================================================
# CONFIGURATION
# ============================================================================
class Config:
    """Configuration parameters matching original structured analysis"""
    PROJECT_DIR = '/content/drive/MyDrive/Colab Notebooks/FTSE Data'
    DATA_PATH = os.path.join(PROJECT_DIR, 'FTSE data.xlsx')
    OUTPUT_DIR = os.path.join(PROJECT_DIR, 'lag_model_outputs_corrected')

    # Match original structured analysis settings
    ESG_COMPONENTS = ['E', 'S', 'G']
    CONTROL_VARS = ['ROA', 'Size', 'Leverage']

    # Consumer Staples firms
    CONSUMER_STAPLES_FIRMS = [
        'Associated British Foods', 'British American Tobacco', 'Coca Cola HBC AG',
        'Diageo', 'J. Sainsbury', 'Reckitt Benckiser', 'Tesco', 'Unilever', 'Haleon Plc'
    ]

# ============================================================================
# DATA LOADER - MATCHING ORIGINAL
# ============================================================================
class DataLoader:
    """Load and prepare data exactly as in structured analysis"""

    def __init__(self, config):
        self.config = config
        self.df = None
        self.df_cs = None  # Consumer Staples only
        self.df_all = None  # All sectors

    def load_and_prepare_data(self):
        """Load data exactly as in original structured analysis"""
        print("Loading and preparing data...")

        # 1. Load data
        self.df = pd.read_excel(self.config.DATA_PATH)
        self.df = self.df.set_index(['Firm', 'Year']).sort_index()
        self.df = self.df.rename(columns={"Tobin's Q": "Tobin_Q"})

        # 2. Create variables exactly as in original
        self.df['Size'] = np.log(self.df['Total Assets'])
        self.df['Leverage'] = self.df['Total Liabilities'] / self.df['Total Assets']
        self.df['Tobin_Q_log'] = np.log(self.df['Tobin_Q'] + 0.001)

        # 3. Create lagged ESG variables (t-1)
        self.df['E_lag'] = self.df.groupby(level=0)['E'].shift(1)
        self.df['S_lag'] = self.df.groupby(level=0)['S'].shift(1)
        self.df['G_lag'] = self.df.groupby(level=0)['G'].shift(1)

        # 4. Create lagged control variables
        self.df['ROA_lag'] = self.df.groupby(level=0)['ROA'].shift(1)
        self.df['Size_lag'] = self.df.groupby(level=0)['Size'].shift(1)
        self.df['Leverage_lag'] = self.df.groupby(level=0)['Leverage'].shift(1)

        # 5. Create sector variable
        sectors = []
        for firm in self.df.index.get_level_values(0):
            if firm in self.config.CONSUMER_STAPLES_FIRMS:
                sectors.append('Consumer Staples')
            else:
                sectors.append('Other')

        self.df['Sector'] = sectors
        self.df['ConsumerStaples'] = (self.df['Sector'] == 'Consumer Staples').astype(int)

        # 6. Create datasets
        self.df_cs = self.df[self.df['ConsumerStaples'] == 1].copy()
        self.df_all = self.df.copy()

        print(f"✓ Total observations: {len(self.df)}")
        print(f"✓ Consumer Staples: {len(self.df_cs)} observations")
        print(f"✓ All sectors: {len(self.df_all)} observations")

        return self.df, self.df_cs, self.df_all

    def create_lagged_variables_for_model(self, df, min_lag=1, max_lag=5, use_original_controls=True):
        """
        Create lagged variables exactly as in original distributed lag model

        Parameters:
        -----------
        df : DataFrame
            Input dataframe
        min_lag : int
            Minimum lag (1)
        max_lag : int
            Maximum lag (1-5)
        use_original_controls : bool
            If True, use original controls (not lagged) as in structured analysis
        """
        df_copy = df.copy()

        # Create distributed lags for ESG - INDIVIDUAL LAGS
        print(f"Creating distributed lags {min_lag}-{max_lag} for ESG...")
        for component in ['E', 'S', 'G']:
            for lag in range(min_lag, max_lag + 1):
                df_copy[f'{component}_lag{lag}'] = df_copy.groupby(level=0)[component].shift(lag)

        # Also create summed versions as in original
        for component in ['E', 'S', 'G']:
            # Initialize with first lag
            if max_lag >= 1:
                lag1 = df_copy.groupby(level=0)[component].shift(1)
                df_copy[f'{component}_lag1'] = lag1

            # Create summed versions
            if max_lag >= 2:
                lag2 = df_copy.groupby(level=0)[component].shift(2)
                df_copy[f'{component}_lag12'] = lag1 + lag2

            if max_lag >= 3:
                lag3 = df_copy.groupby(level=0)[component].shift(3)
                df_copy[f'{component}_lag13'] = lag1 + lag2 + lag3

            if max_lag >= 4:
                lag4 = df_copy.groupby(level=0)[component].shift(4)
                df_copy[f'{component}_lag14'] = lag1 + lag2 + lag3 + lag4

            if max_lag >= 5:
                lag5 = df_copy.groupby(level=0)[component].shift(5)
                df_copy[f'{component}_lag15'] = lag1 + lag2 + lag3 + lag4 + lag5

        # Use control variables as in original structured analysis
        if use_original_controls:
            control_vars = ['ROA', 'Size', 'Leverage']
        else:
            control_vars = ['ROA_lag', 'Size_lag', 'Leverage_lag']

        return df_copy, control_vars

# ============================================================================
# LAG MODEL ANALYZER - MATCHING ORIGINAL
# ============================================================================
class LagModelAnalyzer:
    """Analyzer that matches original structured analysis"""

    def __init__(self, config):
        self.config = config
        self.loader = DataLoader(config)
        self.results = {}

    def run_analysis(self, max_lag=5):
        """Run analysis matching original structured analysis"""
        print("\n" + "="*80)
        print(f"RUNNING LAG MODEL ANALYSIS (1-{max_lag} years)")
        print("="*80)

        # Load data
        df, df_cs, df_all = self.loader.load_and_prepare_data()

        # Run models for different sectors
        sectors = [
            ('cs', df_cs, 'Consumer Staples Only'),
            ('all', df_all, 'All Sectors')
        ]

        for sector_code, sector_df, sector_name in sectors:
            print(f"\n{'='*60}")
            print(f"ANALYZING: {sector_name}")
            print(f"{'='*60}")

            # Create lagged variables
            df_lagged, control_vars = self.loader.create_lagged_variables_for_model(
                sector_df, min_lag=1, max_lag=max_lag, use_original_controls=True
            )

            # Run individual distributed lag model (Model 4 in original)
            print(f"\n1. DISTRIBUTED LAG MODEL (Individual Lags 1-{max_lag})")
            self._run_distributed_lag_individual(df_lagged, max_lag, sector_code,
                                                sector_name, control_vars)

            # Run summed distributed lag model (Model 5 in original)
            print(f"\n2. DISTRIBUTED LAG MODEL (Summed Lags 1-{max_lag})")
            self._run_distributed_lag_summed(df_lagged, max_lag, sector_code,
                                           sector_name, control_vars)

    def _run_distributed_lag_individual(self, df, max_lag, sector_code,
                                       sector_name, control_vars):
        """Run distributed lag model with individual lags (Model 4 in original)"""
        # Prepare lag variables
        lag_vars = []
        for component in ['E', 'S', 'G']:
            for lag in range(1, max_lag + 1):
                lag_vars.append(f'{component}_lag{lag}')

        # Add ConsumerStaples dummy for all sectors
        if sector_code == 'all':
            all_vars = lag_vars + control_vars + ['ConsumerStaples', 'Tobin_Q']
        else:
            all_vars = lag_vars + control_vars + ['Tobin_Q']

        # Clean data
        df_clean = df.dropna(subset=all_vars)

        if len(df_clean) < 10:
            print(f"⚠ Insufficient data: {len(df_clean)} observations")
            return None

        print(f"Observations: {len(df_clean)}")
        print(f"Firms: {df_clean.index.get_level_values(0).nunique()}")

        # Prepare regression
        y = df_clean['Tobin_Q']
        X_vars = lag_vars + control_vars
        if sector_code == 'all':
            X_vars = X_vars + ['ConsumerStaples']

        X = df_clean[X_vars]
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

        # Run regression
        try:
            model = PanelOLS(y, X, entity_effects=False, time_effects=False)
            results = model.fit(cov_type='robust')

            # Store results
            key = f"distributed_individual_{max_lag}_{sector_code}"
            self.results[key] = {
                'results': results,
                'type': 'distributed_individual',
                'sector': sector_code,
                'max_lag': max_lag,
                'nobs': results.nobs,
                'r2': results.rsquared,
                'data': df_clean
            }

            # Print results
            self._print_results(results, f"Distributed Lag (Individual) - {sector_name}")

            # Create lag structure plot
            self._create_lag_structure_plot(results, max_lag,
                                          f"Distributed Lag Individual - {sector_name}")

            return results

        except Exception as e:
            print(f"Error running model: {e}")
            return None

    def _run_distributed_lag_summed(self, df, max_lag, sector_code,
                                   sector_name, control_vars):
        """Run distributed lag model with summed lags (Model 5 in original)"""
        # Use summed variables (e.g., E_lag15 for sum of lags 1-5)
        summed_vars = [f'{component}_lag1{max_lag}' for component in ['E', 'S', 'G']]

        # Add ConsumerStaples dummy for all sectors
        if sector_code == 'all':
            all_vars = summed_vars + control_vars + ['ConsumerStaples', 'Tobin_Q']
        else:
            all_vars = summed_vars + control_vars + ['Tobin_Q']

        # Clean data
        df_clean = df.dropna(subset=all_vars)

        if len(df_clean) < 10:
            print(f"⚠ Insufficient data: {len(df_clean)} observations")
            return None

        print(f"Observations: {len(df_clean)}")

        # Prepare regression
        y = df_clean['Tobin_Q']
        X_vars = summed_vars + control_vars
        if sector_code == 'all':
            X_vars = X_vars + ['ConsumerStaples']

        X = df_clean[X_vars]
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

        # Run regression
        try:
            model = PanelOLS(y, X, entity_effects=False, time_effects=False)
            results = model.fit(cov_type='robust')

            # Store results
            key = f"distributed_summed_{max_lag}_{sector_code}"
            self.results[key] = {
                'results': results,
                'type': 'distributed_summed',
                'sector': sector_code,
                'max_lag': max_lag,
                'nobs': results.nobs,
                'r2': results.rsquared,
                'data': df_clean
            }

            # Print results
            self._print_results(results, f"Distributed Lag (Summed) - {sector_name}")

            return results

        except Exception as e:
            print(f"Error running model: {e}")
            return None

    def _print_results(self, results, title):
        """Print formatted results matching original style"""
        print(f"\n{title}")
        print("-" * 60)
        print(f"R-squared: {results.rsquared:.4f}")
        print(f"Observations: {results.nobs}")

        if hasattr(results, 'f_statistic'):
            print(f"F-statistic: {results.f_statistic.stat:.4f} (p={results.f_statistic.pval:.4f})")

        # Print ESG coefficients with significance
        print("\nESG Coefficients:")
        print("-" * 40)

        # Group by component
        for component in ['E', 'S', 'G']:
            component_vars = [v for v in results.params.index if component in str(v)]
            if component_vars:
                print(f"\n{component} component:")
                for var in component_vars:
                    if var in results.params and var != 'const':
                        coeff = results.params[var]
                        pval = results.pvalues.get(var, 1.0)
                        sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                        print(f"  {var:<15} {coeff:>10.4f}{sig:<3} (p={pval:.4f})")

    def _create_lag_structure_plot(self, results, max_lag, title):
        """Create lag structure plot"""
        fig, axes = plt.subplots(1, 3, figsize=(15, 4))
        fig.suptitle(title, fontsize=14, fontweight='bold')

        for idx, component in enumerate(['E', 'S', 'G']):
            lags = list(range(1, max_lag + 1))
            coeffs = []
            errors = []

            for lag in lags:
                var = f'{component}_lag{lag}'
                if var in results.params:
                    coeffs.append(results.params[var])
                    if hasattr(results, 'std_errors') and var in results.std_errors:
                        errors.append(results.std_errors[var])
                    else:
                        errors.append(0)
                else:
                    coeffs.append(0)
                    errors.append(0)

            axes[idx].errorbar(lags, coeffs, yerr=errors, fmt='o-',
                             capsize=5, linewidth=2, markersize=8)
            axes[idx].axhline(y=0, color='r', linestyle='--', alpha=0.5)
            axes[idx].set_xlabel('Years Lag')
            axes[idx].set_ylabel('Coefficient')
            axes[idx].set_title(f'{component} Component')
            axes[idx].set_xticks(lags)
            axes[idx].grid(True, alpha=0.3)

        plt.tight_layout()

        # Save plot
        os.makedirs(self.config.OUTPUT_DIR, exist_ok=True)
        plot_filename = f"lag_structure_{title.replace(' ', '_').replace('-', '_')}.png"
        plt.savefig(os.path.join(self.config.OUTPUT_DIR, plot_filename),
                   dpi=300, bbox_inches='tight')
        plt.show()

        return fig

    def compare_with_original_simple(self, max_lag=5):
        """
        Run simple lag model matching the ORIGINAL simple approach
        (not the structured analysis, but the simple lag model)
        """
        print("\n" + "="*80)
        print("SIMPLE LAG MODEL (Matching Original Simple Approach)")
        print("="*80)

        # Load data
        df, df_cs, df_all = self.loader.load_and_prepare_data()

        # For Consumer Staples
        print("\nCONSUMER STAPLES ONLY:")
        df_cs_lagged, control_vars = self.loader.create_lagged_variables_for_model(
            df_cs, min_lag=1, max_lag=max_lag, use_original_controls=True
        )

        # Use simple approach: just E_lag, S_lag, G_lag
        simple_vars = ['E_lag', 'S_lag', 'G_lag'] + control_vars + ['Tobin_Q']
        df_simple = df_cs_lagged.dropna(subset=simple_vars)

        if len(df_simple) > 0:
            y = df_simple['Tobin_Q']
            X = df_simple[['E_lag', 'S_lag', 'G_lag'] + control_vars]
            X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

            model = PanelOLS(y, X, entity_effects=False, time_effects=False)
            results = model.fit(cov_type='robust')

            print(f"\nSimple Lag Model (t-1 only):")
            print(f"Observations: {results.nobs}")
            print(f"R-squared: {results.rsquared:.4f}")

            for var in ['E_lag', 'S_lag', 'G_lag']:
                if var in results.params:
                    coeff = results.params[var]
                    pval = results.pvalues.get(var, 1.0)
                    sig = "***" if pval < 0.01 else "**" if pval < 0.05 else "*" if pval < 0.10 else ""
                    print(f"  {var}: {coeff:.4f}{sig} (p={pval:.4f})")

        return results

# ============================================================================
# RESULTS COMPARISON
# ============================================================================
class ResultsComparator:
    """Compare results from different model specifications"""

    def __init__(self, analyzer):
        self.analyzer = analyzer
        self.comparison_df = None

    def create_comparison_table(self):
        """Create comparison table of all results"""
        comparison_data = []

        for key, model_info in self.analyzer.results.items():
            results = model_info['results']

            row = {
                'Model_Type': model_info['type'],
                'Sector': model_info['sector'],
                'Max_Lag': model_info['max_lag'],
                'N': results.nobs,
                'R2': results.rsquared,
            }

            # Extract key coefficients
            for component in ['E', 'S', 'G']:
                # For individual models, get average across lags
                if 'individual' in model_info['type']:
                    coeffs = []
                    for var in results.params.index:
                        if component in str(var) and 'lag' in str(var):
                            coeffs.append(results.params[var])
                    if coeffs:
                        row[f'{component}_avg'] = np.mean(coeffs)
                        row[f'{component}_min'] = min(coeffs)
                        row[f'{component}_max'] = max(coeffs)

                # For summed models
                elif 'summed' in model_info['type']:
                    var = f'{component}_lag1{model_info["max_lag"]}'
                    if var in results.params:
                        row[f'{component}_sum'] = results.params[var]

            comparison_data.append(row)

        self.comparison_df = pd.DataFrame(comparison_data)
        return self.comparison_df

    def print_comparison(self):
        """Print comparison of results"""
        if self.comparison_df is None:
            self.create_comparison_table()

        print("\n" + "="*80)
        print("MODEL COMPARISON")
        print("="*80)

        # Sort by R2
        self.comparison_df = self.comparison_df.sort_values('R2', ascending=False)

        for _, row in self.comparison_df.iterrows():
            print(f"\n{row['Model_Type'].replace('_', ' ').title()}")
            print(f"Sector: {row['Sector'].upper()}, Max Lag: {row['Max_Lag']}")
            print(f"R²: {row['R2']:.4f}, N: {row['N']}")

            if 'individual' in row['Model_Type']:
                print("Average ESG effects:")
                for component in ['E', 'S', 'G']:
                    if f'{component}_avg' in row and not pd.isna(row[f'{component}_avg']):
                        print(f"  {component}: {row[f'{component}_avg']:.4f} " +
                              f"[{row[f'{component}_min']:.4f}, {row[f'{component}_max']:.4f}]")

# ============================================================================
# MAIN EXECUTION
# ============================================================================
def main():
    """Main execution function"""
    print("\n" + "="*80)
    print("LAG MODEL ANALYSIS - CORRECTED TO MATCH ORIGINAL")
    print("="*80)

    # Initialize
    config = Config()
    analyzer = LagModelAnalyzer(config)

    # Run analysis with different lag specifications
    lag_ranges = [2, 3, 4, 5]  # Test different ranges

    for max_lag in lag_ranges:
        print(f"\n{'='*80}")
        print(f"ANALYZING LAG RANGE: 1-{max_lag} years")
        print(f"{'='*80}")

        analyzer.run_analysis(max_lag=max_lag)

    # Compare results
    comparator = ResultsComparator(analyzer)
    comparator.create_comparison_table()
    comparator.print_comparison()

    # Also run simple model for comparison
    analyzer.compare_with_original_simple(max_lag=5)

    print("\n" + "="*80)
    print("ANALYSIS COMPLETE")
    print("="*80)

    return analyzer, comparator

# ============================================================================
# EXECUTION
# ============================================================================
if __name__ == "__main__":
    analyzer, comparator = main()

# ============================================================================
# ROBUSTNESS CHECK FOR DISTRIBUTED LAG MODEL
# ============================================================================
print("\n" + "="*80)
print("ROBUSTNESS CHECK: DISTRIBUTED LAG MODEL")
print("="*80)

# First, capture the original results
print("\n" + "-"*40)
print("ORIGINAL MODEL RESULTS")
print("-"*40)

original_results = {}

# Get results for each lag range and sector
lag_ranges = [2, 3, 4, 5]
for max_lag in lag_ranges:
    for sector_code, sector_name in [('cs', 'Consumer Staples'), ('all', 'All Sectors')]:
        # Individual lags model
        ind_key = f"distributed_individual_{max_lag}_{sector_code}"
        if ind_key in analyzer.results:
            results = analyzer.results[ind_key]['results']

            model_key = f"{sector_name} - Individual Lags 1-{max_lag}"
            original_results[model_key] = {
                'Type': 'Individual Lags',
                'Window': max_lag,
                'Sector': sector_name,
                'R2': results.rsquared,
                'N': analyzer.results[ind_key]['nobs'],
                'Firms': analyzer.results[ind_key]['data'].index.get_level_values(0).nunique()
            }

            # Calculate average coefficients for each pillar
            for pillar in ['E', 'S', 'G']:
                coeffs = []
                pvals = []
                for lag in range(1, max_lag+1):
                    var = f'{pillar}_lag{lag}'
                    if var in results.params:
                        coeffs.append(results.params[var])
                        pvals.append(results.pvalues.get(var, 1.0))

                if coeffs:
                    original_results[model_key][f'{pillar}_AvgCoeff'] = np.mean(coeffs)
                    original_results[model_key][f'{pillar}_SigCount'] = sum(1 for p in pvals if p < 0.10)
                    original_results[model_key][f'{pillar}_AnySig'] = any(p < 0.10 for p in pvals)

            print(f"\n{model_key}:")
            print(f"  R² = {results.rsquared:.4f}")
            print(f"  N = {analyzer.results[ind_key]['nobs']}")
            for pillar in ['E', 'S', 'G']:
                avg = original_results[model_key].get(f'{pillar}_AvgCoeff', 0)
                sig_count = original_results[model_key].get(f'{pillar}_SigCount', 0)
                print(f"  {pillar}: avg = {avg:.4f} ({sig_count}/{max_lag} lags significant)")

# ============================================================================
# CREATE TRANSFORMATION COLUMNS
# ============================================================================
print("\n" + "="*80)
print("CREATING TRANSFORMATION COLUMNS")
print("="*80)

# Get the dataframe
df, df_cs, df_all = analyzer.loader.load_and_prepare_data()

from scipy import stats
import warnings
warnings.filterwarnings('ignore')

# Create transformation columns
print("\nCreating transformation columns...")

# Log transformation
if 'Tobin_Q_log' not in df.columns:
    df['Tobin_Q_log'] = np.log(df['Tobin_Q'].clip(lower=0.001))
    print("✓ Created Tobin_Q_log")

# Square root transformation
if 'Tobin_Q_sqrt' not in df.columns:
    df['Tobin_Q_sqrt'] = np.sqrt(df['Tobin_Q'].clip(lower=0))
    print("✓ Created Tobin_Q_sqrt")

# Inverse transformation
if 'Tobin_Q_inv' not in df.columns:
    with np.errstate(divide='ignore', invalid='ignore'):
        df['Tobin_Q_inv'] = 1 / df['Tobin_Q']
        df['Tobin_Q_inv'] = df['Tobin_Q_inv'].replace([np.inf, -np.inf], np.nan)
    print("✓ Created Tobin_Q_inv")

# Box-Cox transformation
try:
    if (df['Tobin_Q'].dropna() > 0).all():
        boxcox_data, lambda_param = stats.boxcox(df['Tobin_Q'].dropna())
        df['Tobin_Q_boxcox'] = np.nan
        df.loc[df['Tobin_Q'].dropna().index, 'Tobin_Q_boxcox'] = boxcox_data
        print(f"✓ Created Tobin_Q_boxcox (lambda = {lambda_param:.4f})")
    else:
        df['Tobin_Q_boxcox'] = np.nan
        print("⚠ Tobin_Q_boxcox not created (non-positive values)")
except Exception as e:
    df['Tobin_Q_boxcox'] = np.nan
    print(f"⚠ Tobin_Q_boxcox creation failed: {str(e)[:50]}")

# Update dataframes
df_cs = df[df['ConsumerStaples'] == 1].copy()
df_all = df.copy()

# ============================================================================
# ROBUSTNESS CHECK 1: ALTERNATIVE TRANSFORMATIONS
# ============================================================================
print("\n" + "="*80)
print("ROBUSTNESS CHECK 1: ALTERNATIVE DEPENDENT VARIABLE TRANSFORMATIONS")
print("="*80)

# Focus on Consumer Staples, 5-year distributed lag as example
print("\n" + "-"*60)
print("TESTING TRANSFORMATIONS - 5-YEAR DISTRIBUTED LAG (Consumer Staples)")
print("-"*60)

# Prepare data with 5-year lags
max_lag_test = 5
df_cs_test = df_cs.copy()

# Create all lagged variables
for pillar in ['E', 'S', 'G']:
    for lag in range(1, max_lag_test+1):
        df_cs_test[f'{pillar}_lag{lag}'] = df_cs_test.groupby(level=0)[pillar].shift(lag)

# Control variables
if 'Size' not in df_cs_test.columns:
    df_cs_test['Size'] = np.log(df_cs_test['Total Assets'])
if 'Leverage' not in df_cs_test.columns:
    df_cs_test['Leverage'] = df_cs_test['Total Liabilities'] / df_cs_test['Total Assets']

# Define transformations
transformations = [
    ('Original', 'Tobin_Q'),
    ('Log', 'Tobin_Q_log'),
    ('Square Root', 'Tobin_Q_sqrt'),
    ('Inverse', 'Tobin_Q_inv'),
]

if 'Tobin_Q_boxcox' in df_cs_test.columns and not df_cs_test['Tobin_Q_boxcox'].isna().all():
    transformations.append(('Box-Cox', 'Tobin_Q_boxcox'))

transformation_results = {}

for trans_name, trans_var in transformations:
    print(f"\nTesting {trans_name} transformation...")

    # Create list of all lag variables
    lag_vars = []
    for pillar in ['E', 'S', 'G']:
        for lag in range(1, max_lag_test+1):
            lag_vars.append(f'{pillar}_lag{lag}')

    # Drop NA
    required_vars = lag_vars + ['ROA', 'Size', 'Leverage', trans_var]
    df_clean = df_cs_test.dropna(subset=required_vars)

    print(f"  Observations: {len(df_clean)}")

    if len(df_clean) >= 10:
        # Run regression
        y = df_clean[trans_var]
        X = df_clean[lag_vars + ['ROA', 'Size', 'Leverage']]
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

        try:
            model = PanelOLS(y, X, entity_effects=False, time_effects=False)
            results = model.fit(cov_type='robust')

            # Calculate average effects by pillar
            trans_results = {
                'R2': results.rsquared,
                'N': results.nobs
            }

            for pillar in ['E', 'S', 'G']:
                coeffs = []
                pvals = []
                for lag in range(1, max_lag_test+1):
                    var = f'{pillar}_lag{lag}'
                    if var in results.params:
                        coeffs.append(results.params[var])
                        pvals.append(results.pvalues.get(var, 1.0))

                if coeffs:
                    trans_results[f'{pillar}_Avg'] = np.mean(coeffs)
                    trans_results[f'{pillar}_SigCount'] = sum(1 for p in pvals if p < 0.10)
                    trans_results[f'{pillar}_AnySig'] = any(p < 0.10 for p in pvals)

            transformation_results[trans_name] = trans_results

            print(f"  ✓ Completed: R² = {results.rsquared:.4f}")
            for pillar in ['E', 'S', 'G']:
                avg = trans_results.get(f'{pillar}_Avg', 0)
                sig_count = trans_results.get(f'{pillar}_SigCount', 0)
                print(f"    {pillar} avg = {avg:.4f} ({sig_count}/{max_lag_test} sig)")

        except Exception as e:
            print(f"  ✗ Failed: {str(e)[:100]}")
            transformation_results[trans_name] = None
    else:
        print(f"  ✗ Insufficient data: {len(df_clean)} observations")
        transformation_results[trans_name] = None

# ============================================================================
# ROBUSTNESS CHECK 2: INDUSTRY FIXED EFFECTS - FIXED VERSION
# ============================================================================
print("\n" + "="*80)
print("ROBUSTNESS CHECK 2: INDUSTRY FIXED EFFECTS")
print("="*80)

if 'Sector' in df.columns:
    # Check unique sectors
    unique_sectors = df['Sector'].unique()
    print(f"\nIndustry sectors found: {unique_sectors}")

    if len(unique_sectors) > 1:
        # Create industry dummies - drop_first=True to avoid multicollinearity
        industry_dummies = pd.get_dummies(df['Sector'], prefix='ind', drop_first=True)
        df_with_industry = pd.concat([df, industry_dummies], axis=1)
        industry_cols = [col for col in df_with_industry.columns if col.startswith('ind_')]

        print(f"Industry dummies created: {len(industry_cols)}")
        print(f"Dummy columns: {industry_cols}")

        print("\n" + "-"*60)
        print("INDUSTRY FE - 3-YEAR DISTRIBUTED LAG (All Sectors)")
        print("-"*60)

        # Prepare data
        df_all_test = df_with_industry.copy()
        max_lag_test = 3

        # Create lagged variables
        for pillar in ['E', 'S', 'G']:
            for lag in range(1, max_lag_test+1):
                df_all_test[f'{pillar}_lag{lag}'] = df_all_test.groupby(level=0)[pillar].shift(lag)

        # Create control variables if needed
        if 'Size' not in df_all_test.columns:
            df_all_test['Size'] = np.log(df_all_test['Total Assets'])
        if 'Leverage' not in df_all_test.columns:
            df_all_test['Leverage'] = df_all_test['Total Liabilities'] / df_all_test['Total Assets']

        # Create list of lag variables
        lag_vars = []
        for pillar in ['E', 'S', 'G']:
            for lag in range(1, max_lag_test+1):
                lag_vars.append(f'{pillar}_lag{lag}')

        # ===== MODEL WITHOUT INDUSTRY FE =====
        base_vars = lag_vars + ['ROA', 'Size', 'Leverage', 'ConsumerStaples', 'Tobin_Q']
        df_clean_no_fe = df_all_test.dropna(subset=base_vars)

        if len(df_clean_no_fe) >= 10:
            y_no_fe = df_clean_no_fe['Tobin_Q']
            X_no_fe = df_clean_no_fe[lag_vars + ['ROA', 'Size', 'Leverage', 'ConsumerStaples']]
            X_no_fe = pd.concat([pd.Series(1, index=X_no_fe.index, name='const'), X_no_fe], axis=1)

            model_no_fe = PanelOLS(y_no_fe, X_no_fe, entity_effects=False, time_effects=False)
            results_no_fe = model_no_fe.fit(cov_type='robust')

            print(f"\nWithout Industry FE:")
            print(f"  R² = {results_no_fe.rsquared:.4f}")
            print(f"  N = {results_no_fe.nobs}")
            for pillar in ['E', 'S', 'G']:
                coeffs = [results_no_fe.params.get(f'{pillar}_lag{lag}', 0) for lag in range(1, max_lag_test+1)]
                print(f"  {pillar} avg = {np.mean(coeffs):.4f}")

        # ===== MODEL WITH INDUSTRY FE =====
        # IMPORTANT FIX: Check if we have enough variation in industry dummies
        if industry_cols:
            # Check correlation between dummies to avoid multicollinearity
            dummy_data = df_all_test[industry_cols].dropna()
            if len(dummy_data) > 0 and dummy_data.shape[1] > 0:

                # Create variable list for WITH industry FE
                fe_vars = lag_vars + ['ROA', 'Size', 'Leverage', 'ConsumerStaples'] + industry_cols + ['Tobin_Q']
                df_clean_with_fe = df_all_test.dropna(subset=fe_vars)

                print(f"\nWith Industry FE - Observations after cleaning: {len(df_clean_with_fe)}")

                if len(df_clean_with_fe) >= 30:  # Need more observations with dummies
                    try:
                        y_with_fe = df_clean_with_fe['Tobin_Q']
                        X_with_fe = df_clean_with_fe[lag_vars + ['ROA', 'Size', 'Leverage', 'ConsumerStaples'] + industry_cols]
                        X_with_fe = pd.concat([pd.Series(1, index=X_with_fe.index, name='const'), X_with_fe], axis=1)

                        # Check if X has full rank - if not, we'll handle it
                        import numpy.linalg as la
                        X_matrix = X_with_fe.values
                        if np.linalg.matrix_rank(X_matrix) < X_matrix.shape[1]:
                            print("  ⚠ Warning: Perfect multicollinearity detected. Dropping one dummy.")
                            # Drop the last dummy column to fix multicollinearity
                            X_with_fe = X_with_fe.drop(columns=[industry_cols[-1]])

                        model_with_fe = PanelOLS(y_with_fe, X_with_fe, entity_effects=False, time_effects=False)
                        results_with_fe = model_with_fe.fit(cov_type='robust')

                        print(f"With Industry FE:")
                        print(f"  R² = {results_with_fe.rsquared:.4f}")
                        print(f"  N = {results_with_fe.nobs}")
                        for pillar in ['E', 'S', 'G']:
                            coeffs = [results_with_fe.params.get(f'{pillar}_lag{lag}', 0) for lag in range(1, max_lag_test+1)]
                            print(f"  {pillar} avg = {np.mean(coeffs):.4f}")

                    except Exception as e:
                        print(f"  ✗ Industry FE model failed: {str(e)[:100]}")
                        print("  This can happen if industry dummies perfectly predict some observations")
                else:
                    print(f"  ✗ Insufficient data for industry FE: {len(df_clean_with_fe)} observations (need ≥30)")
            else:
                print("  ⚠ No variation in industry dummies after dropping NAs")
    else:
        print(f"  ⚠ Only one sector found: {unique_sectors[0]}. Cannot estimate industry FE.")
else:
    print("Sector variable not available for industry fixed effects")

# ============================================================================
# ROBUSTNESS CHECK 3: NON-LINEAR THRESHOLDS (FOCUS ON S PILLAR)
# ============================================================================
print("\n" + "="*80)
print("ROBUSTNESS CHECK 3: NON-LINEAR THRESHOLDS")
print("="*80)

print("\n" + "-"*60)
print("THRESHOLD MODELS - SOCIAL PILLAR (Consumer Staples, Lag 1)")
print("-"*60)

df_cs_thresh = df_cs.copy()
df_cs_thresh['S_lag1'] = df_cs_thresh.groupby(level=0)['S'].shift(1)

if 'Size' not in df_cs_thresh.columns:
    df_cs_thresh['Size'] = np.log(df_cs_thresh['Total Assets'])
if 'Leverage' not in df_cs_thresh.columns:
    df_cs_thresh['Leverage'] = df_cs_thresh['Total Liabilities'] / df_cs_thresh['Total Assets']

thresholds = {
    'Top 25%': 0.75,
    'Median': 0.50,
    'Top 10%': 0.90
}

threshold_results = {}

for thresh_name, percentile in thresholds.items():
    print(f"\nTesting {thresh_name}...")

    thresh_value = df_cs_thresh['S_lag1'].quantile(percentile)
    df_cs_thresh['S_high'] = (df_cs_thresh['S_lag1'] > thresh_value).astype(int)

    model_vars = ['S_high', 'ROA', 'Size', 'Leverage', 'Tobin_Q']
    df_clean = df_cs_thresh.dropna(subset=model_vars)

    if len(df_clean) >= 10:
        y = df_clean['Tobin_Q']
        X = df_clean[['S_high', 'ROA', 'Size', 'Leverage']]
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

        try:
            model = PanelOLS(y, X, entity_effects=False, time_effects=False)
            results = model.fit(cov_type='robust')

            threshold_results[thresh_name] = {
                'Coeff': results.params.get('S_high', 0),
                'Pval': results.pvalues.get('S_high', 1.0),
                'R2': results.rsquared,
                'N': results.nobs
            }

            sig = "***" if results.pvalues.get('S_high', 1.0) < 0.01 else \
                  "**" if results.pvalues.get('S_high', 1.0) < 0.05 else \
                  "*" if results.pvalues.get('S_high', 1.0) < 0.10 else ""

            print(f"  ✓ S_high coeff = {results.params.get('S_high', 0):.4f}{sig} (p={results.pvalues.get('S_high', 1.0):.4f})")

        except Exception as e:
            print(f"  ✗ Failed: {str(e)[:100]}")
            threshold_results[thresh_name] = None

# ============================================================================
# ROBUSTNESS CHECK 4: MODERATING EFFECTS (FOCUS ON S PILLAR)
# ============================================================================
print("\n" + "="*80)
print("ROBUSTNESS CHECK 4: MODERATING EFFECTS")
print("="*80)

print("\n" + "-"*60)
print("MODERATING EFFECTS - SOCIAL PILLAR (Consumer Staples, Lag 1)")
print("-"*60)

df_cs_mod = df_cs.copy()
df_cs_mod['S_lag1'] = df_cs_mod.groupby(level=0)['S'].shift(1)

if 'Size' not in df_cs_mod.columns:
    df_cs_mod['Size'] = np.log(df_cs_mod['Total Assets'])
if 'Leverage' not in df_cs_mod.columns:
    df_cs_mod['Leverage'] = df_cs_mod['Total Liabilities'] / df_cs_mod['Total Assets']

moderators = {
    'Size': 'Size',
    'ROA': 'ROA',
    'Leverage': 'Leverage'
}

moderation_results = {}

for mod_name, mod_var in moderators.items():
    print(f"\nTesting moderator: {mod_name}")

    df_cs_mod[f'S_x_{mod_var}'] = df_cs_mod['S_lag1'] * df_cs_mod[mod_var]

    model_vars = ['S_lag1', mod_var, f'S_x_{mod_var}', 'ROA', 'Size', 'Leverage', 'Tobin_Q']
    model_vars = list(dict.fromkeys(model_vars))

    df_clean = df_cs_mod.dropna(subset=model_vars)

    if len(df_clean) >= 10:
        y = df_clean['Tobin_Q']
        X = df_clean[[v for v in model_vars if v != 'Tobin_Q']]
        X = pd.concat([pd.Series(1, index=X.index, name='const'), X], axis=1)

        try:
            model = PanelOLS(y, X, entity_effects=False, time_effects=False)
            results = model.fit(cov_type='robust')

            inter_var = f'S_x_{mod_var}'
            if inter_var in results.params:
                moderation_results[mod_name] = {
                    'Inter_Coeff': results.params.get(inter_var, 0),
                    'Inter_Pval': results.pvalues.get(inter_var, 1.0),
                    'R2': results.rsquared,
                    'N': results.nobs
                }

                sig = "***" if results.pvalues.get(inter_var, 1.0) < 0.01 else \
                      "**" if results.pvalues.get(inter_var, 1.0) < 0.05 else \
                      "*" if results.pvalues.get(inter_var, 1.0) < 0.10 else ""

                print(f"  ✓ S × {mod_name} = {results.params.get(inter_var, 0):.4f}{sig} (p={results.pvalues.get(inter_var, 1.0):.4f})")

        except Exception as e:
            print(f"  ✗ Failed: {str(e)[:100]}")
            moderation_results[mod_name] = None

# ============================================================================
# SUMMARY COMPARISON
# ============================================================================
print("\n" + "="*100)
print("ROBUSTNESS CHECK SUMMARY - DISTRIBUTED LAG MODEL")
print("="*100)

comparison_data = []

# Original results (3-year individual lags, Consumer Staples)
original_3yr_key = "Consumer Staples - Individual Lags 1-3"
if original_3yr_key in original_results:
    orig = original_results[original_3yr_key]
    comparison_data.append({
        'Model': 'Original - 3yr Distributed',
        'Type': 'Original',
        'R2': orig['R2'],
        'E_Avg': orig.get('E_AvgCoeff', 0),
        'E_Sig': orig.get('E_AnySig', False),
        'S_Avg': orig.get('S_AvgCoeff', 0),
        'S_Sig': orig.get('S_AnySig', False),
        'G_Avg': orig.get('G_AvgCoeff', 0),
        'G_Sig': orig.get('G_AnySig', False),
        'N': orig['N']
    })

# Add transformation results
for trans_name, res in transformation_results.items():
    if res:
        comparison_data.append({
            'Model': f"Transformation: {trans_name}",
            'Type': 'Robustness',
            'R2': res['R2'],
            'E_Avg': res.get('E_Avg', 0),
            'E_Sig': res.get('E_AnySig', False),
            'S_Avg': res.get('S_Avg', 0),
            'S_Sig': res.get('S_AnySig', False),
            'G_Avg': res.get('G_Avg', 0),
            'G_Sig': res.get('G_AnySig', False),
            'N': res['N']
        })

# Add industry FE results
if 'results_with_fe' in locals():
    # Calculate averages for with-FE model
    e_avg = np.mean([results_with_fe.params.get(f'E_lag{lag}', 0) for lag in range(1,4)])
    s_avg = np.mean([results_with_fe.params.get(f'S_lag{lag}', 0) for lag in range(1,4)])
    g_avg = np.mean([results_with_fe.params.get(f'G_lag{lag}', 0) for lag in range(1,4)])

    comparison_data.append({
        'Model': 'With Industry FE',
        'Type': 'Robustness',
        'R2': results_with_fe.rsquared,
        'E_Avg': e_avg,
        'S_Avg': s_avg,
        'G_Avg': g_avg,
        'N': results_with_fe.nobs
    })

# Add best threshold result
valid_thresholds = {k: v for k, v in threshold_results.items() if v}
if valid_thresholds:
    best_thresh = max(valid_thresholds.items(), key=lambda x: x[1]['R2'])
    comparison_data.append({
        'Model': f"Threshold: {best_thresh[0]} (S)",
        'Type': 'Robustness',
        'R2': best_thresh[1]['R2'],
        'S_Avg': best_thresh[1]['Coeff'],
        'S_Sig': best_thresh[1]['Pval'] < 0.10,
        'N': best_thresh[1]['N']
    })

# Add best moderation result
valid_mods = {k: v for k, v in moderation_results.items() if v}
if valid_mods:
    best_mod = max(valid_mods.items(), key=lambda x: abs(x[1]['Inter_Coeff']))
    comparison_data.append({
        'Model': f"Moderation: S × {best_mod[0]}",
        'Type': 'Robustness',
        'R2': best_mod[1]['R2'],
        'S_Avg': best_mod[1]['Inter_Coeff'],
        'S_Sig': best_mod[1]['Inter_Pval'] < 0.10,
        'N': best_mod[1]['N']
    })

# Create DataFrame
if comparison_data:
    comparison_df = pd.DataFrame(comparison_data)
    print("\n" + comparison_df.to_string(index=False))

    # Robustness assessment
    print("\n" + "="*60)
    print("ROBUSTNESS ASSESSMENT")
    print("="*60)

    orig_row = comparison_df[comparison_df['Type'] == 'Original'].iloc[0]
    robustness_rows = comparison_df[comparison_df['Type'] == 'Robustness']

    for pillar in ['E', 'S', 'G']:
        if f'{pillar}_Avg' in orig_row:
            print(f"\n{pillar} PILLAR:")
            orig_avg = orig_row[f'{pillar}_Avg']
            orig_sig = orig_row.get(f'{pillar}_Sig', False)

            valid_rows = robustness_rows[robustness_rows[f'{pillar}_Avg'].notna()]
            if len(valid_rows) > 0:
                sign_consistent = sum(np.sign(valid_rows[f'{pillar}_Avg']) == np.sign(orig_avg)) / len(valid_rows) * 100
                print(f"  Sign consistency: {sign_consistent:.1f}%")

# Export to Excel
try:
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    robustness_path = os.path.join(Config.OUTPUT_DIR, f'distributed_lag_robustness_{timestamp}.xlsx')

    with pd.ExcelWriter(robustness_path, engine='openpyxl') as writer:
        if original_results:
            pd.DataFrame([{'Model': k, **v} for k, v in original_results.items()]).to_excel(
                writer, sheet_name='Original_Results', index=False)
        if transformation_results:
            pd.DataFrame([{'Transformation': k, **v} for k, v in transformation_results.items() if v]).to_excel(
                writer, sheet_name='Transformations', index=False)
        if comparison_data:
            comparison_df.to_excel(writer, sheet_name='Summary_Comparison', index=False)

    print(f"\n✓ Results exported to: {robustness_path}")
except Exception as e:
    print(f"\n⚠ Could not export: {str(e)}")

print("\n" + "="*80)
print("ROBUSTNESS CHECK COMPLETE")
print("="*80)
```
